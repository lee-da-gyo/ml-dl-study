### 실행 도우미

`input()`이 필요한 문제를 미리 정한 입력값으로 실행하기 위한 도구입니다. 직접 입력해 보려면 `with fake_inputs(...)` 없이 함수만 호출하면 됩니다.

````python
import builtins
from contextlib import contextmanager


@contextmanager
def fake_inputs(*values):
    """블록 안에서 input()이 values를 차례로 돌려주게 했다가, 블록이 끝나면 원래대로 되돌린다."""
    remaining = iter(values)
    real_input = builtins.input

    def _fake_input(prompt=""):
        value = next(remaining)
        print(f"{prompt}{value}")  # 사용자가 직접 입력한 것처럼 화면에 보여 준다
        return value

    builtins.input = _fake_input
    try:
        yield
    finally:
        builtins.input = real_input
````

### 문제1. 3의 배수 출력

- **문제 설명**  
1부터 n까지의 숫자 중에서 3의 배수만 출력하는 함수를 작성하세요.

- **함수 설명**  
`print_multiples_of_three(n: int) -> int`:  
  - n: 3의 배수를 구하고자 하는 범위의 최대값입니다.

- **입출력 예시**   

    - 입력1:

    ```
    7
    ```

    - 출력1:

    ```
    3
    6
    ```

    - 입력2:
    
    ```
    30
    ```

    - 출력2:

    ```
    3
    6
    9
    12
    15
    18
    21
    24
    27
    30
    ```

````python
def print_multiples_of_three(n: int) -> None:
    """1부터 n까지의 수 중 3의 배수를 한 줄에 하나씩 출력한다."""
    # 3부터 3씩 건너뛰면 나머지 검사 없이 3의 배수만 돈다
    for i in range(3, n + 1, 3):
        print(i)
````

### 문제2. 모음 제거하기

- **문제 설명**  
문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환하는 함수를 작성하세요.

- **함수 설명**  
`remove_vowels(s: string) -> string`:  
  - s: 영어 소문자 알파벳으로 이루어진 문자열입니다.

- **입출력 예시**   

    - 입력1:

    ```
    "codeit"
    ```

    - 출력1:

    ```
    "cdt"
    ```

    - 입력2:
    
    ```
    "python programming"
    ```

    - 출력2:

    ```
    "pythn prgrmmng"
    ```



````python
VOWELS = set("aeiouAEIOU")


def remove_vowels(s: str) -> str:
    """문자열에서 모음(a, e, i, o, u)을 제거한 문자열을 반환한다."""
    # set 검색은 O(1)이고, 문자열을 +=로 이어 붙이는 대신 join으로 한 번에 합친다
    return "".join(ch for ch in s if ch not in VOWELS)
````

````python
print(remove_vowels("codeit"))
print(remove_vowels("python programming"))
````

### 문제3. 수 크기 비교하기

- **문제 설명**  
두 개의 숫자를 입력받아, 큰 숫자와 작은 숫자를 각각 반환하는 함수를 작성하세요.
(단, 동일한 숫자를 2회 입력하는 경우에는 순서 없이 한 번만 출력되도록 해 주세요.)

- **함수 설명**

  `compare_numbers(num1, num2)`:  
  - num1, num2 : 정수형

- **입출력 예시**

  - 입력 1:
    ```
    10, 20
    ```

  - 출력 1:
    ```
    (20, 10)
    ```

  - 입력 2:
    ```
    30, 30
    ```

  - 출력 2:
    ```
    30
    ```

````python
def compare_numbers(num1: int, num2: int) -> tuple[int, int] | int:
    """(큰 수, 작은 수)를 반환한다. 두 수가 같으면 그 수 하나만 반환한다."""
    if num1 == num2:
        return num1
    return max(num1, num2), min(num1, num2)
````

````python
print(compare_numbers(10, 20))
print(compare_numbers(30, 30))
````

### 문제4. 문자열 길이 반환하기

문자열의 길이를 반환하되, 문자열이 빈 문자열이면 “문자열이 비어 있습니다.“를 반환하는 함수를 작성하세요.

- **함수 설명**
`check_string_length(string)`:  
  - string : 영어 알파벳 및 공백이 포함된 문자열입니다.

- **입출력 예시**

### 문제5. 총 합계 구하기

- **문제 설명**  
0부터 사용자가 입력한 숫자 n까지의 합을 구하는 함수를 작성하세요.

- **함수 설명**  
`sum_up_to_n(n)`:  
  - n: 0보다 큰 정수

- **입출력 예시**   

    - 입력1:

    ```
    3
    ```

    - 출력1:

    ```
    6
    ```

    - 입력2:
    
    ```
    10
    ```

    - 출력2:

    ```
    55
    ```



````python
def sum_up_to_n(n: int) -> int:
    """0부터 n까지의 합. 가우스 공식 n(n+1)/2로 반복 없이 O(1)에 계산한다."""
    return n * (n + 1) // 2
````

````python
print(sum_up_to_n(3))
print(sum_up_to_n(10))

# 공식이 반복문으로 더한 값과 같은지 확인
assert all(sum_up_to_n(n) == sum(range(n + 1)) for n in range(1, 1000))
````

### 문제6. 짝수 합 구하기

- **문제 설명**

  사용자가 입력한 숫자 n까지의 짝수 합을 구하는 함수를 작성하되, 0을 입력하면 프로그램을 종료하는 함수를 작성하세요. (단, 음수는 입력하지 않는다고 가정합니다.)

- **함수 설명**

  `sum_even_numbers(n: int)`

  - 사용자에게 숫자 입력값을 받아, 1부터 n까지의 짝수 합은 ___ 입니다." 가 출력됩니다.
  - 0 입력시 종료됨을 안내합니다.

- **입출력 예시**

    - 입력1:

    ```
    9
    ```

    - 출력1:

    ```
    "1부터 9까지의 짝수 합은 20입니다."
    ```

    - 입력2:
    
    ```
    0
    ```

    - 출력2:

    ```
    종료
    ```

````python
def sum_even_numbers(n: int) -> None:
    """1부터 n까지의 짝수 합을 출력한다. 0이면 '종료'를 출력한다."""
    if n == 0:
        print("종료")
        return
    even_sum = sum(range(2, n + 1, 2))
    print(f"1부터 {n}까지의 짝수 합은 {even_sum}입니다.")


def even_sum_program() -> None:
    """0을 입력할 때까지 숫자를 입력받아 짝수 합을 출력한다."""
    while True:
        n = int(input("숫자를 입력하세요(0 입력 시 종료): "))
        sum_even_numbers(n)
        if n == 0:
            break
````

````python
sum_even_numbers(9)
sum_even_numbers(0)

print("\n[0을 입력할 때까지 반복]")
with fake_inputs("9", "3", "0"):
    even_sum_program()
````

### 문제7. 두 숫자 나누기

- **문제 설명**
사용자가 입력한 두 숫자를 나누는 프로그램을 작성하세요. 예외가 발생하면 "잘못된 입력입니다."를 반환하는 함수를 작성하세요.

- **함수 설명**

  `divide_numbers(a, b)`
  - 사용자에게 두개의 숫자를 차례대로 입력 받습니다.
  - 첫번째 숫자를 두번째 숫자로 나눕니다.
  - 결과값을 반환합니다. 단, 예외가 발생하면 "잘못된 입력입니다."를 반환합니다.
  - `try-except` 구문을 활용하여 함수를 작성해보세요. 이 문법에 대해서 구글링을 해도 좋습니다.

- **입출력 예시**

- 입력1:
    ```
    42, 7
    ```

- 출력1:
    ```
    6
    ```

- 입력2:
    ```
    135, 0
    ```

- 출력2:
    ```
    "잘못된 입력입니다."
    ```

````python
def divide_numbers(a=None, b=None) -> int | float | str:
    """a를 b로 나눈 결과를 반환한다.

    인자를 주지 않으면 input()으로 두 숫자를 차례대로 입력받는다.
    숫자가 아니거나 0으로 나누면 "잘못된 입력입니다."를 반환한다.
    """
    try:
        if a is None or b is None:
            a = float(input("첫 번째 숫자를 입력하세요: "))
            b = float(input("두 번째 숫자를 입력하세요: "))
        result = a / b  # //는 몫만 남기므로(7 // 2 → 3) 일반 나눗셈을 쓴다
    except (ValueError, TypeError, ZeroDivisionError):
        return "잘못된 입력입니다."

    # 42 / 7 = 6.0처럼 나누어떨어지면 예시와 같이 6으로 보여 준다
    return int(result) if float(result).is_integer() else result
````

````python
print(divide_numbers(42, 7))
print(divide_numbers(135, 0))
print(divide_numbers(7, 2))
print(divide_numbers("abc", 3))

print("\n[input()으로 입력받기]")
with fake_inputs("42", "7"):
    print(divide_numbers())
with fake_inputs("10", "zero"):
    print(divide_numbers())
````

### 문제1. 문자열 압축 게임



- **문제 설명**

  주어진 문자열에서 연속해서 같은 문자가 반복되는 경우, 그 문자와 반복 횟수를 이용해 새로운 형태로 문자열을 압축합니다. 예를 들어, "aabcccccaaa" 문자열은 "a2b1c5a3"로 압축될 수 있습니다.

- **함수 설명**  
`compress_string(s: str) -> str`:  
  - s: 압축하고자 하는 문자열입니다.

- **입출력 예시**

    - 입력1:

    ```
    "aaabbaaa"
    ```

    - 출력1:

    ```
    "a3b2a3"
    ```

    - 입력2:
    
    ```
    "aabcccccaaa"
    ```

    - 출력2:

    ```
    "a2b1c5a3"
    ```

````python
from itertools import groupby


def compress_string(s: str) -> str:
    """연속해서 반복되는 문자를 '문자+반복 횟수'로 압축한다. 예: "aabcccccaaa" → "a2b1c5a3" """
    if not s:
        return ""

    compressed = []
    count = 1
    for prev, cur in zip(s, s[1:]):  # 이웃한 두 글자씩 비교
        if cur == prev:
            count += 1
        else:
            compressed.append(f"{prev}{count}")
            count = 1
    compressed.append(f"{s[-1]}{count}")  # 마지막 묶음은 반복문 안에서 기록되지 않으므로 따로 추가
    return "".join(compressed)


def compress_string_groupby(s: str) -> str:
    """같은 동작을 itertools.groupby로 짧게 쓴 버전 (결과 비교용)."""
    return "".join(f"{ch}{len(list(group))}" for ch, group in groupby(s))
````

````python
print(compress_string("aaabbaaa"))
print(compress_string("aabcccccaaa"))

# 두 구현이 경계 사례까지 같은 결과를 내는지 확인
for case in ["aaabbaaa", "aabcccccaaa", "", "a", "abc", "zzzz"]:
    assert compress_string(case) == compress_string_groupby(case), case
````

### 문제2. 덧셈 프로그램



- **문제 설명**
이 프로그램은 두 개의 숫자를 입력받아 그 합을 계산합니다. 입력받은 값이 숫자 형식이 아니면, 사용자에게 숫자 형식의 입력을 요청합니다. 숫자 입력이 완료되면 두 숫자의 합을 출력합니다.

- **함수 설명**

  `add_numbers()`
  - 이 함수는 `input()`을 사용하여 사용자로부터 두 개의 숫자를 입력받습니다.
  - 입력된 값이 숫자가 아닐 경우, "올바른 숫자를 입력하세요."라는 메시지를 출력하고 다시 입력을 요청합니다.
  - 두 숫자의 합을 계산하고 결과를 출력합니다.

- **입출력 예시**

    - 입력:
        ```
        첫 번째 숫자를 입력하세요: 10
        첫 번째 숫자를 입력하세요: twenty
        첫 번째 숫자를 입력하세요: 10
        두 번째 숫자를 입력하세요: 20
        ```

    - 출력:
        ```
        올바른 숫자를 입력하세요.
        10 + 20 = 30
        ```

````python
def read_number(prompt: str) -> int:
    """정수가 입력될 때까지 다시 묻고, 입력받은 정수를 반환한다."""
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("올바른 숫자를 입력하세요.")


def add_numbers() -> None:
    """두 숫자를 입력받아 합을 출력한다."""
    num1 = read_number("첫 번째 숫자를 입력하세요: ")
    num2 = read_number("두 번째 숫자를 입력하세요: ")
    print(f"{num1} + {num2} = {num1 + num2}")
````

````python
scenarios = {
    "문제 예시: 첫 번째 숫자에 문자 입력": ["twenty", "10", "20"],
    "두 번째 숫자에 문자 입력": ["10", "twenty", "20"],
    "실수·빈 값·기호·쉼표는 모두 다시 입력": ["3.14", "", "!!", "1,000", "10", "20"],
    "음수도 정수로 처리": ["-10", "20"],
}
for title, inputs in scenarios.items():
    print(f"[{title}]")
    with fake_inputs(*inputs):
        add_numbers()
    print()
````

### 문제3. 카운트다운 타이머

- **문제 설명**  
  사용자로부터 시간(초)을 입력받아 해당 시간만큼 카운트 다운을 진행한 후, "타이머 종료!"를 출력하고 프로그램이 종료됩니다. 매 초마다 남은 시간을 화면에 출력합니다.

- **함수 설명**     
  
  `count_down(seconds: int)`:  
  - seconds: 카운트 다운할 시간(초)입니다.
  - 입력받은 초만큼 카운트 다운을 진행하며, 매 초마다 남은 시간을 출력합니다.
  - 시간이 종료되면 "타이머 종료!"를 출력하고 프로그램을 종료합니다.

- **입출력 예시**

  - 입력:
    ```
    타이머를 시작할 시간(초)을 입력하세요: 5
    ```

  - 출력:
    ```

    타이머 종료!
    ```

````python
import time


def countdown_timer(seconds: int) -> None:
    """seconds초부터 1초마다 남은 시간을 출력하고, 끝나면 '타이머 종료!'를 출력한다."""
    if isinstance(seconds, bool) or not isinstance(seconds, int) or seconds < 0:
        print("0 이상의 정수(초)를 입력하세요.")
        return

    print("타이머 시작!")
    # range를 쓰면 음수가 들어와도 무한 루프에 빠지지 않는다
    for remaining in range(seconds, 0, -1):
        print(remaining)
        time.sleep(1)
    print("타이머 종료!")


def timer_program() -> None:
    """사용자에게 시간(초)을 입력받아 타이머를 실행한다."""
    countdown_timer(read_number("타이머를 시작할 시간(초)을 입력하세요: "))
````

### 문제1. 시간 추적 클래스(TimeTracker)

- **실습 설명**

  시간을 관리하고 추적하는 `TimeTracker` 클래스를 구현하는 프로젝트를 시작합니다. 시간 관리 기능은 특히 프로젝트 작업, 운동, 공부 시간 등 다양한 활동의 지속 시간을 측정하는 데 유용합니다.

  `TimeTracker` 클래스는 다음 기능을 제공해야 합니다:

  1. **시작 시간 설정**: 사용자가 활동을 시작할 때의 시간을 기록합니다.
  2. **종료 시간 설정**: 사용자가 활동을 종료할 때의 시간을 기록합니다.
  3. **경과 시간 계산**: 활동의 시작과 종료 사이의 시간 차이를 계산합니다.

  이 클래스의 인스턴스를 사용하여 각각의 활동에 대해 별도의 시간 추적을 할 수 있어야 합니다.

- **구현해야 할 메소드**

  - `start`: 현재 시간을 시작 시간으로 설정합니다.
  - `stop`: 현재 시간을 종료 시간으로 설정하고 경과 시간을 계산합니다.
  - `get_elapsed_time`: 마지막으로 기록된 시작 시간과 종료 시간 사이의 경과 시간을 분 단위로 반환합니다.

- **실습 결과 예시**

  다음과 같은 코드를 실행했을 때의 출력 예시입니다:

  ```python
  study_session = TimeTracker()
  study_session.start()
  # 1시간 30분 경과(가정)
  study_session.stop()

  print("활동 시간:", study_session.get_elapsed_time(), "분")
  ```

  예상 출력:

  ```
  공부한 시간: 90 분
  ```

- **요구 사항**

  1. 실제 시간을 추적하려면 Python의 `datetime` 모듈을 사용하여 현재 시간을 `datetime.now()`로 가져올 수 있습니다.
  2. 경과 시간은 분 단위로 반환해야 합니다.

````python
from datetime import datetime, timedelta


class TimeTracker:
    """활동의 시작·종료 시각을 기록하고 경과 시간을 분 단위로 계산한다."""

    def __init__(self) -> None:
        # 아직 측정하지 않았다는 뜻으로 None을 둔다 (start 전에 호출해도 AttributeError가 나지 않는다)
        self.start_time: datetime | None = None
        self.end_time: datetime | None = None

    def start(self) -> None:
        self.start_time = datetime.now()
        self.end_time = None  # 다시 시작하면 이전 종료 기록을 지운다

    def stop(self) -> None:
        if self.start_time is None:
            print("start()를 먼저 호출하세요.")
            return
        self.end_time = datetime.now()

    def get_elapsed_time(self) -> int:
        """마지막 시작~종료 사이의 경과 시간(분). 측정이 끝나지 않았으면 0."""
        if self.start_time is None or self.end_time is None:
            return 0
        # timedelta.seconds는 하루를 넘는 부분(days)을 버리므로 total_seconds()를 쓴다
        return int((self.end_time - self.start_time).total_seconds() // 60)
````

````python
# 활동 시작
study_session = TimeTracker()
study_session.start()

# 1시간 30분이 지났다고 가정: 시작 시각을 90분 앞으로 옮긴다
study_session.start_time -= timedelta(hours=1, minutes=30)
````

````python
# 활동 종료
study_session.stop()
print("공부한 시간:", study_session.get_elapsed_time(), "분")
````

````python
# 경계 사례 확인
tracker = TimeTracker()
print("측정 전:", tracker.get_elapsed_time(), "분")
tracker.stop()

tracker.start()
tracker.start_time -= timedelta(days=1, minutes=5)  # 하루가 넘는 활동
tracker.stop()
print("하루 5분:", tracker.get_elapsed_time(), "분")  # .seconds를 썼다면 5분이 나온다
````

### 문제1. 투표 시스템 클래스

- **실습 설명**

  간단한 투표 시스템을 위한 `VoteSystem` 클래스를 구현하는 프로젝트를 시작합니다. 이 시스템은 후보자 목록을 관리하고, 각 후보자에 대한 투표를 집계하는 기능을 제공합니다.

  `VoteSystem` 클래스는 다음 기능을 제공해야 합니다:

  - 후보자 등록
  - 투표 기능
  - 투표 결과 조회

- **구현해야 할 메소드**

  - `add_candidate`: 후보자를 등록합니다. 후보자의 이름을 입력받아 목록에 추가합니다.
  - `vote`: 특정 후보자에게 투표합니다. 투표하려는 후보자의 이름을 입력받습니다.
  - `get_results`: 각 후보자의 투표 수를 출력합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    voting_system = VoteSystem()
    voting_system.add_candidate("Alice")
    voting_system.add_candidate("Bob")
    voting_system.add_candidate("Charlie")

    voting_system.vote("Alice")
    voting_system.vote("Alice")
    voting_system.vote("Bob")

    voting_system.get_results()
    ```

  - 예상 출력:

    ```
    Alice 후보가 성공적으로 등록되었습니다.
    Bob 후보가 성공적으로 등록되었습니다.
    Charlie 후보가 성공적으로 등록되었습니다.
    Alice에게 투표하였습니다.
    Alice에게 투표하였습니다.
    Bob에게 투표하였습니다.
    투표 결과:
    Alice: 2 votes
    Bob: 1 vote
    Charlie: 0 votes
    ```

- **요구 사항**

  - 후보자는 중복 등록될 수 없습니다.
  - 등록되지 않은 후보자에게 투표할 수 없습니다.
  - 각 후보자의 이름과 투표 수는 사전(dictionary)을 사용하여 관리합니다.

````python
class VoteSystem:
    """후보자 등록, 투표, 결과 조회를 제공하는 투표 시스템."""

    def __init__(self) -> None:
        # 후보 이름 → 득표 수. 인스턴스 변수로 두어 투표 시스템끼리 결과가 섞이지 않게 한다
        self.candidates: dict[str, int] = {}

    def add_candidate(self, name: str) -> bool:
        if name in self.candidates:
            print(f"{name} 후보는 이미 등록되어 있습니다.")
            return False
        self.candidates[name] = 0
        print(f"{name} 후보가 성공적으로 등록되었습니다.")
        return True

    def vote(self, name: str) -> bool:
        if name not in self.candidates:
            print(f"{name} 후보는 등록되지 않아 투표할 수 없습니다.")
            return False
        self.candidates[name] += 1
        print(f"{name}에게 투표하였습니다.")
        return True

    def get_results(self) -> None:
        print("투표 결과:")
        for name, votes in self.candidates.items():
            unit = "vote" if votes == 1 else "votes"  # 1표만 단수 (0 votes, 2 votes)
            print(f"{name}: {votes} {unit}")
````

````python
# 사용 예시
voting_system = VoteSystem()
voting_system.add_candidate("Alice")
voting_system.add_candidate("Bob")
voting_system.add_candidate("Charlie")

voting_system.vote("Alice")
voting_system.vote("Alice")
voting_system.vote("Bob")

voting_system.get_results()
````

### 문제2. 은행 계좌 클래스

- **실습 설명**

  간단한 은행 계좌 관리 시스템을 위한 `BankAccount` 클래스를 구현하는 프로젝트를 시작합니다. 이 클래스는 개인의 은행 계좌 정보를 관리하고 기본적인 은행 거래 기능을 제공합니다.

  `BankAccount` 클래스는 다음 정보와 기능을 제공해야 합니다:

  - 계좌 번호(account number)
  - 소유자 이름(account holder)
  - 현재 잔액(balance)

- **구현해야 할 메소드**

  - `__init__`: 객체 생성 시 계좌 번호, 소유자 이름, 초기 잔액을 설정합니다.
  - `deposit`: 계좌에 금액을 입금합니다. 입금할 금액을 인자로 받고, 잔액을 업데이트합니다.
  - `withdraw`: 계좌에서 금액을 출금합니다. 출금할 금액을 인자로 받고, 잔액이 충분할 경우에만 출금을 허용하고 잔액을 업데이트합니다.
  - `get_balance`: 현재 계좌 잔액을 반환합니다.

- **실습 결과 예시**

  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    my_account = BankAccount("123-456-789", "김철수", 100000)
    my_account.deposit(50000)
    my_account.withdraw(20000)
    print(f"현재 잔액: {my_account.get_balance()}원")
    ```

  - 예상 출력:

    ```
    김철수님의 계좌 123-456-789가 개설되었습니다. 초기 잔액: 100000원
    50000원이 입금되었습니다. 현재 잔액: 150000원
    20000원이 출금되었습니다. 현재 잔액: 130000원
    현재 잔액: 130000원
    ```


- **요구 사항**

  - 계좌에서 출금 시도 시 잔액보다 많은 금액을 출금하려고 하면, 출금을 거부하고 경고 메시지를 출력해야 합니다.
  - 모든 금액은 정수 또는 실수로 처리될 수 있어야 하며, 화폐 단위로만 입력받습니다.
  - 계좌 생성, 입금, 출금 및 잔액 조회 기능을 모두 구현해야 합니다.

````python
class BankAccount:
    """계좌 번호, 소유자, 잔액을 관리하고 입금·출금·잔액 조회를 제공한다."""

    def __init__(self, account_number: str, account_holder: str, initial_balance: int | float = 0) -> None:
        if not self._is_valid_amount(initial_balance, allow_zero=True):
            raise ValueError("초기 잔액은 0 이상의 숫자여야 합니다.")

        self.account_number = account_number
        self.account_holder = account_holder
        self._balance = initial_balance  # 밖에서 직접 바꾸지 않고 deposit/withdraw로만 바꾼다
        print(f"{account_holder}님의 계좌 {account_number}가 개설되었습니다. 초기 잔액: {initial_balance}원")

    @staticmethod
    def _is_valid_amount(amount, allow_zero: bool = False) -> bool:
        # bool은 int의 하위 클래스라서 True가 1원으로 처리되지 않게 따로 막는다
        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            return False
        return amount >= 0 if allow_zero else amount > 0

    def deposit(self, amount: int | float) -> bool:
        if not self._is_valid_amount(amount):
            print(f"입금 실패: 입금액은 0보다 큰 숫자여야 합니다. (입력: {amount!r})")
            return False
        self._balance += amount
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self._balance}원")
        return True

    def withdraw(self, amount: int | float) -> bool:
        if not self._is_valid_amount(amount):
            print(f"출금 실패: 출금액은 0보다 큰 숫자여야 합니다. (입력: {amount!r})")
            return False
        if amount > self._balance:
            print(f"출금 실패: 잔액이 부족합니다. (요청: {amount}원, 현재 잔액: {self._balance}원)")
            return False
        self._balance -= amount
        print(f"{amount}원이 출금되었습니다. 현재 잔액: {self._balance}원")
        return True

    def get_balance(self) -> int | float:
        return self._balance
````

````python
# 사용 예시
my_account = BankAccount("123-456-789", "김철수", 100000)
my_account.deposit(50000)
my_account.withdraw(20000)
print(f"현재 잔액: {my_account.get_balance()}원")
````

````python
# 요구 사항 확인: 잔액보다 많은 출금, 0·음수·문자 금액은 거부하고 잔액은 그대로
my_account.withdraw(1_000_000)
my_account.deposit(-500)
my_account.deposit("10000")
print(f"현재 잔액: {my_account.get_balance()}원")
````

  - 예상 출력:

    ```
    홍길동 님이 추가되었습니다. 급여: 50000원
    김철수 님이 추가되었습니다. 급여: 60000원
    전체 직원의 평균 급여: 55000원
    ```

- **요구 사항**

  - 직원 정보는 클래스 변수 `employees`에 저장되어 전체 `EmployeeManager` 인스턴스에서 접근 가능해야 합니다.
  - `calculate_average_salary` 메서드는 저장된 모든 직원의 급여를 합산하여 평균을 출력하고, 직원이 없는 경우 0을 반환해야 합니다.

````python
class EmployeeManager:
    """직원 정보를 클래스 변수에 모아 전체 평균 급여를 계산한다."""

    # 모든 인스턴스가 공유하는 클래스 변수.
    # 이름이 아니라 직원 객체를 저장해서 동명이인도 따로 집계된다.
    employees: list["EmployeeManager"] = []

    def __init__(self, name: str, salary: int | float) -> None:
        if isinstance(salary, bool) or not isinstance(salary, (int, float)) or salary < 0:
            raise ValueError("급여는 0 이상의 숫자여야 합니다.")
        self.name = name
        self.salary = salary
        EmployeeManager.employees.append(self)
        print(f"{name} 님이 추가되었습니다. 급여: {salary}원")

    @classmethod
    def calculate_average_salary(cls) -> float:
        """전체 직원의 평균 급여를 출력하고 반환한다. 직원이 없으면 0을 반환한다."""
        if not cls.employees:
            print("등록된 직원이 없습니다.")
            return 0
        average = sum(emp.salary for emp in cls.employees) / len(cls.employees)
        shown = int(average) if average.is_integer() else round(average, 2)
        print(f"전체 직원의 평균 급여: {shown}원")
        return average
````

### 문제4. 프랜차이즈 레스토랑 관리 클래스


- **실습 설명**

  당신은 여러 지점을 가진 레스토랑 체인의 IT 팀에서 일하며, 각 지점의 예약을 관리하고 중앙에서 예약 현황을 파악할 수 있는 시스템을 개발할 임무를 맡았습니다. 이 시스템은 각 지점의 예약 상황을 관리하고, 고객의 예약 요청을 효과적으로 처리할 수 있는 기능을 제공해야 합니다.

  `ReservationSystem` 클래스는 각 레스토랑 지점의 예약을 관리하며, 다음 기능을 제공해야 합니다:

  - **예약 추가**: 고객이 특정 지점, 예약 일시, 인원 수에 대한 예약을 요청하면 시스템에 추가합니다.
  - **예약 취소**: 고객이 예약을 취소할 수 있으며, 해당 예약을 시스템에서 제거합니다.
  - **예약 조회**: 특정 지점의 모든 예약 상황을 확인할 수 있습니다.
  - **예약 집계**: 모든 지점의 예약 수를 합산합니다. 이 메서드는 모든 `ReservationSystem` 인스턴스의 예약 수를 합산하여 보여줍니다.

- **구현해야 할 메소드**

  - `__init__`: 레스토랑 지점의 이름을 초기화하고 예약 리스트를 관리합니다.
  - `add_reservation`: 새로운 예약을 추가합니다. 이 메서드는 예약자 이름, 예약 일시, 인원 수를 받아 저장합니다.
  - `list_reservations`: 현재 지점의 모든 예약 상태를 출력합니다.
  - `sum_reservations`: 주어진 `ReservationSystem` 인스턴스 리스트에서 모든 예약 수를 합산합니다.

- **실습 결과 예시**
  
  - 다음과 같은 코드를 실행했을 때의 출력 예시입니다:

    ```python
    restaurant1 = ReservationSystem("강남점")
    restaurant2 = ReservationSystem("홍대점")

    restaurant1.add_reservation("홍길동", "2024-05-20", 4)
    restaurant2.add_reservation("김철수", "2024-05-21", 2)

    restaurant1.list_reservations()
    restaurant2.list_reservations()

    total_reservations = ReservationSystem.sum_reservations([restaurant1, restaurant2])
    print(f"전체 레스토랑 예약 수: {total_reservations}")
    ```

    ```
    - 김철수, 2024-05-21, 2명
    전체 레스토랑 예약 수: 2
    ```

- **요구 사항**

  - 모든 출력 메시지는 한국어로 제공되어야 합니다.
  - 각 메서드는 적절한 입력 검증과 예외 처리를 포함해야 합니다.
  - `sum_reservations` 클래스 메서드는 모든 지점에서의 예약 수를 효과적으로 합산하여 전체 예약 상태를 중앙에서 확인할 수 있게 합니다.

````python
from datetime import datetime


class ReservationSystem:
    """레스토랑 지점 하나의 예약을 추가·취소·조회하고, 여러 지점의 예약 수를 합산한다."""

    DATE_FORMATS = ("%Y-%m-%d", "%Y/%m/%d", "%Y.%m.%d")

    def __init__(self, branch_name: str) -> None:
        if not isinstance(branch_name, str) or not branch_name.strip():
            raise ValueError("지점 이름은 비어 있지 않은 문자열이어야 합니다.")
        self.branch_name = branch_name.strip()
        # 지점마다 자기 예약 목록을 가진다 (인스턴스 변수)
        self.reservations: list[dict] = []

    @classmethod
    def _normalize_date(cls, date: str) -> str:
        """2024-05-20, 2024/05/20, 2024.05.20을 모두 YYYY-MM-DD로 통일한다."""
        for fmt in cls.DATE_FORMATS:
            try:
                return datetime.strptime(str(date).strip(), fmt).strftime("%Y-%m-%d")
            except ValueError:
                continue
        raise ValueError(f"날짜 형식이 올바르지 않습니다: {date!r} (예: 2024-05-20)")

    def add_reservation(self, name: str, date: str, guests: int) -> bool:
        try:
            if not isinstance(name, str) or not name.strip():
                raise ValueError("예약자 이름을 입력하세요.")
            if isinstance(guests, bool) or not isinstance(guests, int) or guests <= 0:
                raise ValueError(f"인원 수는 1 이상의 정수여야 합니다. (입력: {guests!r})")
            date = self._normalize_date(date)
            if any(r["name"] == name and r["date"] == date for r in self.reservations):
                raise ValueError(f"{name}님은 {date}에 이미 예약이 있습니다.")
        except ValueError as e:
            print(f"[{self.branch_name}] 예약 실패: {e}")
            return False

        self.reservations.append({"name": name.strip(), "date": date, "guests": guests})
        return True

    def cancel_reservation(self, name: str, date: str) -> bool:
        try:
            date = self._normalize_date(date)
        except ValueError as e:
            print(f"[{self.branch_name}] 취소 실패: {e}")
            return False

        for r in self.reservations:
            if r["name"] == name and r["date"] == date:
                self.reservations.remove(r)
                print(f"[{self.branch_name}] {name}님의 {date} 예약이 취소되었습니다.")
                return True
        print(f"[{self.branch_name}] 취소 실패: {name}님의 {date} 예약을 찾을 수 없습니다.")
        return False

    def list_reservations(self) -> None:
        print(f"{self.branch_name} 예약 목록:")
        if not self.reservations:
            print("- 예약 없음")
        for r in self.reservations:
            print(f"- {r['name']}, {r['date']}, {r['guests']}명")

    @classmethod
    def sum_reservations(cls, branches: list["ReservationSystem"]) -> int:
        """주어진 지점들의 예약 수를 모두 더한다 (지점 수가 아니라 예약 건수)."""
        if not all(isinstance(b, cls) for b in branches):
            raise TypeError("ReservationSystem 인스턴스 리스트를 전달하세요.")
        return sum(len(b.reservations) for b in branches)
````

````python
# 사용 예시
restaurant1 = ReservationSystem("강남점")
restaurant2 = ReservationSystem("홍대점")

restaurant1.add_reservation("홍길동", "2024-05-20", 4)
restaurant2.add_reservation("김철수", "2024-05-21", 2)

restaurant1.list_reservations()
restaurant2.list_reservations()

total_reservations = ReservationSystem.sum_reservations([restaurant1, restaurant2])
print(f"전체 레스토랑 예약 수: {total_reservations}")
````

````python
import csv
from itertools import combinations


def load_tasks(filename: str) -> list[tuple[str, int]]:
    """CSV 파일에서 (할 일, 소요 시간) 목록을 읽는다."""
    with open(filename, "r", newline="", encoding="utf-8") as csvfile:
        return [(row["할일"], int(row["소요시간"])) for row in csv.DictReader(csvfile)]


def suggest_tasks(tasks: list[tuple[str, int]], remaining_time: int) -> list[tuple[str, int]]:
    """남은 시간 안에 할 수 있는 할 일을 고른다.

    기준 1) 할 일 개수가 최대: 짧은 것부터 담는 그리디가 최대 개수 k를 보장한다.
    기준 2) 개수가 같으면 남는 시간이 최소: k개짜리 조합 중 총 시간이 가장 긴 것을 고른다.
    결과는 소요 시간이 짧은 순으로 반환한다.
    """
    k, used = 0, 0
    for _, duration in sorted(tasks, key=lambda t: t[1]):
        if used + duration > remaining_time:
            break
        used += duration
        k += 1
    if k == 0:
        return []

    feasible = (c for c in combinations(tasks, k) if sum(d for _, d in c) <= remaining_time)
    best = max(feasible, key=lambda c: sum(d for _, d in c))
    return sorted(best, key=lambda t: t[1])


def main() -> None:
    filename = input("CSV 파일 경로를 입력하세요(엔터: tasks.csv): ").strip() or "tasks.csv"
    try:
        tasks = load_tasks(filename)
    except FileNotFoundError:
        print(f"파일을 찾을 수 없습니다: {filename}")
        return

    remaining_time = read_number("남은 시간(분): ")
    if remaining_time < 0:
        print("남은 시간은 0 이상이어야 합니다.")
        return

    selected = suggest_tasks(tasks, remaining_time)
    if not selected:
        print("남은 시간 안에 할 수 있는 일이 없습니다.")
        return

    print("선택된 할 일 목록:")
    for idx, (task_name, duration) in enumerate(selected, start=1):
        print(f"{idx}. {task_name} - 예상 소요 시간: {duration}분")
    print(f"(총 {sum(d for _, d in selected)}분 사용, {remaining_time - sum(d for _, d in selected)}분 남음)")
````

이러한 선택지 중에서 어떤 기준이 당신의 상황에 가장 적합한지 고민해보고, 이를 코드로 반영해보세요.

**선택 기준**

1. **가장 많은 할 일을 포함하는 조합**을 1순위로 했습니다. 소요 시간이 짧은 것부터 담는 그리디 방식은 개수를 항상 최대로 만듭니다. 더 긴 일을 짧은 일로 바꿔도 개수는 줄지 않기 때문입니다.
2. 개수가 같은 조합이 여러 개면 **가장 적은 시간이 남는 조합**을 골랐습니다.

예를 들어 남은 시간이 200분이면 그리디만으로는 `가계부정리 + 운동하기`(90분)를 고릅니다. 같은 2개라도 `운동하기 + 공부하기`(180분)를 고르면 시간을 더 알차게 쓸 수 있습니다.

### 문제2. 수학 퍼즐: 소수의 합

- **실습 설명**  
  이 프로그램은 사용자로부터 어떤 수 이하의 소수의 합을 구할 것인지 입력 받아, 해당 수까지의 모든 소수의 합을 계산하여 출력합니다. 소수(prime number)는 1과 자기 자신만을 약수로 가지는 자연수를 말합니다. 소수를 찾는 방법은 자유롭게 선택할 수 있지만, 효율적인 방법을 고려하여 구현해야 합니다.

- **기능 설명**
  - 소수 판별: 주어진 수까지의 모든 자연수에 대해 소수 여부를 판별합니다. (힌트 : 에라토스테네스의 체)
  - 소수 합 계산: 판별된 소수들의 합을 계산합니다.
  - 결과 출력: 계산된 소수의 합을 출력합니다. 출력 형식은 사용자가 입력한 수까지의 소수 합과 그 값을 포함합니다.

- **입출력 예시**

  - 입력:

  ```
  어떤 수까지의 소수의 합을 구하시겠습니까?: 10
  ```
  - 출력:

  ```
  10 이하 소수의 합은 17입니다.
  ```

````python
def is_prime(num: int) -> bool:
    """num이 소수인지 판별한다. √num까지만 나눠 보면 충분하다."""
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


def sum_of_primes(n: int) -> int:
    """n 이하 소수의 합. 에라토스테네스의 체로 O(n log log n)에 계산한다."""
    if n < 2:
        return 0
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if sieve[i]:
            # i*i 미만의 배수는 더 작은 소수가 이미 지웠으므로 i*i부터 지운다
            sieve[i * i :: i] = [False] * len(range(i * i, n + 1, i))
    return sum(i for i, prime in enumerate(sieve) if prime)


def main() -> None:
    n = read_number("어떤 수까지의 소수의 합을 구하시겠습니까?: ")
    print(f"{n} 이하 소수의 합은 {sum_of_primes(n)}입니다.")
````

### 문제3. 도서관 관리 시스템

  - 시스템은 사용자의 행동에 따라 적절한 피드백을 제공해야 합니다. (예: 도서가 없을 때, 회원 정보가 없을 때)


````python
from datetime import date, timedelta
from functools import wraps


# ───────────── 사용자 정의 예외 ─────────────
class LibraryError(Exception):
    """도서관 시스템 예외의 부모 클래스."""


class BookNotFoundError(LibraryError):
    def __init__(self, isbn: str):
        super().__init__(f"ISBN '{isbn}'에 해당하는 도서를 찾을 수 없습니다.")


class BookNotAvailableError(LibraryError):
    def __init__(self, title: str):
        super().__init__(f"'{title}' 도서는 모두 대여 중입니다.")


class MemberNotFoundError(LibraryError):
    def __init__(self, name: str):
        super().__init__(f"회원 '{name}'님을 찾을 수 없습니다.")


class NotRentedError(LibraryError):
    def __init__(self, name: str, title: str):
        super().__init__(f"'{name}' 회원님은 '{title}' 도서를 대여하고 있지 않습니다.")


def report_errors(method):
    """도서관 기능 실행 중 생긴 오류를 '오류: ...'로 알려 주고 False를 반환한다.

    성공하면 메서드의 반환값(없으면 True)을 돌려준다.
    """
    @wraps(method)
    def wrapper(*args, **kwargs):
        try:
            result = method(*args, **kwargs)
        except (LibraryError, ValueError) as e:
            print(f"오류: {e}")
            return False
        return True if result is None else result
    return wrapper


# ───────────── 데이터 클래스 ─────────────
class Book:
    """도서 정보(제목, 저자, 출판년도, ISBN)와 보유·대여 가능 권수."""

    def __init__(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        if not all(isinstance(v, str) and v.strip() for v in (title, author, isbn)):
            raise ValueError("제목, 저자, ISBN은 비어 있지 않은 문자열이어야 합니다.")
        if isinstance(publication_year, bool) or not isinstance(publication_year, int):
            raise ValueError(f"출판년도는 정수여야 합니다. (입력: {publication_year!r})")
        self.title = title
        self.author = author
        self.publication_year = publication_year
        self.isbn = isbn
        self.total_copies = 1
        self.available_copies = 1

    def __str__(self) -> str:
        return f"{self.title} (저자: {self.author}, 출판년도: {self.publication_year})"


class Member:
    """회원 정보(이름, 회원번호)와 대여 중인 도서 목록."""

    def __init__(self, name: str, member_id: int) -> None:
        self.name = name
        self.member_id = member_id
        self.rented_books: list[Book] = []

    def __str__(self) -> str:
        titles = ", ".join(b.title for b in self.rented_books) or "없음"
        return f"{self.name} (대여 중인 도서: {titles})"


class Rental:
    """대여 기록(도서, 회원, 대여일, 반납 예정일, 반납일)."""

    LOAN_DAYS = 14

    def __init__(self, book: Book, member: Member, rental_date: date | None = None) -> None:
        self.book = book
        self.member = member
        self.rental_date = rental_date or date.today()
        self.due_date = self.rental_date + timedelta(days=self.LOAN_DAYS)
        self.return_date: date | None = None  # 반납 전에는 None

    @property
    def is_returned(self) -> bool:
        return self.return_date is not None

    def mark_returned(self, return_date: date | None = None) -> None:
        self.return_date = return_date or date.today()


# ───────────── 관리 시스템 ─────────────
class LibraryManagement:
    """도서·회원·대여 정보를 관리한다."""

    def __init__(self) -> None:
        self.books: dict[str, Book] = {}      # ISBN → Book
        self.members: dict[str, Member] = {}  # 회원 이름 → Member (예시처럼 이름으로 회원을 찾는다)
        self.rentals: list[Rental] = []       # 반납된 기록까지 모두 보관
        self._next_member_id = 1

    # 내부 도우미: 찾지 못하면 예외를 던진다
    def _get_book(self, isbn: str) -> Book:
        if isbn not in self.books:
            raise BookNotFoundError(isbn)
        return self.books[isbn]

    def _get_member(self, name: str) -> Member:
        if name not in self.members:
            raise MemberNotFoundError(name)
        return self.members[name]

    def _find_active_rental(self, book: Book, member: Member) -> Rental | None:
        for rental in self.rentals:
            if rental.book is book and rental.member is member and not rental.is_returned:
                return rental
        return None

    # ── 도서 ──
    @report_errors
    def add_book(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        if isbn in self.books:  # 같은 책이 또 들어오면 권수를 늘린다
            book = self.books[isbn]
            book.total_copies += 1
            book.available_copies += 1
            print(f"'{book.title}' 도서가 1권 더 추가되었습니다. (보유 {book.total_copies}권)")
            return
        book = Book(title, author, publication_year, isbn)
        self.books[isbn] = book
        print(f"'{title}' (저자: {author}, 출판년도: {publication_year}) 도서가 추가되었습니다.")

    @report_errors
    def remove_book(self, isbn: str) -> None:
        book = self._get_book(isbn)
        if book.available_copies < book.total_copies:
            raise LibraryError(f"'{book.title}' 도서는 대여 중인 책이 있어 삭제할 수 없습니다.")
        del self.books[isbn]
        print(f"'{book.title}' 도서가 삭제되었습니다.")

    @report_errors
    def search_books(self, keyword: str) -> list[Book]:
        """제목·저자·ISBN에 keyword가 들어간 도서를 찾는다."""
        keyword = keyword.strip().lower()
        found = [b for b in self.books.values()
                 if keyword in b.title.lower() or keyword in b.author.lower() or keyword in b.isbn]
        print(f"'{keyword}' 검색 결과: {len(found)}건")
        for b in found:
            print(f"- {b} [대여 가능 {b.available_copies}/{b.total_copies}권]")
        return found

    def print_books(self) -> None:
        if not self.books:
            print("등록된 도서가 없습니다.")
            return
        print("도서 목록:")
        for book in self.books.values():
            print(f"- {book}")

    # ── 회원 ──
    @report_errors
    def add_member(self, name: str) -> None:
        if not isinstance(name, str) or not name.strip():
            raise ValueError("회원 이름을 입력하세요.")
        if name in self.members:
            raise LibraryError(f"회원 '{name}'님은 이미 등록되어 있습니다.")
        self.members[name] = Member(name, self._next_member_id)
        self._next_member_id += 1
        print(f"회원 '{name}'님이 등록되었습니다.")

    @report_errors
    def get_member_info(self, name: str) -> None:
        member = self._get_member(name)
        print(f"회원번호 {member.member_id}: {member}")
        for rental in self.rentals:
            if rental.member is member and not rental.is_returned:
                print(f"  - '{rental.book.title}' 대여일 {rental.rental_date}, 반납 예정일 {rental.due_date}")

    def print_members(self) -> None:
        if not self.members:
            print("등록된 회원이 없습니다.")
            return
        print("회원 목록:")
        for member in self.members.values():
            print(f"- {member}")

    # ── 대여·반납 ──
    @report_errors
    def rent_book(self, isbn: str, member_name: str) -> None:
        book = self._get_book(isbn)
        member = self._get_member(member_name)
        if self._find_active_rental(book, member):
            raise LibraryError(f"'{member_name}' 회원님은 이미 '{book.title}' 도서를 대여 중입니다.")
        if book.available_copies == 0:
            raise BookNotAvailableError(book.title)

        book.available_copies -= 1
        member.rented_books.append(book)
        self.rentals.append(Rental(book, member))
        print(f"'{member_name}' 회원님이 '{book.title}' 도서를 대여하였습니다.")

    @report_errors
    def return_book(self, isbn: str, member_name: str) -> None:
        book = self._get_book(isbn)
        member = self._get_member(member_name)
        rental = self._find_active_rental(book, member)
        if rental is None:
            raise NotRentedError(member_name, book.title)

        # 모든 검사를 통과한 뒤에 상태를 바꾼다 (중간에 실패해도 기록이 어긋나지 않게)
        rental.mark_returned()
        book.available_copies += 1
        member.rented_books.remove(book)
        print(f"'{member_name}' 회원님이 '{book.title}' 도서를 반납하였습니다.")
````

````python
# 사용 예시
library = LibraryManagement()
library.add_book("1984", "조지 오웰", 1949, "978-0451524935")
library.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")
library.add_member("홍길동")

library.rent_book("978-0451524935", "홍길동")
library.print_books()
library.print_members()
library.return_book("978-0451524935", "홍길동")
library.print_members()
````
