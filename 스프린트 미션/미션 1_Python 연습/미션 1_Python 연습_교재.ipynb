## 0. 준비

### 0-1. 입력 자동화 · 출력 캡처 · 숫자 입력 도구

**요구사항** (노트북 운영용) `input()`이 필요한 문제를 사람 손 없이 재실행하고, `print()` 출력을 예상 출력과 비교할 수 있어야 함

**설계**
- `fake_inputs`: `input()`을 정해 둔 값으로 잠시 교체, `try · finally`로 반드시 복구
- `printed`: `redirect_stdout`으로 출력을 문자열로 받아 `assert`와 6-1 자동 대조에 사용
- `parse_number` · `read_number`: 문자열 → 숫자 변환(정수 우선 · 소수 허용 옵션)과 숫자가 나올 때까지 다시 묻는 입력을 한 번만 정의해 1-6 · 2-2 · 2-3 · 5장에서 재사용
- 준비한 입력값이 모자라면 원인을 알리는 `RuntimeError`
- `is_number` · `require_non_negative_int`: 숫자 검사(bool · inf · nan 제외)와 0 이상 정수 검사를 한 번만 정의해 1-5 · 1-6 · 2-3 · 4-2 · 4-3에서 재사용

**실행 흐름**

1. **입력**: 미리 정해 둔 입력값, 출력을 확인할 함수
2. **처리**: 원래 `input()` 보관 → 정해 둔 값을 차례로 돌려주는 가짜 `input()`으로 교체 → with 블록 실행 → 복구 / `print()` 방향을 문자열 버퍼로 바꿔 출력 수집 / 문자열 → 숫자 변환
3. **저장**: `NOTEBOOK_START` · `fake_inputs` · `printed` · `parse_number` · `read_number` · `is_number` · `require_non_negative_int`
4. **출력**: 질문 + 입력값, 안내 문구, 동작 확인 결과

**핵심 개념**
- **모듈 속 함수 바꿔 끼우기**: `input()`은 `builtins` 모듈에 있는 함수라 `builtins.input = 새 함수`로 바꾸면 이후 모든 `input()` 호출이 새 함수로 감
- **`@` 데코레이터 문법**: `@contextmanager`처럼 함수 바로 위에 쓰면 아래 함수를 그 함수로 감싸 기능을 덧붙임 (5-3에서 직접 만들어 봄)
- **is · assert**: `a is b`는 같은 객체인지(`is None`처럼 None 확인에 사용), `assert 조건`은 조건이 거짓이면 `AssertionError`로 멈춰 결과가 틀렸음을 알림
- **iter · next**: `iter()`로 만든 반복자에서 `next()`가 값을 하나씩 꺼내고, 다 꺼낸 뒤에는 두 번째 인자(`None`)를 돌려줌
- **클로저**: 안쪽 함수 `_fake_input`이 바깥 함수의 변수 `remaining` · `echo`를 기억해 호출될 때마다 다음 값을 꺼냄
- **contextmanager**: `yield` 앞은 with 블록 시작 때, `finally` 뒤는 끝날 때 실행 → 블록에서 오류가 나도 원래 `input()` 복구
- **redirect_stdout**: `print()`가 화면 대신 `StringIO`에 쓰이게 해 출력 형식을 문자열로 비교
- **정수 먼저**: `int(text)`를 먼저 시도해야 아주 큰 정수가 float로 바뀌며 끝자리가 틀어지지 않음, `math.isfinite`로 inf · nan 거부
- **bool을 먼저 거르는 이유**: `True`는 `int`의 일종이라 `isinstance(True, int)`가 True → `isinstance(n, bool)`을 먼저 확인해야 `True`가 1로 계산되지 않음
- **타입 힌트**: `text: str` · `-> int`는 어떤 자료형을 주고받는지 적은 안내일 뿐 실행할 때 검사하지 않음, `int | None`은 int 또는 None (노트북 전체에서 사용)
- **f-string · `!r`**: `f"{prompt}{value}"`는 값을 문장에 넣고, `{prompt!r}`은 `'질문: '`처럼 따옴표까지 보여 빈 값 · 공백을 드러냄

````python
# 모듈 불러오기
import builtins  # 내장 input()을 잠시 바꿔 끼우기 위해
import io  # print() 결과를 문자열로 모으기 위해
import math  # inf · nan 같은 비정상 숫자를 걸러내기 위해
import time  # 2-3 대기 · 5-2 속도 · 6-1 전체 실행 시간 측정
from contextlib import contextmanager, redirect_stdout  # with 블록 정의, print() 출력 방향 바꾸기

# 전체 실행 시간 기준점
NOTEBOOK_START = time.perf_counter()  # 6-1에서 처음부터 걸린 시간을 출력하기 위해

# 입력 자동화 함수
@contextmanager
def fake_inputs(*values, echo=True):  # 입력값 개수와 상관없이 받기 위해 가변 인자
    """with 블록 안의 input()을 values로 차례 대체 (echo=True면 질문과 입력값도 표시)"""
    remaining = iter(values)  # 정해 둔 입력값을 하나씩 꺼내기 위해
    real_input = builtins.input  # 블록이 끝나면 되돌리기 위해 원래 input() 보관
    def _fake_input(prompt=""):  # 원래 input()과 같은 모양으로 호출되도록
        value = next(remaining, None)  # 호출될 때마다 다음 입력값을 주기 위해
        if value is None:  # 준비한 입력이 모자라면 원인을 바로 알 수 있도록
            raise RuntimeError(f"준비한 입력값이 부족합니다 (질문: {prompt!r})")
        if echo:  # 실행 화면에서는 실제 입력처럼 보이고, 자동 대조 때는 출력만 남기기 위해
            print(f"{prompt}{value}")
        return value
    builtins.input = _fake_input  # 블록 안의 input()이 정해 둔 값을 돌려주도록
    try:  # 블록에서 오류가 나도 복구 단계가 실행되도록
        yield
    finally:
        builtins.input = real_input  # 오류가 나도 원래 input()으로 복구되도록

# 출력 캡처 함수
def printed(func, *args):
    """func(*args)가 print()한 내용을 문자열로 반환 (출력 형식 검증용)"""
    buffer = io.StringIO()  # 화면 대신 출력을 모아 둘 곳
    with redirect_stdout(buffer):  # 블록 안의 print()가 buffer로 가도록
        func(*args)
    return buffer.getvalue()

# 숫자 변환 · 입력 함수 (1-6 · 2-2 · 2-3 · 5장에서 재사용)
def parse_number(text: str, allow_float: bool = False) -> int | float | None:
    """문자열을 숫자로 변환, 숫자가 아니면 None"""
    text = text.strip()
    try:
        return int(text)  # 정수를 먼저 시도해 큰 정수도 정확하게
    except ValueError:
        pass
    if not allow_float:  # 정수만 받는 모드
        return None
    try:
        value = float(text)
    except ValueError:
        return None
    if not math.isfinite(value):  # inf · nan은 숫자로 받지 않음
        return None
    if value.is_integer() and abs(value) < 2 ** 53:  # 10.0은 10으로 (실수가 정수를 정확히 담는 범위에서만)
        return int(value)
    return value
def read_number(prompt: str, allow_float: bool = False, minimum: int | None = None) -> int | float:
    """숫자가 입력될 때까지 다시 질문, minimum보다 작으면 다시 질문"""
    while True:  # 올바른 값이 나올 때까지 반복하기 위해
        value = parse_number(input(prompt), allow_float)
        if value is None:
            print("올바른 숫자를 입력하세요.")  # 문제지 2-2의 안내 문구 그대로
        elif minimum is not None and value < minimum:  # 음수 시간 · 음수 범위처럼 의미 없는 값은 다시 질문
            print(f"{minimum} 이상의 숫자를 입력하세요.")
        else:
            return value

# 공통 검사 함수 (1-5 · 1-6 · 2-3 · 4-2 · 4-3에서 재사용)
def is_number(x) -> bool:
    """int · float이면서 bool이 아니고, inf · nan이 아닌 값인지"""
    if isinstance(x, bool) or not isinstance(x, (int, float)):  # True · False는 숫자로 보지 않음
        return False
    return not isinstance(x, float) or math.isfinite(x)  # 정수는 항상 유한
def require_non_negative_int(n) -> None:
    """0 이상의 정수가 아니면 ValueError"""
    if not is_number(n) or not isinstance(n, int) or n < 0:
        raise ValueError(f"0 이상의 정수를 입력하세요: {n!r}")

# 동작 확인
original_input = builtins.input
with fake_inputs("hello"):
    assert input("입력: ") == "hello"  # 정해 둔 값이 반환되는지 확인하기 위해
assert builtins.input is original_input  # 블록이 끝나면 원래 input()으로 돌아왔는지 확인하기 위해
try:
    with fake_inputs():
        input("질문: ")
except RuntimeError as e:
    assert "부족" in str(e) and builtins.input is original_input  # 입력이 모자라도 원인 안내 · input() 복구
assert printed(print, "a", "b") == "a b\n"  # print() 결과가 문자열로 잡히는지 확인하기 위해
with fake_inputs("abc", "inf", "7"):
    assert read_number("정수: ") == 7  # 숫자가 아닌 입력 · inf를 거르고 7을 받는지 확인하기 위해
assert parse_number("2.5", True) == 2.5 and parse_number("2.5") is None  # 소수는 허용 모드에서만
assert parse_number("10.0", True) == 10 and parse_number("nan", True) is None  # 10.0은 정수 10, nan 거부
assert parse_number("99999999999999999999", True) == 99999999999999999999  # 큰 정수도 정확하게
with fake_inputs("-3", "4", echo=False):
    assert printed(read_number, "", False, 0) == "0 이상의 숫자를 입력하세요.\n"  # 최솟값보다 작으면 다시 질문
assert is_number(3) and is_number(2.5) and not any(is_number(x) for x in [True, "3", float("inf"), None])
for bad in [-1, 2.5, True, "3"]:  # 음수 · 소수 · bool · 문자열 거부
    try:
        require_non_negative_int(bad)
        raise AssertionError("검사 누락")
    except ValueError:
        pass
print("✅ 통과")
````

### 1-1. 3의 배수 출력

**요구사항** 1부터 n까지의 3의 배수를 한 줄씩 출력

**설계**
- `range(3, n + 1, 3)`으로 3의 배수만 생성 → 나머지 검사 불필요

**실행 흐름**

1. **입력**: `n` = 7 · 30, 검증용 0 · 1 · 3 · 100
2. **처리**: 3부터 n까지 3씩 증가하는 수 생성 → 출력 / 나머지 검사로 만든 답과 출력 비교
3. **저장**: 없음 (바로 출력)
4. **출력**: 3의 배수 한 줄씩, 검증 결과

**핵심 개념**
- **range(시작, 끝, 간격)**: 끝 값은 포함하지 않으므로 `n + 1`까지 지정
- **출력 검증**: 출력은 `printed`로 문자열로 받아 정답 문자열과 비교, 반환값(출력한 개수)은 출력을 숨긴 채 따로 확인
- **제너레이터 식**: `(f"{i}\n" for i in ... if ...)`는 조건에 맞는 값을 하나씩 만들어 `join`에 넘김 (리스트를 따로 만들지 않음)

````python
# 함수 정의
def print_multiples_of_three(n: int) -> int:  # 문제지 시그니처 그대로, 출력한 개수를 반환
    count = 0
    for i in range(3, n + 1, 3):  # 나머지 검사 없이 3의 배수만 만들기 위해 3씩 증가
        print(i)  # 문제 형식대로 한 줄에 하나씩 출력하기 위해
        count += 1
    return count

# 문제지 예시 1 · 2
print_multiples_of_three(7)
print("---")  # 두 예시의 출력을 구분하기 위해
print_multiples_of_three(30)

# 검증: 나머지 검사(i % 3 == 0)로 구한 답과 출력 비교
for n in [0, 1, 3, 7, 30, 100]:  # 경계(0 · 1 · 3)와 문제지 예시를 함께 확인하기 위해
    expected = "".join(f"{i}\n" for i in range(1, n + 1) if i % 3 == 0)
    assert printed(print_multiples_of_three, n) == expected
with redirect_stdout(io.StringIO()):  # 출력은 숨기고 반환값만 확인
    assert print_multiples_of_three(30) == 10  # 30 이하 3의 배수 10개를 반환
print("✅ 통과")
````

### 1-2. 모음 제거하기

**요구사항** 문자열에서 모음(a, e, i, o, u)을 제거해 반환

**설계**
- 모음을 `set`에 담아 글자마다 포함 여부 확인, 대문자 모음까지 처리

**실행 흐름**

1. **입력**: `"codeit"` · `"python programming"`, 검증용 빈 문자열 · 대문자
2. **처리**: 글자마다 모음 집합에 있는지 검사 → 모음이 아닌 글자만 이어 붙이기
3. **저장**: `VOWELS` 모음 집합
4. **출력**: 모음을 뺀 문자열, 검증 결과

**핵심 개념**
- **set 포함 검사**: `in`으로 확인할 때 리스트는 처음부터 차례로 보지만, set은 바로 찾음
- **제너레이터 식**: `(ch for ch in s if ...)`는 조건에 맞는 글자를 하나씩 만들어 `join`에 넘김

````python
# 모음 집합
VOWELS = set("aeiouAEIOU")  # 대 · 소문자 모음을 한 번에 빠르게 검사하기 위해 set으로

# 함수 정의
def remove_vowels(s: str) -> str:
    return "".join(ch for ch in s if ch not in VOWELS)  # 모음이 아닌 글자만 남기기 위해

# 문제지 예시 1 · 2
print(remove_vowels("codeit"))
print(remove_vowels("python programming"))

# 검증
assert remove_vowels("codeit") == "cdt"
assert remove_vowels("python programming") == "pythn prgrmmng"  # 공백은 그대로 남는지 확인하기 위해
assert remove_vowels("") == "" and remove_vowels("AEIOU") == ""  # 빈 문자열 · 대문자 모음
print("✅ 통과")
````

### 1-3. 수 크기 비교하기

**요구사항** (큰 수, 작은 수) 반환, 같은 수면 그 수 하나만 반환

**설계**
- `max` · `min`으로 입력 순서와 상관없이 정렬, 같은 수는 튜플이 아닌 숫자 하나

**실행 흐름**

1. **입력**: 두 수 (10, 20) · (30, 30), 검증용 (20, 10) · (-5, 3)
2. **처리**: 같으면 그 수 하나, 다르면 `max` · `min`으로 (큰 수, 작은 수)
3. **저장**: 없음
4. **출력**: 비교 결과, 검증 결과

**핵심 개념**
- **튜플 반환**: `return a, b`는 `(a, b)` 튜플 하나를 돌려줌
- **연쇄 비교**: `x == y == z`는 `x == y and y == z`와 같음

````python
# 함수 정의
def compare_numbers(num1: int, num2: int) -> tuple[int, int] | int:
    if num1 == num2:  # 문제 조건대로 같은 수는 하나만 반환하기 위해
        return num1
    return max(num1, num2), min(num1, num2)  # 입력 순서와 상관없이 (큰 수, 작은 수)가 되도록

# 문제지 예시 1 · 2
print(compare_numbers(10, 20))
print(compare_numbers(30, 30))

# 검증
assert compare_numbers(10, 20) == compare_numbers(20, 10) == (20, 10)  # 입력 순서와 무관한지 확인하기 위해
assert compare_numbers(30, 30) == 30  # 같은 수는 튜플이 아닌 숫자 하나
assert compare_numbers(-5, 3) == (3, -5)  # 음수도 같은 규칙
print("✅ 통과")
````

### 1-4. 문자열 길이 반환하기

**요구사항** 길이 반환, 빈 문자열이면 `"문자열이 비어 있습니다."` 반환

**설계**
- 빈 문자열은 `False`로 판단되는 성질로 먼저 처리

**실행 흐름**

1. **입력**: `"Python"` · `""`, 검증용 `"a b"`
2. **처리**: 비어 있으면 안내 문구, 아니면 `len()`
3. **저장**: 없음
4. **출력**: 길이 또는 안내 문구, 검증 결과

**핵심 개념**
- **빈 값의 진릿값**: `""` · `0` · `[]` · `None`은 `if` 조건에서 False → `if not string`으로 빈 문자열 판별

````python
# 함수 정의
def check_string_length(string: str) -> int | str:
    if not string:  # 빈 문자열은 False로 판단되므로 따로 처리하기 위해
        return "문자열이 비어 있습니다."
    return len(string)  # 공백도 글자 수에 포함

# 문제지 예시 1 · 2
print(check_string_length("Python"))
print(check_string_length(""))

# 검증
assert check_string_length("Python") == 6
assert check_string_length("") == "문자열이 비어 있습니다."
assert check_string_length("a b") == 3  # 문제지 설명대로 공백이 포함된 문자열
print("✅ 통과")
````

### 1-5. 총 합계 구하기

**요구사항** 0부터 n까지의 합 계산

**설계**
- 반복문 대신 가우스 공식 `n(n+1)/2` → 계산량이 n과 무관 (O(1))
- 조건(0 이상의 정수) 밖의 값은 그럴듯한 답 대신 오류로 거부

**실행 흐름**

1. **입력**: `n` = 3 · 10, 잘못된 값 -5 · 2.5
2. **처리**: 자료형 · 범위 검사 → `n × (n + 1) // 2` / 반복문 합과 비교 (n = 0 ~ 999)
3. **저장**: 없음
4. **출력**: 합계, 거부 안내, 검증 결과

**핵심 개념**
- **가우스 공식**: 1 + 2 + … + n = n(n+1)/2 → n이 커져도 계산 한 번
- **공통 검사 재사용**: 0-1의 `require_non_negative_int`가 음수 · 소수 · bool을 `ValueError`로 거부 → 1-6 · 2-3도 같은 함수 사용
- **raise · except**: 함수가 처리할 수 없는 입력을 호출한 쪽에 알리고, 호출한 쪽은 `try · except`로 받아 안내

````python
# 함수 정의
def sum_up_to_n(n: int) -> int:
    require_non_negative_int(n)  # 조건 밖의 값(음수 · 소수)이 그럴듯한 답을 내지 않도록 0-1 공통 검사
    return n * (n + 1) // 2  # 반복 없이 한 번에 계산하는 가우스 공식 (// 로 정수 유지)

# 문제지 예시 1 · 2
print(sum_up_to_n(3))
print(sum_up_to_n(10))

# 추가 사례: 조건 밖의 입력
for bad in [-5, 2.5]:
    try:
        sum_up_to_n(bad)
    except ValueError as e:
        print("거부:", e)

# 검증: 공식 결과 = 반복문으로 더한 결과
assert all(sum_up_to_n(n) == sum(range(n + 1)) for n in range(1000))  # n = 0 ~ 999
print("✅ 통과")
````

### 1-6. 짝수 합 구하기

**요구사항** n까지의 짝수 합 출력, 0을 입력하면 종료

**설계**
- 인자가 있으면 한 번 계산, 문제지 실행 코드처럼 인자 없이 부르면 0이 나올 때까지 입력 반복
- 숫자가 아닌 입력 · 음수(문제지 가정 밖)는 `read_number`가 다시 질문 → 프로그램이 멈추지 않음
- 인자로 직접 부를 때는 1-5와 같은 기준(0 이상의 정수)으로 검사
- 짝수 합은 공식 k(k + 1), k = n // 2 (1-5처럼 반복 없이 계산, 검증은 짝수를 하나씩 더한 합과 비교)

**실행 흐름**

1. **입력**: 자동 입력 9 → nine → -2 → 4 → 0, 실행 코드 인자 3, 잘못된 인자 2.5
2. **처리**: 인자 없으면 0이 나올 때까지 입력 반복 (숫자 아님 · 음수는 다시 질문) → 짝수 개수 k = n // 2 → 공식 k(k + 1) / 검증: 짝수를 하나씩 더한 합과 비교
3. **저장**: 없음
4. **출력**: 짝수 합 문장, 안내 문구, 종료 문구, 검증 결과

**핵심 개념**
- **기본값 None**: 인자를 생략할 수 있게 하고, `if n is None`으로 생략 여부를 구분
- **짝수 합 공식**: 2 + 4 + … + 2k = 2 × (1 + … + k) = k(k + 1), k = n // 2 → 1-5처럼 반복 없이 계산하고 반복문 합과 비교
- **재귀 호출**: 입력받은 n으로 같은 함수를 다시 불러 한 번 계산하는 부분을 재사용
- **None != 0**: 처음 `n`은 None이고 None은 0과 다르므로 `while n != 0`에 처음 한 번은 반드시 들어감

````python
# 함수 정의
def sum_even_numbers(n: int | None = None) -> None:
    """n을 주면 한 번 계산, 문제지 실행 코드처럼 인자 없이 부르면 0이 나올 때까지 입력받아 반복"""
    if n is None:
        while n != 0:  # 0이 입력되면 반복 종료
            n = read_number("숫자를 입력하세요 (0: 종료): ", minimum=0)  # 숫자가 아닌 입력 · 음수는 다시 질문
            sum_even_numbers(n)
        return
    require_non_negative_int(n)  # 1-5와 같은 0-1 공통 검사
    if n == 0:  # 0은 종료 신호
        print("종료")
        return
    k = n // 2  # n 이하 짝수 개수
    print(f"1부터 {n}까지의 짝수 합은 {k * (k + 1)}입니다.")  # 2 + 4 + … + 2k = 2 × (1 + … + k) = k(k + 1), 1-5처럼 공식으로

# 문제지 예시: 9 → 0 (사이에 숫자가 아닌 입력 · 음수 · 4)
with fake_inputs("9", "nine", "-2", "4", "0"):
    sum_even_numbers()

# 문제지 실행 코드 · 잘못된 인자
sum_even_numbers(3)
try:
    sum_even_numbers(2.5)
except ValueError as e:
    print("거부:", e)

# 검증: 공식 결과 = 짝수를 하나씩 더한 결과
for n in range(1, 101):
    assert printed(sum_even_numbers, n) == f"1부터 {n}까지의 짝수 합은 {sum(range(2, n + 1, 2))}입니다.\n"
assert printed(sum_even_numbers, 0) == "종료\n"
print("✅ 통과")
````

### 1-7. 두 숫자 나누기

**요구사항** 두 숫자를 입력받아 나눈 값 반환, 예외가 나면 `"잘못된 입력입니다."` 반환 (`try-except`)

**설계**
- 문제지 실행 코드처럼 인자 없이 부르면 `input()`, 인자를 주면 그 값 사용 (생략한 값만 입력)
- 입력은 0-1의 `parse_number`로 변환(숫자가 아니면 None), 숫자 아님 · 잘못된 자료형(`TypeError`) · 0으로 나누기 · 너무 큰 수 3종류를 한 `except`에서 처리, inf · nan 결과도 거부
- 나누어떨어지면 `6.0` 대신 `6` (예상 출력 형식)

<img src="images/m1_divide.png" alt="divide_numbers의 정상 경로와 예외 경로" width="800">

**실행 흐름**

1. **입력**: 자동 입력 (42, 7) · (135, 0) · (10, zero), 인자 (7, 2) · (10의 400제곱, 3)
2. **처리**: bool 거부 → 생략한 값만 입력(parse_number) → 나눗셈 → 오류 3종류는 안내 문구 → inf · nan 거부 → 나누어떨어지면 int로
3. **저장**: 없음
4. **출력**: 몫 · 안내 문구, 검증 결과

**핵심 개념**
- **except (A, B, C)**: 여러 오류를 하나의 처리로 묶음 (숫자가 아닌 입력은 `parse_number`가 None으로 돌려줘 나눗셈에서 `TypeError`)
- **조건 표현식**: `A if 조건 else B`는 조건이 참이면 A, 거짓이면 B인 한 줄 if-else
- **OverflowError**: 아주 큰 정수를 float로 바꿀 수 없을 때 발생 (`10 ** 400 / 3`)

````python
# 함수 정의
def divide_numbers(a=None, b=None) -> int | float | str:
    """문제지 실행 코드처럼 인자 없이 부르면 두 숫자를 차례로 입력 (생략한 값만 입력)"""
    if isinstance(a, bool) or isinstance(b, bool):  # True · False는 숫자로 보지 않음 (1-5 · 1-6과 같은 기준)
        return "잘못된 입력입니다."
    try:
        if a is None:
            a = parse_number(input("첫 번째 숫자: "), allow_float=True)  # 0-1과 같은 변환, 숫자가 아니면 None → 나눗셈에서 TypeError
        if b is None:
            b = parse_number(input("두 번째 숫자: "), allow_float=True)
        result = a / b  # 7 ÷ 2 = 3.5처럼 소수를 유지하기 위해 몫(//) 대신 일반 나눗셈
    except (TypeError, ZeroDivisionError, OverflowError):  # 숫자 아님(None) · 잘못된 자료형 · 0으로 나누기 · 너무 큰 수
        return "잘못된 입력입니다."
    if not math.isfinite(result):  # inf · nan 입력이 결과로 새어 나가지 않도록
        return "잘못된 입력입니다."
    return int(result) if result.is_integer() and abs(result) < 2 ** 53 else result  # 6.0을 6으로 보이기 위해

# 문제지 예시 1 · 2 (입력받는 방식)
with fake_inputs("42", "7"):
    print(divide_numbers())
with fake_inputs("135", "0"):
    print(divide_numbers())

# 추가 사례: 소수 결과, 숫자가 아닌 입력, 인자로 호출
print(divide_numbers(7, 2))
with fake_inputs("10", "zero"):
    print(divide_numbers())
print(divide_numbers(10 ** 400, 3))  # float 범위를 넘는 수

# 검증
assert divide_numbers(42, 7) == 6 and isinstance(divide_numbers(42, 7), int)  # 6.0이 아닌 정수 6
assert divide_numbers(7, 2) == 3.5
assert divide_numbers(135, 0) == divide_numbers("a", 1) == divide_numbers(float("inf"), 1) == divide_numbers(float("nan"), 1) == divide_numbers(True, 2) == "잘못된 입력입니다."
with fake_inputs("4", echo=False):
    assert divide_numbers(10) == 2.5  # 첫 번째 값은 그대로 두고 두 번째만 입력
print("✅ 통과")
````

**실행 흐름**

1. **입력**: `"aaabbaaa"` · `"aabcccccaaa"`, 검증용 경계 사례 4개
2. **처리**: 이웃한 두 글자 비교 → 같으면 개수 + 1, 다르면 조각 저장 → 마지막 묶음 추가 → `groupby` 결과와 비교
3. **저장**: 압축 조각 목록 `result`
4. **출력**: 압축 문자열, 검증 결과

**핵심 개념**
- **튜플 언패킹**: `result, count = [], 1`은 두 변수에 한 번에 대입, `for prev, cur in ...`은 짝마다 두 값을 꺼냄
- **zip(s, s[1:])**: (0번, 1번), (1번, 2번) … 처럼 이웃한 두 글자를 짝지음
- **groupby**: 연속된 같은 값끼리 묶음, 묶음 `g`는 한 번만 읽을 수 있어 `len(list(g))`로 개수를 셈

````python
# 모듈 불러오기
from itertools import groupby  # 다른 방법으로 구한 답과 비교하기 위해 (검증용)

# 함수 정의
def compress_string(s: str) -> str:
    if not s:  # 빈 문자열에서 s[-1] 오류가 나지 않도록
        return ""
    result, count = [], 1  # 압축 조각과 현재 글자 반복 수를 기록하기 위해
    for prev, cur in zip(s, s[1:]):  # 이웃한 두 글자를 비교하기 위해
        if cur == prev:  # 같은 글자가 이어지면 개수만 늘리기 위해
            count += 1
        else:
            result.append(f"{prev}{count}")  # 묶음이 끝났으므로 "문자 + 개수"로 저장
            count = 1  # 새 묶음을 1부터 세기 위해
    result.append(f"{s[-1]}{count}")  # 반복문 뒤에 남은 마지막 묶음 추가
    return "".join(result)  # 문자열 덧셈 반복 대신 한 번에 이어 붙이기 위해

# 문제지 예시 1 · 2
print(compress_string("aaabbaaa"))
print(compress_string("aabcccccaaa"))

# 검증: itertools.groupby로 구한 결과와 비교
for case in ["aaabbaaa", "aabcccccaaa", "", "a", "abc", "aaaaaaaaaaaa"]:  # 경계 사례 · 두 자리 개수까지
    assert compress_string(case) == "".join(f"{ch}{len(list(g))}" for ch, g in groupby(case))
print("✅ 통과")
````

**결과**

```
✅ 통과           : 빈 문자열 · 한 글자 · 반복 없음 · 12번 반복까지 groupby 결과와 같음
```

### 2-2. 덧셈 프로그램

**요구사항** 두 숫자를 입력받아 합 출력, 숫자가 아니면 `"올바른 숫자를 입력하세요."` 출력 후 다시 입력

**설계**
- 두 번째 숫자가 틀리면 첫 번째부터 다시 → 문제지 입력 4줄(`10 → twenty → 10 → 20`)을 그대로 넣어 예상 출력 재현
- 정수와 소수 모두 숫자로 받고, `10.0`은 `10`으로 표시

**실행 흐름**

1. **입력**: 자동 입력 10 → twenty → 10 → 20, twenty → 10 → 20, 1.5 → 2
2. **처리**: 첫 번째는 숫자가 될 때까지 다시 질문 → 두 번째가 틀리면 안내 후 첫 번째부터 다시 → 두 수 더하기
3. **저장**: 없음
4. **출력**: 안내 문구, 덧셈 식, 검증 결과

**핵심 개념**
- **입력 함수 재사용**: 다시 묻는 반복은 0-1의 `read_number`, 한 번만 확인할 때는 `parse_number`

### 2-3. 카운트다운 타이머

**요구사항** 입력한 초부터 1초마다 남은 시간 출력, 끝나면 `"타이머 종료!"`

**설계**
- `range(seconds, 0, -1)`로 무한 루프 없이 감소, `time.sleep(1)`로 1초 간격
- `interval` 인자(기본 1초)로 검증 때는 대기 없이 실행, 입력 단계에서 음수는 다시 질문
- 함수에 직접 넣은 소수 · 음수는 1-5 · 1-6과 같은 기준으로 `ValueError`

**실행 흐름**

1. **입력**: 자동 입력 -1 → 5, 0초, 잘못된 인자 -3 · 2.5
2. **처리**: 음수 재질문 → 정수 검사 → 남은 시간 출력 + 1초 대기 반복 → 실제 걸린 시간 측정
3. **저장**: `count_down` (같은 함수의 다른 이름)
4. **출력**: 시작 · 남은 시간 · 종료 문구, 거부 안내, 검증 결과

**핵심 개념**
- **함수도 값**: `count_down = countdown_timer`처럼 다른 이름으로 같은 함수를 가리킬 수 있음
- **perf_counter**: 경과 시간 측정용 시계 (두 시점의 차이만 의미 있음)
- **형식 지정 `:.1f`**: `f"{elapsed:.1f}"`는 소수 첫째 자리까지 표시

````python
# 함수 정의 (time은 0-1에서 불러옴)
def countdown_timer(seconds: int, interval: float = 1) -> None:  # 문제지 실행 코드의 이름, interval은 검증 때 대기 없이 돌리기 위해
    require_non_negative_int(seconds)  # 1-5 · 1-6과 같은 0-1 공통 검사로 소수 · 음수 시간 거부
    print("타이머 시작!")
    for remaining in range(seconds, 0, -1):  # seconds부터 1까지 줄이기 위해
        print(remaining)  # 매초 남은 시간을 보이기 위해
        time.sleep(interval)
    print("타이머 종료!")
count_down = countdown_timer  # 문제 설명의 이름 count_down으로도 호출되도록

# 문제지 예시: 5초 (실제 대기 시간도 측정)
start = time.perf_counter()
with fake_inputs("-1", "5"):  # 음수는 다시 질문
    countdown_timer(read_number("타이머를 시작할 시간(초)을 입력하세요: ", minimum=0))  # 숫자가 아닌 입력 · 음수는 다시 질문
elapsed = time.perf_counter() - start

# 검증
assert elapsed >= 5  # 1초 간격 5번을 실제로 기다렸는지 확인 (느린 PC에서도 깨지지 않도록 하한만 검사)
assert printed(count_down, 0, 0) == "타이머 시작!\n타이머 종료!\n"  # 0초는 바로 종료
for bad in [-3, 2.5]:  # 음수 · 소수 거부
    try:
        countdown_timer(bad, 0)
    except ValueError as e:
        print("거부:", e)
print(f"실제 걸린 시간 {elapsed:.1f}초 · ✅ 통과")
````

## 3. 클래스 기초

### 3-1. 시간 추적 클래스 `TimeTracker`

**요구사항** `start` · `stop`으로 시작 · 종료 시각 기록, `stop`에서 경과 시간 계산, `get_elapsed_time`은 분 단위로 반환

**설계**
- `datetime.now()`로 시각 기록, `total_seconds()`로 하루를 넘는 시간까지 계산 (`.seconds`는 하루 넘는 부분을 버림)
- 90분을 실제로 기다리지 않도록 시작 시각을 1시간 30분 앞당겨 확인
- 시작 전 종료 · 두 번 종료는 안내만 하고 기록 유지, 측정 중 다시 시작하면 안내 후 새로 측정

<img src="images/m1_timer.png" alt="TimeTracker로 90분 경과 시간을 확인하는 순서" width="912">

**실행 흐름**

1. **입력**: 현재 시각
2. **처리**: 시작 시각 기록 → 시작 시각을 앞당김 → 종료 시각 기록 · (종료 - 시작) 초 // 60 계산 → 계산해 둔 분 반환
3. **저장**: `study_session` · `workout` · `long_session`의 시각과 경과 분
4. **출력**: 공부 · 운동 시간(분), 잘못된 순서 안내, 검증 결과

**핵심 개념**
- **클래스 · 객체**: `class TimeTracker:`는 설계도, `TimeTracker()`로 만든 것이 객체 → 만들 때 `__init__`(생성자)이 실행됨
- **메서드 · self**: 메서드는 클래스 안의 함수, `study_session.start()`처럼 부르면 `self`에 그 객체가 들어감
- **인스턴스 변수**: `self.start_time`은 객체마다 따로 있어 두 활동이 섞이지 않음
- **timedelta**: 시간 간격, `datetime`끼리 빼면 나오고 `total_seconds()`로 초 단위 변환 (`.seconds`는 하루를 넘는 부분을 버림)

````python
# 모듈 불러오기
from datetime import datetime, timedelta  # 현재 시각과 시간 간격을 다루기 위해

# 클래스 정의
class TimeTracker:
    def __init__(self) -> None:
        self.start_time: datetime | None = None  # 측정 전임을 None으로 구분하기 위해
        self.end_time: datetime | None = None
        self.elapsed_minutes = 0  # 마지막으로 계산한 경과 시간(분)

    # 측정 시작 · 종료
    def start(self) -> None:
        if self.start_time is not None and self.end_time is None:  # 측정 중 기록이 말없이 사라지지 않도록
            print("측정 중이던 기록을 버리고 다시 시작합니다.")
        self.start_time = datetime.now()  # 시작 시각 기록
        self.end_time, self.elapsed_minutes = None, 0  # 다시 시작하면 이전 기록이 남지 않도록
    def stop(self) -> None:
        if self.start_time is None:  # 시작 전 종료 방지
            print("start()를 먼저 호출하세요.")
            return
        if self.end_time is not None:  # 두 번 종료해 기록이 덮어써지지 않도록
            print("이미 종료된 측정입니다.")
            return
        self.end_time = datetime.now()  # 종료 시각 기록
        seconds = (self.end_time - self.start_time).total_seconds()  # .seconds와 달리 하루를 넘는 시간도 포함
        self.elapsed_minutes = max(0, int(seconds // 60))  # 문제지 요구대로 stop()에서 계산, 시계가 뒤로 가도 음수가 되지 않도록

    # 경과 시간
    def get_elapsed_time(self) -> int:
        return self.elapsed_minutes  # 마지막으로 기록된 시작 · 종료 사이의 분 (측정 전 · 측정 중은 0)

# 문제지 예시: 1시간 30분 경과 (기다리지 않도록 시작 시각을 앞당김)
study_session = TimeTracker()
study_session.start()
study_session.start_time -= timedelta(hours=1, minutes=30)
study_session.stop()
print("공부한 시간:", study_session.get_elapsed_time(), "분")

# 추가 사례: 다른 활동은 따로 측정, 잘못된 순서
workout = TimeTracker()
workout.start()
workout.start_time -= timedelta(minutes=45)
workout.start()  # 측정 중 다시 시작
workout.start_time -= timedelta(minutes=45)
workout.stop()
print("운동한 시간:", workout.get_elapsed_time(), "분")
workout.stop()
TimeTracker().stop()

# 검증
assert study_session.get_elapsed_time() == 90 and workout.get_elapsed_time() == 45  # 인스턴스마다 따로 기록
assert TimeTracker().get_elapsed_time() == 0  # 측정 전은 0분
long_session = TimeTracker()
long_session.start()
long_session.start_time -= timedelta(days=1, minutes=5)
long_session.stop()
assert long_session.get_elapsed_time() == 24 * 60 + 5  # 하루를 넘어도 1445분 (.seconds였다면 5분)
print("✅ 통과")
````

### 4-1. 투표 시스템 `VoteSystem`

**요구사항** 후보 등록 · 투표 · 결과 조회, 중복 등록과 미등록 후보 투표 금지, 사전으로 관리

**설계**
- 득표는 인스턴스 변수 사전(후보 → 득표 수) → 투표 시스템끼리 섞이지 않음
- 이름은 앞뒤 공백을 뺀 값으로 저장 · 비교 → `" Alice "`를 다른 후보로 등록하지 않음
- 1표일 때만 `vote`, 나머지는 `votes` (예상 출력 형식)

**실행 흐름**

1. **입력**: 후보 3명, 투표 3회, 잘못된 시도 4건
2. **처리**: 이름 정리 · 등록 여부 검사 후 등록 · 투표 → 거부된 시도는 집계 제외
3. **저장**: `candidates` 사전 (후보 → 득표 수)
4. **출력**: 처리 안내, 투표 결과, 검증 결과

**핵심 개념**
- **사전(dict)**: 키(후보 이름)로 값(득표 수)을 바로 찾고 바꿈, 넣은 순서 유지
- **@staticmethod**: `self`가 필요 없는 도우미 메서드 → 이름 정리 `_clean`처럼 객체 상태와 무관한 일
- **정리한 값으로 저장**: 검사에 쓴 `strip()` 결과를 그대로 저장 · 비교해야 `" Alice "`와 `"Alice"`가 같은 후보가 됨

````python
# 클래스 정의
class VoteSystem:
    def __init__(self) -> None:
        self.candidates: dict[str, int] = {}  # 문제지 요구대로 사전으로 관리 (후보 → 득표 수), 시스템마다 따로

    # 이름 정리
    @staticmethod
    def _clean(name) -> str | None:
        if not isinstance(name, str) or not name.strip():  # 빈 이름 · 문자열이 아닌 값은 None
            return None
        return name.strip()  # " Alice "와 "Alice"를 같은 후보로 보기 위해

    # 후보 등록
    def add_candidate(self, name: str) -> None:
        clean = self._clean(name)
        if clean is None:
            print(f"후보 이름이 올바르지 않습니다: {name!r}")
            return
        if clean in self.candidates:  # 같은 후보가 두 번 등록되지 않도록
            print(f"{clean} 후보는 이미 등록되어 있습니다.")
            return
        self.candidates[clean] = 0  # 0표에서 시작
        print(f"{clean} 후보가 성공적으로 등록되었습니다.")

    # 투표
    def vote(self, name: str) -> None:
        clean = self._clean(name)
        if clean is None:  # 등록 때와 같은 안내
            print(f"후보 이름이 올바르지 않습니다: {name!r}")
            return
        if clean not in self.candidates:  # 없는 후보 표가 집계되지 않도록
            print(f"{clean} 후보는 등록되지 않아 투표할 수 없습니다.")
            return
        self.candidates[clean] += 1
        print(f"{clean}에게 투표하였습니다.")

    # 결과 조회
    def get_results(self) -> None:
        print("투표 결과:")
        if not self.candidates:  # 4-4 · 5-3처럼 빈 목록도 안내
            print("- 등록된 후보 없음")
        for name, votes in self.candidates.items():  # 등록 순서대로 보이기 위해
            print(f"{name}: {votes} {'vote' if votes == 1 else 'votes'}")  # 1표일 때만 단수형

# 문제지 예시
voting_system = VoteSystem()
for name in ["Alice", "Bob", "Charlie"]:
    voting_system.add_candidate(name)
for name in ["Alice", "Alice", "Bob"]:
    voting_system.vote(name)
voting_system.get_results()

# 추가 사례: 중복 등록(공백 포함) · 빈 이름 · 미등록 후보 투표
voting_system.add_candidate(" Alice ")
voting_system.add_candidate(" ")
voting_system.vote("Dave")
voting_system.vote(None)
VoteSystem().get_results()  # 후보가 없을 때

# 검증
assert voting_system.candidates == {"Alice": 2, "Bob": 1, "Charlie": 0}  # 거부된 시도는 집계에서 빠짐
print("✅ 통과")
````

```
" Alice " · 빈 이름 · Dave · None : 공백만 다른 중복 · 빈 이름 · 미등록 후보 · 이름이 아닌 값 거부
후보 없는 시스템                  : "- 등록된 후보 없음" 안내
✅ 통과                           : 거부된 시도 뒤에도 득표 2 · 1 · 0 그대로
```

### 4-2. 은행 계좌 `BankAccount`

**요구사항** 입금 · 출금 · 잔액 조회, 잔액보다 많은 출금은 거부하고 경고, 금액은 정수 · 실수 모두 처리

**설계**
- 잔액은 `_balance`에 두고 입금 · 출금 메서드로만 변경
- 금액 검사를 `_is_valid_amount` 한곳에 모음: 숫자가 아님 · `True` · inf · 0 이하 · 소수 셋째 자리 이하 거부
- 잔액은 소수 둘째 자리로 정리하고 정수로 떨어지면 `int`로 표시 → 실수 오차가 쌓이지 않고 `130000.0` 같은 표기도 없음

<img src="images/m1_bank.png" alt="문제지 예시(개설 · 입금 · 출금), 거부된 4건, 실수 금액 입금 · 출금의 잔액 변화" width="900">

**실행 흐름**

1. **입력**: 초기 잔액 10만 원, 입금 · 출금 8건, 잘못된 개설 2건
2. **처리**: 금액 검사 → 잔액 초과 출금 거부 → 정상 거래만 반영 → 소수 둘째 자리 정리 · 정수면 int
3. **저장**: `_balance` (입금 · 출금 메서드로만 변경)
4. **출력**: 거래별 안내, 현재 잔액, 검증 결과

**핵심 개념**
- **앞 밑줄 `_balance`**: 클래스 밖에서 직접 바꾸지 말라는 약속
- **금액 검사 한곳에**: `_is_valid_amount` 하나를 입금 · 출금 · 개설이 함께 써서 기준이 어긋나지 않음
- **실수 오차**: `0.1 + 0.2`가 `0.30000000000000004`가 되므로 `round(값, 2)`로 정리
- **호출할 때의 `*`**: `BankAccount(*args)`는 튜플 `("000", "오류", -1)`을 인자 3개로 풀어 전달 (0-1의 `*values`는 반대로 여러 인자를 하나로 모음)

````python
# 클래스 정의
class BankAccount:
    def __init__(self, account_number: str, account_holder: str, initial_balance: int | float = 0) -> None:
        for value, label in [(account_number, "계좌번호"), (account_holder, "예금주")]:  # 빈 값으로 계좌가 만들어지지 않도록
            if not isinstance(value, str) or not value.strip():
                raise ValueError(f"{label}가 올바르지 않습니다: {value!r}")
        if not self._is_valid_amount(initial_balance, allow_zero=True):  # 초기 잔액은 0도 허용
            raise ValueError("초기 잔액은 0 이상, 소수 둘째 자리까지의 숫자여야 합니다.")  # 잘못된 계좌가 만들어지지 않도록
        self.account_number = account_number.strip()  # 4-1 · 4-4처럼 앞뒤 공백 정리
        self.account_holder = account_holder.strip()
        self._balance = self._tidy(initial_balance)  # 밖에서 직접 바꾸지 않고 입금 · 출금으로만 바꾸기 위해, 100.0은 100으로
        print(f"{self.account_holder}님의 계좌 {self.account_number}가 개설되었습니다. 초기 잔액: {self._balance}원")  # 정리된 값으로 안내 (100.0 → 100)

    # 금액 검사 · 정리
    @staticmethod
    def _is_valid_amount(amount, allow_zero: bool = False) -> bool:
        if not is_number(amount):  # True · 문자열 · inf · nan 거부 (0-1 공통 검사)
            return False
        if round(amount, 2) != amount:  # 0.001원처럼 표시할 수 없는 단위는 입금됐다고 안내한 뒤 사라지지 않도록 거부
            return False
        return amount >= 0 if allow_zero else amount > 0  # 상황에 따라 0 허용 여부를 바꾸기 위해
    @staticmethod
    def _tidy(value: float) -> int | float:
        value = round(value, 2)  # 0.1 + 0.2 같은 실수 오차가 잔액에 쌓이지 않도록
        return int(value) if value == int(value) else value  # 130000.0은 130000으로

    # 입금 · 출금 · 조회
    def deposit(self, amount: int | float) -> None:
        if not self._is_valid_amount(amount):
            print(f"입금 실패: 0보다 크고 소수 둘째 자리까지인 금액을 입력하세요. (입력: {amount!r})")
            return
        self._balance = self._tidy(self._balance + amount)
        print(f"{amount}원이 입금되었습니다. 현재 잔액: {self._balance}원")
    def withdraw(self, amount: int | float) -> None:
        if not self._is_valid_amount(amount):
            print(f"출금 실패: 0보다 크고 소수 둘째 자리까지인 금액을 입력하세요. (입력: {amount!r})")
        elif amount > self._balance:  # 문제지 요구: 잔액보다 많은 출금은 거부하고 경고
            print(f"출금 실패: 잔액이 부족합니다. (요청 {amount}원, 잔액 {self._balance}원)")
        else:
            self._balance = self._tidy(self._balance - amount)
            print(f"{amount}원이 출금되었습니다. 현재 잔액: {self._balance}원")
    def get_balance(self) -> int | float:
        return self._balance  # 밖에서는 읽기만 하도록

# 문제지 예시
my_account = BankAccount("123-456-789", "김철수", 100000)
my_account.deposit(50000)
my_account.withdraw(20000)
print(f"현재 잔액: {my_account.get_balance()}원")

# 추가 사례: 잔액 초과 · 음수 · 문자열 · 0.001원 거부, 실수 금액 입금 · 출금
my_account.withdraw(1_000_000)
my_account.deposit(-500)
my_account.deposit("1000")
my_account.deposit(0.001)
my_account.deposit(0.5)  # 문제지 요구: 실수 금액도 처리
my_account.withdraw(0.5)
print(f"현재 잔액: {my_account.get_balance()}원")

# 검증
assert my_account.get_balance() == 130000 and isinstance(my_account.get_balance(), int)  # 거부된 거래는 반영 안 됨, 130000.0이 아닌 130000
for args in [("000", "오류", -1), (None, "", 10)]:  # 음수 초기 잔액 · 빈 계좌번호
    try:
        BankAccount(*args)
    except ValueError as e:
        print("개설 거부:", e)  # 잘못된 값이면 계좌 자체를 만들지 않음
print("✅ 통과")
````

### 4-3. 직원 관리 `EmployeeManager`

**요구사항** 직원 정보를 클래스 변수 `employees`에 저장, 클래스 메서드로 평균 급여 계산 · 출력, 직원이 없으면 0 반환

**설계**
- 생성자에서 직원 객체 자체를 클래스 변수 `employees`에 추가
- 평균이 정수면 `55000`으로 표시 (예상 출력 형식), 잘못된 급여는 목록에 넣기 전에 거부

**실행 흐름**

1. **입력**: 직원 0명 → 직원 2명 (급여 50000 · 60000) → 잘못된 급여
2. **처리**: 클래스 변수 비우기 → 생성자에서 검사 후 자동 추가 → 클래스 메서드로 평균
3. **저장**: 클래스 변수 `employees`
4. **출력**: 추가 안내, 평균 급여, 거부 안내, 검증 결과

**핵심 개념**
- **클래스 변수**: 클래스에 하나만 있어 모든 인스턴스가 공유 (`EmployeeManager.employees`)
- **@classmethod · cls**: 객체 없이 클래스 이름으로 호출, 클래스 변수에 접근

````python
# 클래스 정의
class EmployeeManager:
    employees: list["EmployeeManager"] = []  # 문제지 요구: 모든 인스턴스가 공유하는 클래스 변수
    def __init__(self, name: str, salary: int | float) -> None:
        if not isinstance(name, str) or not name.strip():  # 빈 이름 거부
            raise ValueError(f"직원 이름이 올바르지 않습니다: {name!r}")
        if not is_number(salary) or salary < 0:  # 잘못된 급여가 평균에 섞이지 않도록 (0-1 공통 검사)
            raise ValueError(f"급여는 0 이상의 숫자여야 합니다: {salary!r}")
        self.name, self.salary = name.strip(), salary  # 4-1 · 4-4처럼 앞뒤 공백 정리
        EmployeeManager.employees.append(self)  # 이름 · 급여를 함께 보관하도록 직원 객체 자체를 저장
        print(f"{self.name} 님이 추가되었습니다. 급여: {salary}원")

    # 평균 급여
    @classmethod
    def calculate_average_salary(cls) -> int | float:
        if not cls.employees:  # 문제지 요구: 직원이 없으면 0 반환 (0으로 나누기 방지)
            return 0
        average = sum(emp.salary for emp in cls.employees) / len(cls.employees)
        average = round(average, 2)  # 소수 둘째 자리까지
        average = int(average) if average == int(average) else average  # 예상 출력처럼 55000.0이 아닌 55000으로
        print(f"전체 직원의 평균 급여: {average}원")
        return average

# 직원이 없을 때
EmployeeManager.employees.clear()  # 셀을 다시 실행해도 직원이 중복 저장되지 않도록
assert EmployeeManager.calculate_average_salary() == 0

# 문제지 예시
emp1 = EmployeeManager("홍길동", 50000)
emp2 = EmployeeManager("김철수", 60000)
average = EmployeeManager.calculate_average_salary()

# 추가 사례: 잘못된 급여는 추가되지 않음
for args in [("이영희", "6만"), (" ", 40000)]:  # 숫자가 아닌 급여 · 빈 이름
    try:
        EmployeeManager(*args)
    except ValueError as e:
        print("추가 거부:", e)

# 검증
assert average == 55000 and len(EmployeeManager.employees) == 2
print("✅ 통과")
````

```
'6만' · 빈 이름   : 숫자가 아닌 급여 · 이름 없는 직원은 추가 거부
```

### 4-4. 레스토랑 예약 `ReservationSystem` · 클래스

**요구사항** 지점별 예약 추가 · 취소 · 조회와 전체 지점 예약 수 합산(클래스 메서드), 각 메서드에 입력 검증과 예외 처리

**설계**
- 날짜는 `2024-05-20` · `2024/05/20` · `2024.05.20`을 모두 받아 `2024-05-20`으로 통일, 뒤에 시각(`18:30`)을 붙이면 일시로 저장 → 추가 · 취소가 같은 기준으로 비교
- 이름 · 인원 · 날짜 · 중복 예약 오류를 `ValueError` 하나로 모아 지점 이름과 함께 안내
- 날짜와 시각을 따로 저장(시각이 없으면 `None`), 시각이 없는 예약은 그날 하루 전체로 보고 같은 날 시각 예약과 겹친다고 판단
- 취소는 시각을 주면 그 시각 예약만, 날짜만 주면 그날 예약 (여러 건이면 시각을 요청)
- 합산은 지점 객체 목록만 받고, 같은 지점이 두 번 들어와도 한 번만 셈 (`dict.fromkeys`: 순서를 지키며 중복 제거)
- 지난 날짜는 허용: 문제지 예시가 2024년 날짜라 막으면 예시가 실패

**실행 흐름**

1. **입력**: 지점 이름, 예약 (이름 · 일시 · 인원)
2. **처리**: 이름 · 인원 · 일시 검사 → 일시를 (날짜, 시각)으로 나눠 통일 → 겹치는 예약 확인 → 예약 추가 · 취소 · 조회 · 합산
3. **저장**: `ReservationSystem` 클래스
4. **출력**: 없음 (다음 셀에서 실행)

**핵심 개념**
- **strptime · strftime**: 문자열 → 날짜(형식 지정) · 날짜 → 문자열, 형식이 맞지 않으면 ValueError
- **형식 코드**: `%Y` 연 · `%m` 월 · `%d` 일 · `%H` 시 · `%M` 분 → `"%Y-%m-%d %H:%M"`은 `2024-05-23 18:30`
- **중첩 for**: 날짜 표기 3가지 × (날짜만 · 날짜 + 시각) 조합을 차례로 시도
- **날짜 · 시각 따로 저장**: `_parse_when`이 `("2024-05-23", "18:30")`처럼 둘로 나눠 돌려주고, 시각이 없으면 `None` → `None`이면 그날 하루 전체와 겹친다고 판단
- **dict.fromkeys(목록)**: 순서를 유지하며 중복을 없앰 (같은 지점 두 번 방지)
- **리스트 컴프리헨션**: `[r for r in 목록 if 조건]`은 조건에 맞는 것만 모은 새 목록 (for · if · append를 한 줄로)

````python
# 클래스 정의 (datetime은 3-1에서 불러옴)
class ReservationSystem:
    DATE_FORMATS = ("%Y-%m-%d", "%Y/%m/%d", "%Y.%m.%d")  # 허용할 날짜 표기 3가지 (뒤에 " 18:00"처럼 시각을 붙여도 됨)
    def __init__(self, branch_name: str) -> None:
        if not isinstance(branch_name, str) or not branch_name.strip():  # 이름 없는 지점이 만들어지지 않도록
            raise ValueError("지점 이름은 비어 있지 않은 문자열이어야 합니다.")
        self.branch_name = branch_name.strip()
        self.reservations: list[dict] = []  # 예약 1건 = 딕셔너리 1개 (날짜 · 시각을 따로 저장)

    # 입력 검증
    @staticmethod
    def _check_name(name: str) -> str:
        if not isinstance(name, str) or not name.strip():  # 빈 이름 · 문자열이 아닌 값 거부
            raise ValueError(f"예약자 이름이 올바르지 않습니다: {name!r}")
        return name.strip()
    @classmethod
    def _parse_when(cls, when: str) -> tuple[str, str | None]:
        """2024/05/20 → ("2024-05-20", None), 2024.05.20 18:00 → ("2024-05-20", "18:00") (문제지의 "예약 일시")"""
        for fmt in cls.DATE_FORMATS:  # 표기를 하나씩 시도하기 위해
            for with_time in (False, True):  # 날짜만 · 날짜 + 시각
                try:
                    parsed = datetime.strptime(str(when).strip(), fmt + " %H:%M" if with_time else fmt)
                except ValueError:  # 이 표기가 아니면 다음 표기
                    continue
                return parsed.strftime("%Y-%m-%d"), parsed.strftime("%H:%M") if with_time else None
        raise ValueError(f"날짜 형식이 올바르지 않습니다: {when!r}")  # 모두 실패
    @staticmethod
    def _overlaps(r: dict, date: str, time: str | None) -> bool:
        if r["date"] != date:
            return False
        return r["time"] is None or time is None or r["time"] == time  # 시각이 없는 예약은 그날 하루 전체와 겹침
    @staticmethod
    def _label(date: str, time: str | None) -> str:
        return f"{date} {time}" if time else date  # 출력용 일시 문자열

    # 예약 추가 · 취소 · 조회
    def add_reservation(self, name: str, date: str, guests: int) -> None:
        try:  # 이름 · 인원 · 일시 오류를 한곳에서 안내하기 위해
            name = self._check_name(name)
            if isinstance(guests, bool) or not isinstance(guests, int) or guests <= 0:  # 인원은 1 이상의 정수만
                raise ValueError(f"인원 수는 1 이상의 정수여야 합니다: {guests!r}")
            day, time = self._parse_when(date)
            if any(r["name"] == name and self._overlaps(r, day, time) for r in self.reservations):  # 같은 사람 · 겹치는 일시 중복 방지
                raise ValueError(f"{name}님의 {self._label(day, time)} 예약이 이미 있습니다.")
        except ValueError as e:
            print(f"[{self.branch_name}] 예약 실패 - {e}")
            return
        self.reservations.append({"name": name, "date": day, "time": time, "guests": guests})  # 검사를 통과한 예약만 저장
    def cancel_reservation(self, name: str, date: str) -> None:
        try:
            name, (day, time) = self._check_name(name), self._parse_when(date)  # 추가할 때와 같은 기준으로 비교하기 위해
        except ValueError as e:
            print(f"[{self.branch_name}] 취소 실패 - {e}")
            return
        found = [r for r in self.reservations if r["name"] == name and r["date"] == day and (time is None or r["time"] == time)]  # 시각을 주면 그 시각 예약만, 날짜만 주면 그날 예약 전부
        if len(found) > 1:  # 날짜만 줬는데 그날 예약이 여러 건이면 어느 것인지 알 수 없음
            print(f"[{self.branch_name}] 취소 실패 - {name}님의 {day} 예약이 {len(found)}건이라 시각을 함께 입력하세요.")
            return
        if not found:
            print(f"[{self.branch_name}] 취소 실패 - {name}님의 {self._label(day, time)} 예약이 없습니다.")
            return
        self.reservations.remove(found[0])
        print(f"[{self.branch_name}] {name}님의 {self._label(found[0]['date'], found[0]['time'])} 예약이 취소되었습니다.")
    def list_reservations(self) -> None:
        print(f"{self.branch_name} 예약 목록:")
        if not self.reservations:  # 빈 목록도 결과가 보이도록
            print("- 예약 없음")
        for r in self.reservations:
            print(f"- {r['name']}, {self._label(r['date'], r['time'])}, {r['guests']}명")

    # 전체 예약 수
    @classmethod
    def sum_reservations(cls, branches: list["ReservationSystem"]) -> int:
        if not isinstance(branches, (list, tuple)) or not all(isinstance(branch, cls) for branch in branches):  # 지점 객체 목록만 받기 위해
            raise TypeError("ReservationSystem 객체 목록만 합산할 수 있습니다.")
        unique = list(dict.fromkeys(branches))  # 같은 지점을 두 번 넣어도 한 번만 세기 위해
        return sum(len(branch.reservations) for branch in unique)  # 지점 수가 아닌 예약 건수 합
````

### 4-4. 레스토랑 예약 `ReservationSystem` · 실행

**요구사항** 문제지 예시를 그대로 실행한 뒤 입력 검증 · 취소 시나리오 실행

**설계**
- 문제지 예시를 먼저 실행해 예상 출력(합계 2)과 바로 비교할 수 있게 함

**실행 흐름**

1. **입력**: 문제지 예시 2건, 추가 예약 8건 · 취소 3건 · 잘못된 합산 1건
2. **처리**: 예시 실행 → 입력 검증 · 시각이 있는 예약 · 날짜만 예약 → 취소 → 합산 거부 → 검증
3. **저장**: `restaurant1` · `restaurant2`
4. **출력**: 지점별 목록, 전체 예약 수, 실패 · 취소 안내, 검증 결과

**핵심 개념**
- **예시 먼저**: 문제지 예시를 그대로 먼저 실행해 예상 출력과 바로 비교할 수 있게 함
- **클래스 메서드 호출**: `ReservationSystem.sum_reservations([...])`처럼 객체 없이 클래스 이름으로 호출

````python
# 문제지 예시
restaurant1 = ReservationSystem("강남점")
restaurant2 = ReservationSystem("홍대점")
restaurant1.add_reservation("홍길동", "2024-05-20", 4)
restaurant2.add_reservation("김철수", "2024-05-21", 2)
restaurant1.list_reservations()
restaurant2.list_reservations()
total_reservations = ReservationSystem.sum_reservations([restaurant1, restaurant2])
print(f"전체 레스토랑 예약 수: {total_reservations}")
assert total_reservations == 2
print()  # 예시와 추가 사례의 출력을 구분하기 위해

# 추가 사례: 입력 검증
restaurant2.add_reservation("이영희", "2024/05/22", 3)  # 다른 날짜 표기 → 통일해서 저장
restaurant2.add_reservation("박민수", "5월 23일", 2)  # 잘못된 날짜
restaurant2.add_reservation("박민수", "2024-05-23", 0)  # 잘못된 인원
restaurant2.add_reservation(" ", "2024-05-23", 2)  # 빈 이름
restaurant2.add_reservation("김철수", "2024.05.21", 3)  # 같은 사람 · 같은 날짜 중복
restaurant1.add_reservation("박민수", "2024-05-23 12:00", 2)  # 시각이 있는 예약 일시
restaurant1.add_reservation("박민수", "2024/05/23 18:30", 4)  # 같은 날 다른 시각 → 별도 예약
restaurant1.add_reservation("박민수", "2024-05-23", 1)  # 날짜만 = 그날 하루 → 같은 날 시각 예약과 겹침
restaurant1.cancel_reservation("박민수", "2024-05-23")  # 그날 2건이라 시각을 함께 요청

# 추가 사례: 취소
restaurant2.cancel_reservation("이영희", "2024.05.22")  # 표기가 달라도 같은 날짜로 찾음
restaurant2.cancel_reservation("이영희", "2024-05-22")  # 이미 취소된 예약
restaurant2.list_reservations()
try:
    ReservationSystem.sum_reservations([restaurant1, "홍대점"])
except TypeError as e:
    print("합산 거부:", e)

# 검증
restaurant1.list_reservations()
assert ReservationSystem.sum_reservations([restaurant1, restaurant2]) == 4  # 실패한 추가 · 취소는 예약 수를 바꾸지 않음 (강남 3 + 홍대 1)
assert ReservationSystem.sum_reservations([restaurant1, restaurant1]) == 3  # 같은 지점은 한 번만
assert restaurant2.reservations == [{"name": "김철수", "date": "2024-05-21", "time": None, "guests": 2}]
assert [r["time"] for r in restaurant1.reservations] == [None, "12:00", "18:30"]  # 날짜 · 시각을 따로 저장
print("✅ 통과")
````

**실행 흐름**

2. **처리**: `DictWriter`로 머리글 + 4행 쓰기 → 저장된 바이트를 utf-8 · cp949로 읽어 보기
3. **저장**: `tasks.csv`
4. **출력**: 저장 인코딩

**핵심 개념**
- **인코딩**: `open()`에 `encoding`을 주지 않으면 OS 기본값(한국어 Windows는 cp949)으로 저장 → 읽는 쪽이 대비해야 함

````python
### 문제를 위한 파일 생성코드입니다. 문제를 풀기 전 이 코드를 반드시 실행해주세요.
import csv
# (시뮬레이션을 위한) 예시 파일 작성코드
def write_tasks_to_csv(filename, tasks):
    with open(filename, 'w', newline='') as csvfile:
        fieldnames = ['할일', '소요시간']
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for task, time in tasks:
            writer.writerow({'할일': task, '소요시간': time})

# 할 일과 소요 시간 데이터
tasks = [
    ('공부하기', 120),
    ('운동하기', 60),
    ('가계부정리', 30),
    ('영화보기', 150)
]

# CSV 파일로 저장
write_tasks_to_csv('tasks.csv', tasks)

# 저장 인코딩 확인 (제공 코드가 인코딩을 지정하지 않아 OS마다 다름)
with open("tasks.csv", "rb") as f:
    raw = f.read()
try:
    raw.decode("utf-8")
    saved_as = "utf-8"
except UnicodeDecodeError:  # 한국어 Windows 기본값이면 utf-8로 읽을 수 없음
    raw.decode("cp949")  # cp949로는 읽히는지 확인 (아니면 오류로 드러남)
    saved_as = "cp949 (한국어 Windows 기본값)"
print("tasks.csv 저장 인코딩:", saved_as)
````

### 5-1. 시간 관리 프로그램 · 함수

**요구사항** CSV의 할 일 중 남은 시간 안에 할 수 있는 일을 골라 소요 시간이 짧은 순으로 출력, 여러 조합이 가능할 때의 선택 기준 결정

**설계**
- **기준 1 · 할 일 개수 최대**: 짧은 일부터 담는 그리디로 최대 개수 `k` 결정 (짧은 일부터 담아야 가장 많이 담을 수 있음)
- **기준 2 · 남는 시간 최소**: 개수가 `k`인 조합 중 시간 합이 가장 큰 조합
- **기준 3 · 동점**: CSV에서 앞쪽에 적힌 할 일이 포함된 조합 = 가장 큰 시간 합인 조합 중 처음 만들어진 것 (`combinations`는 CSV 순서대로 조합을 만듦, 파일 순서 = 사용자 우선순위)
- 제공 코드는 인코딩 없이 저장 → 읽을 때 `utf-8-sig` · `cp949`를 차례로 시도 (`for · else`: 두 인코딩 모두 실패해 반복이 break 없이 끝날 때만 오류) (바로 위 셀 출력, 한국어 Windows는 cp949 · 그 밖의 OS는 대개 utf-8)
- 기준 1 · 2는 문제지의 "가능한 최대한 많은 할 일" · "시간을 최적화"에서 나옴, "특정 할 일 우선" · "긴 일 포함"은 CSV에 선호 정보가 없어 동점 처리(기준 3)에만 CSV 순서로 반영
- 문제지 틀의 `main()`은 고유 이름 `run_time_manager()`로 만들고 `main`으로도 연결

**실행 흐름**

1. **입력**: `tasks.csv` (5-1 제공 코드)
2. **처리**: 인코딩을 바꿔 가며 읽기 → 행마다 검사 → 기준 1(짧은 순으로 최대 개수 k) → 기준 2(k개 조합 중 시간 합 최대) → 입력 · 출력 → `main` 이름 연결
3. **저장**: 함수 3개 · `main`
4. **출력**: 없음 (다음 셀에서 실행)

**핵심 개념**
- **lambda**: `lambda t: t[1]`은 이름 없는 한 줄 함수 (t를 받아 t[1]을 돌려줌)
- **key=lambda**: `sorted(tasks, key=lambda t: t[1])`는 각 할 일의 두 번째 값(시간)으로 정렬, `max(..., key=...)`도 같은 방식으로 비교 기준을 정함
- **combinations(목록, k)**: 순서 없이 k개를 고른 모든 조합 (4개 중 2개 → 6가지), CSV 순서대로 만들어짐
- **동점 처리 두 단계**: 가장 큰 시간 합 `most`를 구한 뒤, 그 시간 합인 조합 중 첫 번째(= CSV 앞쪽)를 고름
- **`A or B`**: A가 빈 값 · None이면 B → 빈 칸 · 빈 입력의 기본값 (`row["할일"] or ""`)
- **그리디**: 매번 가장 짧은 일부터 담으면 담을 수 있는 개수가 최대
- **for · else**: 반복이 `break` 없이 끝났을 때만 `else` 실행 (두 인코딩 모두 실패)
- **except 순서**: `FileNotFoundError`는 `OSError`의 한 종류라 더 구체적인 쪽을 먼저 써야 "파일 없음"이 따로 안내됨

````python
# 모듈 불러오기
import os  # 시연용 파일 삭제
from itertools import combinations  # 개수가 k인 조합을 모두 만들기 위해

# 함수 정의
def load_tasks(filename: str) -> list[tuple[str, int]]:
    for encoding in ("utf-8-sig", "cp949"):  # 제공 코드는 인코딩을 지정하지 않아 OS마다 저장 방식이 다르므로 둘 다 시도
        try:
            with open(filename, newline="", encoding=encoding) as f:
                reader = csv.DictReader(f)  # 열 순서가 바뀌어도 읽히도록 머리글 이름 사용
                header, rows = reader.fieldnames or [], list(reader)
            break
        except UnicodeDecodeError:
            continue
    else:
        raise ValueError("파일 인코딩을 읽을 수 없습니다.")
    if "할일" not in header or "소요시간" not in header:  # 행이 없어도 머리글부터 확인
        raise ValueError("머리글 '할일,소요시간'이 필요합니다.")
    if not rows:  # 머리글만 있는 파일은 "시간 부족"과 구분해 안내
        raise ValueError("머리글만 있고 할 일 행이 하나도 없습니다.")
    tasks = []
    for line, row in enumerate(rows, start=2):  # 1번째 줄은 머리글
        name, minutes = (row["할일"] or "").strip(), (row["소요시간"] or "").strip()
        if not name or not minutes.isdecimal() or int(minutes) <= 0:  # 이름 없음 · 숫자 아님(³ 같은 위첨자 포함) · 0분 이하 거부
            raise ValueError(f"{line}번째 줄을 읽을 수 없습니다 (할일 {name!r}, 소요시간 {minutes!r})")
        tasks.append((name, int(minutes)))
    return tasks
def suggest_tasks(tasks: list[tuple[str, int]], remaining_time: int) -> list[tuple[str, int]]:
    k, used = 0, 0  # 기준 1: 담은 개수와 사용 시간
    for _, duration in sorted(tasks, key=lambda t: t[1]):  # 개수를 최대로 하려면 짧은 일부터 담음
        if used + duration > remaining_time:
            break
        used += duration
        k += 1
    if k == 0:  # 가장 짧은 일도 못 하면 빈 목록
        return []
    feasible = [c for c in combinations(tasks, k) if sum(d for _, d in c) <= remaining_time]  # 기준 2: 시간 안에 드는 k개 조합만
    most = max(sum(d for _, d in c) for c in feasible)  # 기준 2: 쓸 수 있는 가장 많은 시간
    best = [c for c in feasible if sum(d for _, d in c) == most][0]  # 기준 3: 동점이면 먼저 만들어진 조합 = CSV 앞쪽 할 일 (combinations는 CSV 순서대로 만듦)
    return sorted(best, key=lambda t: t[1])  # 문제지 형식대로 짧은 순 정렬
def run_time_manager() -> None:  # 문제지 틀의 main(), 5-2의 main()과 이름이 겹치지 않도록 이름만 바꿈
    filename = input("CSV 파일 경로: ").strip() or "tasks.csv"  # 빈 입력이면 기본 파일
    try:
        tasks = load_tasks(filename)
    except FileNotFoundError:  # 없는 파일에도 프로그램이 멈추지 않도록
        print(f"파일을 찾을 수 없습니다: {filename}")
        return
    except OSError as e:  # 폴더 경로 · 권한 문제
        print(f"파일을 열 수 없습니다: {e}")
        return
    except ValueError as e:  # 머리글 누락 · 빈 칸 · 숫자 아님 (load_tasks가 한국어로 안내)
        print(f"CSV 형식이 올바르지 않습니다: {e}")
        return
    remaining_time = read_number("남은 시간(분): ", minimum=0)  # 숫자가 아닌 입력 · 음수는 다시 질문
    selected = suggest_tasks(tasks, remaining_time)
    if not selected:
        print("남은 시간 안에 할 수 있는 일이 없습니다.")
        return
    print("선택된 할 일 목록:")
    for i, (name, duration) in enumerate(selected, start=1):  # 1번부터 번호
        print(f"{i}. {name} - 예상 소요 시간: {duration}분")
main = run_time_manager  # 문제지 실행 코드의 main()으로도 호출되도록
````

**실행 흐름**

1. **입력**: 자동 입력 (파일 경로, 남은 시간 210 · 200 · -5 → 20), 없는 파일 · 폴더 · 잘못된 CSV
2. **처리**: 문제지 실행 코드 `main()` → `run_time_manager()` 6번 실행 (200분 1회 + 음수 · 파일 오류 5회) → 문제지 기준 4가지를 200분에 적용 → 동점 처리 · 전체 조합 비교 검증
3. **저장**: `loaded` · `every200` · `options`
4. **출력**: 선택된 할 일 목록, 오류 안내, 기준 비교, 검증 결과

**핵심 개념**
- **`if __name__ == "__main__":`**: 파일을 직접 실행할 때만 참, 노트북에서도 참이라 문제지 실행 코드가 그대로 동작
- **`*answers`**: 4-2의 `*args`와 같이 튜플을 인자 여러 개로 풀어 전달
- **for 두 번 쓰는 컴프리헨션**: `[c for k in ... for c in combinations(loaded, k) if ...]`은 `for k` 안에 `for c`가 있는 이중 반복문과 같음 (0개 ~ 4개 조합 전부)
- **전체 탐색으로 검증**: 할 일 4개는 조합이 16개뿐이라 모두 비교한 답을 정답으로 사용
- **튜플 비교**: `(개수, 시간 합)`은 앞 값부터 비교 → 개수가 같을 때만 시간 합을 봄, 동점 조합은 CSV 위치 목록이 가장 작은 것을 `min`으로 따로 구해 검증

````python
# 이 셀에서 쓰는 할 일 목록 (제공 코드의 전역 tasks와 섞이지 않도록 다른 이름)
loaded = load_tasks("tasks.csv")

# 문제지 예시: 남은 시간 210분 (문제지 실행 코드 그대로)
with fake_inputs("tasks.csv", "210"):
    if __name__ == "__main__":
        main()

# 추가 사례: 기준 2가 그리디와 다른 답을 고르는 200분
with fake_inputs("tasks.csv", "200"):
    run_time_manager()

# 추가 사례: 음수 시간 · 할 수 있는 일 없음 · 파일 오류 4종
with open("bad_tasks.csv", "w", newline="", encoding="utf-8") as f:
    f.write("할일,소요시간\n산책,30분\n")  # 소요시간이 숫자가 아닌 행
with open("empty_tasks.csv", "w", newline="", encoding="utf-8") as f:
    f.write("할일,소요시간\n")  # 머리글만 있는 파일
for answers in [("tasks.csv", "-5", "20"), ("none.csv",), (".",), ("bad_tasks.csv",), ("empty_tasks.csv",)]:  # 음수 재질문 후 할 일 없음 · 없는 파일 · 폴더 경로 · 잘못된 CSV · 빈 CSV
    with fake_inputs(*answers):
        run_time_manager()
os.remove("bad_tasks.csv")  # 시연용 파일은 남기지 않기 위해
os.remove("empty_tasks.csv")

# 문제지가 제시한 기준 4가지를 200분에 적용해 비교
every200 = [c for k in range(len(loaded) + 1) for c in combinations(loaded, k) if sum(d for _, d in c) <= 200]  # 200분 안에 드는 모든 조합
greedy, used = [], 0
for task in sorted(loaded, key=lambda t: t[1]):  # 짧은 일부터 담기
    if used + task[1] <= 200:
        greedy.append(task)
        used += task[1]
options = {
    "① 개수 최대 (짧은 순으로 담기)": greedy,
    "② 남는 시간 최소": max(every200, key=lambda c: sum(d for _, d in c)),
    "③ 특정 일 우선 (공부하기)": max([c for c in every200 if ("공부하기", 120) in c], key=lambda c: sum(d for _, d in c)),
    "④ 긴 일 포함 (영화보기)": max([c for c in every200 if ("영화보기", 150) in c], key=lambda c: sum(d for _, d in c)),
    "채택: ① 개수 → ② 시간 → CSV 순서": suggest_tasks(loaded, 200),
}
for label, combo in options.items():
    print(f"{label}: {' + '.join(n for n, _ in combo)} = {sum(d for _, d in combo)}분 · {len(combo)}개")

# 검증: 200분 동점 처리, 채택 결과 = 모든 조합을 직접 비교한 결과
assert len([c for c in every200 if len(c) == 2 and sum(d for _, d in c) == 180]) == 2  # 200분에서 2개 · 180분인 조합이 실제로 2개 (동점)
assert suggest_tasks(loaded, 200) == [("운동하기", 60), ("공부하기", 120)]  # 동점이면 CSV에서 앞쪽인 공부하기 · 운동하기 쪽 선택
positions = {task: i for i, task in enumerate(loaded)}  # 할 일 → CSV 위치 (0부터)
for t in range(401):  # 0 ~ 400분, 1분 간격 전부
    every = [c for k in range(len(loaded) + 1) for c in combinations(loaded, k) if sum(d for _, d in c) <= t]
    top = max((len(c), sum(d for _, d in c)) for c in every)  # 개수 → 시간 합의 최댓값
    tied = [c for c in every if (len(c), sum(d for _, d in c)) == top]  # 최댓값이 같은 조합들
    expected = min(tied, key=lambda c: sorted(positions[x] for x in c))  # 그중 CSV 위치가 가장 앞선 조합
    assert suggest_tasks(loaded, t) == sorted(expected, key=lambda x: x[1])  # 고른 조합까지 같은지
print("✅ 통과")
````

### 5-2. 수학 퍼즐: 소수의 합

**요구사항** 입력한 수 이하 소수의 합 출력, 효율적인 방법 사용

**설계**
- 에라토스테네스의 체: i의 배수를 슬라이스 대입으로 한 번에 지움, √n까지만 반복 (O(n log log n))
- 검증용 `is_prime`(√num까지 나눠 보기)으로 같은 답인지 교차 확인
- 문제지 틀의 `main()`은 고유 이름 `run_prime_sum()`으로 만들고 `main`으로도 연결, 음수 범위는 다시 질문

**실행 흐름**

1. **입력**: 자동 입력 ten → -5 → 10, n = 10만 · 200만
2. **처리**: 문제지 실행 코드 `main()` → 0 ~ n을 소수 후보로 표시 → 2부터 √n까지 소수의 배수 지우기 → 남은 수 합계 → 하나씩 판별과 비교
3. **저장**: 체 목록 `sieve`
4. **출력**: 소수 합 문장, 속도 비교, 200만 이하 합, 검증 결과

**핵심 개념**
- **에라토스테네스의 체**: 작은 소수부터 그 배수를 지우면 남는 수가 소수
- **`a = b = 값`**: `sieve[0] = sieve[1] = False`처럼 두 곳에 같은 값을 한 번에 대입
- **i × i부터 지우기**: i보다 작은 배수(2i, 3i …)는 더 작은 소수를 처리할 때 이미 지워졌으므로 i²부터
- **√n까지만**: n 이하 합성수는 √n 이하의 약수를 반드시 가짐
- **간격 슬라이스 대입**: `a[s::k] = 목록`은 오른쪽 목록 길이가 바뀔 칸 수와 같아야 하므로 `len(range(s, n + 1, k))`로 개수를 맞춤
- **0이 아닌 수는 True**: `all(num % i for ...)`는 나머지가 하나라도 0(False)이면 False → 나누어떨어지는 수가 없을 때만 소수

````python
# 소수 판별 함수 (검증용)
def is_prime(num: int) -> bool:
    """√num까지만 나눠 보면 소수 여부 확인 가능"""
    if num < 2:  # 0 · 1은 소수가 아님
        return False
    return all(num % i for i in range(2, int(num ** 0.5) + 1))

# 에라토스테네스의 체
def sum_of_primes(n: int) -> int:
    """에라토스테네스의 체: O(n log log n)"""
    require_non_negative_int(n)  # 1-5와 같은 0-1 공통 검사
    if n < 2:  # 소수가 없는 범위
        return 0
    sieve = [True] * (n + 1)  # 인덱스를 수로 쓰기 위해 n + 1칸 (True = 소수 후보)
    sieve[0] = sieve[1] = False
    for i in range(2, int(n ** 0.5) + 1):  # √n 이후의 배수는 이미 지워졌으므로 √n까지만
        if sieve[i]:  # 아직 남아 있으면 소수 → 배수를 지움
            sieve[i * i :: i] = [False] * len(range(i * i, n + 1, i))  # i의 배수를 한 번에 지우기 위해 슬라이스 대입
    return sum(i for i, prime in enumerate(sieve) if prime)

# 실행 함수
def run_prime_sum() -> None:  # 문제지 틀의 main(), 5-1과 이름이 겹치지 않도록 이름만 바꿈
    n = read_number("어떤 수까지의 소수의 합을 구하시겠습니까?: ", minimum=0)  # 숫자가 아닌 입력 · 음수는 다시 질문
    print(f"{n} 이하 소수의 합은 {sum_of_primes(n)}입니다.")
main = run_prime_sum  # 문제지 실행 코드의 main()으로도 호출되도록

# 문제지 예시: 10 (문제지 실행 코드 그대로, 앞에 숫자가 아닌 입력 · 음수)
with fake_inputs("ten", "-5", "10"):
    if __name__ == "__main__":
        main()

# 효율 비교: n = 10만에서 체 vs 하나씩 판별 (시간은 실행마다 조금 다름)
start = time.perf_counter()
by_sieve = sum_of_primes(100_000)
t_sieve = time.perf_counter() - start
start = time.perf_counter()
by_trial = sum(x for x in range(100_001) if is_prime(x))
t_trial = time.perf_counter() - start
print(f"n = 100,000 · 체 {t_sieve:.3f}초 · 하나씩 판별 {t_trial:.3f}초 · 결과 같음: {by_sieve == by_trial}")

# 큰 입력: 200만 이하 (널리 알려진 값 142,913,828,922와 비교)
start = time.perf_counter()
big = sum_of_primes(2_000_000)
elapsed = time.perf_counter() - start
print(f"200만 이하 소수의 합: {big:,} ({elapsed:.2f}초)")

# 검증
assert big == 142_913_828_922  # 널리 알려진 값과 일치
assert by_sieve == by_trial  # 두 방법의 답이 같음 (속도는 PC 부하에 따라 달라 출력으로만 비교)
assert all(sum_of_primes(n) == sum(x for x in range(n + 1) if is_prime(x)) for n in range(500))  # n = 0 ~ 499 하나씩 판별한 합과 일치
print("✅ 통과")
````

### 5-3. 도서관 관리 시스템 · 예외와 정보 클래스

**요구사항** `Book` · `Member` · `Rental` · `LibraryManagement` 구현, 모든 클래스 · 메서드에 입력 검증과 예외 처리, 도서 · 회원이 없을 때 등 상황별 피드백

**설계**
- 오류 종류마다 `LibraryError`를 상속한 예외 → 메서드는 오류를 던지기만 하고, `@report_errors`가 `"오류: ..."`로 안내 (메서드마다 같은 try · except 반복 방지)
- `require_text`로 빈 값 · 문자열이 아닌 값 검사를 한곳에서 처리, 출판년도는 1 ~ 올해만 허용
- `Book`은 ISBN 기호만 입력 거부, `Member`는 회원번호(1 이상 정수) · 빌리지 않은 책 반납 거부, `Rental`은 도서 · 회원 객체인지 · 두 번 반납 거부, 문제지대로 회원 ID · 도서 ISBN을 함께 기록, 반납 예정일 = 대여일 + 14일

<img src="images/m1_library.png" alt="예외 5개 · 정보 클래스 3개 · 관리 클래스의 관계" width="837">

**실행 흐름**

1. **입력**: 없음
2. **처리**: 오류 종류별 예외 클래스 → 예외를 안내 문구로 바꾸는 데코레이터 → ISBN 정리 · 빈 값 검사 함수 → `Book` · `Member` · `Rental`
3. **저장**: 예외 5개 · `report_errors` · `isbn_key` · `require_text` · 정보 클래스 3개
4. **출력**: 없음 (정의 셀)

**핵심 개념**
- **사용자 정의 예외**: `Exception`을 상속해 오류 종류에 이름을 붙임 → 부모 `LibraryError` 하나로 모두 잡을 수 있음
- **데코레이터**: 함수를 받아 기능을 덧붙인 함수를 돌려줌 → `@report_errors` 한 줄로 메서드마다 같은 오류 처리
- **\*args · \*\*kwargs · @wraps**: 받은 인자를 그대로 넘겨 어떤 메서드에도 씌울 수 있고, `@wraps`로 감싼 뒤에도 원래 메서드 이름이 유지됨
- **getattr(객체, 이름, 기본값)**: 속성이 없으면 기본값 → `give_back`에 Book이 아닌 값이 와도 오류 문구에 표시
- **역할 나누기**: 대여 목록 변경은 `Member.borrow` · `give_back`, 반납 처리는 `Rental.close` (문제지가 정한 클래스별 역할)

````python
# 모듈 불러오기
from datetime import date  # 대여일 (timedelta는 3-1에서 불러옴)
from functools import wraps  # 데코레이터를 씌워도 원래 메서드 이름이 유지되도록

# 사용자 정의 예외
class LibraryError(Exception):  # 도서관 오류만 골라 처리하기 위해 공통 부모
    """도서관 예외의 부모 클래스"""
class BookNotFoundError(LibraryError):
    def __init__(self, isbn):
        super().__init__(f"ISBN '{isbn}' 도서를 찾을 수 없습니다.")
class BookNotAvailableError(LibraryError):
    def __init__(self, title):
        super().__init__(f"'{title}' 도서는 모두 대여 중입니다.")
class MemberNotFoundError(LibraryError):
    def __init__(self, name):
        super().__init__(f"회원 '{name}'님을 찾을 수 없습니다.")
class NotRentedError(LibraryError):
    def __init__(self, name, title):
        super().__init__(f"'{name}' 회원님은 '{title}' 도서를 대여하고 있지 않습니다.")

# 오류 안내 데코레이터
def report_errors(method):  # 메서드마다 try를 반복하지 않도록
    """실행 중 생긴 LibraryError를 '오류: ...'로 안내"""
    @wraps(method)
    def wrapper(*args, **kwargs):  # 어떤 메서드에도 씌울 수 있도록 인자를 그대로 전달
        try:
            return method(*args, **kwargs)
        except LibraryError as e:  # 도서관 오류만 안내하고 다른 오류는 그대로 드러나도록
            print(f"오류: {e}")
    return wrapper

# 입력 검사 함수
def isbn_key(isbn) -> str:
    key = require_text(isbn, "ISBN").replace("-", "").replace(" ", "")  # 978-0451524935와 9780451524935를 같은 책으로
    if not key:  # "---"처럼 기호만 있는 값 거부
        raise LibraryError(f"ISBN 값이 올바르지 않습니다: {isbn!r}")
    return key
def require_text(value, label: str) -> str:
    if not isinstance(value, str) or not value.strip():  # 빈 값 · 문자열이 아닌 값 거부
        raise LibraryError(f"{label} 값이 올바르지 않습니다: {value!r}")
    return value.strip()

# 도서 · 회원 · 대여 기록
class Book:
    def __init__(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        self.title, self.author = require_text(title, "제목"), require_text(author, "저자")
        is_year = isinstance(publication_year, int) and not isinstance(publication_year, bool)
        if not is_year or not 1 <= publication_year <= date.today().year:  # 숫자가 아닌 연도 · 미래 · 0 이하 연도 거부
            raise LibraryError(f"출판년도가 올바르지 않습니다: {publication_year!r}")
        isbn_key(isbn)  # 빈 값 · 기호만 있는 ISBN은 Book에서도 거부
        self.publication_year, self.isbn = publication_year, isbn.strip()
        self.total_copies = self.available_copies = 1  # 보유 권수와 대여 가능 권수를 따로 세기 위해
    def __str__(self) -> str:
        return f"{self.title} (저자: {self.author}, 출판년도: {self.publication_year})"
class Member:
    def __init__(self, name: str, member_id: int) -> None:
        if isinstance(member_id, bool) or not isinstance(member_id, int) or member_id < 1:  # 회원번호는 1 이상의 정수
            raise LibraryError(f"회원번호가 올바르지 않습니다: {member_id!r}")
        self.name, self.member_id = require_text(name, "회원 이름"), member_id
        self.rented_books: list[Book] = []  # 대여 중인 도서 목록
    def borrow(self, book: Book) -> None:  # 문제지: 회원별 대여 목록은 Member가 관리
        if not isinstance(book, Book):  # 도서가 아닌 값이 목록에 섞이지 않도록
            raise LibraryError("대여 목록에는 도서만 넣을 수 있습니다.")
        if book in self.rented_books:  # 같은 책 중복 방지
            raise LibraryError(f"'{self.name}' 회원님은 이미 '{book.title}' 도서를 대여 중입니다.")
        self.rented_books.append(book)
    def give_back(self, book: Book) -> None:
        if book not in self.rented_books:  # 빌리지 않은 책은 목록에서 뺄 수 없음
            raise NotRentedError(self.name, getattr(book, "title", book))
        self.rented_books.remove(book)
    def __str__(self) -> str:
        titles = ", ".join(b.title for b in self.rented_books) or "없음"
        return f"{self.name} (대여 중인 도서: {titles})"
class Rental:
    LOAN_DAYS = 14  # 대여 기간을 한곳에서 바꿀 수 있도록
    def __init__(self, book: Book, member: Member) -> None:
        if not isinstance(book, Book) or not isinstance(member, Member):  # 도서 · 회원 객체로만 대여 기록 생성
            raise LibraryError("대여 기록에는 도서와 회원 정보가 필요합니다.")
        self.book, self.member = book, member
        self.isbn, self.member_id = book.isbn, member.member_id  # 문제지 요구: 회원 ID와 도서 ISBN 연결
        self.rental_date = date.today()
        self.due_date = self.rental_date + timedelta(days=self.LOAN_DAYS)  # 반납 예정일
        self.return_date: date | None = None  # 반납 전임을 None으로 구분하기 위해
    def close(self) -> None:  # 문제지: 반납 처리는 Rental이 담당
        if self.return_date is not None:  # 두 번 반납해 반납일이 덮어써지지 않도록
            raise LibraryError(f"'{self.book.title}' 대여 기록은 이미 반납 처리되었습니다.")
        self.return_date = date.today()
````

### 5-3. 도서관 관리 시스템 · 관리 클래스

**요구사항** 도서 추가 · 삭제 · 검색, 회원 등록 · 정보 조회, 대여 · 반납

**설계**
- ISBN은 하이픈 · 공백을 뺀 값으로 비교 → `978-0451524935`와 `9780451524935`는 같은 책
- 같은 ISBN은 권수만 증가, ISBN이 같은데 제목 · 저자 · 출판년도가 다르면 거부
- 도서 · 회원이 없으면 목록에 `등록된 도서 없음` · `등록된 회원 없음` 안내 (문제지 피드백 예시)
- `search_books`는 오류여도 빈 목록을 돌려주도록 데코레이터 대신 직접 처리 → 반환 자료형 일정
- 대여 · 반납은 모든 검사를 통과한 뒤에만 재고 · 기록 변경 → 실패한 시도가 상태를 바꾸지 않음
- 대여 목록 변경은 `Member.borrow` · `give_back`, 반납 처리는 `Rental.close`가 담당 (문제지의 클래스별 역할)

**실행 흐름**

1. **입력**: 없음
2. **처리**: 도서 · 회원 조회 도우미 → 도서 추가 · 삭제 · 검색 · 목록(빈 목록 안내) → 회원 등록 · 조회 · 목록 → 대여 · 반납
3. **저장**: `LibraryManagement` 클래스
4. **출력**: 없음 (정의 셀)

**핵심 개념**
- **검사 후 변경**: 대여 · 반납은 모든 검사가 끝난 뒤에만 재고 · 기록을 바꿔, 실패한 시도가 상태를 망가뜨리지 않음
- **is**: 값이 같은지가 아니라 같은 객체인지 비교
- **튜플 비교 !=**: (제목, 저자, 출판년도) 셋 중 하나라도 다르면 다른 책

````python
# 도서관 관리 클래스 (문제지 틀의 LibraryManagement)
class LibraryManagement:
    def __init__(self) -> None:
        self.books: dict[str, Book] = {}  # ISBN으로 바로 찾기 위해
        self.members: dict[str, Member] = {}  # 이름으로 바로 찾기 위해
        self.rentals: list[Rental] = []  # 대여 이력을 남기기 위해 반납된 기록까지 보관

    # 내부 도우미: 찾지 못하면 예외
    def _get_book(self, isbn: str) -> Book:
        key = isbn_key(isbn)  # None · 숫자 · 빈 값도 안내 문구로, 하이픈 · 공백은 무시
        if key not in self.books:
            raise BookNotFoundError(isbn)
        return self.books[key]
    def _get_member(self, name: str) -> Member:
        name = require_text(name, "회원 이름")
        if name not in self.members:
            raise MemberNotFoundError(name)
        return self.members[name]
    def _active_rental(self, book: Book, member: Member) -> Rental | None:
        for r in self.rentals:  # 반납 전 대여 기록 찾기
            if r.book is book and r.member is member and r.return_date is None:
                return r
        return None

    # 도서 추가 · 삭제 · 검색 · 목록
    @report_errors
    def add_book(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        book = Book(title, author, publication_year, isbn)  # 잘못된 도서 정보는 여기서 예외
        key = isbn_key(book.isbn)
        if key in self.books:  # 같은 책이면 권수만 늘리기 위해
            saved = self.books[key]
            if (saved.title, saved.author, saved.publication_year) != (book.title, book.author, book.publication_year):  # ISBN이 같은데 다른 책이면 거부
                raise LibraryError(f"ISBN '{book.isbn}'에 등록된 도서 정보({saved})와 다릅니다.")
            saved.total_copies += 1
            saved.available_copies += 1
            print(f"'{saved.title}' 도서가 1권 더 추가되었습니다. (보유 {saved.total_copies}권)")
            return
        self.books[key] = book
        print(f"'{book.title}' (저자: {book.author}, 출판년도: {book.publication_year}) 도서가 추가되었습니다.")
    @report_errors
    def remove_book(self, isbn: str) -> None:
        book = self._get_book(isbn)
        if book.available_copies < book.total_copies:  # 대여 중인 책이 사라지지 않도록
            raise LibraryError(f"'{book.title}' 도서는 대여 중이라 삭제할 수 없습니다.")
        del self.books[isbn_key(book.isbn)]
        print(f"'{book.title}' 도서가 삭제되었습니다.")
    def search_books(self, keyword: str) -> list[Book]:
        try:
            keyword = require_text(keyword, "검색어")  # 빈 검색어가 모든 책을 돌려주지 않도록
        except LibraryError as e:  # 오류여도 반환 자료형이 list로 같도록 데코레이터 대신 직접 처리
            print(f"오류: {e}")
            return []
        found = [b for b in self.books.values() if keyword in b.title or keyword in b.author]  # 제목 · 저자 모두 검색
        print(f"'{keyword}' 검색 결과: {len(found)}건")
        for b in found:
            print(f"- {b} [대여 가능 {b.available_copies}/{b.total_copies}권]")
        return found
    def print_books(self) -> None:
        print("도서 목록:")
        if not self.books:  # 문제지 요구: 도서가 없을 때도 안내
            print("- 등록된 도서 없음")
        for book in self.books.values():
            print(f"- {book}")

    # 회원 등록 · 조회 · 목록
    @report_errors
    def add_member(self, name: str) -> None:
        name = require_text(name, "회원 이름")
        if name in self.members:  # 같은 회원이 두 번 등록되지 않도록
            raise LibraryError(f"회원 '{name}'님은 이미 등록되어 있습니다.")
        self.members[name] = Member(name, len(self.members) + 1)  # 등록 순서를 회원번호로
        print(f"회원 '{name}'님이 등록되었습니다.")
    @report_errors
    def find_member(self, name: str) -> None:
        member = self._get_member(name)
        print(f"회원번호 {member.member_id} · {member}")
    def print_members(self) -> None:
        print("회원 목록:")
        if not self.members:  # 문제지 요구: 회원 정보가 없을 때도 안내
            print("- 등록된 회원 없음")
        for member in self.members.values():
            print(f"- {member}")

    # 대여 · 반납
    @report_errors
    def rent_book(self, isbn: str, member_name: str) -> None:
        book, member = self._get_book(isbn), self._get_member(member_name)  # 없는 도서 · 회원은 예외로 안내
        if self._active_rental(book, member):  # 같은 책을 두 번 빌리지 않도록
            raise LibraryError(f"'{member.name}' 회원님은 이미 '{book.title}' 도서를 대여 중입니다.")
        if book.available_copies == 0:
            raise BookNotAvailableError(book.title)
        book.available_copies -= 1  # 모든 검사를 통과한 뒤에만 상태를 바꾸기 위해
        member.borrow(book)
        self.rentals.append(Rental(book, member))
        print(f"'{member.name}' 회원님이 '{book.title}' 도서를 대여하였습니다.")
    @report_errors
    def return_book(self, isbn: str, member_name: str) -> None:
        book, member = self._get_book(isbn), self._get_member(member_name)
        rental = self._active_rental(book, member)
        if rental is None:  # 빌리지 않은 책 반납 거부
            raise NotRentedError(member.name, book.title)
        rental.close()  # 모든 검사를 통과한 뒤에만 상태를 바꾸기 위해
        book.available_copies += 1
        member.give_back(book)
        print(f"'{member.name}' 회원님이 '{book.title}' 도서를 반납하였습니다.")
````

````python
# 빈 도서관
LibraryManagement().print_books()
LibraryManagement().print_members()
print()

# 문제지 예시
library = LibraryManagement()
library.add_book("1984", "조지 오웰", 1949, "978-0451524935")
library.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")
library.add_member("홍길동")
library.rent_book("978-0451524935", "홍길동")
library.print_books()
library.print_members()
library.return_book("978-0451524935", "홍길동")
library.print_members()
print()  # 예시와 추가 사례의 출력을 구분하기 위해

# 추가 사례: 잘못된 입력
library.add_book("", "작가 미상", 2000, "111")  # 빈 제목
library.add_book("미래의 책", "누군가", 3000, "222")  # 미래 출판년도
library.add_book("동물농장", "조지 오웰", 1945, "978-0451524935")  # ISBN이 같은데 다른 책
library.add_book("1984", "조지 오웰", 1950, "978-0451524935")  # ISBN이 같은데 출판년도가 다름
library.add_book("무제", "누군가", 2000, "---")  # 기호만 있는 ISBN
library.add_member("홍길동")  # 중복 회원
library.search_books(" ")  # 빈 검색어

# 추가 사례: 대여 오류
library.add_member("김철수")
library.rent_book("000-0000000000", "홍길동")  # 없는 도서
library.rent_book("978-0451524935", "박민수")  # 없는 회원
library.rent_book("978-0451524935", "홍길동")
library.rent_book("978-0451524935", "홍길동")  # 중복 대여
library.rent_book("978-0451524935", "김철수")  # 1권뿐이라 재고 없음
library.return_book("978-0446310789", "김철수")  # 빌리지 않은 책 반납
library.remove_book("978-0451524935")  # 대여 중인 책 삭제

# 추가 사례: 권수 추가 → 대여 · 조회 · 검색 · 반납 · 삭제
library.add_book("1984", "조지 오웰", 1949, "9780451524935")  # 하이픈 없이 써도 같은 책 → 1권 더
library.rent_book("978-0451524935", "김철수")
library.find_member("김철수")
library.rent_book(None, "김철수")  # ISBN 자리에 잘못된 값
library.search_books("오웰")
library.return_book("978-0451524935", "홍길동")
library.return_book("978-0451524935", "김철수")
library.remove_book("978-0451524935")

# 검증
assert len(library.rentals) == 3  # 정상 대여 3건만 기록, 실패한 시도는 기록 없음
assert all(r.return_date is not None for r in library.rentals)  # 모두 반납 처리
assert all(r.due_date - r.rental_date == timedelta(days=14) for r in library.rentals)  # 반납 예정일 = 대여일 + 14일
assert list(library.books) == ["9780446310789"]  # 삭제된 1984만 빠짐 (키는 하이픈을 뺀 ISBN)
closed = library.rentals[0]
for make in [lambda: Member("a", 0), lambda: Rental(None, None), lambda: Book("t", "a", 0, "1"),
             lambda: Book("t", "a", 2000, "---"), lambda: Member("a", 1).give_back(closed.book), closed.close,
             lambda: Member("a", 1).borrow(None)]:  # 회원번호 0 · 객체 아님 · 출판년도 0 · 기호 ISBN · 안 빌린 책 · 두 번 반납 · 도서 아닌 값
    try:
        make()
        raise AssertionError("검사 누락")  # LibraryError가 나지 않으면 실패
    except LibraryError:
        pass
print("✅ 통과")
````
