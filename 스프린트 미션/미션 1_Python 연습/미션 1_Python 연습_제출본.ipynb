## 0. 공통 도구

대조 · 검증 함수, 여러 문제가 함께 쓰는 숫자 입력 함수, 값 검사 · 표시 함수.

```python
import builtins
import io
import math
import re
import sys
from collections.abc import Callable
from contextlib import contextmanager
from fractions import Fraction

RESULTS = {}  # (문제, 항목) → 통과 여부, 다시 실행하면 같은 항목을 덮어씀


@contextmanager
def fake_inputs(*values: str):
    """with 블록 안의 input()에 values를 차례로 넣고, 질문과 입력값을 화면에 함께 표시"""
    remaining = list(values)
    real_input = builtins.input

    def _fake_input(prompt=""):
        if not remaining:
            raise RuntimeError(f"입력값 부족: '{prompt}' 질문에 줄 값이 없음")
        value = remaining.pop(0)
        print(f"{prompt}{value}")
        return value
    builtins.input = _fake_input
    try:
        yield
    finally:
        builtins.input = real_input  # 오류가 나도 원래 input()으로 복구
    if remaining:
        raise RuntimeError(f"쓰지 않은 입력값: {remaining}")


class _Tee(io.StringIO):
    """출력을 대조용으로 모으면서, 완성된 줄 단위로 화면에 바로 보냄 (타이머의 1초 간격 확인용)"""

    def __init__(self, screen):
        super().__init__()
        self.screen = screen
        self.pending = ""

    def write(self, text: str) -> int:
        self.pending += text
        if "\n" in self.pending:
            done, self.pending = self.pending.rsplit("\n", 1)
            self.screen.write(done + "\n")
            self.screen.flush()
        return super().write(text)

    def finish(self) -> None:
        if self.pending:
            self.screen.write(self.pending)
            self.pending = ""


def check_output(key: str, label: str, run: Callable[[], object], expected: str) -> None:
    """run()의 출력이 expected와 정확히 같은지 대조 (expected 끝에 print의 줄바꿈 하나만 붙여 비교)"""
    screen, err_screen = sys.stdout, sys.stderr
    out, err = _Tee(screen), _Tee(err_screen)
    sys.stdout, sys.stderr = out, err
    try:
        run()
    except Exception:
        RESULTS[(key, label)] = False
        raise
    finally:
        out.finish()
        err.finish()
        sys.stdout, sys.stderr = screen, err_screen
    actual = out.getvalue()
    wanted = expected + "\n" if expected else ""  # 출력이 없어야 하면 빈 문자열
    ok = actual == wanted and err.getvalue() == ""
    RESULTS[(key, label)] = ok
    print(f"{'✅' if ok else '❌'} [{label}] 예상 출력과 {'일치' if ok else '불일치'}")
    if not ok:
        raise AssertionError(f"예상:\n{expected!r}\n실제:\n{actual!r}\n오류 출력:\n{err.getvalue()!r}")


def verify(key: str, label: str, condition) -> None:
    """추가 검증 결과를 기록하고, 실패하면 멈춤"""
    RESULTS[(key, label)] = bool(condition)
    print(f"{'✅' if condition else '❌'} {label}")
    if not condition:
        raise AssertionError(label)


def raises(error: type[BaseException], func: Callable[..., object], *args) -> bool:
    """func(*args)가 error를 내면 True"""
    try:
        func(*args)
    except error:
        return True
    return False


NUMBER = re.compile(r"[+-]?[0-9]{1,30}(\.[0-9]{1,30})?")  # 부호, 아라비아 숫자, 소수점 / 정수부 · 소수부 각 30자리 이하


def to_number(text) -> int | Fraction:
    """'10' → 10, '1.5' → Fraction(3, 2), '2.0' → 2 (입력한 10진수를 분수로 정확히 보관)

    형식이 다르면(nan, inf, 1e3, 1_000, .5, 31자리 이상 등) ValueError
    """
    text = str(text).strip()
    if not NUMBER.fullmatch(text):
        raise ValueError(f"숫자 형식이 아님: {text!r}")
    value = Fraction(text)
    return value.numerator if value.denominator == 1 else value


def to_fraction(value) -> Fraction:
    """int · float · Fraction → Fraction (float은 화면에 보이는 값 그대로, 예: 0.1 → 1/10)"""
    return Fraction(repr(value)) if isinstance(value, float) else Fraction(value)


def show(value) -> str:
    """10진수로 끝나는 값은 지수 표기 없이 표시 (예: 1/10 → 0.1, 1e-07 → 0.0000001, 2.0 → 2)

    10진수로 끝나지 않는 분수(예: 1/3)는 float의 기본 표기 (0.3333333333333333)
    """
    if isinstance(value, int):
        return str(value)
    fr = to_fraction(value)
    d, twos, fives = fr.denominator, 0, 0
    while d % 2 == 0:
        d, twos = d // 2, twos + 1
    while d % 5 == 0:
        d, fives = d // 5, fives + 1
    if d != 1:  # 10진수로 끝나지 않는 분수
        return repr(float(fr))
    k = max(twos, fives)  # 소수점 아래 자릿수
    scaled = abs(fr.numerator) * 10**k // fr.denominator
    digits = str(scaled).rjust(k + 1, "0")
    whole, frac = digits[:-k] if k else digits, digits[-k:].rstrip("0") if k else ""
    return ("-" if fr < 0 else "") + whole + (f".{frac}" if frac else "")


def plain(value: int | Fraction) -> int | float:
    """계산 결과를 돌려줄 때: 정수면 int, 아니면 float (문제지의 '정수 또는 실수')

    float 범위를 넘거나(OverflowError), 0이 아닌데 너무 작아 0.0이 되면 OverflowError
    """
    fr = to_fraction(value)
    if fr.denominator == 1:
        return fr.numerator
    result = float(fr)
    if result == 0.0:  # 0이 아닌 값이 0.0으로 사라지는 경우
        raise OverflowError("float으로 나타내기에 너무 작은 값")
    return result


def short(value) -> str:
    """오류 문구용 값 표시 (30자리를 넘는 정수는 자릿수로 요약, 4300자리를 넘어도 오류 없음)"""
    if isinstance(value, int) and not isinstance(value, bool):
        sign = "-" if value < 0 else ""  # 음수였다는 사실도 보이도록
        try:
            digits = len(str(abs(value)))
        except ValueError:  # Python이 문자열로 바꾸지 않는 4300자리 초과 정수
            return f"<{sign}약 {int(abs(value).bit_length() * 0.30103) + 1}자리 정수>"
        if digits > 30:
            return f"<{sign}{digits}자리 정수>"
    return repr(value)


def number_or_warn(text: str) -> int | Fraction | None:
    """숫자로 바꾸고, 숫자가 아니면 안내 후 None (자릿수만 넘은 숫자는 따로 안내)"""
    try:
        return to_number(text)
    except ValueError:
        if re.fullmatch(r"[+-]?[0-9]+(\.[0-9]+)?", text.strip()):
            print("정수부와 소수부는 각 30자리 이하로 입력하세요.")
        else:
            print("올바른 숫자를 입력하세요.")
        return None


def read_number(prompt: str) -> int | Fraction:
    """숫자가 입력될 때까지 다시 질문"""
    while True:
        value = number_or_warn(input(prompt))
        if value is not None:
            return value


def read_int(prompt: str, low: int = 0, high: int | None = None) -> int:
    """low 이상(high 이하)의 정수가 입력될 때까지 다시 질문"""
    while True:
        value = read_number(prompt)
        if isinstance(value, int) and value >= low and (high is None or value <= high):
            return value
        limit = f"{low} 이상" + (f" {high:,} 이하" if high is not None else "")
        print(f"{limit}의 정수를 입력하세요.")


def is_number(value) -> bool:
    """bool이 아닌 정수 · 분수, 또는 유한한 실수인지 (큰 정수도 넘침 없이 판정)"""
    if isinstance(value, bool):
        return False
    if isinstance(value, (int, Fraction)):
        return True
    if isinstance(value, float):
        return value == value and value not in (float("inf"), float("-inf"))  # nan은 자기 자신과 다름
    return False


MONEY_LIMIT = 10**13  # 10조 원 미만: 정수부 13자리 + 소수 2자리 = 유효숫자 15자리라 float으로 정확히 반환 가능


def to_money(value) -> Fraction | None:
    """금액을 0.01원 단위로 사사오입한 분수 (예: 0.1 + 0.2 → 0.3, 1000.005 → 1000.01)

    숫자가 아니거나 사사오입한 금액이 10조 원 이상이면 None
    """
    if not is_number(value):
        return None
    fr = to_fraction(value)
    if abs(fr) >= MONEY_LIMIT:
        return None
    money = Fraction(math.floor(fr * 100 + Fraction(1, 2)), 100)
    return money if abs(money) < MONEY_LIMIT else None


def is_money(value, allow_zero: bool = False) -> bool:
    """음수가 아니고, 사사오입한 금액이 10조 원 미만이며 0원 초과(allow_zero면 0원 이상)인지

    음수는 사사오입 전에 거부 (예: -0.004가 0원으로 바뀌어 통과하지 않도록)
    """
    money = to_money(value)
    if money is None or to_fraction(value) < 0:
        return False
    return money >= 0 if allow_zero else money > 0


def rounding_note(value) -> str:
    """사사오입으로 금액이 바뀌었으면 안내 (예: ' (입력 1000.005원을 0.01원 단위로 사사오입)')"""
    original = to_fraction(value)
    if original == to_money(value):
        return ""
    rest = original.denominator
    for p in (2, 5):
        while rest % p == 0:
            rest //= p
    # 10진수로 끝나는 값은 그대로, 끝나지 않는 분수(예: 1/3)는 근삿값 대신 분수로 표시
    text = show(original) if rest == 1 else f"{original.numerator}/{original.denominator}"
    return f" (입력 {text}원을 0.01원 단위로 사사오입)"


def is_count(value) -> bool:
    """bool이 아닌 0 이상의 정수인지"""
    return isinstance(value, int) and not isinstance(value, bool) and value >= 0
```

### 기초 1. 3의 배수 출력

**요구사항**

- `print_multiples_of_three(n)`: 1부터 n까지 3의 배수를 한 줄에 하나씩 출력

**문제지 예시**: 입력 7 → `3`, `6` / 입력 30 → `3`부터 `30`까지

**구현**

- `range(3, n + 1, 3)`으로 3의 배수만 만들어 나머지 검사 생략
- 0 이상의 정수가 아니면 `ValueError`

```python
def print_multiples_of_three(n: int) -> None:
    if not is_count(n):
        raise ValueError(f"0 이상의 정수가 아님: {n!r}")
    for i in range(3, n + 1, 3):  # 3부터 3씩 증가해 3의 배수만 생성
        print(i)


check_output("기초 1", "입력 7", lambda: print_multiples_of_three(7), "3\n6")
check_output("기초 1", "입력 30", lambda: print_multiples_of_three(30),
             "\n".join(str(i) for i in range(3, 31, 3)))
```

### 기초 2. 모음 제거하기

**문제지 예시**: `"codeit"` → `"cdt"`, `"python programming"` → `"pythn prgrmmng"`

**구현**

- 문제는 소문자 입력이지만 대문자 모음도 함께 제거

```python
VOWELS = set("aeiouAEIOU")


def remove_vowels(s: str) -> str:
    return "".join(ch for ch in s if ch not in VOWELS)  # 모음이 아닌 글자만 이어 붙임


check_output("기초 2", "codeit", lambda: print(remove_vowels("codeit")), "cdt")
check_output("기초 2", "python programming",
             lambda: print(remove_vowels("python programming")), "pythn prgrmmng")
```

### 기초 3. 수 크기 비교하기

**요구사항**

- `compare_numbers(num1, num2)`: `(큰 수, 작은 수)` 반환, 두 수가 같으면 그 수 하나만 반환

**문제지 예시**: `10, 20` → `(20, 10)`, `30, 30` → `30`

**구현**

- 같은 수를 먼저 처리한 뒤 `max`, `min`으로 입력 순서와 상관없이 정렬
- 숫자가 아니거나 nan · inf면 `TypeError`

```python
def compare_numbers(num1: int, num2: int) -> int | tuple[int, int]:
    if not (is_number(num1) and is_number(num2)):
        raise TypeError(f"비교할 수 없는 값: {num1!r}, {num2!r}")
    if num1 == num2:  # 같은 수는 하나만 반환
        return num1
    return max(num1, num2), min(num1, num2)


check_output("기초 3", "10, 20", lambda: print(compare_numbers(10, 20)), "(20, 10)")
check_output("기초 3", "30, 30", lambda: print(compare_numbers(30, 30)), "30")
```

```python
verify("기초 3", "입력 순서가 바뀌어도 (큰 수, 작은 수)", compare_numbers(20, 10) == (20, 10))
verify("기초 3", "음수 비교", compare_numbers(-1, -5) == (-1, -5))
verify("기초 3", "nan · 문자열은 TypeError",
       raises(TypeError, compare_numbers, float("nan"), 1)
       and raises(TypeError, compare_numbers, "a", 1))
```

### 기초 5. 총 합계 구하기

**요구사항**

- `sum_up_to_n(n)`: 0부터 n까지의 합 반환 (n은 0보다 큰 정수)

**문제지 예시**: `3` → `6`, `10` → `55`

**구현**

- 반복문 대신 가우스 공식 `n(n + 1) / 2`로 한 번에 계산 (`//`로 정수 유지)
- 0도 허용해 합 0 반환, 음수 · 실수 · bool은 `ValueError`

```python
def sum_up_to_n(n: int) -> int:
    if not is_count(n):
        raise ValueError(f"0 이상의 정수가 아님: {n!r}")
    return n * (n + 1) // 2


check_output("기초 5", "3", lambda: print(sum_up_to_n(3)), "6")
check_output("기초 5", "10", lambda: print(sum_up_to_n(10)), "55")
```

```python
verify("기초 5", "공식 = 반복문 합 (n = 0 ~ 999)",
       all(sum_up_to_n(n) == sum(range(n + 1)) for n in range(1000)))
verify("기초 5", "음수 · 실수 · bool은 ValueError",
       all(raises(ValueError, sum_up_to_n, bad) for bad in (-3, 2.5, True)))
```

### 기초 6. 짝수 합 구하기

**요구사항**

- `sum_even_numbers(n)`: `"1부터 n까지의 짝수 합은 ___입니다."` 출력
- 0을 입력하면 종료 안내 (음수는 입력하지 않는다고 가정)

**문제지 예시**: 입력 9 → `"1부터 9까지의 짝수 합은 20입니다."`, 입력 0 → `종료`

**구현**

- 계산 · 출력은 `sum_even_numbers(n)`, 반복 입력은 `even_sum_program()`, 문제 셀처럼 인자 없이 `sum_even_numbers()`를 부르면 반복 입력 실행
- 반복 입력은 `read_int`로 받아 숫자가 아니거나 음수 · 실수면 다시 질문

```python
def sum_even_numbers(n: int | None = None) -> None:
    if n is None:  # 문제 셀처럼 인자 없이 부르면 반복 입력
        even_sum_program()
        return
    if not is_count(n):
        raise ValueError(f"0 이상의 정수가 아님: {n!r}")
    if n == 0:  # 0은 종료 신호
        print("종료")
        return
    print(f"1부터 {n}까지의 짝수 합은 {sum(range(2, n + 1, 2))}입니다.")


def even_sum_program() -> None:
    """0이 입력될 때까지 숫자를 받아 짝수 합 출력"""
    while True:
        n = read_int("숫자를 입력하세요 (0: 종료): ")
        sum_even_numbers(n)
        if n == 0:
            break


check_output("기초 6", "입력 9", lambda: sum_even_numbers(9), "1부터 9까지의 짝수 합은 20입니다.")
check_output("기초 6", "입력 0", lambda: sum_even_numbers(0), "종료")


def no_argument():
    with fake_inputs("9", "0"):
        sum_even_numbers()


check_output("기초 6", "인자 없이 호출하면 반복 입력", no_argument,
             "숫자를 입력하세요 (0: 종료): 9\n1부터 9까지의 짝수 합은 20입니다.\n"
             "숫자를 입력하세요 (0: 종료): 0\n종료")
```

```python
def flow():
    with fake_inputs("9", "abc", "-2", "3", "0"):  # 정상 → 숫자 아님 → 음수 → 실행 셀의 3 → 종료
        even_sum_program()


check_output("기초 6", "반복 입력 흐름", flow,
             "숫자를 입력하세요 (0: 종료): 9\n1부터 9까지의 짝수 합은 20입니다.\n"
             "숫자를 입력하세요 (0: 종료): abc\n올바른 숫자를 입력하세요.\n"
             "숫자를 입력하세요 (0: 종료): -2\n0 이상의 정수를 입력하세요.\n"
             "숫자를 입력하세요 (0: 종료): 3\n1부터 3까지의 짝수 합은 2입니다.\n"
             "숫자를 입력하세요 (0: 종료): 0\n종료")
```

### 기초 7. 두 숫자 나누기

**요구사항**

- `divide_numbers(a, b)`: 첫 번째 수를 두 번째 수로 나눈 값 반환, 사용자에게 두 숫자를 차례로 입력받음
- 예외가 발생하면 `"잘못된 입력입니다."` 반환, `try-except` 사용

**문제지 예시**: `42, 7` → `6`, `135, 0` → `"잘못된 입력입니다."`

**구현**

- 두 인자를 주면 바로 계산, 둘 다 생략하면 `input()`으로 입력, 하나만 주면 잘못된 입력
- 숫자 아님, 잘못된 자료형, 0으로 나누기, 계산 범위 초과(`ValueError`, `TypeError`, `ArithmeticError`)를 모두 처리
- 분수(`Fraction`)로 정확히 나눈 뒤, 나누어떨어지면 정확한 정수(`6.0`이 아닌 `6`), 아니면 `float`(`7 / 2` → `3.5`)로 반환
- 결과가 float 범위(약 1.8 × 10^308)를 넘거나 0이 아닌데 0.0이 될 만큼 작으면 잘못된 입력으로 처리

```python
def divide_numbers(a: int | float | None = None,
                   b: int | float | None = None) -> int | float | str:
    try:
        if a is None and b is None:  # 인자를 모두 생략하면 사용자 입력
            a = to_number(input("첫 번째 숫자: "))
            b = to_number(input("두 번째 숫자: "))
        elif a is None or b is None:  # 하나만 준 호출은 잘못된 입력
            raise TypeError("인자는 둘 다 주거나 둘 다 생략")
        if not (is_number(a) and is_number(b)):
            raise TypeError(f"숫자가 아님: {a!r}, {b!r}")
        return plain(to_fraction(a) / to_fraction(b))  # 분수로 정확히 나눈 뒤 int 또는 float
    except (ValueError, TypeError, ArithmeticError):
        return "잘못된 입력입니다."


check_output("기초 7", "42, 7", lambda: print(divide_numbers(42, 7)), "6")
check_output("기초 7", "135, 0", lambda: print(divide_numbers(135, 0)), "잘못된 입력입니다.")
```

```python
def typed():
    with fake_inputs("42", "7"):  # 실행 셀처럼 인자 없이 입력
        print(divide_numbers())


def typed_bad():
    with fake_inputs("10", "zero"):
        print(divide_numbers())


check_output("기초 7", "입력 42, 7", typed, "첫 번째 숫자: 42\n두 번째 숫자: 7\n6")
check_output("기초 7", "입력 10, zero", typed_bad, "첫 번째 숫자: 10\n두 번째 숫자: zero\n잘못된 입력입니다.")
verify("기초 7", "소수 결과는 float (7 / 2 = 3.5)",
       divide_numbers(7, 2) == 3.5 and type(divide_numbers(7, 2)) is float)
verify("기초 7", "나누어떨어지는 큰 정수는 정확한 int (10**400 / 10**399 = 10)",
       divide_numbers(10**400, 10**399) == 10)
verify("기초 7", "나누어떨어지지 않으면 int로 바꾸지 않음 ((10**30 + 1) / 2 → float)",
       type(divide_numbers(10**30 + 1, 2)) is float)
bad_calls = [divide_numbers(42), divide_numbers("a", 1), divide_numbers(1j, 1),
             divide_numbers(float("inf"), 1), divide_numbers(0, 0), divide_numbers(10**400, 3),
             divide_numbers(1, 10**400)]
verify("기초 7", "인자 하나, 문자열, 복소수, inf, 0 / 0, float 범위 초과, 0.0이 될 만큼 작은 몫 → 잘못된 입력",
       all(r == "잘못된 입력입니다." for r in bad_calls))
```

### 응용 1. 문자열 압축

**요구사항**

- `compress_string(s)`: 연속으로 반복되는 문자를 `문자 + 반복 횟수`로 압축

**문제지 예시**: `"aaabbaaa"` → `"a3b2a3"`, `"aabcccccaaa"` → `"a2b1c5a3"`

**구현**

- 이웃한 두 글자를 비교해 같으면 개수 + 1, 다르면 조각 저장, 마지막 묶음 추가

```python
def compress_string(s: str) -> str:
    if not s:  # 빈 문자열은 s[-1] 오류가 나므로 먼저 처리
        return ""
    result, count = [], 1
    for prev, cur in zip(s, s[1:]):  # 이웃한 두 글자 비교
        if cur == prev:
            count += 1
        else:
            result.append(f"{prev}{count}")  # 묶음이 끝나면 저장
            count = 1
    result.append(f"{s[-1]}{count}")  # 마지막 묶음
    return "".join(result)


check_output("응용 1", "aaabbaaa", lambda: print(compress_string("aaabbaaa")), "a3b2a3")
check_output("응용 1", "aabcccccaaa", lambda: print(compress_string("aabcccccaaa")), "a2b1c5a3")
```

```python
from itertools import groupby  # 다른 방법으로 구한 답과 비교하기 위해

verify("응용 1", "groupby 결과와 일치 (빈 문자열 · 한 글자 · 반복 없음 · 한 종류)",
       all(compress_string(c) == "".join(f"{ch}{len(list(g))}" for ch, g in groupby(c))
           for c in ["", "a", "abc", "zzzz"]))
```

### 응용 3. 카운트다운 타이머

**요구사항**

- `count_down(seconds)`: 입력받은 초부터 1초마다 남은 시간 출력, 끝나면 `"타이머 종료!"`

**문제지 예시**: 입력 5 → `타이머 시작!`, `5` ~ `1`, `타이머 종료!`

**구현**

- `range(seconds, 0, -1)` + `time.sleep(1)`로 1초 간격, 출력은 실행 중에 바로 표시
- 입력은 `timer_program()`이 `read_int`로 받아 숫자가 아니거나 음수 · 실수면 다시 질문
- `count_down`에 직접 음수 · 실수 · bool을 넘기면 안내 후 종료

```python
import time


def count_down(seconds: int) -> None:
    if not is_count(seconds):
        print("0 이상의 정수를 입력하세요.")
        return
    print("타이머 시작!")
    for remaining in range(seconds, 0, -1):
        print(remaining)
        time.sleep(1)
    print("타이머 종료!")


countdown_timer = count_down  # 문제지 실행 셀의 함수 이름도 동작하도록


def timer_program() -> None:
    count_down(read_int("타이머를 시작할 시간(초)을 입력하세요: "))


def example():
    with fake_inputs("5"):
        timer_program()


start = time.perf_counter()
check_output("응용 3", "입력 5", example,
             "타이머를 시작할 시간(초)을 입력하세요: 5\n타이머 시작!\n5\n4\n3\n2\n1\n타이머 종료!")
elapsed = time.perf_counter() - start
```

```python
print(f"5초 타이머 실행 시간: {elapsed:.1f}초")
verify("응용 3", "실제 1초 간격 (4.5초 이상 10초 미만)", 4.5 <= elapsed < 10)
check_output("응용 3", "음수 직접 전달", lambda: count_down(-3), "0 이상의 정수를 입력하세요.")
check_output("응용 3", "실수 직접 전달", lambda: count_down(2.5), "0 이상의 정수를 입력하세요.")
check_output("응용 3", "bool 직접 전달", lambda: count_down(True), "0 이상의 정수를 입력하세요.")
check_output("응용 3", "countdown_timer(0)", lambda: countdown_timer(0), "타이머 시작!\n타이머 종료!")
```

## 3. 클래스 기초 (2문제)

### 클래스 기초 1. 시간 추적 클래스 `TimeTracker`

**요구사항**

- `start()`: 현재 시각을 시작 시각으로 기록
- `stop()`: 현재 시각을 종료 시각으로 기록하고 경과 시간 계산
- `get_elapsed_time()`: 경과 시간을 분 단위로 반환
- 현재 시각은 `datetime.now()` 사용

**문제지 예시**: 1시간 30분 경과 가정 → `공부한 시간: 90 분`

**구현**

- `start`, `stop`에서 `datetime.now()`로 기록, `stop`에서 경과 시간을 **분 단위 정수(버림)**로 계산
- 실제로 90분을 기다릴 수 없어 시작 시각을 1시간 30분 앞당겨 같은 상황 재현
- 측정이 끝나기 전에 조회하면 안내 후 `None` 반환 (0분 측정과 구분)
- 측정 중에 다시 `start()`하면 안내 후 새로 시작, 이미 끝난 측정의 `stop()`은 안내 후 기록 유지
- 시스템 시계가 뒤로 바뀌어 종료가 시작보다 앞서면 0분으로 처리

```python
from datetime import datetime, timedelta


class TimeTracker:
    def __init__(self) -> None:
        self.start_time = None
        self.end_time = None
        self.elapsed_minutes = None  # 측정 전에는 None

    def start(self) -> None:
        if self.start_time is not None and self.elapsed_minutes is None:
            print("측정 중이던 기록을 버리고 다시 시작합니다.")
        self.start_time = datetime.now()
        self.end_time = None  # 다시 시작하면 이전 기록 초기화
        self.elapsed_minutes = None

    def stop(self) -> None:
        if self.start_time is None:
            print("start()를 먼저 호출하세요.")
            return
        if self.elapsed_minutes is not None:
            print("이미 종료된 측정입니다.")
            return
        self.end_time = datetime.now()
        seconds = max(0.0, (self.end_time - self.start_time).total_seconds())  # 시계가 뒤로 가도 음수 방지
        self.elapsed_minutes = int(seconds // 60)

    def get_elapsed_time(self) -> int | None:
        if self.elapsed_minutes is None:
            print("아직 측정이 끝나지 않았습니다.")
            return None
        return self.elapsed_minutes


def example():
    study_session = TimeTracker()
    study_session.start()
    study_session.start_time -= timedelta(hours=1, minutes=30)  # 1시간 30분 경과 재현
    study_session.stop()
    print("공부한 시간:", study_session.get_elapsed_time(), "분")


check_output("클래스 기초 1", "90분", example, "공부한 시간: 90 분")
```

```python
tracker = TimeTracker()


def before_start():
    tracker.stop()
    print(tracker.get_elapsed_time())


check_output("클래스 기초 1", "시작 전 종료 · 조회", before_start,
             "start()를 먼저 호출하세요.\n아직 측정이 끝나지 않았습니다.\nNone")
tracker.start()
tracker.start_time -= timedelta(minutes=89, seconds=59)
tracker.stop()
verify("클래스 기초 1", "89분 59초 → 89분 (분 단위 버림)", tracker.get_elapsed_time() == 89)


def stop_twice_restart():
    tracker.stop()  # 이미 끝난 측정
    print(tracker.get_elapsed_time())  # 기록 유지
    tracker.start()  # 끝난 뒤 새 측정은 안내 없음
    tracker.start()  # 측정 중 다시 시작


check_output("클래스 기초 1", "두 번 종료는 기록 유지 · 측정 중 다시 시작은 안내", stop_twice_restart,
             "이미 종료된 측정입니다.\n89\n측정 중이던 기록을 버리고 다시 시작합니다.")
```

### 클래스 기초 2. 주소록 클래스 `Contact`

- 이름은 앞뒤 공백을 지워 저장 · 비교, 빈 이름과 문자열이 아닌 이름은 등록 · 투표 모두 거부
- 1표일 때만 단수형 `vote`

```python
class VoteSystem:
    def __init__(self) -> None:
        self.candidates = {}  # 후보 → 득표 수

    @staticmethod
    def _clean(name) -> str | None:
        """앞뒤 공백을 지운 이름, 올바르지 않으면 None"""
        if isinstance(name, str) and name.strip():
            return name.strip()
        return None

    def add_candidate(self, name: str) -> None:
        clean = self._clean(name)
        if clean is None:
            print(f"후보 이름이 올바르지 않습니다: {name!r}")
        elif clean in self.candidates:
            print(f"{clean} 후보는 이미 등록되어 있습니다.")
        else:
            self.candidates[clean] = 0
            print(f"{clean} 후보가 성공적으로 등록되었습니다.")

    def vote(self, name: str) -> None:
        clean = self._clean(name)
        if clean is None:
            print(f"후보 이름이 올바르지 않습니다: {name!r}")
            return
        name = clean
        if name not in self.candidates:
            print(f"{name} 후보는 등록되지 않아 투표할 수 없습니다.")
            return
        self.candidates[name] += 1
        print(f"{name}에게 투표하였습니다.")

    def get_results(self) -> None:
        print("투표 결과:")
        for name, votes in self.candidates.items():
            print(f"{name}: {votes} {'vote' if votes == 1 else 'votes'}")


def example():
    voting_system = VoteSystem()
    voting_system.add_candidate("Alice")
    voting_system.add_candidate("Bob")
    voting_system.add_candidate("Charlie")
    voting_system.vote("Alice")
    voting_system.vote("Alice")
    voting_system.vote("Bob")
    voting_system.get_results()


check_output("클래스 응용 1", "예시", example,
             "Alice 후보가 성공적으로 등록되었습니다.\nBob 후보가 성공적으로 등록되었습니다.\n"
             "Charlie 후보가 성공적으로 등록되었습니다.\nAlice에게 투표하였습니다.\nAlice에게 투표하였습니다.\n"
             "Bob에게 투표하였습니다.\n투표 결과:\nAlice: 2 votes\nBob: 1 vote\nCharlie: 0 votes")
```

```python
vs = VoteSystem()


def edge():
    vs.add_candidate("Alice")
    vs.vote("Alice")
    vs.add_candidate("Alice")  # 중복 등록
    vs.add_candidate(" Alice ")  # 공백만 다른 중복
    vs.vote("Dave")  # 미등록 후보
    vs.add_candidate("")  # 빈 이름
    vs.vote([])  # 문자열이 아닌 이름


check_output("클래스 응용 1", "중복 · 미등록 · 잘못된 이름 거부", edge,
             "Alice 후보가 성공적으로 등록되었습니다.\nAlice에게 투표하였습니다.\n"
             "Alice 후보는 이미 등록되어 있습니다.\nAlice 후보는 이미 등록되어 있습니다.\n"
             "Dave 후보는 등록되지 않아 투표할 수 없습니다.\n"
             "후보 이름이 올바르지 않습니다: ''\n후보 이름이 올바르지 않습니다: []")
verify("클래스 응용 1", "중복 등록 시도 뒤에도 득표 유지 (Alice 1표)", vs.candidates == {"Alice": 1})
```

### 클래스 응용 2. 은행 계좌 `BankAccount`

**요구사항**

- `__init__`, `deposit`, `withdraw`, `get_balance`
- 잔액보다 많은 출금은 거부하고 경고, 금액은 정수 또는 실수

**문제지 예시**: 10만 원 개설 → 5만 원 입금 → 2만 원 출금 → `현재 잔액: 130000원`

**구현**

- 잔액은 `_balance`에 분수(`Fraction`)로 두고 입금 · 출금 메서드로만 변경 (0.1 + 0.2가 정확히 0.3)
- 금액은 정수 · 실수 모두 받아 0.01원 단위로 사사오입(예: `0.1 + 0.2` → 0.3원, 1000.005 → 1000.01원), 사사오입으로 바뀌면 안내, 0원이 되는 금액 · 10조 원 이상 · 숫자가 아닌 값은 경고 출력 후 `False` 반환
- 잔액도 10조 원 미만 (그 범위에서는 float 반환값이 화면 표시와 정확히 같음)
- 안내 문구를 먼저 만든 뒤 잔액을 바꿔, 실패할 때 상태가 반쯤 바뀌지 않음
- `get_balance()`는 정수 잔액이면 int, 아니면 float 반환
- 계좌번호 · 소유자가 빈 문자열이거나 초기 잔액이 잘못되면 계좌를 만들 수 없으므로 `ValueError`

```python
AMOUNT_RULE = "0.01원 단위로 사사오입해 0원보다 크고 10조 원 미만인 숫자 금액을 입력하세요."


class BankAccount:
    def __init__(self, account_number: str, account_holder: str,
                 initial_balance: int | float = 0) -> None:
        for label, text in (("계좌번호", account_number), ("소유자", account_holder)):
            if not isinstance(text, str) or text.strip() == "":
                raise ValueError(f"{label}가 올바르지 않습니다: {short(text)}")
        if not is_money(initial_balance, allow_zero=True):
            raise ValueError(f"초기 잔액은 0원 이상 10조 원 미만의 숫자여야 합니다: {short(initial_balance)}")
        self.account_number = account_number
        self.account_holder = account_holder
        self._balance = to_money(initial_balance)  # 분수로 보관해 실수 오차 방지
        print(f"{account_holder}님의 계좌 {account_number}가 개설되었습니다. "
              f"초기 잔액: {show(self._balance)}원{rounding_note(initial_balance)}")

    def deposit(self, amount: int | float) -> bool:
        if not is_money(amount):
            print(f"입금 실패: {AMOUNT_RULE} (입력: {short(amount)})")
            return False
        money = to_money(amount)  # 0.01원 단위로 사사오입
        new_balance = self._balance + money
        if new_balance >= MONEY_LIMIT:
            print(f"입금 실패: 잔액이 10조 원 이상이 됩니다. (현재 잔액 {show(self._balance)}원)")
            return False
        message = f"{show(money)}원이 입금되었습니다{rounding_note(amount)}. 현재 잔액: {show(new_balance)}원"
        self._balance = new_balance  # 안내 문구까지 만든 뒤 상태 변경
        print(message)
        return True

    def withdraw(self, amount: int | float) -> bool:
        if not is_money(amount):
            print(f"출금 실패: {AMOUNT_RULE} (입력: {short(amount)})")
            return False
        money = to_money(amount)
        if money > self._balance:
            print(f"출금 실패: 잔액이 부족합니다. (요청 {show(money)}원, 잔액 {show(self._balance)}원)")
            return False
        new_balance = self._balance - money
        message = f"{show(money)}원이 출금되었습니다{rounding_note(amount)}. 현재 잔액: {show(new_balance)}원"
        self._balance = new_balance
        print(message)
        return True

    def get_balance(self) -> int | float:
        return plain(self._balance)


def example():
    my_account = BankAccount("123-456-789", "김철수", 100000)
    my_account.deposit(50000)
    my_account.withdraw(20000)
    print(f"현재 잔액: {my_account.get_balance()}원")


check_output("클래스 응용 2", "예시", example,
             "김철수님의 계좌 123-456-789가 개설되었습니다. 초기 잔액: 100000원\n"
             "50000원이 입금되었습니다. 현재 잔액: 150000원\n20000원이 출금되었습니다. 현재 잔액: 130000원\n"
             "현재 잔액: 130000원")
```

```python
acc = BankAccount("000-000", "테스트", 1000)
results = [acc.withdraw(5000), acc.deposit(-500), acc.deposit("100"), acc.deposit(float("inf")),
           acc.deposit(0.1), acc.deposit(0.2)]
verify("클래스 응용 2", "잔액 초과, 음수, 문자열, inf 거부 / 소수 입금만 반영해 잔액 정확히 1000.3",
       results == [False, False, False, False, True, True] and acc.get_balance() == 1000.3)
verify("클래스 응용 2", "소수 금액 0.3 출금 → 잔액 정확히 1000 (int)",
       acc.withdraw(0.3) and acc.get_balance() == 1000 and type(acc.get_balance()) is int)


def float_amounts():
    acc.deposit(0.1 + 0.2)  # 0.30000000000000004 → 0.3
    acc.deposit(1000.005)  # → 1000.01
    acc.deposit(0.004)  # → 0원이라 거부


check_output("클래스 응용 2", "계산으로 만든 실수 금액은 0.01원 단위로 사사오입하고 안내", float_amounts,
             "0.3원이 입금되었습니다 (입력 0.30000000000000004원을 0.01원 단위로 사사오입). 현재 잔액: 1000.3원\n"
             "1000.01원이 입금되었습니다 (입력 1000.005원을 0.01원 단위로 사사오입). 현재 잔액: 2000.31원\n"
             f"입금 실패: {AMOUNT_RULE} (입력: 0.004)")
verify("클래스 응용 2", "큰 정수 금액도 입금 (2조 원)", acc.deposit(2 * 10**12) is True)
verify("클래스 응용 2", "10조 원 이상 금액 · 잔액이 10조 원 이상이 되는 입금 · 5001자리 금액(10**5000)은 오류 없이 거부",
       all(acc.deposit(bad) is False for bad in (10**13, 9 * 10**12, 10**5000)))
verify("클래스 응용 2", "잔액 전액 출금 → 잔액 0", acc.withdraw(acc.get_balance()) and acc.get_balance() == 0)
edge = BankAccount("999-999", "경계", 9999999999999.99)
verify("클래스 응용 2", "상한 바로 아래 잔액(9999999999999.99원)도 get_balance()의 float이 화면 표시와 같음",
       edge.get_balance() == 9999999999999.99 and show(edge.get_balance()) == "9999999999999.99")
verify("클래스 응용 2", "사사오입하면 정확히 10조 원이 되는 초기 잔액(9999999999999.995)은 ValueError",
       raises(ValueError, BankAccount, "1", "x", Fraction("9999999999999.995")))
verify("클래스 응용 2", "(공통 도구) 오류 문구의 긴 정수는 부호와 자릿수로 요약 (30자리 이하는 그대로)",
       (short(10**40), short(-10**40), short(10**29)) == ("<41자리 정수>", "<-41자리 정수>", repr(10**29)))
verify("클래스 응용 2", "(공통 도구) 10진수로 끝나지 않는 분수 입력(1/3)은 안내에 근삿값 대신 분수로 표시",
       rounding_note(Fraction(1, 3)) == " (입력 1/3원을 0.01원 단위로 사사오입)")
check_output("클래스 응용 2", "초기 잔액도 사사오입하면 안내", lambda: BankAccount("111-222-999", "a", 100.005),
             "a님의 계좌 111-222-999가 개설되었습니다. 초기 잔액: 100.01원 (입력 100.005원을 0.01원 단위로 사사오입)")
verify("클래스 응용 2", "잘못된 초기 잔액(-1, 사사오입하면 0원인 -0.004) · 빈 계좌번호 · 소유자 None은 ValueError",
       raises(ValueError, BankAccount, "1", "x", -1)
       and raises(ValueError, BankAccount, "1", "x", -0.004)
       and raises(ValueError, BankAccount, "", "x", 0)
       and raises(ValueError, BankAccount, "1", None, 0))
```

### 클래스 응용 3. 직원 관리 `EmployeeManager`

**요구사항**

- `__init__`: 이름과 급여를 받아 클래스 변수 `employees`에 저장
- `calculate_average_salary`: 클래스 메서드, 전체 평균 급여 출력, 직원이 없으면 0 반환

**문제지 예시**: 홍길동 50000, 김철수 60000 → `전체 직원의 평균 급여: 55000원`

**구현**

- 모든 인스턴스가 하나의 목록을 공유하는 클래스 변수, 동명이인도 따로 집계되도록 객체 저장
- 급여는 0원 이상 10조 원 미만의 숫자만 허용, 0.01원 단위로 사사오입해 int 또는 float으로 저장 (바뀌면 안내, 검사를 통과하고 안내 문구를 만든 뒤에 저장)
- 평균은 분수로 정확히 계산해 소수 셋째 자리에서 사사오입(예: 100.005 → 100.01), 정수면 int, 아니면 float으로 출력 · 반환

```python
import math


class EmployeeManager:
    employees = []  # 모든 인스턴스가 공유하는 클래스 변수

    def __init__(self, name: str, salary: int | float) -> None:
        if not isinstance(name, str) or name.strip() == "":
            raise ValueError(f"직원 이름이 올바르지 않습니다: {short(name)}")
        if not is_money(salary, allow_zero=True):
            raise ValueError(f"급여는 0원 이상 10조 원 미만의 숫자여야 합니다: {short(salary)}")
        self.name, self.salary = name, plain(to_money(salary))  # 사사오입한 급여를 int 또는 float으로 저장
        message = f"{name} 님이 추가되었습니다. 급여: {show(self.salary)}원{rounding_note(salary)}"
        type(self).employees.append(self)  # 검사를 통과한 직원만 저장
        print(message)

    @classmethod
    def calculate_average_salary(cls) -> int | float:
        if not cls.employees:
            print("등록된 직원이 없습니다.")
            return 0
        total = sum(to_fraction(emp.salary) for emp in cls.employees)
        exact = total / len(cls.employees)  # 분수로 정확한 평균
        average = Fraction(math.floor(exact * 100 + Fraction(1, 2)), 100)  # 소수 셋째 자리 사사오입
        result = plain(average)  # 반환값을 먼저 만들어 출력 뒤에 실패하는 경로를 없앰
        print(f"전체 직원의 평균 급여: {show(average)}원")
        return result


def example():
    EmployeeManager.employees.clear()  # 셀을 다시 실행해도 직원이 중복 저장되지 않도록
    EmployeeManager("홍길동", 50000)  # 문제지의 emp1
    EmployeeManager("김철수", 60000)  # 문제지의 emp2
    EmployeeManager.calculate_average_salary()


check_output("클래스 응용 3", "예시", example,
             "홍길동 님이 추가되었습니다. 급여: 50000원\n김철수 님이 추가되었습니다. 급여: 60000원\n"
             "전체 직원의 평균 급여: 55000원")
```

```python
def two_employees():
    EmployeeManager.employees.clear()  # 이 셀만 다시 실행해도 같은 결과가 나오도록 처음부터 다시 구성
    EmployeeManager("홍길동", 50000)
    EmployeeManager("김철수", 60000)


check_output("클래스 응용 3", "직원 2명 다시 구성", two_employees,
             "홍길동 님이 추가되었습니다. 급여: 50000원\n김철수 님이 추가되었습니다. 급여: 60000원")
verify("클래스 응용 3", "직원 2명이 클래스 변수에 저장", len(EmployeeManager.employees) == 2)
verify("클래스 응용 3", "잘못된 급여('200', -100, -0.004, None, 10조 원 이상, 사사오입하면 10조 원)는 ValueError, 저장 안 됨",
       all(raises(ValueError, EmployeeManager, "x", bad)
           for bad in ("200", -100, -0.004, None, 10**13, Fraction("9999999999999.995")))
       and len(EmployeeManager.employees) == 2)
verify("클래스 응용 3", "저장된 급여는 넣은 그대로의 int (분수 자료형이 드러나지 않음)",
       [type(emp.salary) for emp in EmployeeManager.employees] == [int, int])


def odd_average():
    EmployeeManager("이순신", 60002)  # (50000 + 60000 + 60002) / 3 = 56667.333…
    print(EmployeeManager.calculate_average_salary())


check_output("클래스 응용 3", "평균이 소수일 때 (소수 셋째 자리 사사오입)", odd_average,
             "이순신 님이 추가되었습니다. 급여: 60002원\n전체 직원의 평균 급여: 56667.33원\n56667.33")


def half_up():
    EmployeeManager.employees.clear()
    EmployeeManager("가", 100.01)
    EmployeeManager("나", 100)  # 평균 100.005 → 사사오입 100.01 (은행원 반올림이면 100)
    print(EmployeeManager.calculate_average_salary())


check_output("클래스 응용 3", "경계값 100.005는 100.01로 올림", half_up,
             "가 님이 추가되었습니다. 급여: 100.01원\n나 님이 추가되었습니다. 급여: 100원\n"
             "전체 직원의 평균 급여: 100.01원\n100.01")
EmployeeManager.employees.clear()
check_output("클래스 응용 3", "급여도 사사오입하면 안내", lambda: EmployeeManager("다", 100.005),
             "다 님이 추가되었습니다. 급여: 100.01원 (입력 100.005원을 0.01원 단위로 사사오입)")
rounded = EmployeeManager.employees[0].salary
verify("클래스 응용 3", "사사오입한 급여는 float 100.01로 저장", rounded == 100.01 and type(rounded) is float)
EmployeeManager.employees.clear()
check_output("클래스 응용 3", "직원이 없을 때", lambda: print(EmployeeManager.calculate_average_salary()),
             "등록된 직원이 없습니다.\n0")
```

### 클래스 응용 4. 프랜차이즈 레스토랑 `ReservationSystem`

**요구사항**

- `__init__`, `add_reservation`, `list_reservations`, `sum_reservations`(여러 지점 합산), 예약 취소
- 모든 출력은 한국어, 각 메서드에 입력 검증과 예외 처리

**문제지 예시**: 강남점, 홍대점에 1건씩 → 지점별 목록, `전체 레스토랑 예약 수: 2`

**구현**

- 예약 일시는 날짜(`2024-05-20`, `2024/05/20`, `2024.05.20`) 또는 날짜 + 시각(`2024-05-20 18:30`, `2024-05-20T18:30`, 초 포함)을 받아 분 단위의 한 형식으로 통일
- 이름은 앞뒤 공백을 지워 저장 · 비교, 없는 날짜, 1 이상의 정수가 아닌 인원, 같은 이름의 같은 일시 중복 예약은 거부
- `sum_reservations`는 클래스 메서드로 지점 객체들의 예약 건수 합산, 지점이 아닌 값이 섞이면 `TypeError`

```python
from datetime import datetime


class ReservationSystem:
    DATE_PARTS = ("%Y-%m-%d", "%Y/%m/%d", "%Y.%m.%d")
    TIME_PARTS = (" %H:%M", "T%H:%M", " %H:%M:%S", "T%H:%M:%S")

    def __init__(self, branch_name: str) -> None:
        if not isinstance(branch_name, str) or branch_name.strip() == "":
            raise ValueError(f"지점 이름이 올바르지 않습니다: {branch_name!r}")
        self.branch_name = branch_name.strip()
        self.reservations = []  # 예약 1건 = 딕셔너리 1개

    @classmethod
    def _normalize(cls, when) -> str:
        """날짜만 → 'YYYY-MM-DD', 시각 포함 → 'YYYY-MM-DD HH:MM' (초는 버림), datetime 객체도 허용"""
        if isinstance(when, datetime):
            return when.strftime("%Y-%m-%d %H:%M")
        text = str(when).strip()
        for date_fmt in cls.DATE_PARTS:
            for time_fmt in ("",) + cls.TIME_PARTS:
                try:
                    parsed = datetime.strptime(text, date_fmt + time_fmt)
                except ValueError:
                    continue
                return parsed.strftime("%Y-%m-%d %H:%M" if time_fmt else "%Y-%m-%d")
        raise ValueError(f"예약 일시 형식이 올바르지 않습니다: {when!r}")

    @staticmethod
    def _clean_name(name) -> str:
        """앞뒤 공백을 지운 예약자 이름 (자료형 오류와 빈 값 오류를 나눠 알림)"""
        if not isinstance(name, str):
            raise ValueError(f"예약자 이름은 문자열이어야 합니다: {name!r}")
        if name.strip() == "":
            raise ValueError("예약자 이름이 비어 있습니다.")
        return name.strip()

    def add_reservation(self, name: str, when: str, guests: int) -> None:
        try:
            name = self._clean_name(name)
            if not is_count(guests) or guests == 0:
                raise ValueError(f"인원 수는 1 이상의 정수여야 합니다: {guests!r}")
            when = self._normalize(when)
            if any(r["name"] == name and r["when"] == when for r in self.reservations):
                raise ValueError(f"{name}님의 {when} 예약이 이미 있습니다.")
        except ValueError as e:
            print(f"[{self.branch_name}] 예약 실패 - {e}")
            return
        self.reservations.append({"name": name, "when": when, "guests": guests})

    def cancel_reservation(self, name: str, when: str) -> None:
        try:
            name = self._clean_name(name)
            when = self._normalize(when)
        except ValueError as e:
            print(f"[{self.branch_name}] 취소 실패 - {e}")
            return
        for r in self.reservations:
            if r["name"] == name and r["when"] == when:
                self.reservations.remove(r)
                print(f"[{self.branch_name}] {name}님의 {when} 예약이 취소되었습니다.")
                return
        print(f"[{self.branch_name}] 취소 실패 - {name}님의 {when} 예약이 없습니다.")

    def list_reservations(self) -> None:
        print(f"{self.branch_name} 예약 목록:")
        for r in self.reservations:
            print(f"- {r['name']}, {r['when']}, {r['guests']}명")

    @classmethod
    def sum_reservations(cls, branches) -> int:
        """지점 객체들(리스트 · 튜플 · 제너레이터 등)의 예약 건수 합계, 잘못된 값이면 TypeError"""
        if isinstance(branches, str):
            raise TypeError("지점 목록 대신 문자열이 들어왔습니다.")
        try:
            items = list(branches)
        except TypeError:
            raise TypeError(f"지점 목록이 아닙니다: {branches!r}") from None
        wrong = [b for b in items if not isinstance(b, cls)]
        if wrong:
            raise TypeError(f"지점 객체가 아닌 값이 있습니다: {wrong!r}")
        return sum(len(branch.reservations) for branch in items)


def example():
    restaurant1 = ReservationSystem("강남점")
    restaurant2 = ReservationSystem("홍대점")
    restaurant1.add_reservation("홍길동", "2024-05-20", 4)
    restaurant2.add_reservation("김철수", "2024-05-21", 2)
    restaurant1.list_reservations()
    restaurant2.list_reservations()
    total_reservations = ReservationSystem.sum_reservations([restaurant1, restaurant2])
    print(f"전체 레스토랑 예약 수: {total_reservations}")


check_output("클래스 응용 4", "예시", example,
             "강남점 예약 목록:\n- 홍길동, 2024-05-20, 4명\n홍대점 예약 목록:\n- 김철수, 2024-05-21, 2명\n"
             "전체 레스토랑 예약 수: 2")
```

### 심화 1. 시간 관리 프로그램

**요구사항**

- CSV 파일 경로와 남은 시간을 입력받아 할 수 있는 할 일을 고르고 소요 시간이 짧은 순으로 출력
- 같은 조건의 조합이 여럿일 때 선택 기준을 정해 코드에 반영

**문제지 예시**: 남은 시간 210분 → 가계부정리 30분, 운동하기 60분, 공부하기 120분

**구현**

- **기준 1, 할 일 개수 최대**: 짧은 일부터 담는 그리디로 최대 개수 k 계산 (개수 최대화에는 그리디가 최적)
- **기준 2, 남는 시간 최소**: 개수가 k인 조합 중 시간을 가장 많이 쓰는 조합, 같으면 먼저 만들어진 조합
- CSV는 utf-8 → utf-8-sig → cp949 순서로 읽기 시도 (한국어 Windows 기본 인코딩 cp949로 저장된 파일, Excel이 BOM을 붙여 저장한 파일 대응), 할 일 이름이 비었거나 소요 시간이 1 ~ 999999분(6자리)의 정수가 아닌 줄은 줄 번호와 함께 안내
- 예시 파일은 저장소에 남지 않도록 임시 폴더에 만들고 검증 뒤 삭제 (그래서 이 노트북 밖에서 `tasks.csv`를 쓰려면 문제지의 파일 생성 코드를 먼저 실행)

```python
import csv
import os
import re
import tempfile
from itertools import combinations


def make_tasks_csv(folder: str) -> str:
    """문제지 예시 tasks.csv를 folder에 만들고 경로 반환 (인코딩만 utf-8로 지정)"""
    path = os.path.join(folder, "tasks.csv")
    with open(path, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["할일", "소요시간"])
        writer.writeheader()
        for task, minutes in [("공부하기", 120), ("운동하기", 60), ("가계부정리", 30), ("영화보기", 150)]:
            writer.writerow({"할일": task, "소요시간": minutes})
    return path


def load_tasks(filename: str) -> list[tuple[str, int]]:
    """CSV → [(할 일, 소요 시간)], 열 · 인코딩 · 값이 잘못되면 한국어 안내가 담긴 ValueError"""
    for encoding in ("utf-8", "utf-8-sig", "cp949"):
        try:
            with open(filename, newline="", encoding=encoding) as f:
                reader = csv.DictReader(f)
                header, rows = reader.fieldnames or [], list(reader)
        except UnicodeDecodeError:
            continue
        if not {"할일", "소요시간"} <= set(header):
            continue  # 머리글이 다르면(예: BOM이 붙은 utf-8) 다음 인코딩으로 다시 읽기
        tasks = []
        for line, row in enumerate(rows, start=2):
            name, minutes = (row.get("할일") or "").strip(), (row.get("소요시간") or "").strip()
            valid = name and re.fullmatch(r"[0-9]{1,6}", minutes) and int(minutes) > 0  # 6자리 이하의 분
            if not valid:
                raise ValueError(f"CSV {line}번째 줄이 올바르지 않습니다: {name}, {minutes}")
            tasks.append((name, int(minutes)))
        return tasks
    raise ValueError("CSV를 읽을 수 없습니다 ('할일', '소요시간' 열과 인코딩 확인)")


def greedy_tasks(tasks, remaining_time: int) -> list[tuple[str, int]]:
    """기준 1: 짧은 일부터 시간 안에 담기"""
    chosen, used = [], 0
    for task in sorted(tasks, key=lambda t: t[1]):
        if used + task[1] > remaining_time:
            break
        chosen.append(task)
        used += task[1]
    return chosen


def suggest_tasks(tasks, remaining_time: int) -> list[tuple[str, int]]:
    k = len(greedy_tasks(tasks, remaining_time))
    if k == 0:
        return []
    feasible = [c for c in combinations(tasks, k) if sum(d for _, d in c) <= remaining_time]
    best = max(feasible, key=lambda c: sum(d for _, d in c))  # 기준 2: 남는 시간 최소
    return sorted(best, key=lambda t: t[1])


def time_manager_main() -> None:
    filename = input("CSV 파일 경로: ").strip()
    try:
        tasks = load_tasks(filename)
    except FileNotFoundError:
        print(f"파일을 찾을 수 없습니다: {filename}")
        return
    except OSError:  # 폴더 경로, 권한 없음 등
        print(f"파일을 열 수 없습니다: {filename}")
        return
    except ValueError as e:
        print(e)
        return
    remaining_time = read_int("남은 시간(분): ")
    selected = suggest_tasks(tasks, remaining_time)
    if not selected:
        print("남은 시간 안에 할 수 있는 일이 없습니다.")
        return
    print("선택된 할 일 목록:")
    for i, (name, duration) in enumerate(selected, start=1):
        print(f"{i}. {name} - 예상 소요 시간: {duration}분")


main = time_manager_main  # 문제지 실행 셀의 main()이 이 문제를 실행하도록


def run_in(folder: str, *answers) -> None:
    """folder를 작업 폴더로 삼아 실행 (화면에 절대 경로가 나오지 않도록 상대 경로 사용)"""
    here = os.getcwd()
    os.chdir(folder)
    try:
        with fake_inputs(*answers):
            time_manager_main()
    finally:
        os.chdir(here)


with tempfile.TemporaryDirectory() as folder:  # 저장소 밖 임시 폴더, 블록이 끝나면 삭제
    make_tasks_csv(folder)
    check_output("심화 1", "210분", lambda: run_in(folder, "tasks.csv", "210"),
                 "CSV 파일 경로: tasks.csv\n남은 시간(분): 210\n선택된 할 일 목록:\n"
                 "1. 가계부정리 - 예상 소요 시간: 30분\n2. 운동하기 - 예상 소요 시간: 60분\n"
                 "3. 공부하기 - 예상 소요 시간: 120분")
```

```python
with tempfile.TemporaryDirectory() as folder:
    tasks = load_tasks(make_tasks_csv(folder))
greedy, chosen = greedy_tasks(tasks, 200), suggest_tasks(tasks, 200)
print(f"200분: 그리디만 → {[n for n, _ in greedy]} (남는 시간 {200 - sum(d for _, d in greedy)}분), "
      f"기준 2 적용 → {[n for n, _ in chosen]} (남는 시간 {200 - sum(d for _, d in chosen)}분)")
verify("심화 1", "200분: 개수는 같고(2개) 남는 시간 110분 → 20분",
       len(greedy) == len(chosen) == 2 and [n for n, _ in chosen] == ["운동하기", "공부하기"]
       and 200 - sum(d for _, d in greedy) == 110 and 200 - sum(d for _, d in chosen) == 20)
verify("심화 1", "가장 짧은 일보다 시간이 적으면 빈 목록", suggest_tasks(tasks, 10) == [])


def write_csv(folder, name, text, encoding="utf-8"):
    with open(os.path.join(folder, name), "w", newline="", encoding=encoding) as f:
        f.write(text)


def bad_files(folder):
    for path in ("bad.csv", "minus.csv", "noname.csv", "long.csv", "없는파일.csv", "."):
        run_in(folder, path)


with tempfile.TemporaryDirectory() as folder:
    write_csv(folder, "cp949.csv", "할일,소요시간\n산책,20\n", encoding="cp949")
    write_csv(folder, "bom.csv", "할일,소요시간\n독서,40\n", encoding="utf-8-sig")
    write_csv(folder, "bad.csv", "할일,소요시간\n공부하기,abc\n")
    write_csv(folder, "minus.csv", "할일,소요시간\n휴식,-50\n")
    write_csv(folder, "noname.csv", "할일,소요시간\n ,10\n")
    write_csv(folder, "long.csv", "할일,소요시간\n여행,1000000\n")
    verify("심화 1", "cp949, BOM 붙은 utf-8 CSV도 읽음",
           load_tasks(os.path.join(folder, "cp949.csv")) == [("산책", 20)]
           and load_tasks(os.path.join(folder, "bom.csv")) == [("독서", 40)])
    check_output("심화 1", "잘못된 값 · 음수 시간 · 빈 이름 · 7자리 시간 · 없는 파일 · 폴더 경로",
                 lambda: bad_files(folder),
                 "CSV 파일 경로: bad.csv\nCSV 2번째 줄이 올바르지 않습니다: 공부하기, abc\n"
                 "CSV 파일 경로: minus.csv\nCSV 2번째 줄이 올바르지 않습니다: 휴식, -50\n"
                 "CSV 파일 경로: noname.csv\nCSV 2번째 줄이 올바르지 않습니다: , 10\n"
                 "CSV 파일 경로: long.csv\nCSV 2번째 줄이 올바르지 않습니다: 여행, 1000000\n"
                 "CSV 파일 경로: 없는파일.csv\n파일을 찾을 수 없습니다: 없는파일.csv\n"
                 "CSV 파일 경로: .\n파일을 열 수 없습니다: .")
```

### 심화 2. 수학 퍼즐: 소수의 합

**요구사항**

- 입력한 수 이하의 소수를 모두 더해 출력, 효율적인 방법 사용 (힌트: 에라토스테네스의 체)

**문제지 예시**: 입력 10 → `10 이하 소수의 합은 17입니다.`

**구현**

- 에라토스테네스의 체: 2부터 √n까지 소수의 배수를 지우고 남은 수를 더함
- 입력은 0 이상 1,000만 이하의 정수 (메모리를 과하게 쓰지 않도록 상한)

```python
def is_prime(num: int) -> bool:
    if num < 2:
        return False
    return all(num % i for i in range(2, int(num ** 0.5) + 1))  # √num까지만 나눠 봄


def sum_of_primes(n: int) -> int:
    if n < 2:
        return 0
    sieve = [True] * (n + 1)  # True = 소수 후보
    sieve[0] = sieve[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if sieve[i]:
            sieve[i * i::i] = [False] * len(range(i * i, n + 1, i))  # i의 배수 지우기
    return sum(i for i, prime in enumerate(sieve) if prime)


def prime_sum_main() -> None:
    n = read_int("어떤 수까지의 소수의 합을 구하시겠습니까?: ", high=10_000_000)
    print(f"{n} 이하 소수의 합은 {sum_of_primes(n)}입니다.")


main = prime_sum_main  # 문제지 실행 셀의 main()이 이 문제를 실행하도록


def example():
    with fake_inputs("10"):
        prime_sum_main()


check_output("심화 2", "10", example, "어떤 수까지의 소수의 합을 구하시겠습니까?: 10\n10 이하 소수의 합은 17입니다.")
```

```python
verify("심화 2", "체 결과 = 하나씩 판별한 합 (n = 0 ~ 499)",
       all(sum_of_primes(n) == sum(x for x in range(n + 1) if is_prime(x)) for n in range(500)))


def bad_inputs():
    with fake_inputs("inf", "20000000", "-1", "20"):
        prime_sum_main()


check_output("심화 2", "inf · 상한 초과 · 음수 거부", bad_inputs,
             "어떤 수까지의 소수의 합을 구하시겠습니까?: inf\n올바른 숫자를 입력하세요.\n"
             "어떤 수까지의 소수의 합을 구하시겠습니까?: 20000000\n0 이상 10,000,000 이하의 정수를 입력하세요.\n"
             "어떤 수까지의 소수의 합을 구하시겠습니까?: -1\n0 이상 10,000,000 이하의 정수를 입력하세요.\n"
             "어떤 수까지의 소수의 합을 구하시겠습니까?: 20\n20 이하 소수의 합은 77입니다.")
```

### 심화 3. 도서관 관리 시스템

**요구사항**

- `Book`, `Member`, `Rental`, `LibraryManagement` 클래스
- 도서(추가, 삭제, 검색), 회원(등록, 조회), 대여와 반납
- 모든 메서드에 입력 검증과 예외 처리, 도서 · 회원이 없을 때 등 상황에 맞는 안내

**문제지 예시**: 도서 2권, 회원 1명 → 대여 → 목록 출력 → 반납 → `대여 중인 도서: 없음`

**구현**

- 없는 도서 · 회원, 재고 없음, 빌리지 않은 책 반납은 전용 예외, 나머지 오류는 공통 `LibraryError`
- `@report_errors` 데코레이터가 예외를 `오류: ...` 안내로 바꿈
- `Rental`은 회원 ID, ISBN, 대여일, 반납 예정일(14일), 반납일 기록
- 대여는 기록 객체를 먼저 만든 뒤 재고 · 회원 목록을 바꿔 실패한 시도가 상태를 바꾸지 않음

```python
from datetime import date, timedelta
from functools import wraps


class LibraryError(Exception):
    """도서관 예외의 부모 클래스"""


class BookNotFoundError(LibraryError):
    def __init__(self, isbn):
        super().__init__(f"ISBN '{isbn}' 도서를 찾을 수 없습니다.")


class BookNotAvailableError(LibraryError):
    def __init__(self, title):
        super().__init__(f"'{title}' 도서는 모두 대여 중입니다.")


class MemberNotFoundError(LibraryError):
    def __init__(self, name):
        super().__init__(f"회원 '{name}'님을 찾을 수 없습니다.")


class NotRentedError(LibraryError):
    def __init__(self, name, title):
        super().__init__(f"'{name}' 회원님은 '{title}' 도서를 대여하고 있지 않습니다.")


def report_errors(method):
    """메서드에서 생긴 LibraryError를 '오류: ...' 안내로 변환"""
    @wraps(method)
    def wrapper(*args, **kwargs):
        try:
            return method(*args, **kwargs)
        except LibraryError as e:
            print(f"오류: {e}")
    return wrapper


class Book:
    def __init__(self, title: str, author: str, publication_year: int, isbn: str) -> None:
        self.title = title
        self.author = author
        self.publication_year = publication_year
        self.isbn = isbn
        self.total_copies = self.available_copies = 1

    def same_info(self, title, author, publication_year) -> bool:
        mine = (self.title, self.author, self.publication_year)
        return mine == (title, author, publication_year)

    def __str__(self) -> str:
        return f"{self.title} (저자: {self.author}, 출판년도: {self.publication_year})"


class Member:
    def __init__(self, name: str, member_id: int) -> None:
        self.name = name
        self.member_id = member_id
        self.rented_books = []

    def __str__(self) -> str:
        titles = ", ".join(b.title for b in self.rented_books) or "없음"
        return f"{self.name} (대여 중인 도서: {titles})"


class Rental:
    LOAN_DAYS = 14

    def __init__(self, book: Book, member: Member, rental_date: date | None = None) -> None:
        self.book, self.member = book, member
        self.member_id, self.isbn = member.member_id, book.isbn  # 회원 ID와 도서 ISBN 연결
        self.rental_date = rental_date or date.today()
        self.due_date = self.rental_date + timedelta(days=self.LOAN_DAYS)
        self.return_date = None  # 반납 전

    def close(self, return_date: date | None = None) -> None:
        self.return_date = return_date or date.today()
```

```python
class LibraryManagement:
    def __init__(self) -> None:
        self.books = {}  # ISBN → Book
        self.members = {}  # 이름 → Member
        self.rentals = []  # 반납된 기록까지 보관

    def _get_book(self, isbn) -> Book:
        if isbn not in self.books:
            raise BookNotFoundError(isbn)
        return self.books[isbn]

    def _get_member(self, name) -> Member:
        if name not in self.members:
            raise MemberNotFoundError(name)
        return self.members[name]

    def _active_rental(self, book, member):
        active = (r for r in self.rentals
                  if r.book is book and r.member is member and r.return_date is None)
        return next(active, None)

    @report_errors
    def add_book(self, title, author, publication_year, isbn) -> None:
        if not all(isinstance(v, str) and v.strip() for v in (title, author, isbn)):
            raise LibraryError("제목, 저자, ISBN은 빈 값일 수 없습니다.")
        if isinstance(publication_year, bool) or not isinstance(publication_year, int):
            raise LibraryError(f"출판년도는 정수여야 합니다: {publication_year!r}")
        if isbn in self.books:
            book = self.books[isbn]
            if not book.same_info(title, author, publication_year):  # 같은 ISBN에 다른 정보는 거부
                raise LibraryError(f"ISBN '{isbn}' 번호는 이미 '{book.title}'에 쓰이고 있습니다.")
            book.total_copies += 1
            book.available_copies += 1
            print(f"'{title}' 도서가 1권 더 추가되었습니다. (보유 {book.total_copies}권)")
            return
        self.books[isbn] = Book(title, author, publication_year, isbn)
        print(f"'{title}' (저자: {author}, 출판년도: {publication_year}) 도서가 추가되었습니다.")

    @report_errors
    def remove_book(self, isbn) -> None:
        book = self._get_book(isbn)
        if book.available_copies < book.total_copies:
            raise LibraryError(f"'{book.title}' 도서는 대여 중이라 삭제할 수 없습니다.")
        del self.books[isbn]
        print(f"'{book.title}' 도서가 삭제되었습니다.")

    def search_books(self, keyword) -> list:
        """제목 · 저자에 keyword가 들어간 도서 목록 (검색어가 잘못되면 안내 후 빈 목록)"""
        if not isinstance(keyword, str) or keyword.strip() == "":
            print("오류: 검색어가 비어 있습니다.")
            return []
        found = [b for b in self.books.values() if keyword in b.title or keyword in b.author]
        print(f"'{keyword}' 검색 결과: {len(found)}건")
        for b in found:
            print(f"- {b} [대여 가능 {b.available_copies}/{b.total_copies}권]")
        return found

    def print_books(self) -> None:
        print("도서 목록:")
        for book in self.books.values():
            print(f"- {book}")

    @report_errors
    def add_member(self, name) -> None:
        if not isinstance(name, str) or name.strip() == "":
            raise LibraryError("회원 이름이 비어 있습니다.")
        if name in self.members:  # 이름으로 회원을 찾으므로 같은 이름은 거부
            raise LibraryError(f"회원 '{name}'님은 이미 등록되어 있습니다.")
        self.members[name] = Member(name, len(self.members) + 1)  # 회원번호 = 등록 순서
        print(f"회원 '{name}'님이 등록되었습니다.")

    def print_members(self) -> None:
        print("회원 목록:")
        for member in self.members.values():
            print(f"- {member}")

    @report_errors
    def rent_book(self, isbn, member_name) -> None:
        book, member = self._get_book(isbn), self._get_member(member_name)
        if self._active_rental(book, member):
            raise LibraryError(f"'{member_name}' 회원님은 이미 '{book.title}' 도서를 대여 중입니다.")
        if book.available_copies == 0:
            raise BookNotAvailableError(book.title)
        rental = Rental(book, member)  # 기록을 먼저 만든 뒤 상태 변경
        book.available_copies -= 1
        member.rented_books.append(book)
        self.rentals.append(rental)
        print(f"'{member_name}' 회원님이 '{book.title}' 도서를 대여하였습니다.")

    @report_errors
    def return_book(self, isbn, member_name) -> None:
        book, member = self._get_book(isbn), self._get_member(member_name)
        rental = self._active_rental(book, member)
        if rental is None:
            raise NotRentedError(member_name, book.title)
        rental.close()
        book.available_copies += 1
        member.rented_books.remove(book)
        print(f"'{member_name}' 회원님이 '{book.title}' 도서를 반납하였습니다.")


def example():
    library = LibraryManagement()
    library.add_book("1984", "조지 오웰", 1949, "978-0451524935")
    library.add_book("앵무새 죽이기", "하퍼 리", 1960, "978-0446310789")
    library.add_member("홍길동")
    library.rent_book("978-0451524935", "홍길동")
    library.print_books()
    library.print_members()
    library.return_book("978-0451524935", "홍길동")
    library.print_members()


check_output("심화 3", "예시", example,
             "'1984' (저자: 조지 오웰, 출판년도: 1949) 도서가 추가되었습니다.\n"
             "'앵무새 죽이기' (저자: 하퍼 리, 출판년도: 1960) 도서가 추가되었습니다.\n"
             "회원 '홍길동'님이 등록되었습니다.\n'홍길동' 회원님이 '1984' 도서를 대여하였습니다.\n"
             "도서 목록:\n- 1984 (저자: 조지 오웰, 출판년도: 1949)\n- 앵무새 죽이기 (저자: 하퍼 리, 출판년도: 1960)\n"
             "회원 목록:\n- 홍길동 (대여 중인 도서: 1984)\n'홍길동' 회원님이 '1984' 도서를 반납하였습니다.\n"
             "회원 목록:\n- 홍길동 (대여 중인 도서: 없음)")
```

```python
GRADED = ["기초 1", "기초 2", "기초 3", "기초 4", "기초 5", "기초 6", "기초 7", "응용 1", "응용 2", "응용 3",
          "클래스 기초 1", "클래스 기초 2", "클래스 응용 1", "클래스 응용 2", "클래스 응용 3", "클래스 응용 4"]
EXTRA = ["심화 1", "심화 2", "심화 3"]


def pad(text: str, width: int) -> str:
    """한글은 2칸으로 세어 오른쪽을 공백으로 채움"""
    return text + " " * (width - sum(2 if ord(ch) > 0x1100 else 1 for ch in text))


def checks_of(key: str) -> list[bool]:
    return [ok for (k, _), ok in RESULTS.items() if k == key]


def passed_problem(key: str) -> bool:
    """기록이 1개 이상이고 모두 통과해야 통과"""
    checks = checks_of(key)
    return bool(checks) and all(checks)


print(f"{pad('문제', 16)}{pad('통과 / 항목', 14)}결과")
for key in GRADED + EXTRA:
    checks = checks_of(key)
    note = "  (채점 제외)" if key in EXTRA else ""
    mark = "✅" if passed_problem(key) else "❌"
    print(f"{pad(key, 16)}{pad(f'{sum(checks)} / {len(checks)}', 14)}{mark}{note}")
graded_items = [ok for (k, _), ok in RESULTS.items() if k in GRADED]
passed = sum(passed_problem(key) for key in GRADED)
print(f"\n채점 대상 {len(GRADED)}문제 중 {passed}문제 통과")
print(f"채점 대상 대조 · 검증 {sum(graded_items)} / {len(graded_items)}항목, "
      f"전체 {sum(RESULTS.values())} / {len(RESULTS)}항목")
if passed != len(GRADED):
    raise AssertionError("통과하지 못한 문제가 있음")
# 문제지 예시 대조 (문제지에 예시로 적힌 호출만 따로 집계)
EXAMPLE_CASES = [
    ("기초 1", "입력 7"), ("기초 1", "입력 30"), ("기초 2", "codeit"), ("기초 2", "python programming"),
    ("기초 3", "10, 20"), ("기초 3", "30, 30"), ("기초 4", "Python"), ("기초 4", "빈 문자열"),
    ("기초 5", "3"), ("기초 5", "10"), ("기초 6", "입력 9"), ("기초 6", "입력 0"),
    ("기초 7", "42, 7"), ("기초 7", "135, 0"), ("응용 1", "aaabbaaa"), ("응용 1", "aabcccccaaa"),
    ("응용 2", "10 → twenty → 10 → 20"), ("응용 3", "입력 5"),
    ("클래스 기초 1", "90분"), ("클래스 기초 2", "Jane Doe"),
    ("클래스 응용 1", "예시"), ("클래스 응용 2", "예시"), ("클래스 응용 3", "예시"), ("클래스 응용 4", "예시"),
    ("심화 1", "210분"), ("심화 2", "10"), ("심화 3", "예시"),
]
example_ok = [RESULTS.get(case, False) for case in EXAMPLE_CASES]  # 기록이 없으면 불일치로 셈
problems = len({key for key, _ in EXAMPLE_CASES})
print(f"\n문제지 예시 대조 {sum(example_ok)} / {len(EXAMPLE_CASES)}건 일치 ({problems}문제, 심화 포함)")
if not len(EXAMPLE_CASES) == sum(example_ok) == 27:  # 목록이 줄거나 비어도 통과로 세지 않도록 27건에 고정
    raise AssertionError("문제지 예시 27건 중 다르거나 빠진 것이 있음")
```

### 설계에서 고른 것

- 오류 알림
  - 인자로 값을 받는 함수 · 생성자: 잘못된 인자에 예외(`ValueError`, `TypeError`)
  - 예외 사례: `divide_numbers`는 문제지 요구대로 `"잘못된 입력입니다."` 반환, `count_down`은 타이머 화면에 안내 문구 출력
  - 사용자 입력을 받는 대화형 기능과 계좌 · 투표 · 예약 같은 관리 기능: 안내 문구 출력 (단, 합계를 돌려주는 `sum_reservations`는 잘못된 목록에 `TypeError`)
- 소수 계산: 입력한 소수, 나눗셈, 계좌 금액, 평균 급여는 분수(`Fraction`)로 정확히 계산
  - 화면에는 10진수 그대로(`0.1 + 0.2 = 0.3`), 반환값은 정수면 int, 아니면 float
- 다루는 범위: 사용자가 입력하는 숫자는 정수부 · 소수부 각 30자리 이하, 금액 · 급여 · 잔액은 0.01원 단위로 사사오입해 10조 원 미만 (정수부 13자리 + 소수 2자리 = 유효숫자 15자리라 `get_balance()` 등이 돌려주는 float이 화면 표시와 정확히 같음)
- 답을 다른 방법으로 한 번 더 구해 비교: 총합 공식 ↔ 반복문, 직접 구현한 압축 ↔ `itertools.groupby`, 소수의 체 ↔ 하나씩 판별
- 실패한 시도는 저장된 상태(잔액 · 득표 · 예약 · 대여 기록)를 바꾸지 않음
- 시간 관리는 그리디로 개수를 정한 뒤 조합으로 남는 시간 최소화
