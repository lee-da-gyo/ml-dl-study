### 1-1. `A` 표 만들기와 크기 · `pd.DataFrame()` · `shape` <sub>원본 00 · 1~2장</sub>

> 🎯 **데이터가 10만 행이면 눈으로 세지 않고 크기를 어떻게 확인하는가?**

💡 **쉽게 말하면**

- DataFrame은 **엑셀 시트 한 장**.
- 열마다 이름이 붙고, 행마다 0부터 번호(index)가 붙음.
- `shape`는 시트의 **행 수 × 열 수**.

<img src="images/pandas_dataframe.png" alt="DataFrame = index · 열 이름 · 열마다 자료형" width="659">

**실행 흐름**

1. {열 이름: 값 목록} 사전
2. DataFrame
3. shape · columns · len · size

**먼저 예측** · 학생 6명 × 4개 항목 표의 `shape`는? — 답은 코드 실행 결과 아래에

````python
# 학생 6명 표 만들기: {열 이름: 값 목록}
students = pd.DataFrame({  # 사전 → 표 (키 = 열 이름)
    "name": ["민수", "지수", "영희", "철수", "유진", "현우"],
    "score": [78, 95, 84, 65, 91, 88],
    "city": ["서울", "부산", "서울", "대전", "서울", "부산"],
    "study_hours": [3, 7, 5, 2, 6, 5],
})
print(students)            # 표 전체를 눈으로 확인


# 빈칸 00-① 크기와 열 이름
print("shape:", students.shape)                         # (행 수, 열 수)
print("columns:", students.columns.tolist())            # 열 이름 목록
print("len:", len(students), "| size:", students.size)  # 행 수 · 칸 수
````

**결과** · 예측 정답은 `(6, 4)` — (행 수, 열 수) 순서

- `민수 78 서울 3` — 사전의 **열 하나 = 값 목록 하나**, 왼쪽 0 ~ 5는 자동 번호(index)
- `shape: (6, 4)` — 원본 질문 답: 행 6 · 열 4
- `columns: ['name', 'score', 'city', 'study_hours']` — 열 이름 목록 (오타 확인용)
- `len: 6 | size: 24` — len = 행 수, size = 6 × 4 칸

### 1-2. `A` 자료형과 일부 보기 · `dtypes` · `head()` · `tail()` <sub>원본 00 · 2장</sub>

> 🎯 **점수 `88`과 도시 `"부산"`을 pandas는 같은 종류로 보는가?**

💡 **쉽게 말하면**

- 열마다 **한 가지 규격 상자**를 씀(NumPy 배열과 같음).

**먼저 예측** · `score` 열과 `city` 열의 dtype은? — 답은 코드 실행 결과 아래에

````python
# 열마다 자료형
print(students.dtypes)  # 열마다 자료형 (계산 가능한가)


# 처음 3행
print(students.head(3))  # 기본값 5 → 3으로 지정


# 빈칸 00-② 마지막 3행
print(students.tail(3))  # 뒤쪽 이상한 행이 없는지
````

### 2-1. `A` 열 고르기 · `df["열"]` · `df[["열"]]` <sub>원본 00 · 3장 · 원본 02 · Mission 1</sub>

> 🎯 **한 열을 고르면 Series, 두 열을 고르면 DataFrame이 되는 이유는?**

💡 **쉽게 말하면**

- Series는 **이름표 붙은 한 줄짜리 목록**, DataFrame은 **그 목록 여러 개를 나란히 붙인 표**.
- 대괄호 안에 **목록**을 넣으면 표가 나옴.

**실행 흐름**

1. 한 열 · 두 열 선택
2. type · shape 비교
3. [['city']]
4. 4개 열
5. df[0] 에러

**먼저 예측** · `students[["city"]]`는 Series일까 DataFrame일까? — 답은 코드 실행 결과 아래에

````python
# 빈칸 00-③ city 한 열 · name과 city 두 열
city = students["city"]                 # 열 이름 하나 → Series
name_city = students[["name", "city"]]  # 열 이름 목록 → DataFrame
print(name_city.head(2))                # 두 열이 표로 나옴


# 빈칸 00-④ 두 결과의 type · shape
print(type(city).__name__, city.shape,                        # 1차원 vs 2차원 비교
      "|", type(name_city).__name__, name_city.shape)
one_col = students[["city"]]                                  # 목록이면 한 열도 표
print("[['city']] →", type(one_col).__name__, one_col.shape)  # 한 열도 목록이면 표


# 빈칸 02-⑤ 주문 데이터에서 4개 열만
print(orders[["product_name", "category", "price", "quantity"]].head(2))  # 분석에 필요한 열만


# [] 에 행 번호를 넣으면 열 이름으로 찾음
try:
    students[0]            # 0이라는 열을 찾음
except KeyError as e:      # 없는 열 이름 에러를 잡음
    print("KeyError:", e)  # 0을 열 이름으로 찾았다는 뜻
````

### 2-2. `A` 라벨과 위치로 고르기 · `loc` · `iloc` <sub>원본 00 · 4장</sub>

> 🎯 **`loc`과 `iloc`은 무엇을 기준으로 찾으며, 정렬한 뒤에는 결과가 어떻게 달라지는가?**

💡 **쉽게 말하면**

- `loc`은 **이름표(라벨)로 찾기**, `iloc`은 **앞에서 몇 번째 자리로 찾기**.
- 정렬하면 이름표는 사람을 따라가고 자리는 새로 매겨짐.

<img src="images/pandas_loc_iloc.png" alt="정렬 뒤 loc은 라벨 0, iloc은 첫 줄" width="644">

**실행 흐름**

1. loc[0] · loc[0:2, 열]
2. iloc[0:2, 0:2] · iloc[0:3, 0:2]
3. 정렬 뒤 loc[0] vs iloc[0]

**먼저 예측** · `students.loc[0:2]`는 몇 행? — 답은 코드 실행 결과 아래에

````python
# 빈칸 00-⑤ loc: 라벨 기준 (끝 라벨 포함)
print(students.loc[0].to_dict())             # 라벨 0 행 → Series
print(students.loc[0:2, ["name", "score"]])  # 라벨 0 · 1 · 2 → 3행


# 빈칸 00-⑤ iloc: 위치 기준 (끝 위치 미포함)
print(students.iloc[0:2, 0:2])                                 # 위치 0 · 1행 × 0 · 1열
print("iloc[0:3, 0:2] shape:", students.iloc[0:3, 0:2].shape)  # 끝 위치 3은 빠짐


# 정렬하면 라벨과 위치가 달라짐
ranked = students.sort_values("score", ascending=False)  # 라벨은 그대로
print("ranked 라벨 순서:", ranked.index.tolist())        # 라벨이 사람을 따라감
print("loc[0] →", ranked.loc[0, "name"],                 # 라벨 0
      "| iloc[0] →", ranked.iloc[0]["name"])             # 지금 첫 줄
````

**결과** · 예측 정답은 `3`행 — loc은 끝 라벨 2를 포함

- `{'name': '민수', 'score': 78, 'city': '서울', 'study_hours': 3}` — 원본 빈칸: 첫 번째 학생 = 라벨 0
- `2 영희 84` — `loc[0:2, ...]` → 0 · 1 · 2번 **3행** (끝 포함)
- `iloc[0:3, 0:2] shape: (3, 2)` — `iloc`은 끝 위치 3 미포함 → 3행
- `ranked 라벨 순서: [1, 4, 5, 2, 0, 3]` — 정렬해도 라벨은 원래 번호를 들고 다님

### 3-1. `A` 조건 하나로 행 고르기 · `df[mask]` <sub>원본 00 · 5장 · 원본 02 · Mission 2</sub>

> 🎯 **80점 이상인 학생만 고르려면 어떤 두 단계를 거치는가?**

💡 **쉽게 말하면**

- 조건은 **행마다 O · X를 적은 출석부**(mask).
- 표에 출석부를 대면 O인 줄만 남음 (NumPy mask와 같은 원리).

<img src="images/pandas_mask.png" alt="조건 → True/False → True 행만" width="748">

**실행 흐름**

1. 조건
2. True/False 목록
3. df[mask]
4. 85점 · 부산
5. 주문 5만 원 이상 · 전자기기

**먼저 예측** · `students["score"] >= 80`의 결과는 표일까? — 답은 코드 실행 결과 아래에

````python
# 조건 → 행마다 True/False
mask = students["score"] >= 80  # 행마다 True/False
print("mask:", mask.tolist())   # 6개 답을 눈으로
print(students[mask])           # True 행만


# 빈칸 00-⑥ · ⑦ 85점 이상 · 부산 학생
print("85점 이상:", students[students["score"] >= 85]["name"].tolist())  # 조건 → 고르기 → 이름 열
# 문자열 비교도 같은 흐름
print("부산     :", students[students["city"] == "부산"]["name"].tolist())


# 빈칸 02-⑥ · ⑦ 가격 5만 원 이상 (True/False → 선택)
is_high = orders["price"] >= 50000                                        # 1000개 True/False
print("앞 5개:", is_high.head().tolist(), "| True 개수:", is_high.sum())  # True = 1로 세기
print("5만 원 이상 주문:", orders[is_high].shape)                         # 행만 줄고 열은 그대로


# 빈칸 02-⑧ 전자기기 주문만
electronics = orders[orders["category"] == "전자기기"]  # 카테고리 조건
print("전자기기 주문:", len(electronics))               # 남은 행 수
````

**결과** · 예측 정답은 표가 아니라 6개짜리 True/False Series

- `mask: [False, True, True, False, True, True]` — 학생마다 하나씩, 민수 · 철수만 False
- `85점 이상: ['지수', '유진', '현우']` — 원본 00 빈칸 답
- `부산 : ['지수', '현우']` — 문자열 비교도 같은 흐름

### 3-2. `A` 조건 묶기 · `&` · `|` · `isin()` <sub>원본 00 · 5장 · 원본 02 · Mission 2</sub>

> 🎯 **조건 두 개를 함께 쓸 때 왜 `&` · `|`와 괄호가 필요한가?**

💡 **쉽게 말하면**

- `and`는 **질문 하나에 예/아니오 하나**인데, mask는 행마다 답이 있어서 **줄마다 답하는** `&` · `|`가 필요.
- `&`가 `>=`보다 먼저 계산돼서 괄호로 순서를 정함.

**실행 흐름**

1. 서울 & 80↑
2. 부산 & 85↑ · 서울 | 부산
3. isin
4. 주문 2조건
5. 괄호 없음 에러

**먼저 예측** · 서울 또는 부산 학생은 몇 명? — 답은 코드 실행 결과 아래에

````python
# 서울 그리고 80점 이상 (조건마다 괄호)
is_seoul_80 = (students["city"] == "서울") & (students["score"] >= 80)  # 조건마다 괄호 + &
print("서울 & 80↑:", students[is_seoul_80]["name"].tolist())            # 둘 다 True인 학생


# 빈칸 00-⑧ · ⑨ 부산 & 85점 이상 · 서울 또는 부산
busan_85 = (students["city"] == "부산") & (students["score"] >= 85)    # 그리고
seoul_or_busan = (students["city"] == "서울") | (students["city"] == "부산")  # 또는
print("부산 & 85↑:", students[busan_85]["name"].tolist())              # 빈칸 00-⑧ 답
print("서울 | 부산:", students[seoul_or_busan]["name"].tolist())       # 빈칸 00-⑨ 답
same = students["city"].isin(["서울", "부산"]).equals(seoul_or_busan)  # isin = | 여러 번
print("isin 같은가:", same)                                            # 두 mask가 완전히 같은가


# 빈칸 02-⑨ · ⑩ 전자기기 & 10만 원 이상 · 서울 & 전자기기
is_elec = orders["category"] == "전자기기"                   # 조건을 변수로 재사용
is_seoul = orders["region"] == "서울"                        # '서울 '처럼 공백 붙은 값은 False
print("전자기기 & 10만↑:", len(orders[is_elec & (orders["price"] >= 100000)]))  # 두 조건 동시
print("서울 & 전자기기 :", len(orders[is_seoul & is_elec]))  # 공백 정리 전 개수


# 괄호를 빼면 & 가 먼저 계산됨: "서울" & score 부터
try:
    students[students["city"] == "서울" & students["score"] >= 80]  # 일부러 괄호 없이
except TypeError:                                                   # & 계산 순서 에러를 잡음
    print("괄호 없음 → TypeError")                                  # 괄호가 필요한 이유
````

**결과** · 예측 정답은 `5`명 — 대전의 철수만 빠짐

- `부산 & 85↑: ['지수', '현우']` — 원본 00 빈칸 답
- `서울 | 부산: ['민수', '지수', '영희', '유진', '현우']` — 원본 00 빈칸 답
- `isin 같은가: True` — 같은 열의 "또는"은 `isin` 한 번으로

### 4-1. `B` 기본 통계 · `mean()` · `agg()` <sub>원본 00 · 6장</sub>

> 🎯 **점수를 하나씩 더하지 않고 평균 · 최대 · 합계를 구하려면?**

💡 **쉽게 말하면**

- 열 하나(Series)에 **계산 버튼**이 달려 있음.
- `.mean()` · `.max()` · `.sum()`을 누르면 그 열 전체를 계산.

**실행 흐름**

1. score 통계 4개
2. study_hours 통계 3개

**먼저 예측** · 점수 평균은 80보다 클까? — 답은 코드 실행 결과 아래에

````python
# 한 열의 기본 통계
score = students["score"]                             # 한 열 = Series
print("평균:", score.mean(), "| 최대:", score.max(),  # Series 계산 버튼
      "| 최소:", score.min(), "| 합계:", score.sum())


# 빈칸 00-⑩ 공부 시간 평균 · 최댓값 · 합계
hours = students["study_hours"]                   # 공부 시간 열
print("공부 시간 평균:", round(hours.mean(), 2),  # 소수 둘째 자리까지
      "| 최대:", hours.max(), "| 합계:", hours.sum())


# 여러 통계를 한 번에
# 통계 여러 개를 표로
print(students[["score", "study_hours"]].agg(["mean", "max", "sum"]).round(2))
````

- `공부 시간 평균: 4.67 | 최대: 7 | 합계: 28` — 원본 빈칸 답 (28 ÷ 6 = 4.67)

### 4-3. `B` 정렬 · `sort_values()` <sub>원본 00 · 8장 · 원본 02 · Mission 3</sub>

> 🎯 **점수가 높은 순서로 보고, 동점이면 다른 기준을 쓰려면?**

💡 **쉽게 말하면**

- `sort_values`는 **기준 열로 줄 세우기**.
- 기본은 작은 값부터, `ascending=False`면 큰 값부터.
- 동점이면 **두 번째 기준**을 줌.

**실행 흐름**

### 5-1. `A` 결측 찾기 · `isna().sum()` <sub>원본 02 · Mission 3 ~ 4</sub>

💡 **쉽게 말하면**

- NaN은 **빈 칸**.
- 빈 칸이 들어간 계산은 답도 빈 칸이 되어 조용히 번짐.
- `isna()`로 빈 칸에 O 표시 → 열마다 더하면 개수.

**실행 흐름**

1. 할인 적용 미리 계산

**먼저 예측** · 할인 적용 금액 1000개 중 NaN은 몇 개? — 답은 코드 실행 결과 아래에

````python
# 할인 적용 금액을 미리 계산 → 비는 칸 확인
preview = orders["sales"] * (1 - orders["discount"])  # 저장하지 않고 미리 계산
print("NaN 인 행 번호 앞 5개:", preview[preview.isna()].index[:5].tolist())  # 빈 결과가 어디인가
print("NaN 개수:", preview.isna().sum())              # 빈 결과 수


# 빈칸 02-⑬ 열마다 결측 개수 (결측이 있는 열만)
missing = orders.isna().sum()  # True = 1 로 세어 열마다 합
print(missing[missing > 0])    # 결측이 있는 열만


# NaN 이 섞이면 계산 결과도 NaN
print("10000 × (1 - NaN) =", 10000 * (1 - np.nan))  # NaN은 계산에 번짐
````

### 5-2. `A` 결측 처리 · 채우기와 지우기 · `fillna()` · `dropna()` <sub>원본 02 · Mission 4</sub>

> 🎯 **같은 NaN인데 `discount`와 `region`은 왜 다르게 처리하는가?**

💡 **쉽게 말하면**

- 빈 칸의 **뜻**이 다름.
- `discount` 빈 칸 = 할인 없음(0이 확실), `region` 빈 칸 = 모름(추측하면 거짓말).
- 지우면 그 주문의 매출까지 사라짐.

**실행 흐름**

1. dropna 결과만 확인
2. discount fillna(0)
3. final_sales
4. region은 그대로

**먼저 예측** · `dropna()`를 하면 1000행 중 몇 행이 남을까? — 답은 코드 실행 결과 아래에

````python
# 지우면 얼마나 사라지는가 (실제로 지우지는 않음)
print("dropna() 하면      :", len(orders), "→", len(orders.dropna()), "행")    # 지우면 사라질 행 수
print("discount 기준 dropna:", len(orders.dropna(subset=["discount"])), "행")  # 한 열만 기준


# 빈칸 02-⑭ discount: 비어 있음 = 할인 없음 → 0
orders["discount"] = orders["discount"].fillna(0)         # 빈 칸 = 할인 없음
print("discount 결측:", orders["discount"].isna().sum())  # 0개인지 확인


# 빈칸 02-⑮ 할인 적용 후 금액
orders["final_sales"] = orders["sales"] * (1 - orders["discount"])  # 빈칸 02-⑮ 할인 적용
print("final_sales 결측:", orders["final_sales"].isna().sum())      # 모든 행 계산됨


# region: 알 수 없는 값 → 채우지도 지우지도 않음
print("region 결측 그대로:", orders["region"].isna().sum())  # 모르는 값은 그대로
````

- `to_datetime`으로 진짜 달력 날짜로 바꿔야 요일 · 기간 계산이 됨.

**실행 흐름**

2. to_datetime

**먼저 예측** · 변환 전 `order_date`의 dtype은? — 답은 코드 실행 결과 아래에

````python
# 빈칸 02-⑯ 날짜처럼 보이는 열의 dtype
dates = orders["order_date"]                               # 날짜 열
print("변환 전:", dates.dtype, "| 첫 값:", dates.iloc[0])  # 화면은 날짜, 실제는 str


# 문자열 → 날짜
orders["order_date"] = pd.to_datetime(orders["order_date"])  # 문자열 → 날짜
dates = orders["order_date"]                                 # 날짜 열
print("변환 후:", dates.dtype)                               # datetime64
print("기간:", dates.min().date(), "~", dates.max().date())  # 첫날 · 마지막 날
print("요일 앞 3개:", dates.dt.day_name().head(3).tolist())  # 날짜일 때만 .dt


# 형식이 섞인 날짜 · 잘못된 날짜
mixed = pd.Series(["2026/08/05", "2026-08-06"])  # / 와 - 가 섞임
mixed = pd.to_datetime(mixed, format="mixed")    # 형식이 섞여도 읽음
print(mixed.dt.strftime("%Y-%m-%d").tolist())    # 같은 형식으로 출력
bad = pd.Series(["2026-07-01", "7월 1일"])       # 읽을 수 없는 값 포함
bad = pd.to_datetime(bad, errors="coerce")       # 못 읽으면 NaT
print("NaT 인가:", bad.isna().tolist())          # 못 읽은 자리 찾기
````

### 5-5. `A` 중복 찾기 · `duplicated()` <sub>원본 02 · Mission 4</sub>

> 🎯 **같은 주문이 두 번 저장됐는지 무엇으로 찾고, 행 중복과 Key 중복은 어떻게 다른가?**

💡 **쉽게 말하면**

- **Row Duplicate**는 복사 붙여넣기한 똑같은 줄, **Key Duplicate**는 주문번호만 같고 내용이 다른 줄(어느 쪽이 맞는지 판단 필요).

**실행 흐름**

1. 완전 중복 개수
2. 짝 나란히 보기
3. order_id 중복
4. 값이 충돌하는 행의 상태

**먼저 예측** · 완전 중복 5개, order_id 중복 8개면 내용이 다른 충돌은 몇 개? — 답은 코드 실행 결과 아래에

````python
# 빈칸 02-⑲ 모든 값이 같은 행 (두 번째부터 True)
print("완전 중복:", orders.duplicated().sum())  # 모든 열이 같은 행


# 빈칸 02-⑳ 중복된 행 직접 보기 (keep=False → 짝 모두)
dup_cols = ["order_id", "product_name", "quantity",               # 비교할 열만
            "payment_method", "record_status"]
row_dup = orders.duplicated(keep=False)                           # 짝 모두 True
print(orders[row_dup].sort_values("order_id")[dup_cols].head(4))  # 같은 행끼리 나란히


# 빈칸 02-㉑ order_id 만 같은 행 (Key Duplicate)
id_dup = orders["order_id"].duplicated(keep=False)                 # ID만 같은 행 모두
print("order_id 중복:", orders["order_id"].duplicated().sum(),     # 두 번째부터 셈
      "| 관련 행:", id_dup.sum())
key_only = id_dup & ~row_dup                                       # ID 는 같고 값은 다름
print(orders[key_only].sort_values("order_id")[dup_cols].head(4))  # 값이 다른 짝
status = orders.loc[key_only, "record_status"].value_counts()      # 충돌 행의 상태
print("값이 충돌하는 행의 상태:", status.to_dict())                # 한쪽은 수집 오류
````

### 5-6. `A` 분석용 데이터 확정 · `drop_duplicates()` · `copy()` <sub>원본 02 · Analysis-Ready Gate</sub>

> 🎯 **raw 1000행에서 분석에 쓸 논리 주문 992행을 어떻게 확정하는가?**

💡 **쉽게 말하면**

- **검문소(Gate)**.
- 믿을 수 없는 행(수집 오류)을 먼저 내보내고, 같은 주문은 한 줄만 통과시킨 뒤 "행 수 = 고유 ID 수"를 확인해야 문을 열어 줌.

<img src="images/pandas_gate.png" alt="raw 1000 → verified 997 → logical 992" width="689">

**실행 흐름**

1. ingestion_error 제외(997)
2. order_id 하나당 첫 행(992)
3. assert
4. 공식 매출

**먼저 예측** · Gross Sales는 얼마? — 답은 코드 실행 결과 아래에

````python
# ingestion_error 제외 → order_id 하나당 한 행
is_ok = orders["record_status"].ne("ingestion_error")  # ne = !=
verified_orders = orders.loc[is_ok].copy()             # 믿을 행만 새 표로
logical_orders = verified_orders.drop_duplicates(      # ID당 첫 행만
    subset="order_id", keep="first").copy()
print(pd.Series({                                      # 단계별 행 수 한 번에
    "raw_rows": len(orders),
    "verified_rows": len(verified_orders),
    "logical_rows": len(logical_orders),
    "unique_order_id": logical_orders["order_id"].nunique(),
}).to_dict())


# 게이트 통과 확인 (한 주문 = 한 행) → 공식 매출
assert len(logical_orders) == logical_orders["order_id"].nunique()              # 행 수 = 고유 ID
gross = logical_orders["sales"].sum()                                           # 할인 전 합계
final = logical_orders["final_sales"].sum()                                     # 할인 후 합계
print(f"Gross {gross:,} | Discount {gross - final:,.0f} | Final {final:,.0f}")  # 공식 매출


# 원본 orders 는 그대로
print("orders 행 수 그대로:", len(orders))  # 원본은 안 바뀜
````

**결과** · 예측 정답은 `78,000,000`원 — 할인 4,000,000원, 최종 74,000,000원

### 6-1. `A` 그룹별 합계 · `groupby()` <sub>원본 02 · Mission 5</sub>

> 🎯 **카테고리가 100개여도 카테고리별 매출을 한 줄로 구하려면?**

💡 **쉽게 말하면**

- `groupby`는 **같은 카테고리끼리 바구니에 나눠 담기 → 바구니마다 계산 → 결과를 한 표로**(Split → Apply → Combine).

<img src="images/pandas_groupby.png" alt="groupby: 나누기 → 계산 → 합치기" width="734">

**실행 흐름**

1. 전체 매출
2. 카테고리별 합계
3. 높은 순
4. 건수 1위 vs 총매출 1위

**먼저 예측** · 주문 건수 1위와 총매출 1위는 같은 카테고리일까? — 답은 코드 실행 결과 아래에

````python
# 전체 합계 하나로는 무엇이 매출을 만들었는지 모름
print(f"전체 매출: {logical_orders['final_sales'].sum():,.0f}")  # 전체 하나


# 빈칸 02-㉒ 카테고리별 매출 → 높은 순
by_cat = logical_orders.groupby("category")    # 카테고리별로 나누기
cat_sales = by_cat["final_sales"].sum()        # 열 고르기 → 합계
print(cat_sales.sort_values(ascending=False))  # 빈칸 02-㉒ 높은 순


# 빈칸 02-㉓ · ㉔ 주문 건수 · 총매출 1위
cat_count = by_cat.size()                                # 그룹마다 행 수
print("건수 1위:", cat_count.idxmax(), cat_count.max(),  # 두 1위 비교
      "| 총매출 1위:", cat_sales.idxmax(), f"{cat_sales.max():,.0f}")
````

**결과** · 예측 정답은 다름 — 건수는 식품, 총매출은 전자기기

### 6-2. `A` 여러 지표 · 그룹 결과 되돌리기 · `agg()` · `transform()` <sub>원본 02 · Mission 5 · Applied Challenge</sub>

> 🎯 **건수 · 총매출 · 평균을 한 표로 비교하고, 그룹 결과를 행마다 다시 붙이려면?**

💡 **쉽게 말하면**

- `agg`는 **바구니마다 요약 카드 한 장**(그룹 수만큼), `transform`은 **요약 값을 바구니 안 물건마다 스티커로 붙이기**(원래 행 수 그대로).

<img src="images/pandas_agg_transform.png" alt="agg는 그룹 수만큼, transform은 원래 행 수만큼" width="718">

**실행 흐름**

1. 카테고리 agg
2. 평균 1위
3. 채널 agg
4. transform으로 행마다 비율

**먼저 예측** · App과 Web 중 평균 주문금액이 높은 채널은? — 답은 코드 실행 결과 아래에

````python
# 그룹마다 지표 여러 개
metrics = ["count", "sum", "mean"]                # 건수 · 총매출 · 평균
cat_summary = by_cat["final_sales"].agg(metrics)  # 그룹마다 한 줄
print(cat_summary.round(0))                       # 5개 카테고리 비교표


# 빈칸 02-㉕ 평균 주문금액 1위
best = cat_summary["mean"]                               # 평균 열
print("평균 1위:", best.idxmax(), f"{best.max():,.0f}")  # 빈칸 02-㉕ 답


# 빈칸 02-㉖ 채널별 건수 · 총매출 · 평균
by_channel = logical_orders.groupby("channel")          # App · Web으로 나누기
print(by_channel["final_sales"].agg(metrics).round(0))  # 빈칸 02-㉖ 답


# transform: 그룹 결과를 원래 행마다 다시 붙임
cat_total = by_cat["final_sales"].transform("sum")                    # 그룹 합을 행마다
share = logical_orders["final_sales"] / cat_total                     # 주문 1건의 비율
print("agg:", cat_summary.shape, "| transform:", cat_total.shape)     # 5줄 vs 992줄
share_sum = share.groupby(logical_orders["category"]).sum().round(6)  # 그룹마다 비율 합
print("카테고리마다 비율 합:", share_sum.unique().tolist())           # 모두 1이면 맞음
````

### 7-2. `A` merge how 4종 · `how=` <sub>원본 03 · Mission 3</sub>

> 🎯 **한쪽에만 있는 Key는 `how`에 따라 어떻게 처리되는가?**

💡 **쉽게 말하면**

- 두 명단을 맞대는 **4가지 규칙**.
- left = 왼쪽 명단 전원, inner = 양쪽 모두 있는 사람만, right = 오른쪽 전원, outer = 누구든 한쪽에 있으면.

<img src="images/pandas_merge_how.png" alt="how 4종이 남기는 Key" width="748">

**실행 흐름**

1. C04(주문만) · C05(고객만) 표
2. how 4종 행 수 · Key · 빈 칸
3. Key 중복 시 행 증가

**먼저 예측** · `how="inner"`의 행 수는? — 답은 코드 실행 결과 아래에

````python
# 주문에만 있는 C04 · 고객에만 있는 C05
left_tbl = pd.DataFrame({"customer_id": ["C01", "C02", "C04"],   # C04는 주문에만
                         "sales": [50000, 30000, 90000]})
right_tbl = pd.DataFrame({"customer_id": ["C01", "C02", "C05"],  # C05는 고객에만
                          "membership": ["VIP", "BASIC", "PLUS"]})


# how 4종: 어느 쪽 Key 를 남기는가
for how in ["left", "inner", "right", "outer"]:                   # 4가지를 차례로
    m = left_tbl.merge(right_tbl, on="customer_id", how=how)      # 이번 how로 연결
    keys = m["customer_id"].tolist()                              # 남은 Key
    n_na = int(m.isna().sum().sum())                              # 표 전체 빈 칸 수
    print(f"{how:5} → {len(m)}행 | Key {keys} | 빈 칸 {n_na}개")  # 행 · Key · 빈 칸 비교


# 오른쪽 Key 가 중복이면 행이 늘어남
dup_right = pd.DataFrame({"customer_id": ["C01", "C01"],     # C01이 두 번
                          "membership": ["VIP", "PLUS"]})
m = left_tbl.merge(dup_right, on="customer_id", how="left")  # Key 중복 표와 연결
print("Key 중복 → 행:", len(m))                              # 3행이 4행으로
````

**결과** · 예측 정답은 `2` — 양쪽에 다 있는 C01 · C02만

- `left → 3행 | Key ['C01', 'C02', 'C04'] | 빈 칸 1개` — 왼쪽 전부, C04는 membership NaN
- `inner → 2행 | Key ['C01', 'C02'] | 빈 칸 0개` — 양쪽 공통만 → 주문 C04가 **사라짐**
- `right → 3행 | Key ['C01', 'C02', 'C05'] | 빈 칸 1개` — 오른쪽 전부, C05는 sales NaN
- `outer → 4행 | Key ['C01', 'C02', 'C04', 'C05'] | 빈 칸 2개` — 합집합, 빈 칸 양쪽에

### 7-3. `A` 연결 결과 검증 · `indicator=` · `validate=` <sub>원본 03 · Mission 4 ~ 5</sub>

> 🎯 **merge가 에러 없이 실행되면 모든 고객 정보가 연결된 것인가?**

💡 **쉽게 말하면**

- 택배 송장 번호로 물건을 찾았다고 해서 **전부 찾은 건 아님**.
- `indicator`는 줄마다 "양쪽에서 찾음 / 왼쪽에만 있음" 도장을 찍어 줌.

**실행 흐름**

1. left merge
2. shape 전후
3. membership NaN · 고객 ID
4. indicator · validate

**먼저 예측** · left merge 뒤 행 수는 992보다 늘까? — 답은 코드 실행 결과 아래에

````python
# 빈칸 03-④ · ⑤ 주문 기준 left merge → 행 수 확인
merged = logical_orders.merge(customers, on="customer_id", how="left")  # 주문 기준으로 붙임
print("merge 전:", logical_orders.shape, "→ 후:", merged.shape)         # 행 수가 같은가


# 빈칸 03-⑥ · ⑦ 고객 정보를 못 찾은 주문
no_member = merged["membership"].isna()                                   # 못 찾은 주문 표시
missing_ids = merged.loc[no_member, "customer_id"].unique().tolist()      # 어떤 고객 ID인가
print("membership 빈 주문:", no_member.sum(), "| 고객 ID:", missing_ids)  # 빈칸 03-⑦ 답


# 연결 결과 표시 · 고객 표가 한 행씩인지 검사
merge_check = logical_orders.merge(                    # 검증용으로 한 번 더
    customers, on="customer_id", how="left",
    indicator=True,                                    # _merge 열: both · left_only
    validate="many_to_one")                            # 주문 여러 개 : 고객 1명
print(merge_check["_merge"].value_counts().to_dict())  # 성공 · 실패 개수
````

**결과** · 예측 정답은 그대로 992 — 고객 Key가 한 행씩이라서

- `merge 전: (992, 15) → 후: (992, 21)` — 원본 빈칸 답: 행 같음 · 고객 열 6개 추가 (Key는 하나로)
- `membership 빈 주문: 10 | 고객 ID: ['C9001', 'C9002', 'C9003', 'C9004']` — 원본 빈칸 답: 고객 파일에 없는 ID 4개
- `{'both': 982, 'left_only': 10, 'right_only': 0}` — Merge Evidence: 전체 992 · 성공 982 · 실패 10

### 7-4. `B` 등급별 구매 행동 · `dropna=False` <sub>원본 03 · Mission 6 · Final Mission</sub>

💡 **쉽게 말하면**

- "더 많이 구매"는 **세 가지 질문**(몇 번 · 얼마나 · 한 번에 얼마)으로 나눠야 답이 됨.
- 연결 실패 주문은 숨기지 않고 따로 적음.

**실행 흐름**

1. 등급별 count · sum · mean

**먼저 예측** · 총매출 1위 등급은 VIP일까? — 답은 코드 실행 결과 아래에

````python
# 빈칸 03-⑧ 등급별 건수 · 총매출 · 평균
by_member = merged.groupby("membership")["final_sales"]  # 등급별로 나누기
member_summary = by_member.agg(metrics)                  # 건수 · 총매출 · 평균
print(member_summary.round(0))                           # 빈칸 03-⑧ 답


# 빈칸 03-⑨ 연결 실패 주문까지 포함해서 보기
by_member_all = merged.groupby("membership", dropna=False)["final_sales"]  # NaN도 한 그룹
print(by_member_all.agg(metrics).round(0).tail(1))                         # NaN = 고객 정보 없음
top = member_summary.idxmax()                                              # 열마다 최댓값의 등급
print("1위 → 건수:", top["count"], "| 총매출:", top["sum"],                # 지표마다 1위
      "| 평균:", top["mean"])


# 고객 수로 나누면: 고객 1명당 매출
n_buyers = merged.groupby("membership")["customer_id"].nunique()  # 등급별 고객 수
print("고객 수:", n_buyers.to_dict())                             # VIP는 29명뿐
per_buyer = member_summary["sum"] / n_buyers                      # 고객 1명당 매출
print("1명당 매출:", per_buyer.round(0).to_dict())                # 공정한 비교
````

**먼저 예측** · 상품군 표기는 몇 가지로 적혀 있고, 정리하면 몇 개가 될까? — 답은 코드 실행 결과 아래에

````python
# Mission 1 · 2 처음 보는 판매 파일: 확인 → 검증
sales_data = pd.read_csv("data/unseen.csv")                            # 처음 보는 판매 파일
na = sales_data.isna().sum()                                           # 열마다 결측
print("shape:", sales_data.shape)                                      # 크기
print("결측:", na[na > 0].to_dict())                                   # 결측이 있는 열만
print("item_group 표기:", sales_data["item_group"].nunique(), "가지",  # 표기 · 중복 · 상태
      "| sale_id 중복:", sales_data["sale_id"].duplicated().sum(),
      "| 상태:", sales_data["record_status"].value_counts().to_dict())


# 정리: 원본은 두고 사본에서
cs = sales_data.copy()                                       # cs = clean sales
cs["coupon_rate"] = cs["coupon_rate"].fillna(0)              # 비어 있음 = 쿠폰 미사용
cs["item_group"] = cs["item_group"].str.strip().str.upper()  # 표기 통일
cs["sold_at"] = pd.to_datetime(cs["sold_at"])                # 2026/08/05 → 날짜
cs = cs.loc[cs["record_status"] == "verified"]               # 믿을 행만
cs = cs.drop_duplicates("sale_id", keep="first").copy()      # ID당 한 행


# Mission 3 판매금액 만들기 → 다시 검증
cs["gross_sales"] = cs["unit_price"] * cs["units"]                       # 할인 전 금액
cs["final_sales"] = cs["gross_sales"] * (1 - cs["coupon_rate"])          # 할인 후 금액
clean_sales = cs                                                         # 분석용 표 이름
print("논리 거래:", len(clean_sales),                                    # 행 수 = 고유 ID 확인
      "| 고유 sale_id:", clean_sales["sale_id"].nunique(),
      "| 상품군:", sorted(clean_sales["item_group"].unique()))
print(f"전체 최종 판매금액: {clean_sales['final_sales'].sum():,.0f}원")  # Revenue 답
````

````python
# Mission 4 · 5 상품군 · 매장 비교
by_group = clean_sales.groupby("item_group")["final_sales"]  # 상품군별
by_store = clean_sales.groupby("store")["final_sales"]       # 매장별
print(by_group.agg(metrics).round(0))                        # 건수 · 합 · 평균
print(by_store.agg(metrics).round(0))                        # Store A 주장 검토용


# Mission 6 고객 정보 연결 → 검증
buyers = pd.read_csv("data/unseen_customers.csv")                    # 구매자 파일
sales_buyers = clean_sales.merge(buyers, on="buyer_id", how="left",  # 판매 기준 연결
                                 indicator=True, validate="many_to_one")
print("연결 전후:", len(clean_sales), "→", len(sales_buyers),        # 행 수 · 실패 수
      "|", sales_buyers["_merge"].value_counts().to_dict())


# Mission 7 고객군 비교 (NaN = 연결 실패도 한 줄로)
by_segment = sales_buyers.groupby("buyer_segment", dropna=False)  # NaN = 연결 실패
print(by_segment["final_sales"].agg(metrics).round(0))            # 고객군 비교
````

**결과** · 예측 정답은 `15`가지 → `5`개 (`FOOD` · `Food` · `food`처럼 대소문자만 다른 표기)

### 9-1. `A` 라벨 이름 → 번호 · `map()` · `apply()` <sub>추가</sub>

> 🎯 **`cat` · `dog` 같은 라벨 이름을 모델이 쓰는 번호로 어떻게 바꾸는가?**

💡 **쉽게 말하면**

- `map`은 **번역 사전 대조**.
- 값마다 사전에서 찾아 번호로 바꾸고, 사전에 없는 단어는 빈 칸(NaN)으로 남겨 실수를 드러냄.

**실행 흐름**

1. 라벨 CSV 저장 · 읽기
2. 정렬된 클래스
3. {이름: 번호}
4. map
5. apply로 경로
6. 없는 라벨

**먼저 예측** · `bird` · `cat` · `dog`를 정렬해 번호를 매기면 `dog`는? — 답은 코드 실행 결과 아래에

````python
# 프로젝트에서 받는 형태의 라벨 CSV 만들기 → 읽기
Path("실습파일").mkdir(exist_ok=True)          # 없으면 폴더 만들기
pd.DataFrame({                                 # 파일 이름 · 라벨 표
    "file": [f"img_{i:03d}.jpg" for i in range(8)],
    "label": ["cat", "dog", "bird", "cat", "dog", "cat", "bird", "dog"],
}).to_csv("실습파일/labels.csv", index=False)  # 0, 1, 2 번호 열은 저장 X
labels = pd.read_csv("실습파일/labels.csv")    # 받은 라벨 CSV 읽기


# 클래스 이름 → 번호: 정렬로 번호 고정 → map
classes = sorted(labels["label"].unique())                  # 정렬 → 번호 고정
class_to_idx = {name: i for i, name in enumerate(classes)}  # {이름: 번호}
labels["target"] = labels["label"].map(class_to_idx)        # 사전에서 찾아 바꿈
print("class_to_idx:", class_to_idx)                        # 번호표 확인
print(labels.head(4))                                       # 라벨 옆에 번호


# apply: 값마다 함수 실행
labels["path"] = labels["file"].apply(lambda f: f"images/{f}")  # 값마다 함수
print("path:", labels["path"].head(2).tolist())                 # 경로 확인


# 사전에 없는 라벨은 NaN
print("map 없는 값:", pd.Series(["cat", "fox"]).map(class_to_idx).tolist())  # 없는 라벨은 NaN
````

**결과** · 예측 정답은 `2` — 정렬 순서 bird 0 · cat 1 · dog 2

- `class_to_idx: {'bird': 0, 'cat': 1, 'dog': 2}` — 등장 순서가 아니라 **정렬 순서**로 번호 고정
- `0 img_000.jpg cat 1` — cat → 1
- `path: ['images/img_000.jpg', 'images/img_001.jpg']` — `apply`는 값마다 함수 실행
- `map 없는 값: [1.0, nan]` — 사전에 없는 `fox` → NaN, 정수 열이 실수(1.0)로 바뀜

### 9-2. `A` 모델용 자료형 맞추기 · `astype()` <sub>추가</sub>

> 🎯 **모델에 넣기 전 라벨과 특성의 자료형은 무엇으로 맞추는가?**

💡 **쉽게 말하면**

- 모델은 **정답은 정수(int64), 입력은 실수(float32)** 규격만 받는 기계.
- `astype`은 규격 상자로 옮겨 담기.

**실행 흐름**

1. target int64
2. 특성 float32 · 메모리

**먼저 예측** · float64 → float32로 바꾸면 메모리는? — 답은 코드 실행 결과 아래에

````python
# map 결과의 dtype · 모델이 원하는 dtype
print("target:", labels["target"].dtype)             # map 결과 자료형
labels["target"] = labels["target"].astype("int64")  # 분류 정답 = 정수


# 숫자 특성은 float32
num_cols = ["price", "quantity", "discount"]                                   # 숫자 특성 열
feats = logical_orders[num_cols].astype("float32")                             # 모델 입력 자료형
print(feats.dtypes.to_dict())                                                  # 세 열 모두 float32
before = logical_orders[num_cols].memory_usage(index=False).sum()              # 바꾸기 전 메모리
print("메모리:", before, "→", feats.memory_usage(index=False).sum(), "bytes")  # 절반으로


# NaN 이 있으면 정수로 못 바꿈 → 먼저 확인
odd = pd.Series(["cat", "fox"]).map(class_to_idx)  # NaN이 섞인 라벨
try:
    odd.astype("int64")                            # 정수로 바꾸기 시도
except ValueError as e:                            # 정수 변환 에러를 잡음
    print(type(e).__name__, "→ 결측 먼저 처리")    # 에러 이름 확인


# 글자가 섞인 숫자 열은 to_numeric (못 바꾸면 NaN)
messy = pd.Series(["10", "20", "-"])                   # 글자가 섞인 숫자
print(pd.to_numeric(messy, errors="coerce").tolist())  # 못 바꾸면 NaN
````

### 9-3. `A` 표 이어 붙이기 · `concat()` <sub>추가</sub>

> 🎯 **나눠 받은 데이터 파일 여러 개를 하나의 표로 합치려면?**

💡 **쉽게 말하면**

- `concat`은 **같은 양식의 종이를 위아래로 철하기**(axis=0) 또는 **옆에 이어 붙이기**(axis=1).
- 번호(index)는 새로 매기지 않으면 겹침.

**실행 흐름**

1. 두 조각
2. concat (index 겹침)
3. ignore_index
4. axis=1
5. 열 이름이 다를 때

**먼저 예측** · 따로 읽은 5행 · 3행 표를 그냥 `concat`하면 index는? — 답은 코드 실행 결과 아래에

````python
# 나눠 받은 두 파일 = 같은 열의 표 두 개 (따로 읽으면 각자 0번부터)
part_a = labels.iloc[:5][["file", "target"]]                         # 첫 번째 조각
part_b = labels.iloc[5:][["file", "target"]].reset_index(drop=True)  # 0번부터인 두 번째 조각


# 세로로 잇기 (axis=0): 행이 늘어남
stacked = pd.concat([part_a, part_b])                                      # 위아래로 잇기
print("index 그대로:", stacked.index.tolist())                             # 번호가 겹침
stacked = pd.concat([part_a, part_b], ignore_index=True)                   # 번호 새로 매김
print("ignore_index:", stacked.index.tolist(), "| shape:", stacked.shape)  # 0 ~ 7


# 가로로 붙이기 (axis=1): 같은 index 끼리 옆에
extra = pd.DataFrame({"width": [32] * 8, "height": [32] * 8})  # 같은 8행 특성
print(pd.concat([stacked, extra], axis=1).head(2))             # 옆으로 붙이기


# 열 이름이 다르면 빈 칸이 생김
other = pd.DataFrame({"file": ["x.jpg"], "label": ["cat"]})  # 열 이름이 다른 표
print(pd.concat([part_a.head(1), other]))                    # 없는 칸은 NaN
````

**결과** · 예측 정답은 `[0, 1, 2, 3, 4, 0, 1, 2]` — 번호가 겹침

- `ignore_index: [0, 1, 2, 3, 4, 5, 6, 7] | shape: (8, 2)` — 0부터 다시 매김 → 8행
- `0 img_000.jpg 1 32 32` — axis=1 → 같은 index끼리 옆에
- `0 img_000.jpg 1.0 NaN` — 열 이름이 다르면 없는 칸은 NaN, 정수 열이 실수로

### 9-4. `A` 날짜 index와 날짜 특성 · `set_index()` · `.dt` <sub>추가</sub>

> 🎯 **날짜 열을 시계열 분석이 가능한 형태로 만들려면?**

💡 **쉽게 말하면**

- `set_index`는 **날짜를 시트의 줄 번호로 쓰기**.
- 그러면 "7월 10일부터 12일까지"처럼 달력으로 자를 수 있고, `.dt`로 요일 같은 특성을 뽑음.

**실행 흐름**

1. 문자열 날짜
2. to_datetime
3. set_index · sort_index
4. 주문 데이터 index
5. 날짜 범위 자르기
6. .dt 특성
7. 주말 여부

**먼저 예측** · `ts.loc["2026-07-10":"2026-07-12"]`에 12일 주문이 포함될까? — 답은 코드 실행 결과 아래에

````python
# CSV에서 읽은 날짜 = 문자열 → 날짜 → index → 날짜 순서
raw = pd.DataFrame({"date": ["2026-07-03", "2026-07-01", "2026-07-02"],  # 날짜 문자열 · 섞인 순서
                    "sales": [30, 10, 20]})                              # 순서가 섞인 3일치
raw["date"] = pd.to_datetime(raw["date"])                                # 문자열 → 날짜
by_date = raw.set_index("date")                                          # 날짜가 행 이름(index)
print("정렬 전:", by_date["sales"].tolist(),                             # 정렬 전후 비교
      "| 정렬 후:", by_date.sort_index()["sales"].tolist())


# 주문 데이터: 날짜 열 → 시계열 index (5-3에서 이미 날짜로 변환)
ts = logical_orders.set_index("order_date").sort_index()  # 날짜 index + 정렬
print("index:", type(ts.index).__name__,                  # DatetimeIndex 확인
      "|", ts.index.min().date(), "~", ts.index.max().date())


# 날짜 범위로 자르기 (끝 날짜 포함)
print("7/10~7/12 주문:", len(ts.loc["2026-07-10":"2026-07-12"]))  # 끝 날짜 포함


# .dt 로 날짜 특성 만들기
dates = logical_orders["order_date"]                                 # 날짜 열
date_feats = pd.DataFrame({                                          # 날짜 특성 표
    "date": dates,
    "day": dates.dt.day,                                             # 일
    "dayofweek": dates.dt.dayofweek,                                 # 월 0 ~ 일 6
})
date_feats["is_weekend"] = date_feats["dayofweek"] >= 5              # 토 · 일 = True
print(date_feats.head(3))                                            # 특성 확인
print("주말 주문 비율:", round(date_feats["is_weekend"].mean(), 3))  # True 비율
````

### 9-5. `A` 시계열 묶기 · 이동 평균 · 밀기 · `resample()` · `rolling()` · `shift()` <sub>추가</sub>

> 🎯 **주문 기록을 하루 단위 시계열로 바꾸고, 예측용 입력과 정답을 만들려면?**

💡 **쉽게 말하면**

- `resample`은 **달력 칸마다 모으기**, `rolling`은 **최근 7일 창을 한 칸씩 밀며 평균**, `shift`는 **표를 위아래로 한 칸 밀어** 어제 값 · 내일 값을 옆에 두기.

<img src="images/pandas_shift.png" alt="rolling · shift(1) · shift(-1)" width="733">

**실행 흐름**

1. 일별 합계(31일)
2. 주별 합계
3. rolling(7) 평균
4. shift(1) 어제 · shift(-1) 내일

**먼저 예측** · `rolling(7).mean()`의 첫 값은 며칠째에 나올까? — 답은 코드 실행 결과 아래에

````python
# 일별 매출 시계열 (D = 하루 단위, 빈 날도 0 으로 채움)
daily = ts["final_sales"].resample("D").sum()                              # 하루 단위 합계
print("일별:", daily.shape, "| 첫 3일:", daily.head(3).round(0).tolist())  # 31일


# 주 단위로 묶기 (일요일 끝) → {주가 끝나는 날: 합계}
weekly = daily.resample("W").sum()                                 # 주 단위 합계
print(dict(zip(weekly.index.strftime("%m-%d"), weekly.round(0))))  # 주가 끝나는 날별


# 이동 평균 · 과거 값 · 다음 값
frame = pd.DataFrame({"sales": daily})           # 특성을 붙일 표
frame["ma7"] = frame["sales"].rolling(7).mean()  # 최근 7일 평균
frame["lag1"] = frame["sales"].shift(1)          # 하루 전 값 → 입력 특성
frame["target"] = frame["sales"].shift(-1)       # 다음 날 값 → 정답
print(frame.head(8).round(0))                    # 앞 8일 확인
````

**결과** · 예측 정답은 7일째 (7월 7일) — 앞 6일은 7개가 안 돼서 NaN

- `일별: (31,) | 첫 3일: [1706000.0, 2234575.0, 2271800.0]` — 7월 31일치, 빈 날 없이 하루 한 줄
- `{'07-05': 10750425.0, '07-12': 17354935.0, '07-19': 17057450.0, '07-26': 16268450.0, '08-02': 12568740.0}` — 일요일에 끝나는 주 · 첫 주(1~5일)와 마지막 주(27~31일)는 5일치
- `2026-07-01 1706000.0 NaN NaN 2234575.0` — 첫날: 어제 없음(lag1 NaN), 정답 = 7월 2일 값
- `2026-07-07 2330880.0 2212165.0 2403850.0 2282805.0` — 1~7일 평균 2212165 · 어제 2403850 · 내일 2282805

- 꺼낼 때 규격(dtype)도 함께 지정.

**Q3. merge 후에 무엇을 검증하나요?**
> 행 수가 merge 전과 같은지 확인하고, `indicator=True`로 `left_only`가 몇 건인지 셉니다. 오른쪽 Key가 중복되면 행이 늘어나므로 `validate="many_to_one"`을 걸어 둡니다. 연결 실패 건은 결론에서 숨기지 않고 따로 보고합니다.

**Q4. `groupby`의 `agg`와 `transform`의 차이는?**
> - `agg`는 그룹마다 한 줄로 요약해 그룹 수만큼의 표를 돌려주고, `transform`은 그룹 계산 결과를 원래 행마다 붙여 행 수를 유지합니다.
> - 그룹끼리 비교할 때는 `agg`, 행마다 그룹 평균과의 차이나 비율 같은 특성을 만들 때는 `transform`을 씁니다.

**Q5. pandas 표를 딥러닝 모델 입력으로 바꾸는 과정은?**
> - 라벨 이름은 정렬한 클래스 목록으로 `{이름: 번호}` 사전을 만들어 `map`하고 `int64`로 바꿉니다.
> - 특성은 숫자 열만 골라 `to_numpy(dtype=np.float32)`로 배열을 만듭니다.
> - 시계열은 `resample`로 간격을 맞추고 `shift`로 과거 값과 다음 값을 만든 뒤, 섞지 않고 시간 순서대로 학습 · 검증을 나눕니다.

### 전체 코드

- 1 ~ 9장 정답 코드를 장별 셀로 모음
- **위에서부터 이 셀들만 실행해도 전체가 동작** (노트북 옆 `data/` 필요)
- 각 출력 앞 `[번호]`가 섹션

````python
# ============================================================
# Pandas 전체 정리 — 전체 코드 (data/ 폴더가 있는 곳에서 실행)
# ============================================================
from pathlib import Path  # 실습 파일 경로 (9장)

import numpy as np   # 배열 확인 (9장)
import pandas as pd  # 관례: pd

pd.set_option("display.unicode.east_asian_width", True)  # 한글 열 맞춤
````

````python
# ---------- 1. 표 살펴보기 ----------
# 1-1 · 1-2 DataFrame 만들기 · shape · dtypes
students = pd.DataFrame({                                  # 사전 → 표 (키 = 열 이름)
    "name": ["민수", "지수", "영희", "철수", "유진", "현우"],
    "score": [78, 95, 84, 65, 91, 88],
    "city": ["서울", "부산", "서울", "대전", "서울", "부산"],
    "study_hours": [3, 7, 5, 2, 6, 5],
})
print("[1-1]", students.shape, students.columns.tolist())  # 크기 · 열 이름
print("[1-2]", students.dtypes.astype(str).to_dict(),      # 자료형 · 마지막 3행
      students.tail(3)["name"].tolist())

# 1-3 CSV 읽기
orders = pd.read_csv("data/orders.csv")             # 노트북 기준 상대 경로
na = orders.isna().sum()                            # 열마다 결측
print("[1-3]", orders.shape, na[na > 0].to_dict())  # 크기 · 결측 열
````

````python
# ---------- 2. 선택하기 ----------
# 2-1 한 열 = Series, 열 목록 = DataFrame
print("[2-1]", type(students["city"]).__name__,  # Series vs DataFrame
      type(students[["name", "city"]]).__name__)

# 2-2 loc 라벨(끝 포함) · iloc 위치(끝 미포함)
print("[2-2]", students.loc[0:2, ["name", "score"]].shape,  # 끝 포함 vs 미포함
      students.iloc[0:2, 0:2].shape)
````

````python
# ---------- 3. 조건으로 찾기 ----------
# 3-1 조건 → True/False → 선택
print("[3-1]", students[students["score"] >= 85]["name"].tolist(),  # 조건 → 고르기
      (orders["price"] >= 50000).sum())

# 3-2 & · | + 괄호, isin
busan_85 = (students["city"] == "부산") & (students["score"] >= 85)  # 그리고
seoul_or_busan = students["city"].isin(["서울", "부산"])             # 여러 값 중 하나
print("[3-2]", students[busan_85]["name"].tolist(),                  # & · isin 결과
      students[seoul_or_busan]["name"].tolist())
````

````python
# ---------- 4. 계산 · 새 열 · 정렬 · 개수 ----------
# 4-1 통계
print("[4-1]", students["score"].mean(), students["study_hours"].sum())  # 평균 · 합계

# 4-2 새 열
students["passed"] = students["score"] >= 80                                # 비교 결과를 새 열로
orders["sales"] = orders["price"] * orders["quantity"]                      # 확인 뒤 열로 저장
print("[4-2]", students["passed"].sum(), orders["sales"].head(3).tolist())  # 새 열 확인

# 4-3 정렬
by_hours = students.sort_values(["study_hours", "score"], ascending=False)  # 동점은 점수로
print("[4-3]", by_hours["name"].tolist())                                   # 동점은 점수로 정렬

# 4-4 값별 개수
print("[4-4]", students["passed"].value_counts().to_dict())  # 값별 개수
````

````python
# ---------- 5. 데이터 정리 ----------
# 5-1 · 5-2 결측: discount 는 0 으로, region 은 그대로
orders["discount"] = orders["discount"].fillna(0)                   # 빈 칸 = 할인 없음
orders["final_sales"] = orders["sales"] * (1 - orders["discount"])  # 빈칸 02-⑮ 할인 적용
print("[5-2]", orders["final_sales"].isna().sum(),                  # 결측 처리 결과
      orders["region"].isna().sum())

# 5-3 날짜 변환 · 5-4 공백 제거
orders["order_date"] = pd.to_datetime(orders["order_date"])             # 문자열 → 날짜
orders["region"] = orders["region"].str.strip()                         # 앞뒤 공백 제거
print("[5-4]", orders["order_date"].dtype, orders["region"].nunique())  # 날짜 · 지역 6개

# 5-5 · 5-6 중복 확인 → 분석용 데이터
print("[5-5]", orders.duplicated().sum(),             # 행 중복 · Key 중복
      orders["order_id"].duplicated().sum())
# 오류 제외 → ID당 한 행
logical_orders = (orders.loc[orders["record_status"].ne("ingestion_error")]
                  .drop_duplicates(subset="order_id", keep="first").copy())
total = logical_orders["final_sales"].sum()           # 공식 매출
print("[5-6]", len(logical_orders), f"{total:,.0f}")  # 992행 · 74,000,000
````

````python
# ---------- 6. 그룹 비교 ----------
metrics = ["count", "sum", "mean"]                                          # 건수 · 총매출 · 평균
by_cat = logical_orders.groupby("category")["final_sales"]                  # 카테고리별
cat_summary = by_cat.agg(metrics)                                           # 그룹마다 한 줄
print("[6-1]", cat_summary["count"].idxmax(), cat_summary["sum"].idxmax())  # 건수 1위 · 매출 1위
by_channel = logical_orders.groupby("channel")["final_sales"]               # 채널별
print("[6-2]", by_channel.agg(["count", "sum"]).to_dict())                  # App · Web 비교
````

````python
# ---------- 7. 여러 표 연결 ----------
customers = pd.read_csv("data/customers.csv")                            # 고객 파일
merged = logical_orders.merge(customers, on="customer_id", how="left",   # 주문 기준 연결
                              indicator=True, validate="many_to_one")
print("[7-3]", merged.shape, merged["_merge"].value_counts().to_dict())  # 행 수 · 실패 수
by_member = merged.groupby("membership")["final_sales"]                  # 등급별로 나누기
print("[7-4]", by_member.mean().round(0).to_dict())                      # 등급별 평균
````

````python
# ---------- 8. 종합 실습 ----------
cs = pd.read_csv("data/unseen.csv")                                         # 처음 보는 판매 파일
cs["coupon_rate"] = cs["coupon_rate"].fillna(0)                             # 빈 칸 = 쿠폰 미사용
cs["item_group"] = cs["item_group"].str.strip().str.upper()                 # 표기 통일
cs["sold_at"] = pd.to_datetime(cs["sold_at"])                               # 날짜로
cs = cs.loc[cs["record_status"] == "verified"]                              # 믿을 행만
cs = cs.drop_duplicates("sale_id", keep="first").copy()                     # ID당 한 행
cs["gross_sales"] = cs["unit_price"] * cs["units"]                          # 할인 전 금액
cs["final_sales"] = cs["gross_sales"] * (1 - cs["coupon_rate"])             # 할인 후 금액
clean_sales = cs                                                            # 분석용 표 이름
buyers = pd.read_csv("data/unseen_customers.csv")                           # 구매자 파일
sales_buyers = clean_sales.merge(buyers, on="buyer_id", how="left",         # 판매 기준 연결
                                 indicator=True, validate="many_to_one")
print("[8]", len(clean_sales), f"{clean_sales['final_sales'].sum():,.0f}",  # 거래 · 매출 · 연결
      sales_buyers["_merge"].value_counts().to_dict())
````

````python
# ---------- 9. 딥러닝 데이터로 연결 ----------
# 9-1 · 9-2 라벨 → 번호 (int64)
Path("실습파일").mkdir(exist_ok=True)                                 # 없으면 폴더 만들기
pd.DataFrame({                                                        # 파일 이름 · 라벨 표
    "file": [f"img_{i:03d}.jpg" for i in range(8)],
    "label": ["cat", "dog", "bird", "cat", "dog", "cat", "bird", "dog"],
}).to_csv("실습파일/labels.csv", index=False)
labels = pd.read_csv("실습파일/labels.csv")                           # 받은 라벨 CSV 읽기
classes = sorted(labels["label"].unique())                            # 정렬 → 번호 고정
class_to_idx = {name: i for i, name in enumerate(classes)}            # {이름: 번호}
labels["target"] = labels["label"].map(class_to_idx).astype("int64")  # 이름 → 번호 → 정수
print("[9-1]", class_to_idx, labels["target"].tolist())               # 번호표 · 정답

# 9-3 concat
parts = [labels.iloc[:5], labels.iloc[5:]]                 # 나눠 받은 두 조각
print("[9-3]", pd.concat(parts, ignore_index=True).shape)  # 다시 8행

# 9-4 · 9-5 날짜 index → 일별 매출 → 이동 평균 · 과거 값 · 다음 값
ts = logical_orders.set_index("order_date").sort_index()  # 날짜 index + 정렬
daily = ts["final_sales"].resample("D").sum()             # 하루 단위 합계
frame = pd.DataFrame({"sales": daily})                    # 특성을 붙일 표
frame["ma7"] = frame["sales"].rolling(7).mean()           # 최근 7일 평균
frame["lag1"] = frame["sales"].shift(1)                   # 하루 전 값 → 입력
frame["target"] = frame["sales"].shift(-1)                # 다음 날 값 → 정답
print("[9-5]", daily.shape, frame.dropna().shape)         # 31일 → 24일

# 9-6 시간 순서 분할 → 배열
data = frame.dropna()                                                       # 빈 칸 있는 날 제외
split = int(len(data) * 0.8)                                                # 80% 지점
X_train = data.iloc[:split][["sales", "ma7", "lag1"]].to_numpy(np.float32)  # 입력 배열
print("[9-6]", X_train.shape, X_train.dtype, labels["target"].to_numpy().dtype)  # 모양 · 자료형
````
