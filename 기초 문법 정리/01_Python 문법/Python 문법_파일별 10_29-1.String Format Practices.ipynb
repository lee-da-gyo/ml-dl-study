### 1-1. `A` 정답 · 다른 출력 찾기 · `%` · `str.format()` · `f-string` <sub>원본 문제</sub>

> 🎯 **세 출력 중 결과가 다른 것은 무엇인가?**

💡 **쉽게 말하면**

- 세 방식 모두 **문장 틀에 값을 끼워 넣기**.
- 다만 출력1의 `%x`는 숫자를 **16진수**로 바꿔 넣음.

**실행 흐름**

1. 공통 변수
2. 출력1 · 2 · 3 만들기
3. 세 줄 출력
4. 같은지 비교

**먼저 예측** · 결과가 다른 출력은 몇 번일까? — 답은 코드 실행 결과 아래에

````python
x = 10                # 공통 변수 (원본 그대로)
y = 20                # 더할 값
serialno = 308276567  # 일련번호
n = 'Kim'             # 이름


ex1 = 'n = %s, s = %x, sum=%d' % (n, serialno, (x + y))      # 출력1: % 방식
ex2 = 'n = {n}, s = {serialno}, sum={sum}'.format(n=n, serialno=serialno, sum=x + y)  # 출력2
ex3 = f'n = {n}, s = {serialno}, sum={x + y}'                # 출력3: f-string
print(ex1)                                                   # 출력1
print(ex2)                                                   # 출력2
print(ex3)                                                   # 출력3
print("2 · 3 같음:", ex2 == ex3, "/ 1만 다름:", ex1 != ex2)  # 문자열 비교
````

**결과** · 예측 정답은 출력1 — `s = 125fed57`

- `n = Kim, s = 125fed57, sum=30` — 출력1: `%x`가 일련번호를 16진수로 바꿈
- `n = Kim, s = 308276567, sum=30` — 출력2 · 3: 10진수 그대로
- `2 · 3 같음: True / 1만 다름: True` — 다른 것은 출력1

### 1-2. `B` `%x`와 16진수 · `%d` · `:x` · `int(문자열, 16)` <sub>원본 문제 · 다른 이유</sub>

> 🎯 **`%x`는 무엇이고, 출력1을 같게 고치려면?**

💡 **쉽게 말하면**

- `%d`는 **10진수**, `%x`는 **16진수**(0 ~ 9, a ~ f).
- 같은 숫자를 다른 진법으로 쓴 것 → `int(..., 16)`으로 되돌릴 수 있음.

**실행 흐름**

1. %x · %d 비교
2. f-string :x
3. hex() · 되돌리기
4. %d로 고친 출력1

**먼저 예측** · `int("125fed57", 16)`의 결과는? — 답은 코드 실행 결과 아래에

````python
print("%x →", "%x" % serialno)                             # %x = 16진수
print("%d →", "%d" % serialno)                             # %d = 10진수 정수
print("f-string :x →", f"{serialno:x}")                    # f-string도 :x로 16진수
print("hex() →", hex(serialno))                            # 앞에 0x가 붙은 16진수 문자열
print("되돌리기 →", int("125fed57", 16))                   # 16진수 글자 → 10진수 숫자
fixed = 'n = %s, s = %d, sum=%d' % (n, serialno, (x + y))  # %x → %d로 고침
print("고친 출력1 == 출력3:", fixed == ex3)                # 이제 같음
````

**결과** · 예측 정답은 `308276567` — 원래 일련번호

- 같은 숫자를 16진수 · 10진수로
  - `%x → 125fed57`
  - `%d → 308276567`
- f-string · hex()도 16진수 가능
  - `f-string :x → 125fed57`
  - `hex() → 0x125fed57`
- `%d`로 고치면 세 출력이 같음
  - `되돌리기 → 308276567`
  - `고친 출력1 == 출력3: True`

### 2-1. `A` 숫자 형식 지정 · `:.2f` · `:.1%` · `:,` <sub>원본 힌트 · f-Strings</sub>

> 🎯 **소수 자릿수 · 퍼센트 · 천 단위 · 0 채우기는 어떻게 쓰는가?**

💡 **쉽게 말하면**

- `{값:형식}` — **콜론 뒤가 모양**.
- `.2f` 소수 2자리, `.1%` 퍼센트, `,` 천 단위, `03d` 3칸에 0 채우기.

**실행 흐름**

1. 손실 소수 2자리
2. 정확도 퍼센트
3. 개수 천 단위
4. 0 채우기 · 지수

**먼저 예측** · `f"{0.873:.1%}"`의 결과는? — 답은 코드 실행 결과 아래에

````python
loss, acc, n_img = 0.876543, 0.873, 1234567  # 형식 지정 연습 값
print(f"소수 2자리: {loss:.2f}")             # 반올림해 소수 둘째 자리까지
print(f"퍼센트: {acc:.1%}")                  # 100을 곱하고 % 붙임
print(f"천 단위: {n_img:,}")                 # 세 자리마다 쉼표
print(f"앞에 0 채우기: {5:03d}")             # 3칸 정수, 빈칸은 0
print(f"지수 표기: {0.000123:.1e}")          # 아주 작은 값 (학습률 등)
````

**결과** · 예측 정답은 `87.3%`

- 반올림 · 퍼센트로 바꿔 % 붙임
  - `소수 2자리: 0.88`
  - `퍼센트: 87.3%`
- 쉼표 · 0 채우기
  - `천 단위: 1,234,567`
  - `앞에 0 채우기: 005`
- `지수 표기: 1.2e-04` — 0.000123 = 1.2 × 10⁻⁴

### 2-2. `B` 정렬 · 디버그 출력 · Template · `:>8` · `{x=}` <sub>원본 힌트 · 4가지 방법</sub>

> 🎯 **글자를 칸에 맞춰 정렬하고, 변수 이름과 값을 함께 보려면?**

💡 **쉽게 말하면**

- `>` 오른쪽 · `<` 왼쪽 · `^` 가운데, 숫자는 **칸 수**.
- `{x=}`는 **이름=값**을 한 번에 (디버깅용).

**실행 흐름**

1. 오른쪽 · 왼쪽 · 가운데 정렬
2. {x=} 디버그
3. 힌트 4번 Template

**먼저 예측** · `f"{x=}"`는 무엇을 출력할까? — 답은 코드 실행 결과 아래에

````python
from string import Template  # 원본 힌트의 4번째 방법

print(f"[{'loss':>8}]")                                     # 8칸 오른쪽 정렬
print(f"[{'loss':<8}]")                                     # 8칸 왼쪽 정렬
print(f"[{'loss':^8}]")                                     # 8칸 가운데 정렬
print(f"{x=}, {x + y=}")                                    # = 를 붙이면 '식=값'
print(Template("n = $n, sum=$s").substitute(n=n, s=x + y))  # $이름 자리를 채움
````

**결과** · 예측 정답은 `x=10`

- 같은 8칸 안에서 위치만 다름
  - `[    loss]`
  - `[loss    ]`
  - `[  loss  ]`
- `x=10, x + y=30` — 변수 이름(식)과 값을 함께
- `n = Kim, sum=30` — Template은 `$이름` 자리를 채움

**파이프라인 위치** · ⑤ 학습 → ⑦ 저장

> 🎯 **`(epoch, loss, acc)` 기록 3개를 표로 출력하고, 최저 손실 epoch로 파일 이름을 만들려면?**

💡 **쉽게 말하면**

- 헤더와 값 모두 **같은 칸 폭**으로 정렬.

**먼저 예측** · 최저 손실 epoch는 10 — 파일 이름에는 `ep10`이 들어갈까 `ep010`이 들어갈까? — 답은 코드 실행 결과 아래에

````python
# 1단계: (epoch, loss, acc) 가짜 학습 기록
history = [(1, 0.923456, 0.7123), (2, 0.61, 0.8), (10, 0.4712, 0.8634)]  # 자릿수가 제각각


# 2 · 3단계: 헤더와 값을 같은 폭으로 정렬
print(f"{'epoch':>5} | {'loss':>6} | {'acc':>6}")      # 헤더 (오른쪽 정렬)
for epoch, loss, acc in history:                       # 한 줄씩
    print(f"{epoch:>5} | {loss:>6.4f} | {acc:>6.1%}")  # 폭 + 자릿수 함께 지정


# 4단계: 최저 손실 epoch로 체크포인트 파일 이름 만들기
best = min(history, key=lambda h: h[1])                # 두 번째 값(loss)이 가장 작은 기록
fname = f"model_ep{best[0]:02d}_loss{best[1]:.3f}.pt"  # 2자리 0 채우기 · 소수 3자리
print("저장 파일 이름:", fname)                        # 체크포인트 이름
````

**결과** · 예측 정답은 `ep10` — `:02d`는 최소 2칸이라 10은 그대로

- `epoch |   loss |    acc` — 헤더도 같은 폭으로 오른쪽 정렬
- 자릿수가 달라도 세로 줄이 맞음
  - `    1 | 0.9235 |  71.2%`
  - `   10 | 0.4712 |  86.3%`
- `저장 파일 이름: model_ep10_loss0.471.pt` — 0 채우기 · 소수 3자리

### 전체 코드

1 ~ 3장 코드를 장별 셀로 모음. **위에서부터 이 셀들만 실행해도 전체가 동작**하며, 각 블록 머리 주석의 번호가 섹션.

````python
# ---------- 1-1 ----------
x = 10                # 공통 변수 (원본 그대로)
y = 20                # 더할 값
serialno = 308276567  # 일련번호
n = 'Kim'             # 이름


ex1 = 'n = %s, s = %x, sum=%d' % (n, serialno, (x + y))      # 출력1: % 방식
ex2 = 'n = {n}, s = {serialno}, sum={sum}'.format(n=n, serialno=serialno, sum=x + y)  # 출력2
ex3 = f'n = {n}, s = {serialno}, sum={x + y}'                # 출력3: f-string
print(ex1)                                                   # 출력1
print(ex2)                                                   # 출력2
print(ex3)                                                   # 출력3
print("2 · 3 같음:", ex2 == ex3, "/ 1만 다름:", ex1 != ex2)  # 문자열 비교


# ---------- 1-2 ----------
print("%x →", "%x" % serialno)                             # %x = 16진수
print("%d →", "%d" % serialno)                             # %d = 10진수 정수
print("f-string :x →", f"{serialno:x}")                    # f-string도 :x로 16진수
print("hex() →", hex(serialno))                            # 앞에 0x가 붙은 16진수 문자열
print("되돌리기 →", int("125fed57", 16))                   # 16진수 글자 → 10진수 숫자
fixed = 'n = %s, s = %d, sum=%d' % (n, serialno, (x + y))  # %x → %d로 고침
print("고친 출력1 == 출력3:", fixed == ex3)                # 이제 같음
````

````python
from string import Template  # 원본 힌트의 4번째 방법


# ---------- 2-1 ----------
loss, acc, n_img = 0.876543, 0.873, 1234567  # 형식 지정 연습 값
print(f"소수 2자리: {loss:.2f}")             # 반올림해 소수 둘째 자리까지
print(f"퍼센트: {acc:.1%}")                  # 100을 곱하고 % 붙임
print(f"천 단위: {n_img:,}")                 # 세 자리마다 쉼표
print(f"앞에 0 채우기: {5:03d}")             # 3칸 정수, 빈칸은 0
print(f"지수 표기: {0.000123:.1e}")          # 아주 작은 값 (학습률 등)


# ---------- 2-2 ----------

print(f"[{'loss':>8}]")                                     # 8칸 오른쪽 정렬
print(f"[{'loss':<8}]")                                     # 8칸 왼쪽 정렬
print(f"[{'loss':^8}]")                                     # 8칸 가운데 정렬
print(f"{x=}, {x + y=}")                                    # = 를 붙이면 '식=값'
print(Template("n = $n, sum=$s").substitute(n=n, s=x + y))  # $이름 자리를 채움
````

````python
# ---------- 3. 종합 실습 ----------
# 1단계: (epoch, loss, acc) 가짜 학습 기록
history = [(1, 0.923456, 0.7123), (2, 0.61, 0.8), (10, 0.4712, 0.8634)]  # 자릿수가 제각각


# 2 · 3단계: 헤더와 값을 같은 폭으로 정렬
print(f"{'epoch':>5} | {'loss':>6} | {'acc':>6}")      # 헤더 (오른쪽 정렬)
for epoch, loss, acc in history:                       # 한 줄씩
    print(f"{epoch:>5} | {loss:>6.4f} | {acc:>6.1%}")  # 폭 + 자릿수 함께 지정


# 4단계: 최저 손실 epoch로 체크포인트 파일 이름 만들기
best = min(history, key=lambda h: h[1])                # 두 번째 값(loss)이 가장 작은 기록
fname = f"model_ep{best[0]:02d}_loss{best[1]:.3f}.pt"  # 2자리 0 채우기 · 소수 3자리
print("저장 파일 이름:", fname)                        # 체크포인트 이름
````
