### 1-1. `B` 변수 이름 규칙 · `isidentifier()` <sub>원본 01</sub>

> 🎯 **어떤 이름은 변수로 쓸 수 있고, 어떤 이름은 왜 안 되는가?**

💡 **쉽게 말하면** 변수 이름은 **이름표**. 숫자로 시작하면 파이썬이 숫자로 착각하고, `class` · `for` 같은 예약어는 이미 주인이 있는 이름표.

```
보기 4개 isidentifier() 검사 → 2c 실제 실행 → 예약어 확인
```

**먼저 예측** · `a` · `_b` · `2c` · `_7d` 중 안 되는 것은?

<details><summary>예측 정답</summary>

(3) `2c` — 숫자로 시작

</details>

````python
# 보기 4개가 변수 이름 규칙에 맞는지 검사
candidates = ["a", "_b", "2c", "_7d"]
for name in candidates:
    print(f"{name:>4} → 사용 가능: {name.isidentifier()}")  # 문자·밑줄 시작 → True / 숫자 시작 → False


# 보기 (3)을 실제로 실행해 에러 확인
try:
    exec("2c = int(77)")  # 문자열 → 코드로 실행
except SyntaxError as e:
    print("SyntaxError:", e.msg)  # 에러 메시지만 출력


# 규칙에 맞아도 예약어는 사용 불가
print("class 예약어?", keyword.iskeyword("class"))
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `2c → 사용 가능: False` | 숫자로 시작 → 정답 (3) |
| `SyntaxError: invalid decimal literal` | 2c를 숫자로 읽으려다 실패 |
| `class 예약어? True` | 규칙에 맞아도 예약어는 사용 불가 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `str.isidentifier()` | 이름 **규칙**만 검사 (예약어도 True) | 문자열이 변수 이름 형태인지 확인할 때 |
| `keyword.iskeyword()` | **예약어**인지만 검사 | 규칙엔 맞는데 쓸 수 없는 이름 걸러낼 때 |

> ⚠️ **주의** · 대소문자 구분(`Data` ≠ `data`), 특수문자 `-` `$` `@` 사용 불가

> ✅ **한 줄 정리** · 문자 · 밑줄로 시작하고 숫자 · 밑줄만 이어 쓰며, 예약어는 피한다.

### 1-2. `A` 값 비교와 객체 비교 · `==` · `is` <sub>원본 02</sub>

> 🎯 **`==`와 `is`는 무엇이 다르며, `y = x`는 왜 복사가 아닌가?**

💡 **쉽게 말하면** `==`는 **내용이 같은가**, `is`는 **같은 물건인가**. `y = x`는 물건을 복사한 것이 아니라 같은 물건에 **이름표를 하나 더 붙인 것**.

```
정수 · 같은 리스트 · 내용만 같은 리스트 비교 → y = x 함정 → copy()
```

**먼저 예측** · `b = a` 뒤에 `b.append(4)` 하면 `a`는?

<details><summary>예측 정답</summary>

`[1, 2, 3, 4]` — `a`와 `b`는 같은 객체

</details>

````python
# 예제 1: 서로 다른 정수
x, y = 15, 25
print("예제1 →", x == y, x is y)  # 값 비교 · 객체 비교


# 예제 2: y = x 로 같은 리스트를 가리킴
x = ["orange", "banana", "apple"]
y = x  # 새 리스트 X → 같은 리스트에 이름표 추가
print("예제2 →", x == y, x is y)


# 예제 3: 내용만 같은 새 리스트
y = ["orange", "banana", "apple"]  # [ ] 를 쓰면 새 객체 생성
print("예제3 →", x == y, x is y)


# y = x 함정: b만 바꿨는데 a도 바뀜
a = [1, 2, 3]
b = a
b.append(4)  # b 수정 → 같은 객체인 a도 수정
print("a도 바뀜 →", a)


# copy()로 진짜 복사
c = a.copy()  # 새 객체 생성
c.append(5)  # c만 수정
print("복사본만 바뀜 →", a, c)
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `예제1 → False False` | 값 다름 · 객체 다름 |
| `예제2 → True True` | 같은 객체 |
| `예제3 → True False` | 내용은 같지만 다른 객체 |
| `a도 바뀜 → [1, 2, 3, 4]` | b = a 는 이름표 추가 → 원본까지 바뀜 |
| `복사본만 바뀜 → [1, 2, 3, 4] [1, 2, 3, 4, 5]` | copy()로 만든 c만 바뀜 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `==` | **값(내용)**이 같은가 | 거의 모든 비교 (숫자 · 문자열 · 리스트 내용) |
| `is` | **같은 객체(메모리)**인가 | `None` 확인 (`if x is None:`) · 같은 객체인지 디버깅 |
| `y = x` | 복사 X, **같은 객체에 이름 추가** | 같은 데이터를 다른 이름으로 부르기만 할 때 |
| `x.copy()` | **새 객체** (안쪽 리스트는 공유) | 원본을 보존하고 수정할 때 (1단 구조) |
| `copy.deepcopy(x)` | 안쪽까지 **전부 새로** | 리스트 안 리스트 · 딕셔너리 안 리스트 같은 중첩 구조 |

> ⚠️ **주의** · 중첩 구조(리스트 안 리스트)는 `copy()`로도 안쪽이 공유됨 → `copy.deepcopy()` 필요

> ✅ **한 줄 정리** · `==`는 값, `is`는 같은 객체를 비교하고, `y = x`는 이름표 추가이므로 복사는 `copy()`로 한다.

### 1-3. `A` 자료형 변환 · `str()` · `int()` <sub>원본 03</sub>

> 🎯 **문자열과 숫자는 왜 바로 더할 수 없고, 어떻게 맞추는가?**

💡 **쉽게 말하면** `"Seoul"`은 **글자**, `25`는 **숫자**. 글자 + 숫자는 사과 + 3킬로미터처럼 단위가 달라 한쪽으로 바꿔야 함.

```
에러 메시지 확인 → str() · f-string으로 수정 → int()로 반대 변환
```

**먼저 예측** · `"Seoul" + 25`의 결과는?

<details><summary>예측 정답</summary>

`TypeError` — 문자열과 정수는 바로 더할 수 없음

</details>

````python
x, y = "Seoul", 25


# 문자열 + 정수 → 에러 확인
try:
    z = x + y  # str + int
except TypeError as e:
    print("TypeError:", e)


# 방법 1: 정수를 문자열로 바꿔 이어 붙이기
z1 = x + str(y)  # str + str(int) → str


# 방법 2: f-string으로 끼워 넣기
z2 = f"{x}{y}"  # f"{값}{값}" → 자동으로 문자열


# 두 방법의 결과가 같은지 확인
assert z1 == z2 == "Seoul25"
print("x + y :", z1)


# 자료형 확인 · 반대 방향 변환
print(type(x).__name__, type(y).__name__, "|", int("25") + 10)  # int("25") → 25 → 25 + 10
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `TypeError: can only concatenate str (not "int") to str` | 문자열에는 문자열만 이어 붙일 수 있음 |
| `x + y : Seoul25` | 두 방법 결과 동일 |
| <code>str int &#124; 35</code> | int("25")로 바꾸면 계산 가능 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `x + str(y)` | 자료형을 **직접 변환**해 이어 붙임 | 변환이 한 번뿐인 짧은 결합 |
| `f"{x}{y}"` | 자동 변환 + **형식 지정** 가능 (`{y:03d}`) | 문장 · 로그 · 파일 이름 만들기 (기본 선택) |
| `int()` · `float()` | 문자열 → **숫자** | 파일 · 입력에서 읽은 숫자로 계산할 때 |

> ⚠️ **주의** · CSV에서 읽은 숫자가 문자열인 경우가 많음 → 계산 전 `type()` 확인

> ✅ **한 줄 정리** · 자료형이 다르면 `str()` · `int()` · `float()`로 한쪽으로 맞춘 뒤 계산한다.

### 2-1. `A` 위치로 꺼내기 · `x[i]` <sub>원본 04</sub>

> 🎯 **여러 값 중 원하는 위치의 값 하나를 어떻게 꺼내는가?**

💡 **쉽게 말하면** 리스트는 **번호표 붙은 사물함**. 번호는 **0번부터**, 뒤에서 셀 때는 `-1`부터.

```
앞 번호 · 뒤 번호 · index() 3가지로 꺼내기 → 없는 값 에러
```

**먼저 예측** · 6개 중 `Banana`의 번호는?

<details><summary>예측 정답</summary>

`4` — 0번부터 세므로 다섯 번째가 4번

</details>

````python
x = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]


# 세 가지 방법으로 Banana 꺼내기
print(x[4], x[-2], x[x.index("Banana")])  # 앞에서 4번 · 뒤에서 2번 · 위치 찾아서


# 위치와 전체 길이 확인
print("위치:", x.index("Banana"), "/ 길이:", len(x))  # 마지막 번호 = 길이 - 1


# 없는 값을 찾으면 에러
try:
    x.index("banana")  # 소문자 → 다른 값
except ValueError as e:
    print("ValueError:", e)
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `Banana Banana Banana` | 세 방법 모두 같은 값 |
| `위치: 4 / 길이: 6` | 마지막 번호 = 길이 - 1 |
| `ValueError: list.index(x): x not in list` | 대소문자 구분 → 없는 값 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `x[4]` | **앞**에서 센 번호 | 위치를 알 때 |
| `x[-1]` · `x[-2]` | **뒤**에서 센 번호 | 마지막 · 끝에서 몇 번째 (길이 몰라도 됨) |
| `x.index(값)` | 값으로 **위치를 찾음** (없으면 에러) | 위치를 모르고 값만 알 때 |

> ⚠️ **주의** · 마지막 값은 `x[len(x)]`가 아니라 `x[-1]` (`IndexError` 주의)

> ✅ **한 줄 정리** · 0부터 세는 번호(`x[4]`)나 뒤에서 세는 번호(`x[-2]`)로 꺼낸다.

### 2-2. `A` 구간 자르기 · `x[a:b]` <sub>원본 05</sub>

> 🎯 **연속된 구간을 어떻게 잘라내며, 끝 번호는 왜 포함되지 않는가?**

💡 **쉽게 말하면** `[4:7]`은 **4번 사물함부터 7번 앞까지**. 끝을 빼야 `[0:4]` + `[4:8]`처럼 구간을 겹침 없이 이어 붙일 수 있음.

```
[e, f, g]를 3가지로 자르기 → step · 뒤집기
```

**먼저 예측** · `x[4:7]`은 몇 개?

<details><summary>예측 정답</summary>

`3`개 — 4, 5, 6번 (7번 미포함)

</details>

````python
x = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m"]


# 방법 1: 앞 번호로 자르기
print(x[4:7])  # [start:stop] → 4, 5, 6 (7 미포함)


# 방법 2: 뒤 번호로 자르기
print(x[-9:-6])  # 뒤에서 9번째 ~ 6번째 앞


# 방법 3: 위치를 찾아서 자르기
print(x[x.index("e"):x.index("g") + 1])  # g까지 넣으려면 +1


# 세 방법의 결과가 같은지 확인
assert x[4:7] == x[-9:-6] == ["e", "f", "g"]


# step 사용: 간격 · 뒤집기
print(x[::3], x[::-1][:3])  # [::3] 3칸씩 · [::-1] 뒤집기 → 앞 3개
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `['e', 'f', 'g'] × 3` | 7번은 포함 안 됨 |
| `['a', 'd', 'g', 'j', 'm']` | step 3 → 3칸 간격 |
| `['m', 'l', 'k']` | step -1 → 뒤집기 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `x[i]` | 값 **1개** 반환 | 한 칸만 필요할 때 |
| `x[a:b]` | **리스트** 반환, 범위 밖이어도 에러 없음 | 구간 · 앞 n개 · 학습/검증 나누기 |
| `x[::step]` | 간격 · 방향 지정 | n칸마다 고르기, `[::-1]` 뒤집기 |

> ⚠️ **주의** · stop은 미포함 → 끝값까지 원하면 +1

> ✅ **한 줄 정리** · `[start:stop:step]`으로 자르며, stop은 포함되지 않는다.

<details><summary>딥러닝 코드에서는</summary>

### 2-3. `A` 숫자 범위 만들기 · `range()` <sub>원본 07</sub>

> 🎯 **규칙 있는 숫자 목록을 어떻게 만들고, 끝값은 어떻게 정하는가?**

💡 **쉽게 말하면** `range`는 **번호표 자동 발급기**. 시작 · 끝 · 간격을 주면 차례로 뽑아 주고, 슬라이싱처럼 끝은 미포함.

```
30 → -10, -2씩 생성 → 개수 · 처음 · 끝 검증 → stop 실수
```

**먼저 예측** · `range(30, -10, -2)`의 마지막 값은?

<details><summary>예측 정답</summary>

`-8` — stop인 -10은 포함되지 않음

</details>

````python
# 30부터 -10까지 -2씩 줄어드는 리스트 만들기
r = list(range(30, -11, -2))  # range(시작, 끝 미포함, 간격) → -10을 넣으려면 끝은 -11
print(r)
print("개수:", len(r))


# 처음 · 마지막 값 확인
assert r[0] == 30 and r[-1] == -10


# 흔한 실수: 끝을 -10으로 쓰면
print("stop = -10 이면 마지막:", list(range(30, -10, -2))[-1])  # -10 미포함 → -8에서 멈춤
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `[30, 28, ..., -10]` | -2 간격, 21개 |
| `stop = -10 이면 마지막: -8` | stop은 포함 안 됨 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `range(n)` | 0 ~ n-1 | n번 반복 (`for epoch in range(10)`) |
| `range(a, b)` | a ~ b-1 | 시작이 0이 아닐 때 |
| `range(a, b, step)` | 간격 · **음수면 감소** | 배치 단위로 건너뛰기, 거꾸로 세기 |

> ⚠️ **주의** · 감소할 때는 step을 음수로, stop은 끝값보다 한 칸 더

> ✅ **한 줄 정리** · `range(start, stop, step)`도 슬라이싱처럼 stop을 포함하지 않는다.

### 2-4. `A` 전체 변환 · `[식 for x in 목록]` · `map()` <sub>원본 09</sub>

> 🎯 **리스트 전체에 같은 변환을 한 번에 적용하려면?**

💡 **쉽게 말하면** **공장 컨베이어 벨트**. 원소가 하나씩 지나가며 같은 가공을 받고, `lambda`는 이름 없는 한 줄짜리 가공 기계.

```
1~15 → ×10 → 문자열 → map + lambda / 컴프리헨션 비교
```

**먼저 예측** · `map(...)`을 `list()` 없이 출력하면?

<details><summary>예측 정답</summary>

값이 아니라 `map` 객체가 출력됨

</details>

````python
# 방법 1: map()과 lambda 사용
r1 = list(map(lambda n: str(n * 10), range(1, 16)))  # 1~15 → ×10 → 문자열 변환 → 리스트


# 방법 2: 리스트 컴프리헨션 사용
r2 = [str(n * 10) for n in range(1, 16)]  # 표현식 for 변수 in 반복 범위


# 두 방법의 결과가 같은지 확인
assert r1 == r2
print(r1)


# list()로 감싸지 않으면 map 객체
print("list() 없이:", type(map(str, [1])).__name__)  # 값이 아니라 map 객체
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `['10', '20', ..., '150']` | 15개 모두 문자열 |
| `list() 없이: map` | 값이 아니라 map 객체 → list()로 꺼내야 보임 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `[식 for x in 목록]` | 읽기 쉬움 · **조건(`if`)** 추가 쉬움 | **기본 선택** |
| `map(함수, 목록)` | 이미 있는 함수를 그대로 적용, 결과는 map 객체 | `map(int, "1 2 3".split())`처럼 함수 이름 하나로 끝날 때 |
| `lambda` | 이름 없는 **한 줄 함수** | `sorted(key=...)` · `map`에 잠깐 넘길 때 (길면 `def`) |

> ⚠️ **주의** · `map` 결과는 리스트가 아니며 한 번 꺼내 쓰면 비어 버림

> ✅ **한 줄 정리** · `[변환 for 원소 in 리스트]`로 전체를 한 번에 바꾼다.

### 2-5. `A` 중복 제거 · `set()` <sub>원본 10</sub>

> 🎯 **중복을 없앤 목록을 만들 때 순서는 어떻게 지키는가?**

💡 **쉽게 말하면** `set`은 **주머니**. 같은 것은 하나만 들어가지만 순서가 없어서, 순서가 필요하면 `sorted()`로 다시 줄 세움.

```
set 중복 제거 → 숫자 먼저 · 문자 나중 정렬 → 원래 순서 유지 → 섞어 정렬 에러
```

**먼저 예측** · `set`으로 바꾸면 순서는 유지될까?

<details><summary>예측 정답</summary>

보장되지 않음 — 순서가 필요하면 정렬

</details>

````python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]


# set으로 중복 제거
uniq = set(x)  # 중복 X · 순서 X
print("개수:", len(x), "→", len(uniq))


# 방법 1: 숫자 먼저, 문자 나중으로 정렬
print("정답 형태 :", sorted(uniq, key=lambda v: (isinstance(v, str), v)))  # key = (문자인가, 값) → False(숫자)가 먼저


# 방법 2: 처음 나온 순서 그대로 유지
print("순서 유지 :", list(dict.fromkeys(x)))  # 키 = 값 → 중복 키 자동 제거, 순서 유지


# 숫자와 문자를 그대로 정렬하면 에러
try:
    sorted(uniq)  # 1 < "a" 비교 불가
except TypeError:
    print("TypeError: 숫자와 문자는 크기 비교 불가")
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `개수: 10 → 7` | a · b · 1~5만 남음 |
| `[1, 2, 3, 4, 5, 'a', 'b']` | key = (문자인가, 값) → 숫자가 먼저 |
| `['a', 1, 'b', 2, 3, 4, 5]` | 처음 등장 순서 유지 |
| `TypeError: 숫자와 문자는 크기 비교 불가` | 1 < 'a' 비교 불가 → key 필요 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `set(x)` | 중복 제거 · **순서 없음** · 가장 빠름 | 개수 · 포함 여부만 필요할 때 |
| `sorted(set(x))` | 중복 제거 + **정렬된 순서** | 클래스 목록처럼 순서가 항상 같아야 할 때 |
| `list(dict.fromkeys(x))` | 중복 제거 + **처음 나온 순서** 유지 | 원래 순서가 의미 있을 때 |

> ⚠️ **주의** · `set`의 순서는 보장되지 않음 → 번호를 매길 땐 반드시 `sorted()`

> ✅ **한 줄 정리** · `set`으로 중복을 없애고, 순서는 `sorted()`나 `dict.fromkeys()`로 정한다.

## 3. 딕셔너리

### 3-1. `B` 값 합계 · `values()` <sub>원본 11</sub>

> 🎯 **딕셔너리에서 값만 꺼내 계산하려면?**

💡 **쉽게 말하면** 딕셔너리는 **이름표(키) 붙은 서랍장**. `keys()`는 이름표만, `values()`는 내용물만, `items()`는 둘 다.

```
values() 합계 · 반복 누적 · 키로 접근 → 2327 검증
```

**먼저 예측** · `sum(d)`를 하면?

<details><summary>예측 정답</summary>

`TypeError` — 키(문자열)를 더하려 함

</details>

````python
d = {"a": 17, "b": 114, "c": 247, "d": 362, "e": 220, "f": 728, "g": -283, "h": 922}


# 방법 1: values()로 값만 꺼내 합계
s1 = sum(d.values())  # 값 목록 → 합계


# 방법 2: 반복문으로 하나씩 누적
s2 = 0
for v in d.values():
    s2 += v  # s2 = s2 + v


# 방법 3: 키로 값에 접근해 합계
s3 = sum(d[k] for k in d)  # 키 반복 → d[키] → 합계


# 세 방법의 결과가 같은지 확인
assert s1 == s2 == s3 == 2327
print("합계:", s1)
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `합계: 2327` | 세 방법 동일 (음수 -283 포함) |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `d.keys()` | 키만 | 어떤 항목이 있는지 볼 때 |
| `d.values()` | 값만 | 합계 · 평균 · 최댓값 계산 |
| `d.items()` | (키, 값) 쌍 | 둘 다 필요할 때 (3-3) |
| `for` 누적 | 중간에 조건 · 출력 추가 가능 | 계산 과정에 다른 처리가 섞일 때 |

> ⚠️ **주의** · `sum(d)`는 키(문자열)를 더하려 해서 에러 → `values()` 필요

> ✅ **한 줄 정리** · 값만 필요하면 `d.values()`, 키만 필요하면 `d.keys()`로 꺼낸다.

### 3-2. `B` 항목 추가 · 합치기 · `update()` <sub>원본 12</sub>

> 🎯 **딕셔너리에 항목을 추가하거나 두 딕셔너리를 합치려면?**

💡 **쉽게 말하면** 서랍장에 **새 이름표 서랍**을 끼우는 것. 같은 이름표가 이미 있으면 내용물이 교체됨.

```
copy() 후 4가지 방법 → 결과 동일 → 원본 확인
```

**먼저 예측** · 같은 키가 양쪽에 있으면 어느 값이 남나?

<details><summary>예측 정답</summary>

뒤에 합친 쪽의 값

</details>

````python
base = {"a": "apple", "b": "grape"}
new = {"c": "banana", "d": "kiwi"}


# 방법 1: 키로 직접 넣기
d1 = base.copy()  # 원본 보존용 복사
d1["c"], d1["d"] = "banana", "kiwi"  # d[키] = 값


# 방법 2: update()로 합치기
d2 = base.copy()
d2.update(new)  # d2에 new를 덮어씀 (원본 수정)


# 방법 3: 풀어서 새 딕셔너리 만들기
d3 = {**base, **new}  # {**a, **b} → a 펼침 + b 펼침


# 방법 4: 합치기 연산자 (Python 3.9+)
d4 = base | new  # a | b → 새 딕셔너리


# 네 방법의 결과가 같은지 확인
assert d1 == d2 == d3 == d4
print(d1)
print("원본 그대로:", base)
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `{'a': 'apple', 'b': 'grape', 'c': 'banana', 'd': 'kiwi'}` | 4가지 결과 동일 |
| `원본 그대로: {'a': 'apple', 'b': 'grape'}` | copy() 덕분에 base 유지 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `d[키] = 값` | **1개** 추가 · 수정 | 항목 하나 넣을 때 |
| `d.update(new)` | 여러 개 한 번에, **원본 수정** | 기존 설정에 덮어쓸 때 |
| `{**a, **b}` · a &#124; b | **새 딕셔너리** (원본 유지) | 기본 설정 + 사용자 설정을 합쳐 새로 만들 때 |

> ⚠️ **주의** · `update()`는 원본을 바꾸고, 겹치는 키는 뒤의 값이 남음

> ✅ **한 줄 정리** · 하나는 `d[키] = 값`으로 넣고, 여러 개는 `update()`나 합치기 연산자로 합친다.

### 3-3. `A` 키 · 값 반복 · `items()` <sub>원본 15</sub>

> 🎯 **딕셔너리의 키와 값을 동시에 꺼내며 반복하려면?**

💡 **쉽게 말하면** `items()`는 서랍마다 **(이름표, 내용물) 한 쌍**을 건네주고, `for 키, 값 in`으로 양손에 나눠 받음.

```
range로 값 리스트 3개 → items() 반복 출력 → get()으로 없는 키
```

**먼저 예측** · `for k in d:`는 키와 값 중 무엇을 꺼내나?

<details><summary>예측 정답</summary>

키만

</details>

````python
# 키별로 숫자 리스트를 가진 딕셔너리 만들기
d = dict(one=list(range(1, 11)), two=list(range(11, 23)), three=list(range(23, 37)))  # dict(키=값, ...)


# items()로 키와 값을 함께 꺼내며 출력
for key, values in d.items():  # (키, 값) 쌍 → key, values 로 풀기
    print(f"key '{key}' has values {values} -> total : {len(values)}")


# get()으로 없는 키를 안전하게 조회
print("없는 키:", d.get("four", "없음"))  # 키 없으면 기본값 (d["four"]는 KeyError)
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `key 'one' has values [...] -> total : 10` | total = 값 리스트 길이 (two 12 · three 14) |
| `없는 키: 없음` | get()은 에러 대신 기본값 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `d[키]` | 없는 키 → **KeyError** | 키가 반드시 있어야 할 때 (없으면 버그) |
| `d.get(키, 기본값)` | 없는 키 → **기본값** | 없을 수도 있는 키 (선택 옵션) |
| `for k in d` | **키만** 꺼냄 | 키 목록만 필요할 때 |
| `for k, v in d.items()` | 키 · 값 **쌍** | 둘 다 필요할 때 (기본 선택) |

> ⚠️ **주의** · `for k, v in d:`처럼 `items()`를 빠뜨리면 에러, `for k in d`는 키만 꺼냄

> ✅ **한 줄 정리** · `for 키, 값 in d.items():`로 키와 값을 함께 꺼낸다.

### 3-4. `A` 리스트로 딕셔너리 만들기 · `zip()` · `enumerate()` <sub>원본 24 + 34</sub>

> 🎯 **여러 리스트의 같은 위치 값을 어떻게 묶어 딕셔너리로 만드는가?**

💡 **쉽게 말하면** `zip`은 **두 줄로 선 사람들을 같은 위치끼리 짝짓기**, `enumerate`는 **줄 선 사람들에게 번호표 붙이기**.

```
zip: 이름 · 수량 · 단가 → {이름: 수량×단가} → enumerate: {번호: 색} → 길이 다른 zip
```

**먼저 예측** · 길이가 다른 두 리스트를 `zip`하면?

<details><summary>예측 정답</summary>

짧은 쪽 길이에 맞춰 잘림

</details>

````python
a = ["one", "two", "three", "four"]
b = [30, 20, 15, 75]
c = [5.2, 7.4, 3.6, 4.2]


# 방법 1: 같은 위치의 이름·수량·단가를 묶어 {이름: 수량 × 단가} 생성
r1 = {k: x * y for k, x, y in zip(a, b, c)}  # {키: 값 for 변수들 in zip(...)}


# 방법 2: 값 리스트를 먼저 만들고 dict(zip(키, 값))
r2 = dict(zip(a, [x * y for x, y in zip(b, c)]))  # 수량×단가 리스트 → 이름과 짝짓기 → dict


# 두 방법의 결과가 같은지 확인
assert r1 == r2
print(r1)


# enumerate로 번호를 붙여 딕셔너리 만들기
colors = ["Red", "Green", "Black", "Blue", "Orange", "Purple"]
print(dict(enumerate(colors)))  # (0, Red), (1, Green), ... → dict
print(dict(enumerate(colors, start=100)))  # 번호를 100부터 시작


# 길이가 다른 리스트를 zip하면
print("zip 잘림:", list(zip([1, 2, 3], ["x", "y"])))  # 짧은 쪽(2개)에 맞춰 잘림
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `{'one': 156.0, 'two': 148.0, 'three': 54.0, 'four': 315.0}` | 30×5.2 · 20×7.4 · 15×3.6 · 75×4.2 |
| `{0: 'Red', ...} / {100: 'Red', ...}` | start로 시작 번호 지정 |
| `zip 잘림: [(1, 'x'), (2, 'y')]` | 3은 짝이 없어 사라짐 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `zip(a, b)` | **같은 위치끼리** 묶음, 짧은 쪽에 맞춤 | 이미지 ↔ 라벨처럼 리스트 여러 개를 짝지을 때 |
| `enumerate(a)` | **번호 + 값** | 몇 번째인지 필요할 때 (배치 번호, 클래스 번호) |
| `{k: v for ...}` | 변환 · 조건을 **함께** 적용 | 값을 계산해서 넣을 때 (수량 × 단가) |
| `dict(zip(키, 값))` | 이미 있는 두 리스트를 **그대로** 짝지음 | 키 · 값 리스트가 준비돼 있을 때 |

> ⚠️ **주의** · `zip`은 길이가 다르면 짧은 쪽에 맞춰 조용히 잘림 → 길이 먼저 확인

> ✅ **한 줄 정리** · `zip`은 같은 위치끼리, `enumerate`는 번호와 값을 묶는다.

<details><summary>딥러닝 코드에서는</summary>

`for images, labels in loader:` (짝), `for i, (x, y) in enumerate(loader):` (번호 + 배치)

</details>

#### 3장 정리 · 소속별 최종 요약표

| 함수·문법 | 소속 | 역할 |
| :-- | :-- | :-- |
| `keys()` · `values()` · `items()` · `get()` | dict | 키 · 값 · 쌍 꺼내기, 안전 조회 |
| `update()` · `copy()` · `fromkeys()` | dict | 합치기 · 복사 · 키로 생성 |

### 4-1. `A` 인자 순서와 기본값 · `def` <sub>원본 17</sub>

> 🎯 **기본값이 있는 함수는 인자를 어떤 순서로 정의해야 하는가?**

💡 **쉽게 말하면** 함수는 **주문서**. 꼭 적어야 하는 칸(필수 인자)이 앞, 안 적으면 기본값으로 채워지는 칸이 뒤.

```
원본 함수 → SyntaxError → 순서 수정 → 위치 · 키워드 인자로 호출
```

**먼저 예측** · 기본값 인자 뒤에 기본값 없는 인자를 두면?

<details><summary>예측 정답</summary>

`SyntaxError` — 정의 단계에서 실패

</details>

````python
# 원본 함수: 기본값 인자 뒤에 필수 인자 → 에러
broken = 'def greet(msg="Good morning!", name):\n    return msg'
try:
    exec(broken)
except SyntaxError as e:
    print("SyntaxError:", e.msg)


# 수정: 필수 인자를 앞, 기본값 인자를 뒤로
def greet(name, msg="Good morning!"):  # def 함수명(필수, 기본값=값):
    return "Hi! " + name + ", " + msg


# 세 가지 방식으로 호출
print(greet("Kim"))  # msg 생략 → 기본값 사용
print(greet("Park", "How do you do?"))  # 위치 순서대로 전달
print(greet(msg="Hello", name="Lee"))  # 이름=값 → 순서 무관
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `SyntaxError: parameter without a default follows ...` | 기본값 인자 뒤에 필수 인자 불가 |
| `Hi! Kim, Good morning!` | 생략하면 기본값 |
| `Hi! Park, How do you do?` | 위치 순서대로 전달 |
| `Hi! Lee, Hello` | 이름=값이면 순서 무관 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| 위치 인자 `f(a, b)` | **순서**대로 전달 | 인자가 1~2개로 순서가 뻔할 때 |
| 키워드 인자 `f(b=1, a=2)` | **이름**으로 전달, 순서 무관 | 인자가 많거나 기본값 일부만 바꿀 때 (`lr=1e-3`) |
| 기본값 인자 `def f(a, b=1)` | 생략하면 기본값 | 대부분 같은 값을 쓰는 옵션 |

> ⚠️ **주의** · 기본값으로 `[]` 같은 리스트를 쓰면 호출마다 공유됨 → `None`으로 두고 안에서 생성

> ✅ **한 줄 정리** · 필수 인자 → 기본값 인자 순서로 정의하고, 헷갈리면 `이름=값`으로 호출한다.

### 4-2. `B` 지역 · 전역 변수 · `global` · `return` <sub>원본 20</sub>

> 🎯 **함수 안에서 바꾼 변수는 함수 밖에도 남는가?**

💡 **쉽게 말하면** 함수는 **회의실**. 안에서 쓴 메모(지역 변수)는 나올 때 버려지고, 밖의 게시판(전역 변수)을 고치려면 `global` 허락이 필요.

```
밖: a, b, x, y = 8, 13, 33, 44 → 안: global a 변경 · x, y 교환 · b 생성 → 안 · 밖 비교
```

**먼저 예측** · 함수 실행 뒤 밖의 `a, b, x, y`는?

<details><summary>예측 정답</summary>

`49 13 33 44` — `global`로 선언한 `a`만 바뀜

</details>

````python
# 함수 안에서 변수를 바꾸는 함수 정의
def test(x, y):
    global a  # 밖의 a를 직접 수정하겠다는 선언
    a = 49
    x, y = y, x  # 함수 안 x, y만 교환
    b = 7  # 함수 안 b (밖의 b와 별개)
    c = 135  # 함수가 끝나면 사라짐
    print("Step1 : ", a, b, x, y)


# 밖의 변수 설정 → 함수 실행 → 밖의 값 확인
a, b, x, y = 8, 13, 33, 44
test(23, 7)
print("Step2 : ", a, b, x, y)  # global 쓴 a만 바뀜
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `Step1 :  49 7 7 23` | 함수 안: global a · 안의 b · 교환된 x, y |
| `Step2 :  49 13 33 44` | global로 바꾼 a만 밖에 남음 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| 지역 변수 | 함수 **안에서만** 존재 | 기본 (함수 안 계산용) |
| `global` | 함수 안에서 **밖의 변수 수정** | 거의 쓰지 않음 (추적 어려움) |
| `return` | 결과를 **밖으로 돌려줌** | **기본 선택** (`acc = evaluate(...)`) |

> ⚠️ **주의** · `global`을 남발하면 어디서 값이 바뀌었는지 추적 불가

> ✅ **한 줄 정리** · 함수 안 변수는 함수 안에서만 살아 있으므로, 밖으로 보낼 값은 `return`으로 돌려준다.

<details><summary>딥러닝 코드에서는</summary>

### 5-1. `A` 문장 나누기 · `split()` <sub>원본 21</sub>

> 🎯 **문장을 단어로 나누어 개수를 세려면?**

💡 **쉽게 말하면** `split()`은 **가위**. 공백(또는 지정한 글자)을 기준으로 잘라 조각들을 리스트로 돌려줌.

```
단어 개수 함수 → 10 검증 → 구분자 지정 · 공백 두 칸 비교
```

**먼저 예측** · 공백이 두 칸일 때 `split()`과 `split(" ")`의 차이는?

<details><summary>예측 정답</summary>

`split()`은 빈 문자열 없이 단어만, `split(" ")`은 빈 문자열이 섞임

</details>

````python
in_str = "Suppose we have few words that are separated by spaces."


# 공백으로 나눈 단어 개수를 세는 함수
def count_words(text):
    return len(text.split())  # 문자열 → split() → 단어 리스트 → len()


# 단어 목록 · 개수 확인
print(in_str.split())
print("단어 개수:", count_words(in_str))
assert count_words(in_str) == 10


# 구분자를 지정해서 나누기
print("a,b,,c".split(","))  # 쉼표 기준 → 연속 쉼표 사이는 빈 문자열


# 인자 없는 split() vs 공백 한 칸 split(" ")
print("  a  b ".split(), "  a  b ".split(" "))  # 연속 공백 처리 차이
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `단어 개수: 10` | 마침표는 'spaces.'에 붙어 있음 |
| `['a', 'b', '', 'c']` | 연속 구분자 사이의 빈 문자열도 원소 |
| `['a', 'b'] ['', '', 'a', '', 'b', '']` | 인자 없는 split()이 공백을 깔끔히 처리 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `split()` | **연속 공백을 하나로** 처리, 앞뒤 공백 무시 | 단어 나누기 (기본 선택) |
| `split(" ")` | 공백 **한 칸마다** 자름 → 빈 문자열 생김 | 공백 개수 자체가 의미 있을 때 (드묾) |
| `split(",")` | 지정한 문자로 자름 | CSV 한 줄, `"a,b,c"` 같은 구분 데이터 |

> ⚠️ **주의** · 공백 기준이면 `split(" ")`이 아니라 인자 없는 `split()`

> ✅ **한 줄 정리** · `split()`으로 문자열을 리스트로 나누고 `len()`으로 센다.

### 5-2. `A` 출력 형식 지정 · `f"{값:형식}"` <sub>원본 29</sub>

> 🎯 **값을 원하는 형식(소수점 · 퍼센트 · 진법)으로 출력하려면?**

💡 **쉽게 말하면** f-string은 **빈칸 채우기 양식**. `{값:형식}`에서 `:` 뒤가 '어떻게 적을지' 지시.

```
% · format() · f-string 3가지 → 다른 것 찾기 → 자주 쓰는 형식
```

**먼저 예측** · `%x`는 숫자를 어떻게 바꾸나?

<details><summary>예측 정답</summary>

16진수 — `308276567` → `125fed57`

</details>

````python
x, y = 10, 20
serialno = 308276567
n = "Kim"


# 방법 1: % 연산자 (옛 방식)
ex1 = "n = %s, s = %x, sum=%d" % (n, serialno, x + y)  # %s 문자 · %x 16진수 · %d 정수


# 방법 2: format()
ex2 = "n = {n}, s = {serialno}, sum={sum}".format(n=n, serialno=serialno, sum=x + y)  # {이름} ← 이름=값


# 방법 3: f-string (권장)
ex3 = f"n = {n}, s = {serialno}, sum={x + y}"  # f"{변수 또는 식}"


# 세 결과 비교: 다른 것 찾기
print(ex1, ex2, ex3, sep="\n")
print("출력1만 다름:", ex1 != ex2 and ex2 == ex3)


# 자주 쓰는 형식 지정
print(f"{0.876:.1%} | {1234567:,} | {3.14159:.2f} | {7:03d}")  # 퍼센트 · 천 단위 쉼표 · 소수 2자리 · 0 채워 3자리
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `n = Kim, s = 125fed57, sum=30` | %x = 16진수 → 출력1만 다름 (정답) |
| <code>87.6% &#124; 1,234,567 &#124; 3.14 &#124; 007</code> | 퍼센트 · 천 단위 쉼표 · 소수 2자리 · 0 채우기 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `%` 방식 | 옛 문법 · 읽기 어려움 | 오래된 코드를 읽을 때만 |
| `str.format()` | 틀을 먼저 만들고 **나중에 값 채움** | 같은 틀을 여러 번 재사용 (`tmpl.format(...)`) |
| f-string | 가장 짧고 읽기 쉬움, 식 계산 가능 | **기본 선택** (로그 · 출력 · 파일 이름) |

> ⚠️ **주의** · 소수점 지정 없이 출력하면 `0.8765432109`처럼 읽기 어려움

> ✅ **한 줄 정리** · f-string의 `{값:형식}` 하나로 대부분의 출력 형식을 해결한다.

### 6-1. `B` 파일 쓰기 · `with open()` <sub>원본 26</sub>

> 🎯 **여러 파일을 반복문으로 한 번에 저장하려면?**

💡 **쉽게 말하면** `with open(...)`은 **자동문**. 블록에 들어가면 파일이 열리고, 나오면 저절로 닫힘.

```
이름 · 내용 zip → 파일마다 with open("w") → 목록 · 내용 확인
```

**먼저 예측** · `"w"` 모드로 같은 파일을 다시 열면 기존 내용은?

<details><summary>예측 정답</summary>

지워지고 새로 씀 (덮어쓰기)

</details>

````python
names = ["A", "B", "C", "D", "F", "G"]
contents = ["Python", "JavaScript", "PHP", "Rust", "Solidity", "Assembly"]
folder = BASE / "26-1"


# 이름 · 내용을 짝지어 파일마다 쓰기
for name, text in zip(names, contents):  # (A, Python), (B, JavaScript), ...
    with open(folder / f"{name}.txt", "w", encoding="utf-8") as f:  # with open(경로, 모드) as 파일: → 끝나면 자동 닫힘
        f.write(text)


# 저장된 파일 목록 · 내용 확인
print(sorted(os.listdir(folder)))
print("A.txt:", (folder / "A.txt").read_text(encoding="utf-8"))
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `['A.txt', 'B.txt', 'C.txt', 'D.txt', 'F.txt', 'G.txt']` | 6개 (E는 원본 문제에도 없음) |
| `A.txt: Python` | 이름과 내용이 짝대로 저장 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `"w"` | **덮어쓰기** (없으면 생성) | 결과 파일을 새로 만들 때 |
| `"a"` | 끝에 **이어 쓰기** | 학습 로그처럼 계속 쌓을 때 |
| `"r"` (기본) | 읽기 | 파일 내용 읽을 때 |
| `with open(...)` | 블록이 끝나면 **자동 닫힘** | **항상** (`open()` 단독 사용 X) |

> ⚠️ **주의** · `"w"`는 덮어쓰기, 이어 쓰려면 `"a"`, 한글은 `encoding="utf-8"` 필수

> ✅ **한 줄 정리** · `with open(경로, "w", encoding="utf-8") as f:`로 쓰면 파일은 자동으로 닫힌다.

### 6-2. `A` 여러 파일 읽기 · `glob()` <sub>원본 27</sub>

> 🎯 **폴더 안 파일을 빠짐없이, 항상 같은 순서로 읽으려면?**

💡 **쉽게 말하면** `glob("*.txt")`는 **이름 패턴으로 찾는 검색창**. 결과 순서는 컴퓨터마다 다를 수 있어 `sorted()`로 줄 세움.

```
*.txt 찾기 → 정렬 → 하나씩 읽기 → pathlib 방식과 비교
```

**먼저 예측** · `glob` 결과 순서는 항상 같을까?

<details><summary>예측 정답</summary>

보장되지 않음 — `sorted()`로 고정

</details>

````python
# 방법 1: glob으로 찾고 정렬한 뒤 하나씩 읽기
paths = sorted(glob.glob(str(BASE / "27-1" / "*.txt")))  # *.txt 경로 목록 → 이름순 정렬
result = []
for p in paths:
    with open(p, encoding="utf-8") as f:  # 모드 생략 → "r" 읽기
        result.append(f.read().strip())  # 전체 읽기 → 앞뒤 공백 제거 → 추가


# 방법 2: pathlib으로 한 줄에 읽기
result2 = [p.read_text(encoding="utf-8").strip()  # 읽기 → 공백 제거
           for p in sorted((BASE / "27-1").glob("*.txt"))]  # 경로 찾기 → 정렬


# 두 방법의 결과가 같은지 확인
assert result == result2


# 개수와 일부만 출력
print("파일 개수:", len(result))
print("앞 5개:", result[:5], "/ 마지막:", result[-1])
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `파일 개수: 26` | A.txt ~ Z.txt |
| `앞 5개: ['Python', ..., 'elite'] / 마지막: zigzag` | 파일 이름순 = 원본 정답 순서 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `os.listdir(폴더)` | **모든** 파일 · 폴더 이름 (경로 X) | 폴더 안 전체 목록 |
| `glob.glob("*.txt")` | **패턴**으로 골라 **경로**까지 반환 | 특정 확장자만, 하위 폴더 `**` 검색 |
| `Path(폴더).glob("*.txt")` | 경로 **객체** 반환 → `.read_text()` 등 바로 사용 | 경로를 이어 붙이고 다룰 일이 많을 때 (최근 권장) |
| `f.read()` · `Path.read_text()` | 같은 결과, 열고 닫기를 직접 vs 자동 | 한 번에 읽고 끝이면 `read_text()` |

> ⚠️ **주의** · 정렬하지 않으면 실행 환경마다 파일 순서가 달라짐

> ✅ **한 줄 정리** · `sorted(glob("*.txt"))`로 순서를 고정한 뒤 읽는다.

### 6-3. `B` 확장자별 분류 · `endswith()` <sub>원본 42</sub>

> 🎯 **폴더 안 파일을 확장자별로 골라내려면?**

💡 **쉽게 말하면** 파일 이름 **끝부분(확장자)을 보고 바구니에 나눠 담기**. `endswith`로 끝을 확인.

```
목록 → endswith로 png · py 분류 → glob 검증 → 전체 확장자 개수
```

**먼저 예측** · 정렬하면 `file10`과 `file6` 중 앞은?

<details><summary>예측 정답</summary>

`file10` — 문자열은 한 글자씩 비교 ('1' < '6')

</details>

````python
folder = BASE / "42-1"
files = sorted(os.listdir(folder))  # 폴더 안 파일 이름 → 정렬


# 방법 1: endswith()로 확장자 골라내기
png = [f for f in files if f.endswith(".png")]  # [값 for 변수 in 목록 if 조건]
py = [f for f in files if f.endswith(".py")]
print("PNG file info : ", png, " Count : ", len(png))
print("PY file info : ", py, " Count : ", len(py))


# 방법 2: glob 패턴으로 같은 결과인지 확인
assert png == sorted(glob.glob("*.png", root_dir=folder))  # root_dir 폴더 안에서 *.png 찾기


# 방법 3: suffix로 확장자별 개수 세기
print("확장자별 개수:", dict(Counter(Path(f).suffix for f in files)))  # 파일 → 확장자 → 개수 세기
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `PNG file info : [...] Count : 4` | 원본 정답과 동일 (PY 6개) |
| `['file10.png', 'file11.png', 'file19.png', 'file6.png']` | 숫자 크기가 아니라 글자 순서 |
| `확장자별 개수: {'.jpg': 3, '.png': 4, '.py': 6, ...}` | png 4 · py 6 · txt · jpg · csv 각 3 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `f.endswith(".png")` | 문자열 **끝** 확인, 여러 개는 튜플로 | 목록에서 조건 걸러내기 (`(".jpg", ".png")`) |
| `glob("*.png")` | 찾기와 고르기를 **한 번에** | 처음부터 한 종류만 필요할 때 |
| `Path(f).suffix` | 확장자 **값 자체**를 꺼냄 | 확장자별 개수 · 그룹 나누기 |

> ⚠️ **주의** · 문자열 정렬은 숫자 크기 순서가 아님 → 파일명은 `img_001`처럼 0을 채워 저장

> ✅ **한 줄 정리** · `endswith()` · `glob("*.png")` · `Path.suffix`로 확장자를 골라낸다.

## 7. JSON

> **연결** · 6장에서는 글자만 저장했다 → 딕셔너리 · 리스트 구조를 그대로 저장 · 전달하려면 표준 형식 JSON

**흐름 위치** · ② 불러오기 · ⑦ 저장 — COCO 라벨, 설정 파일, 실험 결과

### 7-1. `A` 저장과 복원 · `json` <sub>원본 32 + 33</sub>

> 🎯 **딕셔너리를 JSON으로 저장하고, 그대로 복원하려면?**

💡 **쉽게 말하면** JSON은 **국제 택배 규격 상자**. 어떤 언어로 보내도 열 수 있지만 규칙이 엄격함 (문자열은 큰따옴표만).

```
dict → 문자열(dumps) · 파일(dump) → 다시 읽기(load) → 작은따옴표 문자열 loads 에러 → 큰따옴표로 바꿔 복원
```

**먼저 예측** · 작은따옴표로 쓴 JSON 문자열을 `loads`하면?

<details><summary>예측 정답</summary>

`JSONDecodeError` — JSON은 큰따옴표만 허용

</details>

````python
d = {"group1": [{"name": "Park", "age": "32", "sex": "Male"},
                {"name": "Cho", "age": "44", "sex": "Female"},
                {"name": "Kang", "age": "39", "sex": "Female", "married": "No"}],
     "group2": [{"name": "Kim", "age": "23", "sex": "Male", "married": "Yes"},
                {"name": "Lee", "age": "37", "sex": "Male", "married": "No"}],
     "type": {"a": "employee", "b": "officer", "c": "director", "d": "manager", "e": "service provider"}}


# 32-1: dict → JSON 문자열
text = json.dumps(d, indent=4)  # dumps(dict) → str, indent = 들여쓰기 칸 수
print(type(d).__name__, "→", type(text).__name__, "/ 앞 4줄:")
print("\n".join(text.splitlines()[:4]))  # 줄 단위로 나눔 → 앞 4줄 → 다시 이어 붙임


# dict → JSON 파일 저장 후 다시 읽어 비교
path = BASE / "people.json"
with open(path, "w", encoding="utf-8") as f:
    json.dump(d, f, indent=4, ensure_ascii=False)  # dump(dict, 파일) · 한글 그대로 저장
with open(path, encoding="utf-8") as f:
    assert json.load(f) == d  # load(파일) → dict → 원본과 비교


# 33-1: JSON 문자열 → dict (원본 문제는 작은따옴표 사용)
s = """{"group1": [{'name': 'Park', 'age': '32', 'sex': 'Male'},
                   {'name': 'Cho', 'age': '44', 'sex': 'Female'},
                   {'name': 'Kang', 'age': '39', 'sex': 'Female', 'married': 'No'}],
        "group2": [{'name': 'Kim', 'age': '23', 'sex': 'Male', 'married': 'Yes'},
                   {'name': 'Lee', 'age': '37', 'sex': 'Male', 'married': 'No'}],
        "type": {"a": "employee", "b": "officer", "c": "director", "d": "manager", "e": "service provider"}}"""


# 작은따옴표 그대로 읽으면 에러
try:
    json.loads(s)  # loads(str) → dict
except json.JSONDecodeError as e:
    print("JSONDecodeError:", e.msg)


# 큰따옴표로 바꾼 뒤 읽고 원본과 비교
loaded = json.loads(s.replace("'", '"'))  # ' → " 치환 → JSON 규칙 충족 → dict
assert loaded == d
print("복원 성공:", loaded["group1"][0])
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `dict → str / 앞 4줄` | dumps 결과는 문자열 (저장 · 전송용) |
| `(assert 통과)` | dump → load 해도 원본과 같음 |
| `JSONDecodeError: Expecting property name enclosed in double quotes` | 작은따옴표는 JSON 규칙 위반 (원본 문제의 함정) |
| `복원 성공: {'name': 'Park', 'age': '32', 'sex': 'Male'}` | 32-1의 d와 완전히 같음 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `json.dumps(d)` | dict → **문자열** | 출력 · 전송 · 다른 문자열에 넣을 때 |
| `json.dump(d, f)` | dict → **파일** | 설정 · 결과 저장 |
| `json.loads(s)` | **문자열** → dict | API 응답 · 문자열로 받은 JSON |
| `json.load(f)` | **파일** → dict | 라벨 파일 (COCO) · 설정 파일 읽기 |

> ⚠️ **주의** · `dumps` / `loads`는 문자열, `dump` / `load`는 파일 (s = string), 한글 저장은 `ensure_ascii=False`

> ✅ **한 줄 정리** · `json.dump()`로 저장하고 `json.load()`로 복원하며, 직접 쓸 때는 큰따옴표만 쓴다.

## 8. 종합 실습 · 폴더 속 이미지 → 학습용 데이터 목록

> **연결** · 1~7장 문법을 한 번에 사용. 실제 딥러닝 프로젝트의 첫 작업과 같은 흐름이며, 결과물(경로 · 번호 목록)이 다음 단계 PyTorch `Dataset`에 그대로 들어감

**흐름 위치** · ① 데이터 준비 → ② 불러오기 → ③ 전처리 → ⑦ 저장

> 🎯 **폴더에 흩어진 이미지로 학습용 목록(경로 · 라벨 · 번호 · 학습/검증 분할)을 어떻게 만드는가?**

💡 **쉽게 말하면** 사진이 `cat` · `dog` · `bird` 폴더에 나뉘어 있으면 **폴더 이름이 곧 정답(라벨)**. 사진 목록을 모으고, 이름을 번호로 바꾸고, 일부를 시험용(검증)으로 떼어 두는 것.

| 단계 | 하는 일 | 쓰는 문법 |
| :-- | :-- | :-- |
| 1 | 가짜 이미지 폴더 만들기 (`img_000.jpg` …) | 3-3 `items()` · 2-3 `range` · 5-2 `:03d` |
| 2 | 이미지 경로만 모으고 정렬 | 6-2 `glob` · 6-3 `suffix` · `sorted()` |
| 3 | 라벨 = 폴더 이름 | 2-4 컴프리헨션 |
| 4 | 클래스 이름 → 번호 | 2-5 `sorted(set())` · 3-4 `enumerate` |
| 5 | 학습 · 검증 나누기 함수 | 4-1 기본값 인자 · 2-2 슬라이싱 · 4-2 `return` |
| 6 | 목록을 JSON으로 저장 · 복원 | 7-1 `dump` · `load` · 5-2 f-string |

**먼저 예측** · `cat` 5장 · `dog` 4장 · `bird` 3장일 때 클래스 번호 `{이름: 번호}`는?

<details><summary>예측 정답</summary>

`{'bird': 0, 'cat': 1, 'dog': 2}` — 폴더를 만든 순서가 아니라 **이름 정렬 순서**로 번호가 정해짐

</details>

````python
import random  # 무작위 섞기 (seed로 결과 고정)


# 1단계: 가짜 이미지 폴더 만들기 (클래스 = 폴더 이름)
counts = {"cat": 5, "dog": 4, "bird": 3}
for cls, n in counts.items():  # 3-3 items()
    folder = BASE / "pets" / cls
    folder.mkdir(parents=True, exist_ok=True)
    for i in range(n):  # 2-3 range
        (folder / f"img_{i:03d}.jpg").touch()  # 5-2 :03d → img_000.jpg
(BASE / "pets" / "cat" / "memo.txt").touch()  # 이미지가 아닌 파일 섞기


# 2단계: 이미지 경로만 모으기 (확장자 거르기 + 정렬)
paths = sorted(p for p in (BASE / "pets").glob("*/*") if p.suffix == ".jpg")  # 6-2 glob · 6-3 suffix
print("이미지 수:", len(paths), "/ 첫 경로:", paths[0].as_posix())


# 3단계: 라벨 = 이미지가 들어 있는 폴더 이름
labels = [p.parent.name for p in paths]  # 2-4 컴프리헨션 · .parent.name → 폴더 이름
print("첫 라벨:", labels[0])


# 4단계: 클래스 이름 → 번호 (정렬해서 번호 고정)
classes = sorted(set(labels))  # 2-5 중복 제거 + 정렬
class_to_idx = {name: i for i, name in enumerate(classes)}  # 3-4 enumerate → {이름: 번호}
targets = [class_to_idx[label] for label in labels]  # 라벨 이름 → 번호 목록
print("클래스 번호:", class_to_idx)
print("클래스별 개수:", dict(Counter(labels)))
````

````python
# 5단계: 학습 · 검증으로 나누는 함수 (같은 seed → 항상 같은 분할)
def split_dataset(paths, targets, val_ratio=0.25, seed=42):  # 4-1 필수 → 기본값 인자
    idx = list(range(len(paths)))
    random.Random(seed).shuffle(idx)  # 번호 섞기 (원본 목록은 그대로)
    n_val = int(len(idx) * val_ratio)
    val_idx, train_idx = idx[:n_val], idx[n_val:]  # 2-2 슬라이싱으로 나누기
    train = [(paths[i].as_posix(), targets[i]) for i in sorted(train_idx)]  # (경로, 번호) 쌍
    val = [(paths[i].as_posix(), targets[i]) for i in sorted(val_idx)]
    return train, val  # 4-2 결과는 return으로


train, val = split_dataset(paths, targets)


# 나눈 결과 검증: 개수 합 · 겹침 없음
assert len(train) + len(val) == len(paths)
assert not set(p for p, _ in train) & set(p for p, _ in val)  # 교집합(&)이 비어 있어야 함
print(f"train {len(train):>2}장 | val {len(val):>2}장 | 클래스 {len(classes)}개")  # 5-2 :>2 → 2칸 오른쪽 정렬
print("val 예시:", val[:2])


# 6단계: 데이터 목록을 JSON으로 저장 후 다시 읽기
meta = {"classes": classes, "class_to_idx": class_to_idx, "train": train, "val": val}
with open(BASE / "dataset.json", "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)  # 7-1 dump → 파일
with open(BASE / "dataset.json", encoding="utf-8") as f:
    loaded = json.load(f)  # 7-1 load → dict
print("복원 후 train 첫 항목:", loaded["train"][0], "→", type(loaded["train"][0]).__name__)  # 튜플 → 리스트로 바뀜
````

**결과 해설**

| 출력 | 의미 |
| :-- | :-- |
| `이미지 수: 12 / 첫 경로: 실습파일/pets/bird/img_000.jpg` | `memo.txt`는 `.jpg`가 아니라 제외됨, 정렬로 bird가 먼저 |
| `클래스 번호: {'bird': 0, 'cat': 1, 'dog': 2}` | 이름 정렬 순서로 번호 고정 → 언제 실행해도 같은 번호 |
| `클래스별 개수: {'bird': 3, 'cat': 5, 'dog': 4}` | 클래스 불균형 확인 (cat이 가장 많음) |
| <code>train  9장 &#124; val  3장 &#124; 클래스 3개</code> | 12 × 0.25 = 3장을 검증용으로, 두 목록은 겹치지 않음 (assert 통과) |
| `복원 후 train 첫 항목: [...] → list` | 저장 전에는 튜플 `( )`, JSON을 거치면 리스트 `[ ]`로 바뀜 |

**차이 · 언제 쓰나**

| 방법 | 차이 | 언제 쓰나 |
| :-- | :-- | :-- |
| `random.shuffle(idx)` | 전역 난수 사용 → 다른 코드의 영향을 받음 | 결과 재현이 필요 없을 때 |
| `random.Random(42).shuffle(idx)` | **이 작업 전용** 난수, seed로 항상 같은 결과 | 학습 · 검증 분할처럼 재현이 필요할 때 (기본 선택) |
| 경로를 섞기 | 원본 목록 순서가 바뀜 | 목록을 다시 안 쓸 때 |
| 번호(`idx`)를 섞기 | 원본 `paths` · `targets`는 그대로 | 여러 목록을 같은 순서로 나눌 때 (기본 선택) |

> ⚠️ **주의** · JSON은 튜플을 리스트로 바꿔 저장함 → 불러온 뒤 `(경로, 번호)`를 튜플로 비교하면 `False`

> ✅ **한 줄 정리** · 폴더 이름을 라벨로 쓰고, `sorted(set())`로 번호를 고정한 뒤, seed를 고정해 나눈 목록을 JSON으로 저장한다.

## 10. 면접 30초 답변

**Q1. 리스트를 `y = x`로 복사하면 어떤 문제가 생기나요?**
> `y = x`는 새 리스트가 아니라 같은 객체에 이름을 하나 더 붙이는 것이라, `y`를 수정하면 `x`도 바뀝니다. 독립 복사본은 `x.copy()`, 중첩 구조는 `copy.deepcopy()`를 씁니다. pandas의 `df.copy()`, PyTorch의 `tensor.clone()`이 있는 이유도 같습니다.

**Q2. 이미지 데이터셋을 만들 때 파일 목록을 왜 정렬하나요?**
> `os.listdir`나 `glob`의 반환 순서는 운영체제와 파일 시스템에 따라 보장되지 않습니다. 정렬하지 않으면 실행할 때마다 순서가 달라져 이미지와 라벨의 짝이 어긋나고 재현성이 깨지므로 `sorted()`로 고정합니다.

**Q3. `set`으로 클래스 목록을 만들 때 주의할 점은?**
> `set`은 순서를 보장하지 않아 실행마다 클래스 번호가 달라질 수 있습니다. 그래서 `sorted(set(labels))`로 순서를 고정한 뒤 `enumerate`로 번호를 매겨, 학습과 추론에서 같은 번호 체계를 쓰게 합니다.

**Q4. 학습 · 검증 데이터를 나눌 때 seed를 고정하는 이유는?**
> seed를 고정하지 않으면 실행할 때마다 다른 데이터가 검증용으로 뽑혀, 모델을 바꿔서 성능이 오른 것인지 데이터 분할이 달라서 오른 것인지 구분할 수 없습니다. 그래서 `random.Random(42)`처럼 seed를 고정해 같은 분할에서 비교합니다.

## 11. 전체 코드

1~8장 정답 코드를 한 셀에 모음. **이 셀만 실행해도 전체가 동작**하며(실습 파일 생성 포함), 각 출력 앞 `[번호]`가 섹션.

````python
# ============================================================
# Python 문법 전체 정리 — 전체 코드
# ============================================================
import os
import glob
import json
import keyword
import random
from collections import Counter
from pathlib import Path


# ---------- 0. 준비: 실습 파일 만들기 ----------
BASE = Path("실습파일")
for name in ["26-1", "27-1", "42-1"]:
    (BASE / name).mkdir(parents=True, exist_ok=True)  # 폴더 생성, 있으면 통과

words = ["Python", "JavaScript", "PHP", "Rust", "elite", "Solidity", "Assembly", "hamster", "india",
         "january", "kibana", "lamada", "monster", "notion", "orange", "pokemon", "query", "range",
         "sonic", "telegram", "urban", "village", "world", "x-ray", "yellow", "zigzag"]
for letter, word in zip("ABCDEFGHIJKLMNOPQRSTUVWXYZ", words):
    (BASE / "27-1" / f"{letter}.txt").write_text(word, encoding="utf-8")  # A.txt ~ Z.txt

ext = {6: "png", 10: "png", 11: "png", 19: "png", 3: "py", 4: "py", 8: "py", 12: "py", 13: "py", 17: "py"}
for i in range(1, 20):
    (BASE / "42-1" / f"file{i}.{ext.get(i, ['txt', 'jpg', 'csv'][i % 3])}").touch()  # file1 ~ file19


# ---------- 1. 변수 · 비교 · 자료형 ----------
# 1-1 변수 이름 규칙: 숫자로 시작하는 이름 찾기
print("[1-1]", [n for n in ["a", "_b", "2c", "_7d"] if not n.isidentifier()], keyword.iskeyword("class"))

# 1-2 값 비교(==) vs 객체 비교(is)
x = ["orange", "banana", "apple"]
y = x  # 같은 객체에 이름표 추가
z = ["orange", "banana", "apple"]  # 새 객체
print("[1-2]", x == y, x is y, x == z, x is z)

# 1-3 자료형 맞춰 이어 붙이기
print("[1-3]", "Seoul" + str(25), f"{'Seoul'}{25}")  # str(int) · f-string


# ---------- 2. 리스트 ----------
# 2-1 인덱싱: 앞 번호 · 뒤 번호 · 위치 찾기
fruits = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]
print("[2-1]", fruits[4], fruits[-2], fruits[fruits.index("Banana")])

# 2-2 슬라이싱: [start:stop] stop 미포함
letters = list("abcdefghijklm")
print("[2-2]", letters[4:7], letters[-9:-6])

# 2-3 range: 30 → -10, -2씩
print("[2-3]", list(range(30, -11, -2)))

# 2-4 리스트 컴프리헨션: 1~15 → ×10 → 문자열
print("[2-4]", [str(n * 10) for n in range(1, 16)])

# 2-5 중복 제거: set → 정렬 / 순서 유지
dup = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]
print("[2-5]", sorted(set(dup), key=lambda v: (isinstance(v, str), v)), list(dict.fromkeys(dup)))


# ---------- 3. 딕셔너리 ----------
# 3-1 값 합계: values() → sum
d = {"a": 17, "b": 114, "c": 247, "d": 362, "e": 220, "f": 728, "g": -283, "h": 922}
print("[3-1]", sum(d.values()))

# 3-2 항목 합치기: a | b
print("[3-2]", {"a": "apple", "b": "grape"} | {"c": "banana", "d": "kiwi"})

# 3-3 items()로 키 · 값 함께 꺼내기
nums = dict(one=list(range(1, 11)), two=list(range(11, 23)), three=list(range(23, 37)))
print("[3-3]", {key: len(values) for key, values in nums.items()})  # {키: 값 개수}

# 3-4 zip: 같은 위치의 이름·수량·단가를 묶어 {이름: 수량 × 단가} 생성
names, qty, price = ["one", "two", "three", "four"], [30, 20, 15, 75], [5.2, 7.4, 3.6, 4.2]
print("[3-4]", {k: q * p for k, q, p in zip(names, qty, price)})

# 3-4 enumerate: 번호를 100부터 붙여 딕셔너리 생성
print("[3-4]", dict(enumerate(["Red", "Green", "Black", "Blue", "Orange", "Purple"], start=100)))


# ---------- 4. 함수 ----------
# 4-1 필수 인자 → 기본값 인자 순서로 정의
def greet(name, msg="Good morning!"):
    return "Hi! " + name + ", " + msg


print("[4-1]", greet("Kim"), "|", greet("Park", "How do you do?"))


# 4-2 global로 바꾼 a만 함수 밖에 남음
def test(x, y):
    global a
    a = 49
    x, y = y, x
    b = 7
    print("[4-2] Step1 :", a, b, x, y)


a, b, x, y = 8, 13, 33, 44
test(23, 7)
print("[4-2] Step2 :", a, b, x, y)


# ---------- 5. 문자열 ----------
# 5-1 split()으로 단어 개수 세기
in_str = "Suppose we have few words that are separated by spaces."
print("[5-1]", len(in_str.split()))

# 5-2 %x(16진수) · f-string · 형식 지정
serialno = 308276567
print("[5-2]", "%x" % serialno, f"{serialno}", f"{0.876:.1%} {1234567:,} {3.14159:.2f}")


# ---------- 6. 파일 · 폴더 ----------
# 6-1 이름 · 내용을 짝지어 파일 쓰기
file_names = ["A", "B", "C", "D", "F", "G"]
file_texts = ["Python", "JavaScript", "PHP", "Rust", "Solidity", "Assembly"]
for name, text in zip(file_names, file_texts):
    with open(BASE / "26-1" / f"{name}.txt", "w", encoding="utf-8") as f:
        f.write(text)
print("[6-1]", sorted(os.listdir(BASE / "26-1")))

# 6-2 *.txt 찾기 → 정렬 → 읽기
read_all = [p.read_text(encoding="utf-8").strip() for p in sorted((BASE / "27-1").glob("*.txt"))]
print("[6-2]", len(read_all), read_all[:5])

# 6-3 확장자로 파일 골라내기
files = sorted(os.listdir(BASE / "42-1"))
print("[6-3] PNG", [f for f in files if f.endswith(".png")], "| PY", [f for f in files if f.endswith(".py")])


# ---------- 7. JSON ----------
# 7-1 dict → 파일 저장(dump) → 다시 읽기(load) → 원본과 비교
people = {"group1": [{"name": "Park", "age": "32", "sex": "Male"}], "type": {"a": "employee"}}
with open(BASE / "people_all.json", "w", encoding="utf-8") as f:
    json.dump(people, f, ensure_ascii=False)
with open(BASE / "people_all.json", encoding="utf-8") as f:
    print("[7-1]", json.load(f) == people == json.loads(json.dumps(people)))

# 7-1 작은따옴표 → 큰따옴표로 바꿔 JSON 읽기
print("[7-1]", json.loads("{'name': 'Park'}".replace("'", '"')))


# ---------- 8. 종합 실습: 폴더 속 이미지 → 학습용 목록 ----------
# 8-1 폴더 이름 = 클래스, 이미지 경로 · 라벨 · 번호 만들기
for cls, n in {"cat": 5, "dog": 4, "bird": 3}.items():
    (BASE / "pets" / cls).mkdir(parents=True, exist_ok=True)
    for i in range(n):
        (BASE / "pets" / cls / f"img_{i:03d}.jpg").touch()
paths = sorted(p for p in (BASE / "pets").glob("*/*") if p.suffix == ".jpg")
labels = [p.parent.name for p in paths]
class_to_idx = {name: i for i, name in enumerate(sorted(set(labels)))}
targets = [class_to_idx[label] for label in labels]
print("[8-1]", len(paths), class_to_idx)

# 8-2 seed 고정으로 섞고 슬라이싱으로 학습 · 검증 나누기
idx = list(range(len(paths)))
random.Random(42).shuffle(idx)
n_val = int(len(idx) * 0.25)
train_idx, val_idx = idx[n_val:], idx[:n_val]
print("[8-2]", f"train {len(train_idx)}장 | val {len(val_idx)}장", not set(train_idx) & set(val_idx))
````

## 12. 전체 최종 요약표

### 등급별 핵심

| 등급 | 주제 | 반드시 기억할 것 | 딥러닝 코드에서 |
| :-: | :-- | :-- | :-- |
| `A` | 비교 · 복사 | `==` 값 · `is` 객체, 복사는 `copy()` | `df.copy()`, `tensor.clone()` |
| `A` | 자료형 | 계산 전 자료형 맞추기 | `x.float()`, `labels.long()` |
| `A` | 인덱싱 · 슬라이싱 · range | 0부터, **stop 미포함** | `X[:800]`, `range(num_epochs)` |
| `A` | 컴프리헨션 | `[변환 for 원소 in 리스트]` | `[transform(i) for i in imgs]` |
| `A` | set + sorted | 중복 제거 후 **반드시 정렬** | `sorted(set(labels))` |
| `A` | items · zip · enumerate | 키·값 / 같은 위치 / 번호+값 | `for i, (x, y) in enumerate(loader):` |
| `A` | 함수 | 필수 → 기본값 인자, 결과는 `return` | `def train(model, loader, epochs=10):` |
| `A` | split · f-string | 나누기 · `{값:형식}` | `f"loss {loss:.4f}"` |
| `A` | 파일 목록 | `sorted(glob(...))` | 이미지 경로 목록 |
| `A` | JSON | `dump/load` 파일 · `dumps/loads` 문자열 | COCO 라벨, 결과 저장 |
| `A` | 종합 | 폴더 이름 = 라벨, seed 고정 분할 | `ImageFolder`, 데이터 분할 |
| `B` | 변수 이름 | 문자 · 밑줄로 시작, 예약어 X | `num_classes` |
| `B` | 딕셔너리 합계 · 추가 | `values()`, `update()` | `config.update(...)` |
| `B` | 전역 변수 | `global` 대신 `return` | `best = evaluate(...)` |
| `B` | 파일 쓰기 · 확장자 | `with open(..., "w")`, `endswith()` | 로그 저장, 이미지만 고르기 |

### 할 것 · 하지 말 것

| ✅ 할 것 | ⚠️ 하지 말 것 |
| :-- | :-- |
| 파일 목록은 `sorted()` | `set` 순서를 믿고 번호 매기기 |
| 원본을 바꾸기 전 `copy()` | 길이 확인 없이 `zip` |
| `with open(..., encoding="utf-8")` | `global`로 학습 상태 관리 |
| 결과는 `assert`로 검증 | 작은따옴표 JSON 직접 작성 |
| 데이터 분할은 seed 고정 | JSON에서 읽은 값을 튜플로 비교 |

### 헷갈리는 짝 한눈에 보기

| 짝 | 차이 | 결론 |
| :-- | :-- | :-- |
| `==` vs `is` | 값 비교 vs 같은 객체 | 보통 `==`, `None`만 `is` |
| `y = x` vs `x.copy()` | 이름 추가 vs 새 객체 | 원본 지킬 땐 `copy()` |
| `x[i]` vs `x[a:b]` | 값 1개 vs 리스트 | 구간은 슬라이싱 |
| 컴프리헨션 vs `map` | 읽기 쉬움 · 조건 가능 vs 함수 그대로 적용 | 보통 컴프리헨션 |
| `set` vs `sorted(set)` vs `dict.fromkeys` | 순서 없음 vs 정렬 vs 등장 순서 | 번호 매길 땐 `sorted(set)` |
| `d[k]` vs `d.get(k)` | 없으면 에러 vs 기본값 | 필수 키는 `d[k]` |
| `update` vs a &#124; b | 원본 수정 vs 새로 생성 | 원본 지킬 땐 &#124; 연산자 |
| `zip` vs `enumerate` | 같은 위치끼리 vs 번호 + 값 | 짝짓기 vs 몇 번째 |
| `global` vs `return` | 밖 변수 수정 vs 결과 돌려주기 | `return` |
| `split()` vs `split(" ")` | 연속 공백 하나로 vs 한 칸마다 | `split()` |
| `format` vs f-string | 틀 재사용 vs 즉시 작성 | 보통 f-string |
| `"w"` vs `"a"` | 덮어쓰기 vs 이어 쓰기 | 로그는 `"a"` |
| `listdir` vs `glob` vs `Path.glob` | 전체 이름 vs 패턴 · 경로 vs 경로 객체 | 패턴 필요하면 `glob` |
| `dump/load` vs `dumps/loads` | 파일 vs 문자열 (s = string) | 저장 · 읽기는 `dump/load` |
