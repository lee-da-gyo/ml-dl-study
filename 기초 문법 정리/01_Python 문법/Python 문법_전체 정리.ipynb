### 1-1. `B` 변수 이름 규칙 · `isidentifier()` <sub>원본 01</sub>

> 🎯 **어떤 이름은 변수로 쓸 수 있고, 어떤 이름은 왜 안 되는가?**

💡 **쉽게 말하면**

- 변수 이름은 **이름표**.
- 숫자로 시작하면 파이썬이 숫자로 착각하고, `class` · `for` 같은 예약어는 이미 주인이 있는 이름표.

**실행 흐름**

1. 보기 4개 isidentifier() 검사
2. 2c 실제 실행
3. 예약어 확인

**먼저 예측** · `a` · `_b` · `2c` · `_7d` 중 안 되는 것은? — 답은 코드 실행 결과 아래에

````python
# 보기 4개가 변수 이름 규칙에 맞는지 검사
# 왜 · 원본 문제의 보기를 눈대중이 아니라 isidentifier()로 판정하기 위해
candidates = ["a", "_b", "2c", "_7d"]  # 원본 문제의 보기 4개
for name in candidates:                # 보기마다 검사
    # 문자·밑줄 시작 → True / 숫자 시작 → False
    print(f"{name:>4} → 사용 가능: {name.isidentifier()}")


# 보기 (3)을 실제로 실행해 에러 확인
# 왜 · 규칙 위반이 실제로 SyntaxError가 되는지 셀을 멈추지 않고 보기 위해
try:
    exec("2c = int(77)")          # 문자열 → 코드로 실행
except SyntaxError as e:          # 이름 규칙 위반 = 문법 에러
    print("SyntaxError:", e.msg)  # 에러 메시지만 출력


# 규칙에 맞아도 예약어는 사용 불가
# 왜 · 이름 규칙과 별개로 예약어도 막혀 있다는 것을 함께 확인하기 위해
print("class 예약어?", keyword.iskeyword("class"))  # 규칙엔 맞지만 예약어인지
````

**결과** · 예측 정답은 (3) `2c` — 숫자로 시작

- `2c → 사용 가능: False` — 숫자로 시작 → 정답 (3)
- `SyntaxError: invalid decimal literal` — 2c를 숫자로 읽으려다 실패
- `class 예약어? True` — 규칙에 맞아도 예약어는 사용 불가

### 1-2. `A` 값 비교와 객체 비교 · `==` · `is` <sub>원본 02</sub>

> 🎯 **`==`와 `is`는 무엇이 다르며, `y = x`는 왜 복사가 아닌가?**

💡 **쉽게 말하면**

- `==`는 **내용이 같은가**, `is`는 **같은 물건인가**.
- `y = x`는 물건을 복사한 것이 아니라 같은 물건에 **이름표를 하나 더 붙인 것**.

<img src="images/py1_names.png" alt="y = x는 이름표 추가, copy()는 새 객체" width="763">

**실행 흐름**

1. 정수 · 같은 리스트 · 내용만 같은 리스트 비교
2. y = x 함정
3. copy()

**먼저 예측** · `b = a` 뒤에 `b.append(4)` 하면 `a`는? — 답은 코드 실행 결과 아래에

````python
# 예제 1: 서로 다른 정수
# 왜 · 값도 객체도 다를 때 == 와 is 가 모두 False인 기준을 잡기 위해
x, y = 15, 25                     # 값이 다른 정수 두 개
print("예제1 →", x == y, x is y)  # 값 비교 · 객체 비교


# 예제 2: y = x 로 같은 리스트를 가리킴
# 왜 · = 가 복사가 아니라 이름표 추가임을 확인하기 위해
x = ["orange", "banana", "apple"]  # 리스트 객체 하나 생성
y = x                              # 새 리스트 X → 같은 리스트에 이름표 추가
print("예제2 →", x == y, x is y)   # 값 비교 · 객체 비교


# 예제 3: 내용만 같은 새 리스트
# 왜 · 내용이 같아도 새로 만든 리스트는 다른 객체임을 예제 2와 비교하기 위해
y = ["orange", "banana", "apple"]  # [ ] 를 쓰면 새 객체 생성
print("예제3 →", x == y, x is y)   # 내용 같음 · 객체 다름


# y = x 함정: b만 바꿨는데 a도 바뀜
# 왜 · 이름표만 붙인 사본을 고치면 원본이 망가지는 실수를 눈으로 보기 위해
a = [1, 2, 3]           # 원본 리스트
b = a                   # 복사 X → 같은 객체에 이름 추가
b.append(4)             # b 수정 → 같은 객체인 a도 수정
print("a도 바뀜 →", a)  # b만 고쳤는데 a 확인


# copy()로 진짜 복사
# 왜 · 원본은 그대로 두고 사본만 고치기 위해
c = a.copy()                    # 새 객체 생성
c.append(5)                     # c만 수정
print("복사본만 바뀜 →", a, c)  # a는 그대로인지 확인
````

**결과** · 예측 정답은 `[1, 2, 3, 4]` — `a`와 `b`는 같은 객체

- `예제1 → False False` · `예제2 → True True` · `예제3 → True False` — 값 · 객체 모두 다름 / 같은 객체 / 내용만 같고 다른 객체
- `a도 바뀜 → [1, 2, 3, 4]` — b = a 는 이름표 추가 → 원본까지 바뀜
- `복사본만 바뀜 → [1, 2, 3, 4] [1, 2, 3, 4, 5]` — copy()로 만든 c만 바뀜

### 1-3. `A` 자료형 변환 · `str()` · `int()` <sub>원본 03</sub>

> 🎯 **문자열과 숫자는 왜 바로 더할 수 없고, 어떻게 맞추는가?**

💡 **쉽게 말하면**

- `"Seoul"`은 **글자**, `25`는 **숫자**.
- 글자 + 숫자는 사과 + 3킬로미터처럼 단위가 달라 한쪽으로 바꿔야 함.

**실행 흐름**

1. 에러 메시지 확인
2. str() · f-string으로 수정
3. int()로 반대 변환

**먼저 예측** · `"Seoul" + 25`의 결과는? — 답은 코드 실행 결과 아래에

````python
x, y = "Seoul", 25  # 문자열 · 정수


# 문자열 + 정수 → 에러 확인
# 왜 · 원본 문제의 TypeError를 셀을 멈추지 않고 확인하기 위해
try:
    z = x + y               # str + int
except TypeError as e:      # 자료형이 안 맞으면 TypeError
    print("TypeError:", e)  # 에러 메시지 그대로 읽기


# 방법 1: 정수를 문자열로 바꿔 이어 붙이기
# 왜 · 자료형을 str 하나로 맞춰 + 연산이 되게 하기 위해
z1 = x + str(y)  # str + str(int) → str


# 방법 2: f-string으로 끼워 넣기
# 왜 · 형변환을 직접 쓰지 않고 같은 결과를 더 짧게 얻기 위해
z2 = f"{x}{y}"  # f"{값}{값}" → 자동으로 문자열


# 두 방법의 결과가 같은지 확인
# 왜 · 두 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert z1 == z2 == "Seoul25"  # 두 방법이 같은 결과인지 검증
print("x + y :", z1)          # 이어 붙인 결과


# 자료형 확인 · 반대 방향 변환
# 왜 · 문자열 → 정수 변환도 있다는 것을 함께 보기 위해
print(type(x).__name__, type(y).__name__, "|", int("25") + 10)  # int("25") → 25 → 25 + 10
````

**결과** · 예측 정답은 `TypeError` — 문자열과 정수는 바로 더할 수 없음

- `TypeError: can only concatenate str (not "int") to str` — 문자열에는 문자열만 이어 붙일 수 있음
- `x + y : Seoul25` — 두 방법 결과 동일

**실행 흐름**

**먼저 예측** · 6개 중 `Banana`의 번호는? — 답은 코드 실행 결과 아래에

````python
x = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]  # 번호 0 ~ 5인 6개


# 세 가지 방법으로 Banana 꺼내기
# 왜 · 같은 값을 앞 번호 · 뒤 번호 · index()로 꺼내 번호 규칙을 익히기 위해
print(x[4], x[-2], x[x.index("Banana")])  # 앞에서 4번 · 뒤에서 2번 · 위치 찾아서


# 위치와 전체 길이 확인
print("위치:", x.index("Banana"), "/ 길이:", len(x))  # 마지막 번호 = 길이 - 1


# 없는 값을 찾으면 에러
# 왜 · 대소문자가 다른 값을 찾을 때 ValueError가 나는 것을 미리 보기 위해
try:
    x.index("banana")        # 소문자 → 다른 값
except ValueError as e:      # 없는 값이면 ValueError
    print("ValueError:", e)  # 에러 메시지 확인
````

### 2-2. `A` 구간 자르기 · `x[a:b]` <sub>원본 05</sub>

> 🎯 **연속된 구간을 어떻게 잘라내며, 끝 번호는 왜 포함되지 않는가?**

💡 **쉽게 말하면**

- `[4:7]`은 **4번 사물함부터 7번 앞까지**.
- 끝을 빼야 `[0:4]` + `[4:8]`처럼 구간을 겹침 없이 이어 붙일 수 있음.

<img src="images/py1_slice.png" alt="4:7로 자르면 4 · 5 · 6번 (7번 미포함)" width="763">

**실행 흐름**

1. [e, f, g]를 3가지로 자르기
2. step · 뒤집기

**먼저 예측** · `x[4:7]`은 몇 개? — 답은 코드 실행 결과 아래에

````python
x = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j", "k", "l", "m"]  # 번호 0 ~ 12인 13개


# 방법 1: 앞 번호로 자르기
# 왜 · 끝 번호는 결과에 들어가지 않는다는 슬라이싱 규칙을 확인하기 위해
print(x[4:7])  # [start:stop] → 4, 5, 6 (7 미포함)


# 방법 2: 뒤 번호로 자르기
print(x[-9:-6])  # 뒤에서 9번째 ~ 6번째 앞


# 방법 3: 위치를 찾아서 자르기
print(x[x.index("e"):x.index("g") + 1])  # g까지 넣으려면 +1


# 세 방법의 결과가 같은지 확인
# 왜 · 세 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert x[4:7] == x[-9:-6] == ["e", "f", "g"]  # 세 방법이 같은지 검증


# step 사용: 간격 · 뒤집기
# 왜 · 건너뛰기와 뒤집기도 같은 [시작:끝:간격] 문법임을 보이기 위해
print(x[::3], x[::-1][:3])  # [::3] 3칸씩 · [::-1] 뒤집기 → 앞 3개
````

**결과** · 예측 정답은 `3`개 — 4, 5, 6번 (7번 미포함)

- 시작 · 끝 · 간격을 주면 차례로 뽑아 주고, 슬라이싱처럼 끝은 미포함.

### 2-4. `A` 전체 변환 · `[식 for x in 목록]` · `map()` <sub>원본 09</sub>

> 🎯 **리스트 전체에 같은 변환을 한 번에 적용하려면?**

💡 **쉽게 말하면**

- **공장 컨베이어 벨트**.
- 원소가 하나씩 지나가며 같은 가공을 받고, `lambda`는 이름 없는 한 줄짜리 가공 기계.

**실행 흐름**

1. 1~15
2. ×10
3. 문자열
4. map + lambda / 컴프리헨션 비교

**먼저 예측** · `map(...)`을 `list()` 없이 출력하면? — 답은 코드 실행 결과 아래에

````python
# 방법 1: map()과 lambda 사용
# 왜 · 함수를 모든 값에 적용하는 map 방식을 컴프리헨션과 비교하기 위해
r1 = list(map(lambda n: str(n * 10), range(1, 16)))  # 1~15 → ×10 → 문자열 변환 → 리스트


# 방법 2: 리스트 컴프리헨션 사용
# 왜 · 같은 일을 더 읽기 쉬운 한 줄로 쓰기 위해
r2 = [str(n * 10) for n in range(1, 16)]  # 표현식 for 변수 in 반복 범위


# 두 방법의 결과가 같은지 확인
# 왜 · 두 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert r1 == r2  # 두 방법 결과가 같은지
print(r1)        # 15개 모두 문자열


# list()로 감싸지 않으면 map 객체
# 왜 · map은 바로 값이 아니라 객체를 돌려준다는 함정을 보이기 위해
print("list() 없이:", type(map(str, [1])).__name__)  # 값이 아니라 map 객체
````

**결과** · 예측 정답은 값이 아니라 `map` 객체가 출력됨

- `['10', '20', '30'` · `'140', '150']` — 15개 모두 문자열

### 2-5. `A` 중복 제거 · `set()` <sub>원본 10</sub>

> 🎯 **중복을 없앤 목록을 만들 때 순서는 어떻게 지키는가?**

💡 **쉽게 말하면**

- `set`은 **주머니**.
- 같은 것은 하나만 들어가지만 순서가 없어서, 순서가 필요하면 `sorted()`로 다시 줄 세움.

**실행 흐름**

1. set 중복 제거
2. 숫자 먼저 · 문자 나중 정렬
3. 원래 순서 유지
4. 섞어 정렬 에러

**먼저 예측** · `set`으로 바꾸면 순서는 유지될까? — 답은 코드 실행 결과 아래에

````python
x = ["a", 1, "b", 2, "a", 3, "b", 4, 5, "b"]  # 숫자 · 문자가 섞인 10개


# set으로 중복 제거
# 왜 · 중복을 한 번에 없애고 줄어든 개수로 확인하기 위해
uniq = set(x)                           # 중복 X · 순서 X
print("개수:", len(x), "→", len(uniq))  # 줄어든 개수 확인


# 방법 1: 숫자 먼저, 문자 나중으로 정렬
# 왜 · 순서가 없는 set을 원하는 규칙으로 정렬해 출력을 고정하기 위해
# key = (문자인가, 값) → False(숫자)가 먼저
print("정답 형태 :", sorted(uniq, key=lambda v: (isinstance(v, str), v)))


# 방법 2: 처음 나온 순서 그대로 유지
# 왜 · 정렬 대신 처음 나온 순서를 지켜야 할 때의 방법을 보이기 위해
print("순서 유지 :", list(dict.fromkeys(x)))  # 키 = 값 → 중복 키 자동 제거, 순서 유지


# 숫자와 문자를 그대로 정렬하면 에러
# 왜 · 숫자와 문자를 섞어 정렬하면 TypeError가 나는 이유를 확인하기 위해
try:
    sorted(uniq)                                      # 1 < "a" 비교 불가
except TypeError:                                     # 숫자 < 문자 비교 불가
    print("TypeError: 숫자와 문자는 크기 비교 불가")  # 에러 이유 출력
````

**결과** · 예측 정답은 보장되지 않음 — 순서가 필요하면 정렬

- `개수: 10 → 7` — a · b · 1~5만 남음
- `[1, 2, 3, 4, 5, 'a', 'b']` — key = (문자인가, 값) → 숫자가 먼저
- `['a', 1, 'b', 2, 3, 4, 5]` — 처음 등장 순서 유지
- `TypeError: 숫자와 문자는 크기 비교 불가` — 1 < 'a' 비교 불가 → key 필요

### 3-1. `B` 값 합계 · `values()` <sub>원본 11</sub>

> 🎯 **딕셔너리에서 값만 꺼내 계산하려면?**

💡 **쉽게 말하면**

- 딕셔너리는 **이름표(키) 붙은 서랍장**.
- `keys()`는 이름표만, `values()`는 내용물만, `items()`는 둘 다.

**실행 흐름**

1. values() 합계 · 반복 누적 · 키로 접근
2. 2327 검증

**먼저 예측** · `sum(d)`를 하면? — 답은 코드 실행 결과 아래에

````python
d = {"a": 17, "b": 114, "c": 247, "d": 362, "e": 220, "f": 728, "g": -283, "h": 922}  # 키 8개


# 방법 1: values()로 값만 꺼내 합계
# 왜 · 키는 필요 없고 값만 더하면 될 때 가장 짧게 쓰기 위해
s1 = sum(d.values())  # 값 목록 → 합계


# 방법 2: 반복문으로 하나씩 누적
# 왜 · sum()이 하는 일을 반복문으로 풀어 비교하기 위해
s2 = 0                # 누적할 합계를 0부터
for v in d.values():  # 값만 하나씩
    s2 += v           # s2 = s2 + v


# 방법 3: 키로 값에 접근해 합계
# 왜 · 키로 값을 찾는 방식도 같은 합계를 내는지 비교하기 위해
s3 = sum(d[k] for k in d)  # 키 반복 → d[키] → 합계


# 세 방법의 결과가 같은지 확인
# 왜 · 세 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert s1 == s2 == s3 == 2327  # 세 방법 결과 검증
print("합계:", s1)             # 값 합계
````

**결과** · 예측 정답은 `TypeError` — 키(문자열)를 더하려 함

### 3-2. `B` 항목 추가 · 합치기 · `update()` <sub>원본 12</sub>

> 🎯 **딕셔너리에 항목을 추가하거나 두 딕셔너리를 합치려면?**

💡 **쉽게 말하면**

- 서랍장에 **새 이름표 서랍**을 끼우는 것.
- 같은 이름표가 이미 있으면 내용물이 교체됨.

**실행 흐름**

1. copy() 후 4가지 방법
2. 결과 동일
3. 원본 확인

**먼저 예측** · 같은 키가 양쪽에 있으면 어느 값이 남나? — 답은 코드 실행 결과 아래에

````python
base = {"a": "apple", "b": "grape"}  # 기존 딕셔너리
new = {"c": "banana", "d": "kiwi"}   # 더할 항목


# 방법 1: 키로 직접 넣기
# 왜 · 원본 base는 지키고 사본에 항목을 하나씩 추가하기 위해
d1 = base.copy()                     # 원본 보존용 복사
d1["c"], d1["d"] = "banana", "kiwi"  # d[키] = 값


# 방법 2: update()로 합치기
# 왜 · update()는 원본을 바꾸므로 사본에서 써야 한다는 것을 보이기 위해
d2 = base.copy()  # update는 원본을 바꾸므로 복사본에
d2.update(new)    # d2에 new를 덮어씀 (원본 수정)


# 방법 3: 풀어서 새 딕셔너리 만들기
# 왜 · 원본을 바꾸지 않고 새 딕셔너리로 합치기 위해
d3 = {**base, **new}  # {**a, **b} → a 펼침 + b 펼침


# 방법 4: 합치기 연산자 (Python 3.9+)
# 왜 · 가장 짧은 최신 문법도 같은 결과인지 비교하기 위해
d4 = base | new  # a | b → 새 딕셔너리


# 네 방법의 결과가 같은지 확인
# 왜 · 네 방법이 같은 결과를 내고 원본은 그대로인지 검증하기 위해
assert d1 == d2 == d3 == d4  # 네 방법 결과 검증
print(d1)                    # 합친 결과
print("원본 그대로:", base)  # copy() 덕분에 base 유지
````

**결과** · 예측 정답은 뒤에 합친 쪽의 값

- `{'a': 'apple', 'b': 'grape', 'c': 'banana', 'd': 'kiwi'}` — 4가지 결과 동일
- `원본 그대로: {'a': 'apple', 'b': 'grape'}` — copy() 덕분에 base 유지

### 3-3. `A` 키 · 값 반복 · `items()` <sub>원본 15</sub>

- `items()`는 서랍마다 **(이름표, 내용물) 한 쌍**을 건네주고, `for 키, 값 in`으로 양손에 나눠 받음.

2. items() 반복 출력

**먼저 예측** · `for k in d:`는 키와 값 중 무엇을 꺼내나? — 답은 코드 실행 결과 아래에

````python
# 키별로 숫자 리스트를 가진 딕셔너리 만들기
# 왜 · 원본 문제와 같은 데이터로 정답 출력을 그대로 비교하기 위해
# dict(키=값, ...)
d = dict(one=list(range(1, 11)), two=list(range(11, 23)), three=list(range(23, 37)))


# items()로 키와 값을 함께 꺼내며 출력
# 왜 · d[key]를 다시 찾지 않고 키 · 값을 한 번에 받기 위해
for key, values in d.items():  # (키, 값) 쌍 → key, values 로 풀기
    print(f"key '{key}' has values {values} -> total : {len(values)}")  # 키 · 값 · 값 개수


# get()으로 없는 키를 안전하게 조회
# 왜 · 없는 키를 찾아도 KeyError로 멈추지 않게 하기 위해
print("없는 키:", d.get("four", "없음"))  # 키 없으면 기본값 (d["four"]는 KeyError)
````

**결과** · 예측 정답은 키만

- `-> total : 10` · `-> total : 12` · `-> total : 14` — total = 값 리스트 길이 (one · two · three)

4. 길이 다른 zip

**먼저 예측** · 길이가 다른 두 리스트를 `zip`하면? — 답은 코드 실행 결과 아래에

````python
a = ["one", "two", "three", "four"]  # 이름 (키가 될 것)
b = [30, 20, 15, 75]                 # 수량
c = [5.2, 7.4, 3.6, 4.2]             # 단가


# 방법 1: 같은 위치의 이름·수량·단가를 묶어 {이름: 수량 × 단가} 생성
# 왜 · 세 리스트를 같은 위치끼리 묶어 한 줄로 계산하기 위해
r1 = {k: x * y for k, x, y in zip(a, b, c)}  # {키: 값 for 변수들 in zip(...)}


# 방법 2: 값 리스트를 먼저 만들고 dict(zip(키, 값))
# 왜 · 키 리스트와 값 리스트를 따로 만든 뒤 합치는 방식도 비교하기 위해
r2 = dict(zip(a, [x * y for x, y in zip(b, c)]))  # 수량×단가 리스트 → 이름과 짝짓기 → dict


# 두 방법의 결과가 같은지 확인
# 왜 · 두 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert r1 == r2  # 두 방법 결과 검증
print(r1)        # {이름: 수량 × 단가}


# enumerate로 번호를 붙여 딕셔너리 만들기
# 왜 · 번호 → 이름 딕셔너리를 반복문 없이 만들기 위해
colors = ["Red", "Green", "Black", "Blue", "Orange", "Purple"]  # 번호를 붙일 6개
print(dict(enumerate(colors)))                                  # (0, Red), (1, Green), ... → dict
print(dict(enumerate(colors, start=100)))                       # 번호를 100부터 시작


# 길이가 다른 리스트를 zip하면
# 왜 · 길이가 다르면 짧은 쪽에 맞춰 조용히 잘린다는 함정을 보이기 위해
print("zip 잘림:", list(zip([1, 2, 3], ["x", "y"])))  # 짧은 쪽(2개)에 맞춰 잘림
````

**결과** · 예측 정답은 짧은 쪽 길이에 맞춰 잘림

- `{'one': 156.0, 'two': 148.0, 'three': 54.0, 'four': 315.0}` — 30×5.2 · 20×7.4 · 15×3.6 · 75×4.2
- `{0: 'Red'` · `{100: 'Red'` — start로 시작 번호 지정
- `zip 잘림: [(1, 'x'), (2, 'y')]` — 3은 짝이 없어 사라짐

````python
# 원본 함수: 기본값 인자 뒤에 필수 인자 → 에러
# 왜 · 원본 함수의 SyntaxError를 셀을 멈추지 않고 확인하기 위해
broken = 'def greet(msg="Good morning!", name):\n    return msg'  # 원본 함수 (문자열로 보관)
try:
    exec(broken)                                                  # 정의만 해 봐도 에러
except SyntaxError as e:                                          # 정의 단계 문법 에러
    print("SyntaxError:", e.msg)                                  # 에러 이유만 출력


# 수정: 필수 인자를 앞, 기본값 인자를 뒤로
# 왜 · 기본값 없는 인자가 앞에 와야 한다는 규칙을 지켜 에러를 없애기 위해
def greet(name, msg="Good morning!"):  # def 함수명(필수, 기본값=값):
    return "Hi! " + name + ", " + msg  # 인사 문장을 돌려줌


# 세 가지 방식으로 호출
# 왜 · 기본값 사용 · 위치 인자 · 키워드 인자 호출의 차이를 한 번에 보기 위해
print(greet("Kim"))                     # msg 생략 → 기본값 사용
print(greet("Park", "How do you do?"))  # 위치 순서대로 전달
print(greet(msg="Hello", name="Lee"))   # 이름=값 → 순서 무관
````

- `Hi! Park, How do you do?` — 위치 순서대로 전달

### 4-2. `B` 지역 · 전역 변수 · `global` · `return` <sub>원본 20</sub>

> 🎯 **함수 안에서 바꾼 변수는 함수 밖에도 남는가?**

💡 **쉽게 말하면**

- 함수는 **회의실**.
- 안에서 쓴 메모(지역 변수)는 나올 때 버려지고, 밖의 게시판(전역 변수)을 고치려면 `global` 허락이 필요.

<img src="images/py1_scope.png" alt="함수 안 변수는 사라지고, global로 고친 a만 남음" width="763">

**실행 흐름**

1. 밖: a, b, x, y = 8, 13, 33, 44
2. 안: global a 변경 · x, y 교환 · b 생성
3. 안 · 밖 비교

**먼저 예측** · 함수 실행 뒤 밖의 `a, b, x, y`는? — 답은 코드 실행 결과 아래에

````python
# 함수 안에서 변수를 바꾸는 함수 정의
# 왜 · global을 쓴 변수와 안 쓴 변수가 밖에서 어떻게 달라지는지 비교하기 위해
def test(x, y):                    # 받은 x, y도 함수 안 변수
    global a                       # 밖의 a를 직접 수정하겠다는 선언
    a = 49                         # global이라 밖의 a가 바뀜
    x, y = y, x                    # 함수 안 x, y만 교환
    b = 7                          # 함수 안 b (밖의 b와 별개)
    c = 135                        # 함수가 끝나면 사라짐
    print("Step1 : ", a, b, x, y)  # 함수 안에서 본 값


# 밖의 변수 설정 → 함수 실행 → 밖의 값 확인
# 왜 · 함수를 부르기 전후로 밖의 값을 비교해 범위 규칙을 확인하기 위해
a, b, x, y = 8, 13, 33, 44     # 함수 밖 변수 4개
test(23, 7)                    # x=23, y=7로 호출
print("Step2 : ", a, b, x, y)  # global 쓴 a만 바뀜
````

**결과** · 예측 정답은 `49 13 33 44` — `global`로 선언한 `a`만 바뀜

- `Step1 :  49 7 7 23` — 함수 안: global a · 안의 b · 교환된 x, y
- `Step2 :  49 13 33 44` — global로 바꾼 a만 밖에 남음

### 5-1. `A` 문장 나누기 · `split()` <sub>원본 21</sub>

- 공백(또는 지정한 글자)을 기준으로 잘라 조각들을 리스트로 돌려줌.

<img src="images/py1_split.png" alt="split()과 split(' ')의 차이" width="763">

**실행 흐름**

1. 단어 개수 함수
2. 10 검증
3. 구분자 지정 · 공백 두 칸 비교

**먼저 예측** · 공백이 두 칸일 때 `split()`과 `split(" ")`의 차이는? — 답은 코드 실행 결과 아래에

````python
in_str = "Suppose we have few words that are separated by spaces."  # 공백으로 구분된 문장


# 공백으로 나눈 단어 개수를 세는 함수
# 왜 · 단어 수 세기를 함수로 만들어 다른 문장에도 다시 쓰기 위해
def count_words(text):        # 단어 개수 세는 함수
    return len(text.split())  # 문자열 → split() → 단어 리스트 → len()


# 단어 목록 · 개수 확인
# 왜 · 함수 결과가 실제 단어 목록 길이와 같은지 확인하기 위해
print(in_str.split())                     # 나눈 단어 목록
print("단어 개수:", count_words(in_str))  # 함수로 센 개수
assert count_words(in_str) == 10          # 정답 10개 검증


# 구분자를 지정해서 나누기
# 왜 · 구분자를 주면 빈 문자열도 생긴다는 차이를 보이기 위해
print("a,b,,c".split(","))  # 쉼표 기준 → 연속 쉼표 사이는 빈 문자열


# 인자 없는 split() vs 공백 한 칸 split(" ")
# 왜 · 연속 공백을 다루는 방식이 다르다는 것을 나란히 비교하기 위해
print("  a  b ".split(), "  a  b ".split(" "))  # 연속 공백 처리 차이
````

**결과** · 예측 정답은 `split()`은 빈 문자열 없이 단어만, `split(" ")`은 빈 문자열이 섞임

- `단어 개수: 10` — 마침표는 'spaces.'에 붙어 있음
- `['a', 'b', '', 'c']` — 연속 구분자 사이의 빈 문자열도 원소
- `['a', 'b'] ['', '', 'a', '', 'b', '']` — 인자 없는 split()이 공백을 깔끔히 처리

### 5-2. `A` 출력 형식 지정 · `f"{값:형식}"` <sub>원본 29</sub>

**먼저 예측** · `%x`는 숫자를 어떻게 바꾸나? — 답은 코드 실행 결과 아래에

````python
x, y = 10, 20         # 합계를 낼 두 수
serialno = 308276567  # 16진수로 바꿔 볼 숫자
n = "Kim"             # 끼워 넣을 이름


# 방법 1: % 연산자 (옛 방식)
# 왜 · 옛 코드에서 자주 보이는 % 형식을 읽을 수 있게 하기 위해
ex1 = "n = %s, s = %x, sum=%d" % (n, serialno, x + y)  # %s 문자 · %x 16진수 · %d 정수


# 방법 2: format()
# {이름} ← 이름=값
ex2 = "n = {n}, s = {serialno}, sum={sum}".format(n=n, serialno=serialno, sum=x + y)


# 방법 3: f-string (권장)
# 왜 · 값과 식을 문장 안에 바로 써서 가장 읽기 쉽게 만들기 위해
ex3 = f"n = {n}, s = {serialno}, sum={x + y}"  # f"{변수 또는 식}"


# 세 결과 비교: 다른 것 찾기
# 왜 · 원본 문제의 '다른 출력 하나'를 코드로 찾기 위해
print(ex1, ex2, ex3, sep="\n")                     # 세 결과를 줄마다
print("출력1만 다름:", ex1 != ex2 and ex2 == ex3)  # %x만 16진수라 다름


# 자주 쓰는 형식 지정
# 왜 · 학습 로그에 자주 쓰는 퍼센트 · 쉼표 · 소수 자리를 한 줄에 익히기 위해
# 퍼센트 · 천 단위 쉼표 · 소수 2자리 · 0 채워 3자리
print(f"{0.876:.1%} | {1234567:,} | {3.14159:.2f} | {7:03d}")
````

**결과** · 예측 정답은 16진수 — `308276567` → `125fed57`

- `n = Kim, s = 125fed57, sum=30` — %x = 16진수 → 출력1만 다름 (정답)
- `87.6% | 1,234,567 | 3.14 | 007` — 퍼센트 · 천 단위 쉼표 · 소수 2자리 · 0 채우기

### 6-1. `B` 파일 쓰기 · `with open()` <sub>원본 26</sub>

> 🎯 **여러 파일을 반복문으로 한 번에 저장하려면?**

💡 **쉽게 말하면**

- `with open(...)`은 **자동문**.
- 블록에 들어가면 파일이 열리고, 나오면 저절로 닫힘.

**실행 흐름**

1. 이름 · 내용 zip
2. 파일마다 with open("w")

**먼저 예측** · `"w"` 모드로 같은 파일을 다시 열면 기존 내용은? — 답은 코드 실행 결과 아래에

````python
# 왜 · 원본 문제의 파일 이름과 내용을 그대로 써서 결과를 비교하기 위해
names = ["A", "B", "C", "D", "F", "G"]  # 파일 이름 (E 없음 · 원본 그대로)
contents = ["Python", "JavaScript", "PHP", "Rust", "Solidity", "Assembly"]  # 파일마다 넣을 내용
folder = BASE / "26-1"                  # 저장할 폴더


# 이름 · 내용을 짝지어 파일마다 쓰기
# 왜 · 파일 이름과 내용을 zip으로 짝지어 반복 한 번에 모두 쓰기 위해
for name, text in zip(names, contents):  # (A, Python), (B, JavaScript), ...
    # with open(경로, 모드) as 파일: → 끝나면 자동 닫힘
    with open(folder / f"{name}.txt", "w", encoding="utf-8") as f:
        f.write(text)                    # 내용 쓰기


# 저장된 파일 목록 · 내용 확인
# 왜 · 파일이 제대로 써졌는지 다시 읽어 확인하기 위해
print(sorted(os.listdir(folder)))                                # 만든 파일 목록
print("A.txt:", (folder / "A.txt").read_text(encoding="utf-8"))  # 내용이 맞게 들어갔는지
````

### 6-2. `A` 여러 파일 읽기 · `glob()` <sub>원본 27</sub>

> 🎯 **폴더 안 파일을 빠짐없이, 항상 같은 순서로 읽으려면?**

💡 **쉽게 말하면**

- `glob("*.txt")`는 **이름 패턴으로 찾는 검색창**.
- 결과 순서는 컴퓨터마다 다를 수 있어 `sorted()`로 줄 세움.

<img src="images/py1_glob.png" alt="glob으로 찾고 sorted로 순서 고정" width="763">

**실행 흐름**

1. *.txt 찾기 · 정렬
2. 하나씩 읽기
3. pathlib 방식과 비교

**먼저 예측** · `glob` 결과 순서는 항상 같을까? — 답은 코드 실행 결과 아래에

````python
# 방법 1: glob으로 찾고 정렬한 뒤 하나씩 읽기
# 왜 · glob 결과 순서는 보장되지 않아 sorted()로 고정한 뒤 읽기 위해
paths = sorted(glob.glob(str(BASE / "27-1" / "*.txt")))  # *.txt 경로 목록 → 이름순 정렬
result = []                                              # 읽은 단어를 모을 곳
for p in paths:                                          # 정렬된 순서대로
    with open(p, encoding="utf-8") as f:                 # 모드 생략 → "r" 읽기
        result.append(f.read().strip())                  # 전체 읽기 → 앞뒤 공백 제거 → 추가


# 방법 2: pathlib으로 한 줄에 읽기
# 왜 · 같은 일을 pathlib만으로 더 짧게 할 수 있음을 보이기 위해
result2 = [p.read_text(encoding="utf-8").strip()            # 읽기 → 공백 제거
           for p in sorted((BASE / "27-1").glob("*.txt"))]  # 경로 찾기 → 정렬


# 두 방법의 결과가 같은지 확인
# 왜 · 두 방법이 같은 결과를 내는지 눈이 아니라 코드로 검증하기 위해
assert result == result2  # 두 방법 결과 검증


# 개수와 일부만 출력
# 왜 · 26개를 모두 읽었는지 개수로 확인하고 출력은 짧게 줄이기 위해
print("파일 개수:", len(result))                       # 26개인지
print("앞 5개:", result[:5], "/ 마지막:", result[-1])  # 긴 출력은 일부만
````

**결과** · 예측 정답은 보장되지 않음 — `sorted()`로 고정

- `파일 개수: 26` — A.txt ~ Z.txt
- `앞 5개: ['Python', 'JavaScript', 'PHP', 'Rust', 'elite']` · `마지막: zigzag` — 파일 이름순 = 원본 정답 순서

### 6-3. `B` 확장자별 분류 · `endswith()` <sub>원본 42</sub>

> 🎯 **폴더 안 파일을 확장자별로 골라내려면?**

💡 **쉽게 말하면**

- 파일 이름 **끝부분(확장자)을 보고 바구니에 나눠 담기**.
- `endswith`로 끝을 확인.

**실행 흐름**

1. 목록
2. endswith로 png · py 분류
3. glob 검증
4. 전체 확장자 개수

**먼저 예측** · 정렬하면 `file10`과 `file6` 중 앞은? — 답은 코드 실행 결과 아래에

````python
folder = BASE / "42-1"              # 확장자가 섞인 폴더
files = sorted(os.listdir(folder))  # 폴더 안 파일 이름 → 정렬


# 방법 1: endswith()로 확장자 골라내기
# 왜 · 파일 이름 끝 글자로 확장자를 골라내는 가장 쉬운 방법을 쓰기 위해
png = [f for f in files if f.endswith(".png")]         # [값 for 변수 in 목록 if 조건]
py = [f for f in files if f.endswith(".py")]           # .py만 골라내기
print("PNG file info : ", png, " Count : ", len(png))  # 원본 문제의 출력 형식
print("PY file info : ", py, " Count : ", len(py))     # 원본 문제의 출력 형식


# 방법 2: glob 패턴으로 같은 결과인지 확인
# 왜 · 패턴으로 찾은 결과도 같은지 교차 확인하기 위해
assert png == sorted(glob.glob("*.png", root_dir=folder))  # root_dir 폴더 안에서 *.png 찾기


# 방법 3: suffix로 확장자별 개수 세기
# 왜 · 확장자마다 몇 개인지 한 번에 세어 빠진 파일이 없는지 보기 위해
print("확장자별 개수:", dict(Counter(Path(f).suffix for f in files)))  # 파일 → 확장자 → 개수 세기
````

**결과** · 예측 정답은 `file10` — 문자열은 한 글자씩 비교 ('1' < '6')

- `PNG file info : ['file10.png'` · `Count : 4` — 원본 정답과 같음 (PY는 Count : 6)
- `['file10.png', 'file11.png', 'file19.png', 'file6.png']` — 숫자 크기가 아니라 글자 순서
- `확장자별 개수: {'.jpg': 3, '.png': 4, '.py': 6, '.csv': 3, '.txt': 3}` — png 4 · py 6 · 나머지 각 3

````python
d = {"group1": [{"name": "Park", "age": "32", "sex": "Male"},  # 그룹 · 사람 정보가 중첩된 dict
                {"name": "Cho", "age": "44", "sex": "Female"},
                {"name": "Kang", "age": "39", "sex": "Female", "married": "No"}],
     "group2": [{"name": "Kim", "age": "23", "sex": "Male", "married": "Yes"},
                {"name": "Lee", "age": "37", "sex": "Male", "married": "No"}],
     "type": {"a": "employee", "b": "officer", "c": "director",
              "d": "manager", "e": "service provider"}}


# 32-1: dict → JSON 문자열
# 왜 · 딕셔너리를 파일 · 통신에 쓸 수 있는 문자열로 바꾸기 위해
text = json.dumps(d, indent=4)           # dumps(dict) → str, indent = 들여쓰기 칸 수
print(type(d).__name__, "→", type(text).__name__, "/ 앞 4줄:")  # dict → str 확인
print("\n".join(text.splitlines()[:4]))  # 줄 단위로 나눔 → 앞 4줄 → 다시 이어 붙임


# dict → JSON 파일 저장 후 다시 읽어 비교
# 왜 · 저장했다 다시 읽어도 내용이 그대로인지 확인하기 위해
path = BASE / "people.json"                        # 저장할 파일 경로
with open(path, "w", encoding="utf-8") as f:       # 쓰기 모드로 열기
    json.dump(d, f, indent=4, ensure_ascii=False)  # dump(dict, 파일) · 한글 그대로 저장
with open(path, encoding="utf-8") as f:            # 읽기 모드로 다시 열기
    assert json.load(f) == d                       # load(파일) → dict → 원본과 비교


# 33-1: JSON 문자열 → dict (원본 문제는 작은따옴표 사용 · 괄호 안 줄들은 이어 붙어 문자열 하나)
s = ("""{"group1": [{'name': 'Park', 'age': '32', 'sex': 'Male'},"""  # 원본 문제 문자열
     """{'name': 'Cho', 'age': '44', 'sex': 'Female'},"""
     """{'name': 'Kang', 'age': '39', 'sex': 'Female', 'married': 'No'}],"""
     """"group2": [{'name': 'Kim', 'age': '23', 'sex': 'Male', 'married': 'Yes'},"""
     """{'name': 'Lee', 'age': '37', 'sex': 'Male', 'married': 'No'}],"""
     """"type": {"a": "employee", "b": "officer", "c": "director","""
     """ "d": "manager", "e": "service provider"}}""")


# 작은따옴표 그대로 읽으면 에러
# 왜 · JSON은 큰따옴표만 허용한다는 규칙 때문에 나는 에러를 확인하기 위해
try:
    json.loads(s)                     # loads(str) → dict
except json.JSONDecodeError as e:     # JSON 규칙 위반
    print("JSONDecodeError:", e.msg)  # 에러 이유만


# 큰따옴표로 바꾼 뒤 읽고 원본과 비교
# 왜 · 따옴표를 고치면 원본 딕셔너리와 같아진다는 것을 검증하기 위해
loaded = json.loads(s.replace("'", '"'))  # ' → " 치환 → JSON 규칙 충족 → dict
assert loaded == d                        # 32-1의 d와 같은지 검증
print("복원 성공:", loaded["group1"][0])  # 복원된 첫 사람
````

**결과** · 예측 정답은 `JSONDecodeError` — JSON은 큰따옴표만 허용

- `dict → str / 앞 4줄:` — dumps 결과는 문자열 (저장 · 전송용)
- assert 통과 · dump → load 해도 원본과 같음
- `JSONDecodeError: Expecting property name enclosed in double quotes` — 작은따옴표는 JSON 규칙 위반 (원본 문제의 함정)
- `복원 성공: {'name': 'Park', 'age': '32', 'sex': 'Male'}` — 32-1의 d와 완전히 같음

**먼저 예측** · `cat` 5장 · `dog` 4장 · `bird` 3장일 때 클래스 번호 `{이름: 번호}`는? — 답은 코드 실행 결과 아래에

**실행 흐름**

1. 클래스 폴더에 가짜 이미지 만들기
2. 이미지 경로만 정렬해서 모으기
3. 폴더 이름으로 라벨 만들기
4. 클래스 이름 → 번호 고정

````python
import random  # 무작위 섞기 (seed로 결과 고정)


# 1단계: 가짜 이미지 폴더 만들기 (클래스 = 폴더 이름)
# 왜 · 폴더 이름이 곧 라벨인 실제 이미지 데이터셋 구조를 재현하기 위해
counts = {"cat": 5, "dog": 4, "bird": 3}       # 클래스별 가짜 이미지 수
for cls, n in counts.items():                  # 3-3 items()
    folder = BASE / "pets" / cls               # 클래스 이름 = 폴더 이름
    folder.mkdir(parents=True, exist_ok=True)  # 폴더 만들기 (있으면 통과)
    for i in range(n):                         # 2-3 range
        (folder / f"img_{i:03d}.jpg").touch()  # 5-2 :03d → img_000.jpg
(BASE / "pets" / "cat" / "memo.txt").touch()   # 이미지가 아닌 파일 섞기


# 2단계: 이미지 경로만 모으기 (확장자 거르기 + 정렬)
# 왜 · 이미지가 아닌 파일을 빼고, 실행마다 같은 순서로 모으기 위해
# 6-2 glob · 6-3 suffix
paths = sorted(p for p in (BASE / "pets").glob("*/*") if p.suffix == ".jpg")
print("이미지 수:", len(paths), "/ 첫 경로:", paths[0].as_posix())  # 모은 개수 · 첫 경로


# 3단계: 라벨 = 이미지가 들어 있는 폴더 이름
# 왜 · 라벨 파일 없이 폴더 이름만으로 정답을 만들기 위해
labels = [p.parent.name for p in paths]  # 2-4 컴프리헨션 · .parent.name → 폴더 이름
print("첫 라벨:", labels[0])             # 첫 이미지의 라벨


# 4단계: 클래스 이름 → 번호 (정렬해서 번호 고정)
# 왜 · 실행할 때마다 클래스 번호가 바뀌지 않게 하기 위해
classes = sorted(set(labels))                               # 2-5 중복 제거 + 정렬
class_to_idx = {name: i for i, name in enumerate(classes)}  # 3-4 enumerate → {이름: 번호}
targets = [class_to_idx[label] for label in labels]         # 라벨 이름 → 번호 목록
print("클래스 번호:", class_to_idx)                         # 이름 → 번호 표
print("클래스별 개수:", dict(Counter(labels)))              # 불균형 확인
````

**실행 흐름**

1. `split_dataset()` 정의 (seed로 섞고 나누기)
2. 학습 · 검증으로 나누기
3. 개수 합 · 겹침 없음 검증
4. 목록을 JSON으로 저장하고 다시 읽기

````python
# 5단계: 학습 · 검증으로 나누는 함수 (같은 seed · 같은 파이썬 버전 → 같은 분할)
# 왜 · 모델을 바꿔 비교할 때 데이터 분할은 같게 유지하기 위해
def split_dataset(paths, targets, val_ratio=0.25, seed=42):  # 4-1 필수 → 기본값 인자
    idx = list(range(len(paths)))                            # 경로 대신 번호를 섞음
    random.Random(seed).shuffle(idx)                         # 번호 섞기 (원본 목록은 그대로)
    n_val = int(len(idx) * val_ratio)                        # 검증용 개수 (12 × 0.25 = 3)
    val_idx, train_idx = idx[:n_val], idx[n_val:]            # 2-2 슬라이싱으로 나누기
    train = [(paths[i].as_posix(), targets[i]) for i in sorted(train_idx)]  # (경로, 번호) 쌍
    val = [(paths[i].as_posix(), targets[i]) for i in sorted(val_idx)]  # 검증 목록도 같은 형태
    return train, val                                        # 4-2 결과는 return으로


train, val = split_dataset(paths, targets)  # 기본값 0.25 · seed 42 사용


# 나눈 결과 검증: 개수 합 · 겹침 없음
# 왜 · 학습 · 검증에 빠진 샘플이나 겹친 샘플이 없는지 확인하기 위해
assert len(train) + len(val) == len(paths)                    # 빠진 이미지가 없는지
assert not set(p for p, _ in train) & set(p for p, _ in val)  # 교집합(&)이 비어 있어야 함
# 5-2 :>2 → 2칸 오른쪽 정렬
print(f"train {len(train):>2}장 | val {len(val):>2}장 | 클래스 {len(classes)}개")
print("val 예시:", val[:2])                                   # 검증 목록 앞 2개


# 6단계: 데이터 목록을 JSON으로 저장 후 다시 읽기
# 왜 · 나눈 목록을 파일로 남겨 다음 실행에서도 같은 분할을 쓰기 위해
# 저장할 내용 한 덩어리
meta = {"classes": classes, "class_to_idx": class_to_idx, "train": train, "val": val}
with open(BASE / "dataset.json", "w", encoding="utf-8") as f:  # 쓰기 모드
    json.dump(meta, f, ensure_ascii=False, indent=2)           # 7-1 dump → 파일
with open(BASE / "dataset.json", encoding="utf-8") as f:       # 읽기 모드
    loaded = json.load(f)                                      # 7-1 load → dict
# 튜플 → 리스트로 바뀜
print("복원 후 train 첫 항목:", loaded["train"][0], "→", type(loaded["train"][0]).__name__)
````

**결과** · 예측 정답은 `{'bird': 0, 'cat': 1, 'dog': 2}` — 폴더를 만든 순서가 아니라 **이름 정렬 순서**로 번호가 정해짐

- `이미지 수: 12` · `첫 경로: 실습파일/pets/bird/img_000.jpg` — `memo.txt`는 `.jpg`가 아니라 제외, 정렬로 bird가 먼저
- `클래스 번호: {'bird': 0, 'cat': 1, 'dog': 2}` · `클래스별 개수: {'bird': 3, 'cat': 5, 'dog': 4}` — 이름 정렬 순서로 번호 고정 · cat이 가장 많은 불균형
- `train  9장 | val  3장 | 클래스 3개` — 12 × 0.25 = 3장을 검증용으로, 두 목록은 겹치지 않음 (assert 통과)
- `복원 후 train 첫 항목: ['실습파일/pets/bird/img_000.jpg', 0] → list` — 저장 전 튜플이 JSON을 거치면 리스트로 바뀜

## 면접 30초 답변

**Q1. 리스트를 `y = x`로 복사하면 어떤 문제가 생기나요?**
> - `y = x`는 새 리스트가 아니라 같은 객체에 이름을 하나 더 붙이는 것이라, `y`를 수정하면 `x`도 바뀝니다.
> - 독립 복사본은 `x.copy()`, 중첩 구조는 `copy.deepcopy()`를 씁니다.
> - pandas의 `df.copy()`, PyTorch의 `tensor.clone()`이 있는 이유도 같습니다.

**Q2. 이미지 데이터셋을 만들 때 파일 목록을 왜 정렬하나요?**
> `os.listdir`나 `glob`의 반환 순서는 운영체제와 파일 시스템에 따라 보장되지 않습니다. 정렬하지 않으면 실행할 때마다 순서가 달라져 이미지와 라벨의 짝이 어긋나고 재현성이 깨지므로 `sorted()`로 고정합니다.

**Q3. `set`으로 클래스 목록을 만들 때 주의할 점은?**
> `set`은 순서를 보장하지 않아 실행마다 클래스 번호가 달라질 수 있습니다. 그래서 `sorted(set(labels))`로 순서를 고정한 뒤 `enumerate`로 번호를 매겨, 학습과 추론에서 같은 번호 체계를 쓰게 합니다.

**Q4. 학습 · 검증 데이터를 나눌 때 seed를 고정하는 이유는?**
> - seed를 고정하지 않으면 실행할 때마다 다른 데이터가 검증용으로 뽑혀, 모델을 바꿔서 성능이 오른 것인지 데이터 분할이 달라서 오른 것인지 구분할 수 없습니다.
> - 그래서 `random.Random(42)`처럼 seed를 고정해 같은 분할에서 비교합니다.

**실행 흐름**

1. 인자 순서를 고친 `greet()`
2. `global` 범위 확인
3. 블록마다 결과 출력

````python
# ---------- 4. 함수 ----------
# 4-1 필수 인자 → 기본값 인자 순서로 정의
# 왜 · 기본값 없는 인자가 앞에 와야 SyntaxError가 나지 않기 때문에
def greet(name, msg="Good morning!"):  # 필수 → 기본값 순서
    return "Hi! " + name + ", " + msg  # 인사 문장을 돌려줌


print("[4-1]", greet("Kim"), "|", greet("Park", "How do you do?"))  # 기본값 · 위치 인자


# 4-2 global로 바꾼 a만 함수 밖에 남음
# 왜 · 함수 안 대입이 밖에 남는 경우를 global로만 한정하기 위해
def test(x, y):                         # 받은 x, y도 함수 안 변수
    global a                            # 밖의 a를 고치겠다는 선언
    a = 49                              # global이라 밖의 a가 바뀜
    x, y = y, x                         # 함수 안 x, y만 교환
    b = 7                               # 함수 안 b (밖과 별개)
    print("[4-2] Step1 :", a, b, x, y)  # 함수 안에서 본 값


a, b, x, y = 8, 13, 33, 44          # 함수 밖 변수 4개
test(23, 7)                         # x=23, y=7로 호출
print("[4-2] Step2 :", a, b, x, y)  # global 쓴 a만 바뀜
````

**실행 흐름**

1. 가짜 이미지 폴더 · 경로 · 라벨 만들기
2. 클래스 번호 고정
3. seed로 섞어 학습 · 검증 나누기
4. JSON 저장 · 복원

````python
# ---------- 8. 종합 실습: 폴더 속 이미지 → 학습용 목록 ----------
# 8-1 폴더 이름 = 클래스, 이미지 경로 · 라벨 · 번호 만들기
# 왜 · 폴더 구조만으로 경로 · 라벨 · 번호를 빠짐없이 만들기 위해
for cls, n in {"cat": 5, "dog": 4, "bird": 3}.items():
    (BASE / "pets" / cls).mkdir(parents=True, exist_ok=True)            # 클래스 이름 = 폴더
    for i in range(n):                                                  # 클래스마다 n장
        (BASE / "pets" / cls / f"img_{i:03d}.jpg").touch()              # 빈 가짜 이미지 파일
paths = sorted(p for p in (BASE / "pets").glob("*/*") if p.suffix == ".jpg")  # .jpg만 모아 정렬
labels = [p.parent.name for p in paths]                                 # 폴더 이름 = 라벨
class_to_idx = {name: i for i, name in enumerate(sorted(set(labels)))}  # 정렬 순서로 번호 고정
targets = [class_to_idx[label] for label in labels]                     # 라벨 → 번호
print("[8-1]", len(paths), class_to_idx)                                # 이미지 수 · 번호 표

# 8-2 seed 고정으로 섞고 슬라이싱으로 학습 · 검증 나누기
# 왜 · 모델을 바꿔 비교할 때 데이터 분할은 같게 유지하기 위해
idx = list(range(len(paths)))                   # 경로 대신 번호를 섞음
random.Random(42).shuffle(idx)                  # seed 42 → 같은 버전에서 같은 순서
n_val = int(len(idx) * 0.25)                    # 검증용 25%
train_idx, val_idx = idx[n_val:], idx[:n_val]   # 슬라이싱으로 나누기
no_overlap = not set(train_idx) & set(val_idx)  # 겹치는 번호가 없으면 True
# 나눈 개수 · 겹침 없음
print("[8-2]", f"train {len(train_idx)}장 | val {len(val_idx)}장", no_overlap)
````
