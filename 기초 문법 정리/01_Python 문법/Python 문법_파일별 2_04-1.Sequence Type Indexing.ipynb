### 1-1. `A` 세 가지 방법으로 꺼내기 · `x[i]` · `x[-i]` · `index()` <sub>원본 문제</sub>

> 🎯 **6개 중 `Banana`를 어떤 번호로 꺼내는가?**

💡 **쉽게 말하면**

- 리스트는 **번호표 붙은 사물함**.
- 번호는 **0번부터**, 뒤에서 셀 때는 **-1부터**.
- `index(값)`은 값을 주면 사물함 번호를 찾아 줌.

**실행 흐름**

1. 리스트 만들기
2. 앞 번호 x[4]
3. 뒤 번호 x[-2]
4. index()로 찾은 번호

**먼저 예측** · 6개 중 `Banana`의 번호는? — 답은 코드 실행 결과 아래에

````python
x = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]  # 번호 0 ~ 5인 6개


# 방법 1 · 2 · 3: 앞 번호 · 뒤 번호 · index()로 찾은 번호
print("앞 번호:", x[4])                  # 앞에서 0, 1, 2, 3, 4번째
print("뒤 번호:", x[-2])                 # 뒤에서 -1, -2번째
print("index():", x[x.index("Banana")])  # 값 → 번호 → 꺼내기
````

**결과** · 예측 정답은 `4` (뒤에서는 `-2`)

- `앞 번호: Banana` · `뒤 번호: Banana` — 0부터 세면 4번, 뒤에서 세면 -2번

### 1-2. `A` 길이 · 마지막 번호와 에러 · `len()` · `ValueError` · `IndexError` <sub>원본 힌트 · index 대소문자</sub>

> 🎯 **번호를 잘못 쓰면 어떤 에러가 나고, 둘은 어떻게 다른가?**

💡 **쉽게 말하면**

- 마지막 번호는 **길이 - 1**.
- 없는 값을 찾으면 ValueError, 없는 번호를 열면 IndexError.

**실행 흐름**

1. index() · len() 출력
2. 소문자로 찾기
3. 범위 밖 번호

**먼저 예측** · `x[6]`을 실행하면? — 답은 코드 실행 결과 아래에

````python
print("위치:", x.index("Banana"), "/ 길이:", len(x))  # 마지막 번호 = 길이 - 1


# 없는 값 · 범위 밖 번호
try:
    x.index("banana")        # 소문자 → 다른 값
except ValueError as e:      # 없으면 ValueError
    print("ValueError:", e)  # 에러 문장 출력
try:
    x[6]                     # 0 ~ 5 밖의 번호
except IndexError as e:      # 범위 밖이면 IndexError
    print("IndexError:", e)  # 에러 문장 출력
````

**결과** · 예측 정답은 `IndexError` — 마지막 번호는 5

- `위치: 4 / 길이: 6` — 마지막 번호 = 길이 - 1 = 5
- `ValueError: list.index(x): x not in list` — 대소문자가 다르면 없는 값
- `IndexError: list index out of range` — 범위 밖 번호

### 2-1. `B` 넣고 빼기 · `append()` · `insert()` · `pop()` · `remove()` <sub>원본 힌트 · 리스트 함수</sub>

> 🎯 **리스트의 끝 · 중간에 넣고, 끝 · 값으로 빼려면?**

💡 **쉽게 말하면**

- `append`는 끝에 하나, `extend`는 끝에 여러 개, `insert`는 원하는 자리에.
- `pop`은 번호로 꺼내며 지우고, `remove`는 값으로 찾아 지움.

**실행 흐름**

1. 복사본 만들기
2. append · extend · insert
3. pop · remove
4. count · 원본 확인

**먼저 예측** · `y.pop()`은 어느 값을 꺼낼까? — 답은 코드 실행 결과 아래에

````python
y = x.copy()                                               # 원본 x는 그대로 두고 복사본으로 실습
y.append("Mango")                                          # 끝에 하나 추가
y.extend(["Lime", "Plum"])                                 # 끝에 여러 개 이어 붙이기
y.insert(0, "Fig")                                         # 0번 자리에 끼워 넣기
print("넣은 뒤:", y)                                       # 10개
print("pop():", y.pop(), "/ pop(0):", y.pop(0))            # 끝 · 맨 앞 꺼내며 삭제
y.remove("Kiwi")                                           # 값으로 찾아 삭제 (처음 한 개)
print("뺀 뒤:", y, "/ 개수:", len(y))                      # 남은 목록
print("count:", y.count("Apple"), "/ 원본 길이:", len(x))  # 값 개수 · 원본은 그대로
````

**결과** · 예측 정답은 `Plum` — 마지막에 넣은 값

- `넣은 뒤: ['Fig', 'Orange', 'Cherry', 'Apple', 'Kiwi', 'Banana', 'Strawberry', 'Mango', 'Lime', 'Plum']` — insert는 0번, append · extend는 끝
- `pop(): Plum / pop(0): Fig` — pop()은 끝, pop(0)은 맨 앞을 꺼내며 지움
- `뺀 뒤: ['Orange', 'Cherry', 'Apple', 'Banana', 'Strawberry', 'Mango', 'Lime'] / 개수: 7` · `count: 1 / 원본 길이: 6` — remove는 값으로 삭제 · 복사본이라 x는 6개 그대로

### 2-2. `A` 정렬 · `sorted()` · `sort()` · `reverse()` <sub>원본 힌트 · sort와 sorted</sub>

> 🎯 **`sorted()`와 `sort()`는 무엇이 다른가?**

💡 **쉽게 말하면**

- `sorted()`는 **정렬한 새 줄**을 하나 더 세워 주고, 원래 줄은 그대로.
- `sort()`는 **원래 줄 자체**를 다시 세우고, 돌려주는 값은 없음(None).

**실행 흐름**

1. sorted로 새 리스트
2. sort()로 원본 정렬
3. reverse=True · reverse()
4. key=len

**먼저 예측** · `result = z.sort()` 뒤 `result`는? — 답은 코드 실행 결과 아래에

````python
z = x.copy()                                    # 정렬 실습용 복사본
print("sorted:", sorted(z))                     # 정렬한 새 리스트
print("z 그대로:", z[:3])                       # sorted는 원본을 안 바꿈
result = z.sort()                               # z 자체를 정렬
print("sort() 반환값:", result, "/ z:", z[:3])  # None · z가 바뀜
z.sort(reverse=True)                            # 거꾸로 정렬
print("역순:", z[:3])                           # 큰 글자부터
z.reverse()                                     # 정렬 없이 순서만 뒤집기
print("reverse():", z[:3])                      # 역순을 다시 뒤집음
print("길이순:", sorted(x, key=len)[:3])        # key로 기준 바꾸기
````

**결과** · 예측 정답은 `None` — sort()는 원본을 바꾸고 아무것도 돌려주지 않음

- `sorted: ['Apple', 'Banana', 'Cherry', 'Kiwi', 'Orange', 'Strawberry']` · `z 그대로: ['Orange', 'Cherry', 'Apple']` — sorted는 새 리스트, 원본은 그대로
- `sort() 반환값: None / z: ['Apple', 'Banana', 'Cherry']` — sort()는 z 자체를 정렬하고 None 반환
- `역순: ['Strawberry', 'Orange', 'Kiwi']` · `reverse(): ['Apple', 'Banana', 'Cherry']` — reverse=True는 큰 것부터 · reverse()는 순서만 뒤집음
- `길이순: ['Kiwi', 'Apple', 'Orange']` — `key=len` → 글자 수 기준 (4 · 5 · 6)

**파이프라인 위치** · ⑤ 학습 → ⑥ 평가

> 🎯 **6 epoch 동안 기록한 검증 loss에서 마지막 값 · 가장 낮은 값과 그 epoch를 어떻게 꺼내는가?**

💡 **쉽게 말하면**

- loss 기록은 **epoch 순서대로 줄 선 리스트**.

**먼저 예측** · 가장 낮은 loss는 몇 번째 epoch일까? (사람이 세는 1부터) — 답은 코드 실행 결과 아래에

````python
# 1단계: epoch마다 기록한 검증 loss (가짜 값)
losses = [0.92, 0.61, 0.47, 0.43, 0.45, 0.44]  # epoch 1 ~ 6


# 2단계: 처음 · 마지막 기록
print("처음:", losses[0], "/ 마지막:", losses[-1])  # 앞 · 뒤 번호


# 3단계: 가장 낮은 loss와 그 epoch
best = min(losses)                                   # 최솟값
best_idx = losses.index(best)                        # 그 값의 번호 (0부터)
print("최저 loss:", best, "/ epoch:", best_idx + 1)  # 사람이 세는 epoch은 1부터


# 4단계: 최저 이후 지난 epoch 수 (조기 종료 판단)
after = len(losses) - 1 - best_idx     # 마지막 번호 - 최저 번호
print("최저 이후 지난 epoch:", after)  # 그동안 개선 없음


# 5단계: 좋은 순 3개 (원본 기록 순서는 그대로)
print("좋은 순 3개:", sorted(losses)[:3], "/ 원본 그대로:", losses[:2])  # sorted는 새 리스트
````

**결과** · 예측 정답은 4번째 epoch (번호로는 3)

- `처음: 0.92 / 마지막: 0.44` — 앞 번호 0 · 뒤 번호 -1
- `최저 loss: 0.43 / epoch: 4` — `index`는 0부터라 +1 해서 사람이 세는 epoch
- `최저 이후 지난 epoch: 2` — 마지막 번호 5 - 최저 번호 3 = 2 → 2번 연속 개선 없음
- `좋은 순 3개: [0.43, 0.44, 0.45] / 원본 그대로: [0.92, 0.61]` — sorted는 새 리스트라 기록 순서가 남음

**Q1. 파이썬 리스트에서 음수 인덱스는 어떻게 동작하나요?**
> 뒤에서부터 세는 번호로, `-1`이 마지막, `-2`가 끝에서 두 번째입니다. 길이를 몰라도 마지막 값을 꺼낼 수 있어 학습 기록의 마지막 loss처럼 끝값을 볼 때 자주 씁니다.

**Q2. `sort()`와 `sorted()`의 차이는 무엇인가요?**
> `sorted()`는 정렬한 새 리스트를 돌려주고 원본은 그대로 두며, `sort()`는 원본 리스트 자체를 정렬하고 `None`을 돌려줍니다. 그래서 `y = y.sort()`처럼 쓰면 `y`가 `None`이 됩니다.

**Q3. 가장 좋은 epoch는 어떻게 찾나요?**
> - 검증 loss를 epoch마다 리스트에 쌓아 두고 `min`으로 최저값을, `index`로 그 위치를 찾은 뒤 1을 더해 사람이 세는 epoch로 바꿉니다.
> - 최저 이후 지난 epoch 수로 조기 종료를 판단할 수도 있습니다.

### 전체 코드

1 ~ 3장 코드를 장별 셀로 모음. **위에서부터 이 셀들만 실행해도 전체가 동작**하며, 각 블록 머리 주석의 번호가 섹션.

````python
# ---------- 1-1 ----------
x = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]  # 번호 0 ~ 5인 6개


# 방법 1 · 2 · 3: 앞 번호 · 뒤 번호 · index()로 찾은 번호
print("앞 번호:", x[4])                  # 앞에서 0, 1, 2, 3, 4번째
print("뒤 번호:", x[-2])                 # 뒤에서 -1, -2번째
print("index():", x[x.index("Banana")])  # 값 → 번호 → 꺼내기


# ---------- 1-2 ----------
print("위치:", x.index("Banana"), "/ 길이:", len(x))  # 마지막 번호 = 길이 - 1


# 없는 값 · 범위 밖 번호
try:
    x.index("banana")        # 소문자 → 다른 값
except ValueError as e:      # 없으면 ValueError
    print("ValueError:", e)  # 에러 문장 출력
try:
    x[6]                     # 0 ~ 5 밖의 번호
except IndexError as e:      # 범위 밖이면 IndexError
    print("IndexError:", e)  # 에러 문장 출력
````

````python
# ---------- 2-1 ----------
y = x.copy()                                               # 원본 x는 그대로 두고 복사본으로 실습
y.append("Mango")                                          # 끝에 하나 추가
y.extend(["Lime", "Plum"])                                 # 끝에 여러 개 이어 붙이기
y.insert(0, "Fig")                                         # 0번 자리에 끼워 넣기
print("넣은 뒤:", y)                                       # 10개
print("pop():", y.pop(), "/ pop(0):", y.pop(0))            # 끝 · 맨 앞 꺼내며 삭제
y.remove("Kiwi")                                           # 값으로 찾아 삭제 (처음 한 개)
print("뺀 뒤:", y, "/ 개수:", len(y))                      # 남은 목록
print("count:", y.count("Apple"), "/ 원본 길이:", len(x))  # 값 개수 · 원본은 그대로


# ---------- 2-2 ----------
z = x.copy()                                    # 정렬 실습용 복사본
print("sorted:", sorted(z))                     # 정렬한 새 리스트
print("z 그대로:", z[:3])                       # sorted는 원본을 안 바꿈
result = z.sort()                               # z 자체를 정렬
print("sort() 반환값:", result, "/ z:", z[:3])  # None · z가 바뀜
z.sort(reverse=True)                            # 거꾸로 정렬
print("역순:", z[:3])                           # 큰 글자부터
z.reverse()                                     # 정렬 없이 순서만 뒤집기
print("reverse():", z[:3])                      # 역순을 다시 뒤집음
print("길이순:", sorted(x, key=len)[:3])        # key로 기준 바꾸기
````

````python
# ---------- 3. 종합 실습 ----------
# 1단계: epoch마다 기록한 검증 loss (가짜 값)
losses = [0.92, 0.61, 0.47, 0.43, 0.45, 0.44]  # epoch 1 ~ 6


# 2단계: 처음 · 마지막 기록
print("처음:", losses[0], "/ 마지막:", losses[-1])  # 앞 · 뒤 번호


# 3단계: 가장 낮은 loss와 그 epoch
best = min(losses)                                   # 최솟값
best_idx = losses.index(best)                        # 그 값의 번호 (0부터)
print("최저 loss:", best, "/ epoch:", best_idx + 1)  # 사람이 세는 epoch은 1부터


# 4단계: 최저 이후 지난 epoch 수 (조기 종료 판단)
after = len(losses) - 1 - best_idx     # 마지막 번호 - 최저 번호
print("최저 이후 지난 epoch:", after)  # 그동안 개선 없음


# 5단계: 좋은 순 3개 (원본 기록 순서는 그대로)
print("좋은 순 3개:", sorted(losses)[:3], "/ 원본 그대로:", losses[:2])  # sorted는 새 리스트
````
