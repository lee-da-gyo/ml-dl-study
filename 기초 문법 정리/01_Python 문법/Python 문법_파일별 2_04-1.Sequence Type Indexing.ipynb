## 1. 원리

## 2. 풀이

단계마다 **먼저 예측** → 코드 실행 → **결과** 순서로 확인. 셀은 위에서부터 차례로 실행.

### 2-1. 세 가지 방법으로 꺼내기

**실행 흐름**

1. 리스트 만들기
2. 앞 번호 x[4]
3. 뒤 번호 x[-2]
4. index()로 찾은 번호

**먼저 예측** · 6개 중 `Banana`의 번호는? — 답은 코드 실행 결과 아래에

````python
x = ["Orange", "Cherry", "Apple", "Kiwi", "Banana", "Strawberry"]  # 번호 0 ~ 5인 6개


# 방법 1 · 2 · 3: 앞 번호 · 뒤 번호 · index()로 찾은 번호
print("앞 번호:", x[4])                  # 앞에서 0, 1, 2, 3, 4번째
print("뒤 번호:", x[-2])                 # 뒤에서 -1, -2번째
print("index():", x[x.index("Banana")])  # 값 → 번호 → 꺼내기
````

**결과** · 예측 정답은 `4` (뒤에서는 `-2`)

- `앞 번호: Banana` · `뒤 번호: Banana` — 0부터 세면 4번, 뒤에서 세면 -2번
- `index(): Banana` — `index("Banana")`가 4를 찾아 꺼냄

### 2-2. 번호 · 길이 확인과 에러 두 가지

**실행 흐름**

1. index() · len() 출력
2. 소문자로 찾기
3. 범위 밖 번호

**먼저 예측** · `x[6]`을 실행하면? — 답은 코드 실행 결과 아래에

````python
print("위치:", x.index("Banana"), "/ 길이:", len(x))  # 마지막 번호 = 길이 - 1


# 없는 값 · 범위 밖 번호
try:
    x.index("banana")        # 소문자 → 다른 값
except ValueError as e:      # 없으면 ValueError
    print("ValueError:", e)  # 에러 문장 출력
try:
    x[6]                     # 0 ~ 5 밖의 번호
except IndexError as e:      # 범위 밖이면 IndexError
    print("IndexError:", e)  # 에러 문장 출력
````

**결과** · 예측 정답은 `IndexError` — 마지막 번호는 5

- `위치: 4 / 길이: 6` — 마지막 번호 = 길이 - 1 = 5
- `ValueError: list.index(x): x not in list` — 대소문자가 다르면 없는 값
- `IndexError: list index out of range` — 범위 밖 번호

### 2-3. 원본 힌트의 리스트 함수

**실행 흐름**

1. 복사본 만들기
2. count · insert · pop
3. sorted와 sort 비교

**먼저 예측** · `sorted(y)`와 `y.sort()`의 차이는? — 답은 코드 실행 결과 아래에

````python
# 힌트에 나온 리스트 함수 (원본 x는 건드리지 않게 복사본으로)
y = x.copy()                                # 원본 보호
print("count:", y.count("Kiwi"))            # 값이 몇 개인지
y.insert(0, "Mango")                        # 0번 자리에 넣기
print("pop:", y.pop())                      # 마지막 값 꺼내며 삭제
print("sorted:", sorted(y)[:3])             # 정렬한 새 리스트 (y는 그대로)
y.sort()                                    # y 자체를 정렬
print("y[0]:", y[0], "/ 원본 x[0]:", x[0])  # sort는 y만 바꿈
````

**결과** · 예측 정답은 `sorted`는 새 리스트를 돌려주고, `sort`는 y 자체를 바꿈 (반환값 None)

- `count: 1` · `pop: Strawberry` — Kiwi는 1개 · pop()은 마지막 값을 꺼내며 지움
- `sorted: ['Apple', 'Banana', 'Cherry']` — 정렬된 새 리스트의 앞 3개
- `y[0]: Apple / 원본 x[0]: Orange` — sort()는 y만 바꿈, 복사본이라 x는 그대로
