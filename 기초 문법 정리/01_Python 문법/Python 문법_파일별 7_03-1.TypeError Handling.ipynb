- 글자 + 숫자는 '이어 붙이기'인지 '더하기'인지 정할 수 없어 거절.

### 2-1. `B` 힌트의 TypeError 세 종류 · `y()` · `nums[idx]` · `len(y)` <sub>원본 힌트</sub>

> 🎯 **호출 · 인덱스 · 기타 TypeError는 각각 어떤 에러 문장이 나오는가?**

💡 **쉽게 말하면**

- **호출 불가** · 숫자에 `()`를 붙임.
- **인덱스** · 번호 자리에 글자.
- **기타** · 길이가 없는 값에 `len()`.

**실행 흐름**

1. 정수를 함수처럼 호출
2. 글자 번호로 리스트 꺼내기
3. 정수의 길이
4. 고친 방법

**먼저 예측** · `nums["1"]`의 에러 문장에는 무엇이 와야 한다고 나올까? — 답은 코드 실행 결과 아래에

````python
nums = [10, 20, 30]                      # 리스트
idx = "1"                                # 번호가 문자열로 들어온 상황
cases = [("호출", lambda: y()),          # 정수를 함수처럼 부름
         ("인덱스", lambda: nums[idx]),  # 번호 자리에 문자열
         ("기타", lambda: len(y))]       # 길이가 없는 정수
for name, bad in cases:                  # 세 경우를 차례로
    try:
        bad()                            # 각 잘못된 코드를 실행
    except TypeError as e:               # 셋 다 TypeError
        print(f"{name}: {e}")            # 종류별 에러 문장
print("고친 인덱스:", nums[int(idx)])    # 번호를 정수로 바꾸면 동작
````

**결과** · 예측 정답은 integers or slices (정수나 슬라이스)

- `호출: 'int' object is not callable` — 정수는 함수가 아니라 `()`로 부를 수 없음
- `인덱스: list indices must be integers or slices, not str` — 리스트 번호는 정수만
- `기타: object of type 'int' has no len()` — 정수에는 길이가 없음
- `고친 인덱스: 20` — `int("1")` → 1번 값

### 2-2. `A` `try` · `except`로 처리 · `except (TypeError, ValueError)` <sub>원본 힌트 · 형변환</sub>

> 🎯 **형변환이 실패할 수 있는 값은 어떻게 안전하게 처리하는가?**

💡 **쉽게 말하면**

- **자료형이 틀리면** TypeError, **자료형은 맞는데 값이 이상하면** ValueError.
- `try`에서 시도하고 `except`에서 대신할 일을 정함.

**실행 흐름**

1. 형변환 함수 정의
2. 정상 값
3. 글자 값 (ValueError)
4. None (TypeError)

**먼저 예측** · `to_int(None)`은 어떤 에러를 잡을까? — 답은 코드 실행 결과 아래에

````python
def to_int(value, default=0):             # 정수로 바꾸고 실패하면 기본값
    try:
        return int(value)                 # 정수로 바꿔 보기
    except (TypeError, ValueError) as e:  # 두 에러를 함께 처리
        print(f"  {type(e).__name__}: {value!r} → 기본값 {default}")  # 어떤 에러였는지
        return default                    # 대신 기본값


print("결과:", to_int("25"))   # 숫자 모양 문자열 → 정상
print("결과:", to_int("abc"))  # 숫자 모양이 아님 → ValueError
print("결과:", to_int(None))   # 값이 없음 → TypeError
````

**결과** · 예측 정답은 `TypeError` — None은 int()에 넣을 수 없는 자료형

- `결과: 25` — 정상 변환
- `ValueError: 'abc' → 기본값 0` — 문자열은 맞지만 숫자 모양이 아님

> - 1 · 2장 문법을 한 번에 사용
> - 로그 파일에서 읽은 값처럼 자료형이 섞인 손실 기록을 정리

<img src="images/pf7_practice.png" alt="섞인 값을 float()로 바꾸고 실패한 값은 건너뜀" width="763">

**파이프라인 위치** · ② 불러오기 → ③ 전처리

> 🎯 **`["0.92", 0.61, "0.47", None, "error", 0.43]`에서 숫자만 골라 평균을 내려면?**

💡 **쉽게 말하면**

- 값마다 `float()`로 **바꿔 보고**, 실패하면 **이유와 함께 건너뜀**.
- 남은 숫자로만 평균.

**실행 흐름**

<td style="text-align: left;">4</td>

**먼저 예측** · 건너뛰는 값은 몇 개이고, 각각 어떤 에러일까? — 답은 코드 실행 결과 아래에

````python
# 1단계: 로그에서 읽은 손실 (문자열 · 숫자 · None이 섞임, 가짜 값)
raw = ["0.92", 0.61, "0.47", None, "error", 0.43]  # 자료형이 섞인 기록


# 2 · 3단계: 하나씩 float()로 바꾸고 실패하면 이유와 함께 건너뜀
clean, skipped = [], []                        # 성공 · 실패 목록
for v in raw:                                  # 값 하나씩
    try:
        clean.append(float(v))                 # 문자열 · 정수 · 실수 → 실수
    except (TypeError, ValueError) as e:       # None · 숫자 아닌 글자
        skipped.append((v, type(e).__name__))  # 값과 에러 종류 기록


# 4단계: 결과 보고
print("정리:", clean)                               # 숫자만 남음
print("건너뜀:", skipped)                           # 어떤 값이 왜 빠졌는지
print(f"평균 손실: {sum(clean) / len(clean):.4f}")  # 소수 4자리
````

**결과** · 예측 정답은 2개 — `None`은 TypeError, `"error"`는 ValueError

- `정리: [0.92, 0.61, 0.47, 0.43]` — 문자열 "0.92" · "0.47"도 실수로 바뀜
- `건너뜀: [(None, 'TypeError'), ('error', 'ValueError')]` — 자료형 문제와 값 문제가 구분됨
- `평균 손실: 0.6075` — 남은 4개의 평균

## 면접 30초 답변

**Q1. TypeError와 ValueError는 어떻게 다른가요?**
> - TypeError는 연산에 맞지 않는 자료형을 넣었을 때 납니다.
> - ValueError는 자료형은 맞지만 값이 올바르지 않을 때 납니다.
> - `int(None)`은 TypeError이고 `int("abc")`는 ValueError입니다.

**Q2. 문자열과 숫자를 함께 출력하려면 어떻게 하나요?**
> - `str()`로 숫자를 문자열로 바꿔 이어 붙이거나 f-string을 씁니다.
> - f-string은 자료형을 신경 쓰지 않고 값을 끼워 넣을 수 있어 가장 많이 씁니다.

**Q3. 예외 처리는 어떻게 하나요?**
> - 실패할 수 있는 코드를 `try`에 두고, 잡을 에러 이름을 `except`에 적습니다.
> - `except:`만 쓰면 다른 버그까지 숨기므로 에러 이름을 꼭 적습니다.

### 전체 코드

1 ~ 3장 코드를 장별 셀로 모음. **위에서부터 이 셀들만 실행해도 전체가 동작**하며, 각 블록 머리 주석의 번호가 섹션.

````python
# ---------- 1-1 ----------
x = "Seoul"                                # 문자열 (str)
y = 25                                     # 정수 (int)
print(type(x).__name__, type(y).__name__)  # 두 값의 자료형
try:
    z = x + y                              # 원본 그대로: 문자열 + 정수
except TypeError as e:                     # 자료형이 맞지 않음
    print("TypeError:", e)                 # 에러 문장 출력


# ---------- 1-2 ----------
z = x + str(y)                         # 정수를 문자열로 바꿔 이어 붙이기
print(f"x + y : {z}")                  # 원본 출력 형식
print(f"x + y : {x}{y}")               # f-string은 형변환 없이 끼워 넣음
age = "25"                             # 숫자 모양의 문자열
print("숫자로 더하기:", int(age) + 5)  # 문자열 → 정수로 바꿔 계산
````

````python
# ---------- 2-1 ----------
nums = [10, 20, 30]                      # 리스트
idx = "1"                                # 번호가 문자열로 들어온 상황
cases = [("호출", lambda: y()),          # 정수를 함수처럼 부름
         ("인덱스", lambda: nums[idx]),  # 번호 자리에 문자열
         ("기타", lambda: len(y))]       # 길이가 없는 정수
for name, bad in cases:                  # 세 경우를 차례로
    try:
        bad()                            # 각 잘못된 코드를 실행
    except TypeError as e:               # 셋 다 TypeError
        print(f"{name}: {e}")            # 종류별 에러 문장
print("고친 인덱스:", nums[int(idx)])    # 번호를 정수로 바꾸면 동작


# ---------- 2-2 ----------
def to_int(value, default=0):             # 정수로 바꾸고 실패하면 기본값
    try:
        return int(value)                 # 정수로 바꿔 보기
    except (TypeError, ValueError) as e:  # 두 에러를 함께 처리
        print(f"  {type(e).__name__}: {value!r} → 기본값 {default}")  # 어떤 에러였는지
        return default                    # 대신 기본값


print("결과:", to_int("25"))   # 숫자 모양 문자열 → 정상
print("결과:", to_int("abc"))  # 숫자 모양이 아님 → ValueError
print("결과:", to_int(None))   # 값이 없음 → TypeError
````

````python
# ---------- 3. 종합 실습 ----------
# 1단계: 로그에서 읽은 손실 (문자열 · 숫자 · None이 섞임, 가짜 값)
raw = ["0.92", 0.61, "0.47", None, "error", 0.43]  # 자료형이 섞인 기록


# 2 · 3단계: 하나씩 float()로 바꾸고 실패하면 이유와 함께 건너뜀
clean, skipped = [], []                        # 성공 · 실패 목록
for v in raw:                                  # 값 하나씩
    try:
        clean.append(float(v))                 # 문자열 · 정수 · 실수 → 실수
    except (TypeError, ValueError) as e:       # None · 숫자 아닌 글자
        skipped.append((v, type(e).__name__))  # 값과 에러 종류 기록


# 4단계: 결과 보고
print("정리:", clean)                               # 숫자만 남음
print("건너뜀:", skipped)                           # 어떤 값이 왜 빠졌는지
print(f"평균 손실: {sum(clean) / len(clean):.4f}")  # 소수 4자리
````
