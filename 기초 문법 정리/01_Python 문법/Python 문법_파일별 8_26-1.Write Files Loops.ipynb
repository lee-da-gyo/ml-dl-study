### 1-1. `A` 정답 · 반복문으로 파일 6개 쓰기 · `zip()` · `open(경로, "w")` <sub>원본 문제</sub>

> 🎯 **이름 리스트와 내용 리스트로 파일을 하나씩 쓰려면?**

💡 **쉽게 말하면**

- `zip`으로 **(이름, 내용)** 짝을 만들고,
- 짝마다 파일을 **열고 → 쓰고 → 닫음**.

**실행 흐름**

1. 폴더 만들기
2. 이름 · 내용 zip
3. 파일마다 with open("w")
4. 목록 · 내용 확인

**먼저 예측** · 만들어지는 파일은 몇 개이고, E.txt는 있을까? — 답은 코드 실행 결과 아래에

````python
from pathlib import Path  # 경로를 객체로 다룸

BASE = Path("실습파일")                    # 노트북 옆 실습 폴더 (원본 ../source 대신)
folder = BASE / "26-1"                     # 원본 경로 /source/26-1/
folder.mkdir(parents=True, exist_ok=True)  # 폴더 만들기 (있으면 통과)
names = ["A", "B", "C", "D", "F", "G"]     # 파일 이름 (E 없음 · 원본 그대로)
contents = ["Python", "JavaScript", "PHP", "Rust", "Solidity", "Assembly"]  # 파일마다 내용


# 정답: 이름 · 내용을 짝지어 파일마다 쓰기
for name, text in zip(names, contents):  # (A, Python), (B, JavaScript), ...
    with open(folder / f"{name}.txt", "w", encoding="utf-8") as f:  # 쓰기 모드로 열기
        f.write(text)                    # 내용 쓰기


print(sorted(p.name for p in folder.glob("*.txt")))              # 만든 파일 목록
print("A.txt:", (folder / "A.txt").read_text(encoding="utf-8"))  # 첫 파일 내용
print("G.txt:", (folder / "G.txt").read_text(encoding="utf-8"))  # 마지막 파일 내용
````

**결과** · 예측 정답은 6개 — E.txt는 원본 이름 리스트에 없음

- `['A.txt', 'B.txt', 'C.txt', 'D.txt', 'F.txt', 'G.txt']` — 이름 리스트 6개 그대로 (E 없음)
- 이름과 내용이 짝대로 저장
  - `A.txt: Python`
  - `G.txt: Assembly`

### 1-2. `A` `with`와 인코딩 · `f.closed` · `encoding="utf-8"` <sub>원본 힌트 · open</sub>

> 🎯 **`with`는 무엇을 해 주고, 한글은 왜 인코딩을 적는가?**

💡 **쉽게 말하면**

- `with`는 **자동문** · 블록을 나오면 파일이 저절로 닫힘.
- 인코딩은 **글자를 숫자로 바꾸는 표** · 쓸 때와 읽을 때 같아야 함.

**실행 흐름**

1. with 안 · 밖에서 닫힘 확인
2. with 없이 열고 직접 닫기
3. 한글 쓰고 읽기

**먼저 예측** · `with` 블록이 끝난 뒤 `f.closed`는? — 답은 코드 실행 결과 아래에

````python
with open(folder / "A.txt", encoding="utf-8") as f:  # 읽기 모드("r")는 생략 가능
    print("블록 안 닫힘?", f.closed)                 # 아직 열려 있음
print("블록 밖 닫힘?", f.closed)                     # with가 끝나면 자동으로 닫힘


f = open(BASE / "kor.txt", "w", encoding="utf-8")               # with 없이 열기
f.write("안녕하세요")                                           # 한글 쓰기
f.close()                                                       # 직접 닫아야 저장이 확실함
print("한글:", (BASE / "kor.txt").read_text(encoding="utf-8"))  # 같은 인코딩으로 읽기
````

**결과** · 예측 정답은 `True` — 블록이 끝나면 자동으로 닫힘

- `with`가 끝나는 순간 닫힘
  - `블록 안 닫힘? False`
  - `블록 밖 닫힘? True`
- `한글: 안녕하세요` — 쓸 때 · 읽을 때 모두 utf-8

### 2-1. `A` 덮어쓰기와 이어 쓰기 · `"w"` · `"a"` <sub>원본 힌트 · 쓰기 모드</sub>

> 🎯 **같은 파일을 `"w"`로 두 번 쓰면, `"a"`로 쓰면 어떻게 되는가?**

💡 **쉽게 말하면**

- `"w"`는 **공책을 새로 펴서** 쓰기 · 앞 내용이 사라짐.
- `"a"`는 **마지막 줄 다음에** 이어 쓰기.

**실행 흐름**

1. w로 first
2. 다시 w로 second
3. a로 third
4. 줄 목록 확인

**먼저 예측** · `"w"`로 first · second를 차례로 쓰면 남는 줄은? — 답은 코드 실행 결과 아래에

````python
log = BASE / "log.txt"                                      # 모드 비교용 파일
with open(log, "w", encoding="utf-8") as f:                 # "w": 비우고 새로 쓰기
    f.write("first\n")                                      # 첫 번째 쓰기
with open(log, "w", encoding="utf-8") as f:                 # 다시 "w" → 앞 내용 사라짐
    f.write("second\n")                                     # 두 번째 쓰기
print("w 두 번:", log.read_text(encoding="utf-8").split())  # 남은 줄 목록


with open(log, "a", encoding="utf-8") as f:                # "a": 끝에 이어 쓰기
    f.write("third\n")                                     # 추가
print("a 추가:", log.read_text(encoding="utf-8").split())  # 앞 내용 + 새 줄
````

**결과** · 예측 정답은 `second`만 — 두 번째 "w"가 앞 내용을 지움

- `w 두 번: ['second']` — "w"는 열 때마다 비우고 새로 씀
- `a 추가: ['second', 'third']` — "a"는 앞 내용 뒤에 이어 씀

**딥러닝 코드에서는**

- 학습 시작 때 `"w"`로 헤더를 쓰고, epoch마다 `"a"`로 한 줄씩 기록

### 2-2. `B` 여러 줄 쓰기와 줄바꿈 · `writelines()` · `"\n".join()` <sub>원본 힌트 · writelines</sub>

> 🎯 **`writelines`는 줄바꿈을 넣어 주는가?**

💡 **쉽게 말하면**

- `writelines`는 리스트를 **그대로 이어 붙일 뿐** 줄바꿈은 안 넣음.
- 줄을 나누려면 `\n`을 직접 붙이거나 `join`으로 합침.

**실행 흐름**

1. writelines 그대로
2. 줄마다 \n 붙여 writelines
3. join으로 한 번에 write

**먼저 예측** · `writelines(["Python", "Rust", "PHP"])`로 쓴 파일 내용은? — 답은 코드 실행 결과 아래에

````python
lines = ["Python", "Rust", "PHP"]                                 # 쓸 줄들
p = BASE / "lines.txt"                                            # 비교용 파일
with open(p, "w", encoding="utf-8") as f:                         # 덮어쓰기
    f.writelines(lines)                                           # 리스트를 그대로 이어 붙임
print("writelines 그대로:", repr(p.read_text(encoding="utf-8")))  # 줄바꿈 없이 붙음


with open(p, "w", encoding="utf-8") as f:                          # 다시 덮어쓰기
    f.writelines(line + "\n" for line in lines)                    # 줄마다 줄바꿈 붙이기
print("줄바꿈 붙임:", p.read_text(encoding="utf-8").splitlines())  # 줄 목록


with open(p, "w", encoding="utf-8") as f:            # 다시 덮어쓰기
    f.write("\n".join(lines))                        # 줄바꿈으로 합친 문자열 하나
print("join:", repr(p.read_text(encoding="utf-8")))  # 마지막 줄 뒤에는 줄바꿈 없음
````

**결과** · 예측 정답은 `'PythonRustPHP'` — 줄바꿈 없이 붙음

- `writelines 그대로: 'PythonRustPHP'` — writelines는 줄바꿈을 넣어 주지 않음
- `줄바꿈 붙임: ['Python', 'Rust', 'PHP']` — 줄마다 줄바꿈을 붙이면 3줄
- `join: 'Python\nRust\nPHP'` — join은 사이에만 줄바꿈

**먼저 예측** · 다시 읽으면 줄은 몇 개일까? (헤더 포함) — 답은 코드 실행 결과 아래에

````python
# 1단계: 학습 시작 — "w"로 헤더만 쓰기
log_path = BASE / "train_log.csv"                 # 학습 로그 파일
with open(log_path, "w", encoding="utf-8") as f:  # 새로 시작
    f.write("epoch,loss,acc\n")                   # 열 이름


# 2단계: epoch마다 "a"로 한 줄씩 추가 (가짜 학습 결과)
history = [(1, 0.92, 0.71), (2, 0.61, 0.80), (3, 0.47, 0.86)]  # (epoch, loss, acc)
for epoch, loss, acc in history:                               # epoch마다
    with open(log_path, "a", encoding="utf-8") as f:           # 이어 쓰기
        f.write(f"{epoch},{loss:.2f},{acc:.2f}\n")             # 쉼표로 구분한 한 줄


# 3 · 4단계: 다시 읽어 줄로 나누고 최저 손실 줄 찾기
rows = log_path.read_text(encoding="utf-8").splitlines()    # 줄 목록
print("줄 수:", len(rows))                                  # 헤더 1 + 기록 3
print("헤더:", rows[0])                                     # 열 이름
best = min(rows[1:], key=lambda r: float(r.split(",")[1]))  # 두 번째 칸(loss) 기준
print("최저 loss 줄:", best)                                # epoch 3
````

**결과** · 예측 정답은 4줄 — 헤더 1 + 기록 3

- `줄 수: 4` — 헤더 1줄 + epoch 3줄
- `헤더: epoch,loss,acc` — "w"로 쓴 첫 줄
- `최저 loss 줄: 3,0.47,0.86` — 문자열을 `float()`로 바꿔 비교

> - `with` 블록이 끝나면 에러가 나더라도 파일이 자동으로 닫힙니다.
> - 직접 `close()`를 빠뜨리면 내용이 저장되지 않거나 파일이 잠길 수 있습니다.

**Q2. `"w"`와 `"a"` 모드는 어떻게 다른가요?**
> - `"w"`는 파일을 비우고 새로 쓰고, `"a"`는 기존 내용 끝에 이어 씁니다.
> - 학습 로그는 시작할 때 `"w"`로 헤더를 쓰고 epoch마다 `"a"`로 추가합니다.

**Q3. 한글 파일을 다룰 때 주의할 점은?**
> - 쓸 때와 읽을 때 같은 인코딩을 써야 합니다.
> - 운영체제마다 기본 인코딩이 달라서 `encoding="utf-8"`을 항상 적습니다.

### 전체 코드

1 ~ 3장 코드를 장별 셀로 모음. **위에서부터 이 셀들만 실행해도 전체가 동작**하며, 각 블록 머리 주석의 번호가 섹션.

````python
from pathlib import Path  # 경로를 객체로 다룸


# ---------- 1-1 ----------

BASE = Path("실습파일")                    # 노트북 옆 실습 폴더 (원본 ../source 대신)
folder = BASE / "26-1"                     # 원본 경로 /source/26-1/
folder.mkdir(parents=True, exist_ok=True)  # 폴더 만들기 (있으면 통과)
names = ["A", "B", "C", "D", "F", "G"]     # 파일 이름 (E 없음 · 원본 그대로)
contents = ["Python", "JavaScript", "PHP", "Rust", "Solidity", "Assembly"]  # 파일마다 내용


# 정답: 이름 · 내용을 짝지어 파일마다 쓰기
for name, text in zip(names, contents):  # (A, Python), (B, JavaScript), ...
    with open(folder / f"{name}.txt", "w", encoding="utf-8") as f:  # 쓰기 모드로 열기
        f.write(text)                    # 내용 쓰기


print(sorted(p.name for p in folder.glob("*.txt")))              # 만든 파일 목록
print("A.txt:", (folder / "A.txt").read_text(encoding="utf-8"))  # 첫 파일 내용
print("G.txt:", (folder / "G.txt").read_text(encoding="utf-8"))  # 마지막 파일 내용


# ---------- 1-2 ----------
with open(folder / "A.txt", encoding="utf-8") as f:  # 읽기 모드("r")는 생략 가능
    print("블록 안 닫힘?", f.closed)                 # 아직 열려 있음
print("블록 밖 닫힘?", f.closed)                     # with가 끝나면 자동으로 닫힘


f = open(BASE / "kor.txt", "w", encoding="utf-8")               # with 없이 열기
f.write("안녕하세요")                                           # 한글 쓰기
f.close()                                                       # 직접 닫아야 저장이 확실함
print("한글:", (BASE / "kor.txt").read_text(encoding="utf-8"))  # 같은 인코딩으로 읽기
````

````python
# ---------- 2-1 ----------
log = BASE / "log.txt"                                      # 모드 비교용 파일
with open(log, "w", encoding="utf-8") as f:                 # "w": 비우고 새로 쓰기
    f.write("first\n")                                      # 첫 번째 쓰기
with open(log, "w", encoding="utf-8") as f:                 # 다시 "w" → 앞 내용 사라짐
    f.write("second\n")                                     # 두 번째 쓰기
print("w 두 번:", log.read_text(encoding="utf-8").split())  # 남은 줄 목록


with open(log, "a", encoding="utf-8") as f:                # "a": 끝에 이어 쓰기
    f.write("third\n")                                     # 추가
print("a 추가:", log.read_text(encoding="utf-8").split())  # 앞 내용 + 새 줄


# ---------- 2-2 ----------
lines = ["Python", "Rust", "PHP"]                                 # 쓸 줄들
p = BASE / "lines.txt"                                            # 비교용 파일
with open(p, "w", encoding="utf-8") as f:                         # 덮어쓰기
    f.writelines(lines)                                           # 리스트를 그대로 이어 붙임
print("writelines 그대로:", repr(p.read_text(encoding="utf-8")))  # 줄바꿈 없이 붙음


with open(p, "w", encoding="utf-8") as f:                          # 다시 덮어쓰기
    f.writelines(line + "\n" for line in lines)                    # 줄마다 줄바꿈 붙이기
print("줄바꿈 붙임:", p.read_text(encoding="utf-8").splitlines())  # 줄 목록


with open(p, "w", encoding="utf-8") as f:            # 다시 덮어쓰기
    f.write("\n".join(lines))                        # 줄바꿈으로 합친 문자열 하나
print("join:", repr(p.read_text(encoding="utf-8")))  # 마지막 줄 뒤에는 줄바꿈 없음
````

````python
# ---------- 3. 종합 실습 ----------
# 1단계: 학습 시작 — "w"로 헤더만 쓰기
log_path = BASE / "train_log.csv"                 # 학습 로그 파일
with open(log_path, "w", encoding="utf-8") as f:  # 새로 시작
    f.write("epoch,loss,acc\n")                   # 열 이름


# 2단계: epoch마다 "a"로 한 줄씩 추가 (가짜 학습 결과)
history = [(1, 0.92, 0.71), (2, 0.61, 0.80), (3, 0.47, 0.86)]  # (epoch, loss, acc)
for epoch, loss, acc in history:                               # epoch마다
    with open(log_path, "a", encoding="utf-8") as f:           # 이어 쓰기
        f.write(f"{epoch},{loss:.2f},{acc:.2f}\n")             # 쉼표로 구분한 한 줄


# 3 · 4단계: 다시 읽어 줄로 나누고 최저 손실 줄 찾기
rows = log_path.read_text(encoding="utf-8").splitlines()    # 줄 목록
print("줄 수:", len(rows))                                  # 헤더 1 + 기록 3
print("헤더:", rows[0])                                     # 열 이름
best = min(rows[1:], key=lambda r: float(r.split(",")[1]))  # 두 번째 칸(loss) 기준
print("최저 loss 줄:", best)                                # epoch 3
````
