> 🎯 **값이 다른 두 정수는 `==`와 `is`로 비교하면 어떻게 되는가?**

💡 **쉽게 말하면**

- `==`는 **내용이 같은가**, `is`는 **같은 물건인가**.

### 1-2. `A` `y = x` · `id()` <sub>원본 예제 2</sub>

> 🎯 **`y = x` 뒤 `x == y`와 `x is y`는 왜 둘 다 True인가?**

💡 **쉽게 말하면**

- `y = x`는 물건을 복사한 것이 아니라 같은 물건에 **이름표를 하나 더 붙인 것**.
- 그래서 내용도 같고, 물건도 하나.

**실행 흐름**

1. 리스트 하나 만들기
2. y = x
3. == · is · id 비교

**먼저 예측** · `y = x` 뒤 `x == y`와 `x is y`는? — 답은 코드 실행 결과 아래에

````python
# 예제 2: y = x 로 같은 리스트를 가리킴
x = ["orange", "banana", "apple"]        # 리스트 객체 하나 생성
y = x                                    # 복사 X → 같은 객체에 이름표 추가
print(f"x == y : {x == y}")              # 내용 비교
print(f"x is y : {x is y}")              # 같은 객체인지
print("같은 번호(id):", id(x) == id(y))  # 메모리 속 번호가 같은지
````

**결과** · 예측 정답은 `True` · `True` — x와 y는 같은 리스트 하나

- 같은 객체 → 내용도 당연히 같음
  - `x == y : True`
  - `x is y : True`
- `같은 번호(id): True` — `id()`가 같음 = 메모리 속 같은 물건

### 2-1. `A` `y = x`의 함정과 `copy()` · `append()` <sub>원본 힌트 · 참조 비교</sub>

> 🎯 **`b = a` 뒤 `b`를 고치면 왜 `a`도 바뀌고, 어떻게 막는가?**

💡 **쉽게 말하면**

- `b = a`는 이름표만 하나 더 붙인 것이라, 물건(리스트)은 하나.
- `copy()`는 물건을 새로 하나 만들어 줌.

**실행 흐름**

1. b = a 후 b 수정
2. a 확인
3. copy()로 진짜 복사
4. a · c 확인

**먼저 예측** · `b = a` 뒤 `b.append(4)` 하면 `a`는? — 답은 코드 실행 결과 아래에

````python
# y = x 로 만든 b를 고치면 a도 바뀜
a = [1, 2, 3]    # 원본 리스트
b = a            # 같은 객체에 이름 추가
b.append(4)      # b 수정 = a 수정
print("a →", a)  # a에도 4가 붙음


# 진짜 복사는 copy()
c = a.copy()                 # 새 객체 생성
c.append(5)                  # c만 수정
print("a →", a, "/ c →", c)  # a는 그대로
print("c is a:", c is a)     # 다른 객체
````

**결과** · 예측 정답은 `[1, 2, 3, 4]` — a와 b는 같은 객체

- `a → [1, 2, 3, 4]` — b를 고쳤는데 a도 바뀜
- `a → [1, 2, 3, 4] / c → [1, 2, 3, 4, 5]` — copy()로 만든 c만 바뀜
- `c is a: False` — copy()는 새 객체

### 2-2. `B` 리스트 안 리스트와 `deepcopy()` · `copy.deepcopy()` <sub>보충 · 원본 힌트 확장</sub>

> 🎯 **리스트 안에 리스트가 있으면 `copy()`로 충분한가?**

💡 **쉽게 말하면**

- `copy()`는 **바깥 상자만** 새로 만들고, 안쪽 상자는 같이 씀.
- `deepcopy()`는 안쪽 상자까지 전부 새로 만듦.

**실행 흐름**

1. 2행 2열 리스트
2. copy · deepcopy 두 복사본
3. 안쪽 값 수정
4. 원본 확인

**먼저 예측** · `shallow = table.copy()` 뒤 `shallow[0][0] = 99` 하면 원본 `table`은? — 답은 코드 실행 결과 아래에

````python
import copy  # 깊은 복사 기능


table = [[1, 2], [3, 4]]                                          # 리스트 안 리스트 (2행 2열)
shallow = table.copy()                                            # 바깥만 새로
deep = copy.deepcopy(table)                                       # 안쪽까지 새로
shallow[0][0] = 99                                                # 안쪽 리스트를 고침
print("copy 후 원본:", table)                                     # 안쪽은 공유 → 바뀜
deep[1][1] = -1                                                   # 깊은 복사본을 고침
print("deepcopy 후 원본:", table)                                 # 그대로
print("안쪽 공유?", shallow[0] is table[0], deep[0] is table[0])  # 안쪽 상자가 같은 객체인지
````

**결과** · 예측 정답은 `[[99, 2], [3, 4]]` — 안쪽 리스트는 함께 씀

- `copy 후 원본: [[99, 2], [3, 4]]` — copy()는 바깥만 새로 → 안쪽 수정이 원본에 보임
- `deepcopy 후 원본: [[99, 2], [3, 4]]` — deepcopy 쪽 수정은 원본에 영향 없음
- `안쪽 공유? True False` — copy는 안쪽을 공유, deepcopy는 안쪽도 새 객체

### 2-3. `A` `None`은 `is`로 · `is None` · `is not None` <sub>원본 힌트 · is 사용</sub>

> 🎯 **값이 없는지(`None`) 확인할 때 왜 `is`를 쓰는가?**

💡 **쉽게 말하면**

- `None`은 파이썬 전체에 **하나뿐인 물건**.
- 그래서 '그 물건인가?'를 묻는 `is`가 가장 정확하고 빠름.

**실행 흐름**

1. None 값 만들기
2. is None · == None 비교
3. 선택 인자처럼 쓰기

**먼저 예측** · `value = None`일 때 `value is None`은? — 답은 코드 실행 결과 아래에

````python
value = None                                             # 값이 없음을 뜻하는 객체
print("value is None:", value is None)                   # 권장 방식
print("None은 하나뿐:", value is None and None is None)  # 같은 객체 하나


# 선택 인자처럼 쓰기: 받은 값이 없을 때만 기본 동작
mask = None                           # 아직 정하지 않은 값
if mask is not None:                  # 값이 있을 때만
    print("마스크 적용")              # 여기는 실행 안 됨
else:
    print("마스크 없음 → 기본 처리")  # None이라 이쪽
````

**결과** · 예측 정답은 `True`

- `value is None: True` — None 비교는 `is`
- `None은 하나뿐: True` — None 객체는 하나라 `is`로 정확히 확인
- `마스크 없음 → 기본 처리` — `is not None`으로 값이 있을 때만 처리

> 🎯 **원본 점수를 0 ~ 1로 바꿀 때 원본까지 바뀌는 실수는 어떻게 생기고, 어떻게 막는가?**

**먼저 예측** · 2단계 뒤 `raw[0]`은 72일까 0.72일까? — 답은 코드 실행 결과 아래에

````python
# 1단계: 원본 점수 (학습 데이터라고 생각)
raw = [72, 95, 88, 61, 79]  # 원본 데이터


# 2단계: 잘못된 방법 — 이름표만 붙이고 값을 바꿈
scaled_bad = raw                         # 복사 X
for i in range(len(scaled_bad)):         # 자리마다
    scaled_bad[i] = scaled_bad[i] / 100  # 0 ~ 1로 바꿈
print("잘못된 방법 후 원본:", raw)       # 원본까지 바뀜


# 3단계: 올바른 방법 — 새 리스트로 만들기
raw = [72, 95, 88, 61, 79]          # 원본 다시 준비
scaled = [v / 100 for v in raw]     # 새 리스트 (원본 그대로)
print("올바른 방법 후 원본:", raw)  # 원본 보존
print("변환 결과:", scaled)         # 0 ~ 1 값


# 4단계: is · == 로 확인
print("같은 객체?", scaled is raw, "/ 값 같음?", scaled == raw)  # 다른 객체 · 다른 값
````

````python
# 5단계: 기준값(scale)을 선택 인자로 → 없으면 최댓값 사용
def normalize(data, scale=None):      # scale을 안 주면 None
    if scale is None:                 # 값이 없을 때만
        scale = max(data)             # 데이터의 최댓값으로
    return [v / scale for v in data]  # 새 리스트 반환 (원본 보존)


print("max 기준:", normalize(raw))       # 95로 나눔
print("100 기준:", normalize(raw, 100))  # 100으로 나눔
print("원본 그대로:", raw)               # 함수도 원본을 안 바꿈
````

**결과** · 예측 정답은 `0.72` — 같은 객체라 원본까지 바뀜

- `잘못된 방법 후 원본: [0.72, 0.95, 0.88, 0.61, 0.79]` — 이름표만 붙여 작업 → 원본 손상
- 새 리스트로 만들면 원본 보존
  - `올바른 방법 후 원본: [72, 95, 88, 61, 79]`
  - `변환 결과: [0.72, 0.95, 0.88, 0.61, 0.79]`
- `같은 객체? False / 값 같음? False` — 다른 객체 · 다른 값
- `scale is None`일 때만 최댓값 사용 · 함수도 원본을 바꾸지 않음
  - `max 기준: [0.7578947368421053, 1.0`
  - `100 기준: [0.72`
  - `원본 그대로: [72, 95, 88, 61, 79]`

## 면접 30초 답변

**Q1. 리스트를 `y = x`로 복사하면 어떤 문제가 생기나요?**
> - `y = x`는 새 리스트가 아니라 같은 객체에 이름을 하나 더 붙이는 것이라, `y`를 수정하면 `x`도 바뀝니다.
> - 독립 복사본은 `x.copy()`, 리스트 안 리스트처럼 중첩된 구조는 `copy.deepcopy()`를 씁니다.

**Q2. `==`와 `is`의 차이는 무엇인가요?**
> - `==`는 두 값의 내용이 같은지, `is`는 두 이름이 메모리 속 같은 객체인지 비교합니다.
> - 값 비교는 `==`로 하고, `is`는 `None` 확인처럼 같은 객체인지가 중요할 때만 씁니다.

**Q3. `None`은 왜 `is`로 비교하나요?**
> - `None`은 파이썬 전체에 하나뿐인 객체라 `is None`이 정확하고 빠릅니다.
> - `==`는 객체가 비교 방법을 바꿔 둘 수 있어 예상과 다른 결과가 나올 수 있습니다.

### 전체 코드

1 ~ 3장 코드를 장별 셀로 모음. **위에서부터 이 셀들만 실행해도 전체가 동작**하며, 각 블록 머리 주석의 번호가 섹션.

````python
# ---------- 1-1 ----------
# 예제 1: 값이 다른 정수 두 개
x = 15                       # 정수 객체 15
y = 25                       # 정수 객체 25
print(f"x == y : {x == y}")  # 값 비교
print(f"x is y : {x is y}")  # 같은 객체인지


# ---------- 1-2 ----------
# 예제 2: y = x 로 같은 리스트를 가리킴
x = ["orange", "banana", "apple"]        # 리스트 객체 하나 생성
y = x                                    # 복사 X → 같은 객체에 이름표 추가
print(f"x == y : {x == y}")              # 내용 비교
print(f"x is y : {x is y}")              # 같은 객체인지
print("같은 번호(id):", id(x) == id(y))  # 메모리 속 번호가 같은지


# ---------- 1-3 ----------
# 예제 3: [ ] 를 다시 써서 내용만 같은 리스트를 하나 더 만듦
x = ["orange", "banana", "apple"]        # 리스트 객체 1
y = ["orange", "banana", "apple"]        # 리스트 객체 2 (새로 생성)
print(f"x == y : {x == y}")              # 내용 비교
print(f"x is y : {x is y}")              # 같은 객체인지
print("같은 번호(id):", id(x) == id(y))  # 다른 물건이면 False
````

````python
import copy  # 깊은 복사 기능


# ---------- 2-1 ----------
# y = x 로 만든 b를 고치면 a도 바뀜
a = [1, 2, 3]    # 원본 리스트
b = a            # 같은 객체에 이름 추가
b.append(4)      # b 수정 = a 수정
print("a →", a)  # a에도 4가 붙음


# 진짜 복사는 copy()
c = a.copy()                 # 새 객체 생성
c.append(5)                  # c만 수정
print("a →", a, "/ c →", c)  # a는 그대로
print("c is a:", c is a)     # 다른 객체


# ---------- 2-2 ----------


table = [[1, 2], [3, 4]]                                          # 리스트 안 리스트 (2행 2열)
shallow = table.copy()                                            # 바깥만 새로
deep = copy.deepcopy(table)                                       # 안쪽까지 새로
shallow[0][0] = 99                                                # 안쪽 리스트를 고침
print("copy 후 원본:", table)                                     # 안쪽은 공유 → 바뀜
deep[1][1] = -1                                                   # 깊은 복사본을 고침
print("deepcopy 후 원본:", table)                                 # 그대로
print("안쪽 공유?", shallow[0] is table[0], deep[0] is table[0])  # 안쪽 상자가 같은 객체인지


# ---------- 2-3 ----------
value = None                                             # 값이 없음을 뜻하는 객체
print("value is None:", value is None)                   # 권장 방식
print("None은 하나뿐:", value is None and None is None)  # 같은 객체 하나


# 선택 인자처럼 쓰기: 받은 값이 없을 때만 기본 동작
mask = None                           # 아직 정하지 않은 값
if mask is not None:                  # 값이 있을 때만
    print("마스크 적용")              # 여기는 실행 안 됨
else:
    print("마스크 없음 → 기본 처리")  # None이라 이쪽
````

````python
# ---------- 3. 종합 실습 ----------
# 1단계: 원본 점수 (학습 데이터라고 생각)
raw = [72, 95, 88, 61, 79]  # 원본 데이터


# 2단계: 잘못된 방법 — 이름표만 붙이고 값을 바꿈
scaled_bad = raw                         # 복사 X
for i in range(len(scaled_bad)):         # 자리마다
    scaled_bad[i] = scaled_bad[i] / 100  # 0 ~ 1로 바꿈
print("잘못된 방법 후 원본:", raw)       # 원본까지 바뀜


# 3단계: 올바른 방법 — 새 리스트로 만들기
raw = [72, 95, 88, 61, 79]          # 원본 다시 준비
scaled = [v / 100 for v in raw]     # 새 리스트 (원본 그대로)
print("올바른 방법 후 원본:", raw)  # 원본 보존
print("변환 결과:", scaled)         # 0 ~ 1 값


# 4단계: is · == 로 확인
print("같은 객체?", scaled is raw, "/ 값 같음?", scaled == raw)  # 다른 객체 · 다른 값


# 5단계: 기준값(scale)을 선택 인자로 → 없으면 최댓값 사용
def normalize(data, scale=None):      # scale을 안 주면 None
    if scale is None:                 # 값이 없을 때만
        scale = max(data)             # 데이터의 최댓값으로
    return [v / scale for v in data]  # 새 리스트 반환 (원본 보존)


print("max 기준:", normalize(raw))       # 95로 나눔
print("100 기준:", normalize(raw, 100))  # 100으로 나눔
print("원본 그대로:", raw)               # 함수도 원본을 안 바꿈
````
