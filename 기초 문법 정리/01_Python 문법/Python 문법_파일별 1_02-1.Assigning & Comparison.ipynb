## 1. 원리

> 🎯 **`==`와 `is`는 무엇을 비교하며, `y = x`와 내용이 같은 새 리스트는 어떻게 다른가?**

💡 **쉽게 말하면**

- `==`는 **내용이 같은가**, `is`는 **같은 물건인가**.
- `y = x`는 물건을 복사한 것이 아니라 같은 물건에 **이름표를 하나 더 붙인 것**.
- `[...]`를 다시 쓰면 내용이 같아도 **새 물건**이 하나 더 생김.

## 2. 풀이

단계마다 **먼저 예측** → 코드 실행 → **결과** 순서로 확인. 셀은 위에서부터 차례로 실행.

### 2-1. 예제 1 · 서로 다른 정수

**실행 흐름**

1. x = 15, y = 25
2. == 로 값 비교
3. is 로 객체 비교

**먼저 예측** · `x == y`와 `x is y`는? — 답은 코드 실행 결과 아래에

````python
# 예제 1: 값이 다른 정수 두 개
x = 15                       # 정수 객체 15
y = 25                       # 정수 객체 25
print(f"x == y : {x == y}")  # 값 비교
print(f"x is y : {x is y}")  # 같은 객체인지
````

**결과** · 예측 정답은 `False` · `False` — 값도 다르고 객체도 다름

- `x == y : False` — 15와 25는 값이 다름
- `x is y : False` — 값이 다르면 당연히 다른 객체

### 2-2. 예제 2 · `y = x`

**실행 흐름**

1. 리스트 하나 만들기
2. y = x
3. == · is · id 비교

**먼저 예측** · `y = x` 뒤 `x == y`와 `x is y`는? — 답은 코드 실행 결과 아래에

````python
# 예제 2: y = x 로 같은 리스트를 가리킴
x = ["orange", "banana", "apple"]        # 리스트 객체 하나 생성
y = x                                    # 복사 X → 같은 객체에 이름표 추가
print(f"x == y : {x == y}")              # 내용 비교
print(f"x is y : {x is y}")              # 같은 객체인지
print("같은 번호(id):", id(x) == id(y))  # 메모리 속 번호가 같은지
````

**결과** · 예측 정답은 `True` · `True` — x와 y는 같은 리스트 하나

- `x == y : True` · `x is y : True` — 같은 객체 → 내용도 당연히 같음
- `같은 번호(id): True` — `id()`가 같음 = 메모리 속 같은 물건

### 2-3. 예제 3 · 내용만 같은 새 리스트

**실행 흐름**

1. 리스트 두 번 따로 만들기
2. == · is · id 비교

**먼저 예측** · 내용이 같은 리스트를 두 번 만들면 `x is y`는? — 답은 코드 실행 결과 아래에

````python
# 예제 3: [ ] 를 다시 써서 내용만 같은 리스트를 하나 더 만듦
x = ["orange", "banana", "apple"]        # 리스트 객체 1
y = ["orange", "banana", "apple"]        # 리스트 객체 2 (새로 생성)
print(f"x == y : {x == y}")              # 내용 비교
print(f"x is y : {x is y}")              # 같은 객체인지
print("같은 번호(id):", id(x) == id(y))  # 다른 물건이면 False
````

**결과** · 예측 정답은 `True` · `False` — 내용은 같지만 다른 객체

- `x == y : True` · `x is y : False` — 내용은 같지만 서로 다른 물건
- `같은 번호(id): False` — 메모리 속 번호가 다름

### 2-4. 더 알아보기 · `y = x`의 함정과 `None`

**실행 흐름**

1. b = a 후 b 수정
2. copy()로 진짜 복사
3. None은 is 로 확인

**먼저 예측** · `b = a` 뒤 `b.append(4)` 하면 `a`는? — 답은 코드 실행 결과 아래에

````python
# y = x 로 만든 b를 고치면 a도 바뀜
a = [1, 2, 3]    # 원본 리스트
b = a            # 같은 객체에 이름 추가
b.append(4)      # b 수정 = a 수정
print("a →", a)  # a에도 4가 붙음


# 진짜 복사는 copy()
c = a.copy()                 # 새 객체 생성
c.append(5)                  # c만 수정
print("a →", a, "/ c →", c)  # a는 그대로


# None 확인은 is (None은 하나뿐인 객체)
value = None                            # 값이 없음을 뜻하는 객체
print("value is None:", value is None)  # 권장 방식
````

**결과** · 예측 정답은 `[1, 2, 3, 4]` — a와 b는 같은 객체

- `a → [1, 2, 3, 4]` — b를 고쳤는데 a도 바뀜
- `a → [1, 2, 3, 4] / c → [1, 2, 3, 4, 5]` — copy()로 만든 c만 바뀜
- `value is None: True` — None 비교는 `is`
