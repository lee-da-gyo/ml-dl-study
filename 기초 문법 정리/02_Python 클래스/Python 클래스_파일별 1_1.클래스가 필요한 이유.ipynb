### 1-1. `A` 변수 · 리스트로 관리 · `del` · `zip()` <sub>원본 1 ~ 3절</sub>

> 🎯 **회사 목록에서만 BMW를 지우면 무슨 일이 생기는가?**

💡 **쉽게 말하면**

- 리스트 두 개로 나누면 **이름표와 물건을 다른 상자에 보관**한 것.
- 한 상자에서만 빼면 나머지가 **한 칸씩 밀려** 엉뚱한 짝이 됨.

**실행 흐름**

1. 차 한 대 = 변수 두 개
2. 리스트 두 개로 나눠 저장
3. 회사 목록에서만 BMW 삭제
4. zip으로 짝짓기

**먼저 예측** · 회사 목록에서만 BMW를 지운 뒤 짝지으면 `Audi` 옆에는 누구의 정보가 붙나? — 답은 코드 실행 결과 아래에

````python
# 1절: 차 한 대 = 변수 두 개 (회사 · 상세 정보)
# 왜 · 클래스 없이 변수로만 관리할 때 몇 대만 돼도 이름이 늘어나는 불편을 보이기 위해
car_company_1 = "Ferrari"                                            # 회사
car_detail_1 = {"color": "White", "horsepower": 400, "price": 8000}  # 상세 정보
print(car_company_1, car_detail_1)                                   # 차가 늘면 변수도 계속 늘어남


# 2 · 3절: 회사 · 상세를 리스트 두 개에 나눠 저장 → 한쪽만 지우면 짝이 어긋남
# 왜 · 따로 저장한 두 목록의 짝이 쉽게 깨지는 위험을 직접 재현하기 위해
wrong_company_list = ["Ferrari", "BMW", "Audi"]                             # 회사 목록
wrong_detail_list = [{"color": "White", "horsepower": 400, "price": 8000},  # 상세 목록
                     {"color": "Black", "horsepower": 270, "price": 5000},  # BMW 상세
                     {"color": "Silver", "horsepower": 300, "price": 6000}]  # Audi 상세
del wrong_company_list[1]  # 회사 목록에서만 BMW 삭제
for company, detail in zip(wrong_company_list, wrong_detail_list):          # 같은 위치끼리 짝짓기
    print(company, detail)                                                  # Audi 옆에 누구 정보가?
````

**결과** · 예측 정답은 BMW의 정보(`Black` · `270`) — 에러 없이 조용히 어긋남

- `Ferrari {'color': 'White', 'horsepower': 400, 'price': 8000}` — 차 한 대에 변수 두 개
- `Audi {'color': 'Black', 'horsepower': 270, 'price': 5000}` — Audi 옆에 BMW 정보 → 에러 없는 버그

> ✅ **한 줄 정리** · 같은 물건의 정보를 리스트 여러 개에 나누면 한쪽만 바뀌어도 에러 없이 짝이 어긋난다.

### 1-2. `B` 딕셔너리로 관리와 키 오타 · `car["키"]` · `KeyError` <sub>원본 4 ~ 5절</sub>

> 🎯 **딕셔너리로 묶으면 무엇이 좋아지고, 무엇이 아직 불편한가?**

💡 **쉽게 말하면**

- 딕셔너리는 차 한 대를 **한 봉투**에 담음 → 짝이 안 어긋남.
- 하지만 봉투마다 **이름표 철자**가 달라도 아무도 막지 않음.

**실행 흐름**

1. 한 대 = 딕셔너리 하나
2. 키로 꺼내기
3. 키 오타 KeyError

**먼저 예측** · `horse_power`로 저장한 차에서 `horsepower`를 꺼내면? — 답은 코드 실행 결과 아래에

````python
# 4절: 한 대의 정보를 딕셔너리 하나로 묶기
# 왜 · 한 대의 정보를 한 덩어리로 묶어 짝이 어긋나지 않게 하기 위해
# 한 대 = 한 덩어리
cars_dicts = [{"company": "Ferrari", "color": "White", "horsepower": 400, "price": 8000},
              {"company": "BMW", "color": "Black", "horsepower": 270, "price": 5000},  # 두 번째 차
              # 세 번째 차
              {"company": "Audi", "color": "Silver", "horsepower": 300, "price": 6000}]
for car in cars_dicts:                                                                 # 한 대씩
    print(car["company"], car["price"])                                                # 키로 꺼내기


# 5절: 키 이름 오타는 실행해야 드러남
# 왜 · 딕셔너리는 키 오타를 미리 막지 못한다는 한계를 보여 클래스가 필요한 이유로 잇기 위해
# horsepower 오타
wrong_car = {"company": "Ferrari", "color": "White", "horse_power": 400, "price": 8000}
try:
    print(wrong_car["horsepower"])           # 다른 차와 같은 키로 꺼내기
except KeyError as error:                    # 없는 키
    print("존재하지 않는 키입니다:", error)  # 원본과 같은 메시지
````

**결과** · 예측 정답은 `KeyError` — 키 이름이 다름

- 회사와 상세가 한 덩어리라 짝이 어긋나지 않음
  - `Ferrari 8000`
  - `BMW 5000`
  - `Audi 6000`
- `존재하지 않는 키입니다: 'horsepower'` — 키 오타는 실행해야 발견

**차이 · 언제 쓰나**

**먼저 예측** · `print(car1)`은 무엇을 출력할까? — 답은 코드 실행 결과 아래에

````python
# 왜 · 데이터(회사 · 색 · 마력 · 가격)와 기능을 한 설계도에 묶어 모든 차가 같은 구조를 갖게 하기 위해
class Car:                                                # 설계도 이름 (대문자로 시작)
    """자동차 한 대의 데이터와 기능을 표현하는 클래스"""  # 클래스 설명 (docstring)

    # 왜 · 객체를 만들 때 필요한 값을 빠짐없이 받도록 강제하기 위해
    def __init__(self, company, color, horsepower, price):  # 객체를 만들 때 자동 실행
        self.company = company                              # self = 지금 만들어지는 객체
        self.color = color                                  # 색
        self.horsepower = horsepower                        # 마력
        self.price = price                                  # 가격

    def __str__(self):  # print(객체)에 쓰는 글자
        # 사용자용
        return f"{self.company} / {self.color} / {self.horsepower}마력 / 가격 {self.price:,}원"

    def __repr__(self):                                                # 리스트 안 등 개발자용 글자
        return (f"Car(company='{self.company}', color='{self.color}', "  # 개발자용 표현
                f"horsepower={self.horsepower}, price={self.price})")  # 속성=값 형식

    # 왜 · 잘못된 할인율이 가격을 망가뜨리지 않게 검사한 뒤 적용하기 위해
    def apply_discount(self, discount_rate):                        # 할인율을 받아 가격 변경
        if not 0 <= discount_rate <= 1:                             # 0 ~ 1 밖이면
            raise ValueError("할인율은 0 이상 1 이하여야 합니다.")  # 원본 메시지
        self.price = round(self.price * (1 - discount_rate))        # 이 객체의 가격만

    # 왜 · 판단 기준을 메서드 한 곳에 두어 기준이 바뀌어도 한 줄만 고치기 위해
    def is_high_performance(self):     # 고성능 여부
        return self.horsepower >= 300  # 원본 코드는 200 (설명 · 출력은 300 기준)


# 왜 · 같은 설계도로 만든 객체들이 각자 값을 따로 가진다는 것을 확인하기 위해
car1 = Car("Ferrari", "White", 400, 8000)  # 객체 1
car2 = Car("BMW", "Black", 270, 5000)      # 객체 2
car3 = Car("Audi", "Silver", 300, 6000)    # 객체 3
print(car1)                                # __str__
print(car1.__dict__)                       # 객체에 저장된 값
````

**결과** · 예측 정답은 `__str__`이 돌려준 글자

- `Ferrari / White / 400마력 / 가격 8,000원` — print(객체) → `__str__`
- `{'company': 'Ferrari', 'color': 'White', 'horsepower': 400, 'price': 8000}` — `__init__`이 self에 저장한 값 4개

<img src="images/pfc1_practice.png" alt="TrainLog 객체 두 개가 각자 손실 기록을 가짐" width="763">

**파이프라인 위치** · ⑤ 학습 → ⑦ 기록

> 🎯 **이름 · 손실 기록 · 최솟값 · 출력 기능을 가진 `TrainLog`를 만들어 모델 두 개에 쓰려면?**

💡 **쉽게 말하면**

- **데이터**(이름 · 손실 리스트)와 **기능**(추가 · 최솟값)을 한 클래스에.
- 모델마다 객체를 하나씩 만들면 기록이 안 섞임.

**실행 흐름**

<table width="100%" style="width:100%; table-layout:auto">
<thead><tr><th style="text-align: left;">단계</th><th style="text-align: left;">하는 일</th><th style="text-align: left;">쓰는 문법</th></tr>
</thead>

<tr><td>1</td><td>설계도 TrainLog</td><td>2-1 <code>__init__</code> · <code>self</code></td></tr>

<tr><td>3</td><td>출력 형식</td><td>2-2 <code>__str__</code> · <code>__repr__</code></td></tr>
<tr><td>4</td><td>객체 두 개로 기록</td><td>1-1 흩어진 데이터 대신 한 덩어리</td></tr>
</tbody></table>

**먼저 예측** · `log_a.losses is log_b.losses`는? — 답은 코드 실행 결과 아래에

````python
# 1 ~ 3단계: 데이터(이름 · 손실)와 기능(추가 · 최솟값 · 출력)을 한 클래스에
# 왜 · 학습 기록과 그 기록을 다루는 기능을 한곳에 모아 모델마다 따로 쓰기 위해
class TrainLog:                # 학습 기록 설계도
    def __init__(self, name):  # 모델 이름을 받아
        self.name = name       # 이 기록의 이름
        self.losses = []       # 객체마다 새 리스트

    # 왜 · 기록을 반드시 이 메서드로만 추가해 다른 객체의 기록과 섞이지 않게 하기 위해
    def add(self, loss):          # epoch마다 손실 추가
        self.losses.append(loss)  # 이 객체의 리스트에만

    # 왜 · 가장 좋은 손실을 찾는 방법을 한 곳에 두어 어디서나 같은 기준으로 쓰기 위해
    def best(self):              # 가장 낮은 손실
        return min(self.losses)  # 최솟값

    def __str__(self):                                                            # print용 요약
        return f"{self.name}: {len(self.losses)} epoch · best {self.best():.2f}"  # 사용자용

    def __repr__(self):                                                    # 리스트 안 표시
        return f"TrainLog(name='{self.name}', epochs={len(self.losses)})"  # 개발자용


# 4단계: 모델 두 개의 기록을 따로 쌓기 (가짜 손실 값)
# 왜 · 객체마다 기록이 독립적인지 두 모델로 비교하기 위해
log_a, log_b = TrainLog("cnn"), TrainLog("mlp")    # 객체 두 개
for loss in [0.9, 0.6, 0.5]:                       # cnn 손실
    log_a.add(loss)                                # log_a에만 추가
for loss in [0.8, 0.7]:                            # mlp 손실
    log_b.add(loss)                                # log_b에만 추가
print(log_a)                                       # __str__
print([log_a, log_b])                              # 리스트 안 → __repr__
print("기록 공유?", log_a.losses is log_b.losses)  # 객체마다 따로
````

**결과** · 예측 정답은 `False` — `__init__`에서 객체마다 새 리스트

- `cnn: 3 epoch · best 0.50` — `__str__` · `best()`로 요약
- `[TrainLog(name='cnn', epochs=3), TrainLog(name='mlp', epochs=2)]` — 리스트 안은 `__repr__`
- `기록 공유? False` — 객체마다 자기 손실 리스트

**차이 · 언제 쓰나**

> - PyTorch에서 `print(model)`로 층 구조가 보이는 것도 `__repr__` 덕분입니다.

````python
# ---------- 1-1 ----------
# 1절: 차 한 대 = 변수 두 개 (회사 · 상세 정보)
# 왜 · 클래스 없이 변수로만 관리할 때 몇 대만 돼도 이름이 늘어나는 불편을 보이기 위해
car_company_1 = "Ferrari"                                            # 회사
car_detail_1 = {"color": "White", "horsepower": 400, "price": 8000}  # 상세 정보
print(car_company_1, car_detail_1)                                   # 차가 늘면 변수도 계속 늘어남


# 2 · 3절: 회사 · 상세를 리스트 두 개에 나눠 저장 → 한쪽만 지우면 짝이 어긋남
# 왜 · 따로 저장한 두 목록의 짝이 쉽게 깨지는 위험을 직접 재현하기 위해
wrong_company_list = ["Ferrari", "BMW", "Audi"]                             # 회사 목록
wrong_detail_list = [{"color": "White", "horsepower": 400, "price": 8000},  # 상세 목록
                     {"color": "Black", "horsepower": 270, "price": 5000},  # BMW 상세
                     {"color": "Silver", "horsepower": 300, "price": 6000}]  # Audi 상세
del wrong_company_list[1]  # 회사 목록에서만 BMW 삭제
for company, detail in zip(wrong_company_list, wrong_detail_list):          # 같은 위치끼리 짝짓기
    print(company, detail)                                                  # Audi 옆에 누구 정보가?


# ---------- 1-2 ----------
# 4절: 한 대의 정보를 딕셔너리 하나로 묶기
# 왜 · 한 대의 정보를 한 덩어리로 묶어 짝이 어긋나지 않게 하기 위해
# 한 대 = 한 덩어리
cars_dicts = [{"company": "Ferrari", "color": "White", "horsepower": 400, "price": 8000},
              {"company": "BMW", "color": "Black", "horsepower": 270, "price": 5000},  # 두 번째 차
              # 세 번째 차
              {"company": "Audi", "color": "Silver", "horsepower": 300, "price": 6000}]
for car in cars_dicts:                                                                 # 한 대씩
    print(car["company"], car["price"])                                                # 키로 꺼내기


# 5절: 키 이름 오타는 실행해야 드러남
# 왜 · 딕셔너리는 키 오타를 미리 막지 못한다는 한계를 보여 클래스가 필요한 이유로 잇기 위해
# horsepower 오타
wrong_car = {"company": "Ferrari", "color": "White", "horse_power": 400, "price": 8000}
try:
    print(wrong_car["horsepower"])           # 다른 차와 같은 키로 꺼내기
except KeyError as error:                    # 없는 키
    print("존재하지 않는 키입니다:", error)  # 원본과 같은 메시지
````

2. 2-2 리스트에 담아 같은 메서드 호출
3. 2-3 할인 적용 전후 비교

````python
# ---------- 2-1 ----------
# 왜 · 데이터(회사 · 색 · 마력 · 가격)와 기능을 한 설계도에 묶어 모든 차가 같은 구조를 갖게 하기 위해
class Car:                                                # 설계도 이름 (대문자로 시작)
    """자동차 한 대의 데이터와 기능을 표현하는 클래스"""  # 클래스 설명 (docstring)

    # 왜 · 객체를 만들 때 필요한 값을 빠짐없이 받도록 강제하기 위해
    def __init__(self, company, color, horsepower, price):  # 객체를 만들 때 자동 실행
        self.company = company                              # self = 지금 만들어지는 객체
        self.color = color                                  # 색
        self.horsepower = horsepower                        # 마력
        self.price = price                                  # 가격

    def __str__(self):  # print(객체)에 쓰는 글자
        # 사용자용
        return f"{self.company} / {self.color} / {self.horsepower}마력 / 가격 {self.price:,}원"

    def __repr__(self):                                                # 리스트 안 등 개발자용 글자
        return (f"Car(company='{self.company}', color='{self.color}', "  # 개발자용 표현
                f"horsepower={self.horsepower}, price={self.price})")  # 속성=값 형식

    # 왜 · 잘못된 할인율이 가격을 망가뜨리지 않게 검사한 뒤 적용하기 위해
    def apply_discount(self, discount_rate):                        # 할인율을 받아 가격 변경
        if not 0 <= discount_rate <= 1:                             # 0 ~ 1 밖이면
            raise ValueError("할인율은 0 이상 1 이하여야 합니다.")  # 원본 메시지
        self.price = round(self.price * (1 - discount_rate))        # 이 객체의 가격만

    # 왜 · 판단 기준을 메서드 한 곳에 두어 기준이 바뀌어도 한 줄만 고치기 위해
    def is_high_performance(self):     # 고성능 여부
        return self.horsepower >= 300  # 원본 코드는 200 (설명 · 출력은 300 기준)


# 왜 · 같은 설계도로 만든 객체들이 각자 값을 따로 가진다는 것을 확인하기 위해
car1 = Car("Ferrari", "White", 400, 8000)  # 객체 1
car2 = Car("BMW", "Black", 270, 5000)      # 객체 2
car3 = Car("Audi", "Silver", 300, 6000)    # 객체 3
print(car1)                                # __str__
print(car1.__dict__)                       # 객체에 저장된 값


# ---------- 2-2 ----------
# 왜 · 객체를 리스트에 담아도 같은 메서드로 한꺼번에 처리할 수 있음을 보이기 위해
car_list = [car1, car2, car3]  # 객체도 리스트에 담을 수 있음
print(car_list)                # 리스트 안 → 원소마다 __repr__
for car in car_list:           # 객체마다 같은 메서드
    # 메서드 호출
    print(f"{car.company}: {'고성능' if car.is_high_performance() else '일반'} 자동차")


# ---------- 2-3 ----------
# 왜 · 한 객체를 바꿔도 다른 객체에는 영향이 없다는 것을 확인하기 위해
print("Ferrari 할인 전:", car1)      # 8,000원
car1.apply_discount(0.1)             # car1만 10% 할인
print("Ferrari 10% 할인 후:", car1)  # 7,200원
print("BMW:", car2)                  # 그대로
print("Audi:", car3)                 # 변화 없음
try:
    car1.apply_discount(1.5)         # 범위 밖 할인율
except ValueError as e:              # raise가 만든 에러
    print("ValueError:", e)          # 원본 메시지
````

**실행 흐름**

1. `TrainLog` 클래스 정의
2. 모델 두 개의 기록 따로 쌓기
3. 객체별 결과 출력

````python
# ---------- 3. 종합 실습 ----------
# 1 ~ 3단계: 데이터(이름 · 손실)와 기능(추가 · 최솟값 · 출력)을 한 클래스에
# 왜 · 학습 기록과 그 기록을 다루는 기능을 한곳에 모아 모델마다 따로 쓰기 위해
class TrainLog:                # 학습 기록 설계도
    def __init__(self, name):  # 모델 이름을 받아
        self.name = name       # 이 기록의 이름
        self.losses = []       # 객체마다 새 리스트

    # 왜 · 기록을 반드시 이 메서드로만 추가해 다른 객체의 기록과 섞이지 않게 하기 위해
    def add(self, loss):          # epoch마다 손실 추가
        self.losses.append(loss)  # 이 객체의 리스트에만

    # 왜 · 가장 좋은 손실을 찾는 방법을 한 곳에 두어 어디서나 같은 기준으로 쓰기 위해
    def best(self):              # 가장 낮은 손실
        return min(self.losses)  # 최솟값

    def __str__(self):                                                            # print용 요약
        return f"{self.name}: {len(self.losses)} epoch · best {self.best():.2f}"  # 사용자용

    def __repr__(self):                                                    # 리스트 안 표시
        return f"TrainLog(name='{self.name}', epochs={len(self.losses)})"  # 개발자용


# 4단계: 모델 두 개의 기록을 따로 쌓기 (가짜 손실 값)
# 왜 · 객체마다 기록이 독립적인지 두 모델로 비교하기 위해
log_a, log_b = TrainLog("cnn"), TrainLog("mlp")    # 객체 두 개
for loss in [0.9, 0.6, 0.5]:                       # cnn 손실
    log_a.add(loss)                                # log_a에만 추가
for loss in [0.8, 0.7]:                            # mlp 손실
    log_b.add(loss)                                # log_b에만 추가
print(log_a)                                       # __str__
print([log_a, log_b])                              # 리스트 안 → __repr__
print("기록 공유?", log_a.losses is log_b.losses)  # 객체마다 따로
````
