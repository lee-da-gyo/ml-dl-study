### 1-1. `A` 리스트와 배열 · `np.array()` · `* 2` <sub>원본 Part 0 · 셀 1 · 4</sub>

> 🎯 **리스트에 `* 2`를 하면 왜 가격이 2배가 되지 않고, 배열은 무엇이 다른가?**

💡 **쉽게 말하면**

- 리스트는 **물건을 담는 상자**라 `* 2`를 하면 상자를 하나 더 이어 붙임.
- 배열은 **계산기**라 안의 숫자마다 2를 곱함.

**실행 흐름**

1. 리스트 * 2
2. 리스트 + 리스트 vs 배열 + 배열
3. 타입 비교

**먼저 예측** · `[30000, 25000, 40000] * 2`의 길이는? — 답은 코드 실행 결과 아래에

````python
# 원본 Part 0: 리스트에 2를 곱하면?
# 왜 · 리스트 곱하기가 반복이라는 점을 먼저 보여 배열이 필요한 이유를 만들기 위해
prices_list = [30000, 25000, 40000]         # 상품 가격 3개 (리스트)
print("리스트 * 2 결과:", prices_list * 2)  # 곱하기가 아니라 반복
print("결과 길이:", len(prices_list * 2))   # 3개 → 6개


# 같은 값을 배열로 만들어 비교
# 왜 · 같은 연산이 배열에서는 원소마다 계산된다는 차이를 나란히 보기 위해
numbers_list = [1, 2, 3]                                # 리스트
numbers_array = np.array(numbers_list)                  # 리스트 → 배열
print("리스트 + 리스트:", numbers_list + numbers_list)  # 이어 붙이기
print("배열 + 배열:", numbers_array + numbers_array)    # 같은 위치끼리 더하기
print("타입:", type(numbers_list).__name__, "→", type(numbers_array).__name__)  # list → ndarray
````

**결과** · 예측 정답은 `6` — 리스트의 `*`는 반복(이어 붙이기)

- `리스트 * 2 결과: [30000, 25000, 40000, 30000, 25000, 40000]` — 같은 목록을 두 번 이어 붙임
- `결과 길이: 6` — 3개 → 6개
- `배열 + 배열: [2 4 6]` — 배열은 같은 위치끼리 더함 (리스트는 `[1, 2, 3, 1, 2, 3]`)
- `타입: list → ndarray` — 배열의 정식 이름은 ndarray

### 1-2. `A` 축 · 모양 · 개수 · `ndim` · `shape` · `size` <sub>원본 셀 5 · 6 · 15 · 16</sub>

> 🎯 **배열의 `ndim` · `shape` · `size`는 각각 무엇을 알려 주는가?**

💡 **쉽게 말하면**

- `ndim`은 **방향(축)이 몇 개**인지, `shape`는 **축마다 칸 수**, `size`는 **칸의 총개수**.
- `np.zeros()` 괄호 안 숫자 개수가 곧 축 개수.

**실행 흐름**

1. 1차원 · 2차원 배열
2. 속성 3개 출력
3. zeros로 5차원 ~ 1차원 만들기

**먼저 예측** · 3행 4열 배열의 `size`는? — 답은 코드 실행 결과 아래에

````python
# 1차원 · 2차원 배열의 축 개수 · 모양 · 개수
# 왜 · ndim · shape · size가 각각 무엇을 세는지 구분하기 위해
one_dim = np.array([10, 20, 30, 40])                               # 값 4개 한 줄
two_dim = np.array([[10, 20, 30, 40],                              # 3행 4열 표
                    [50, 60, 70, 80],
                    [90, 100, 110, 120]])
for label, array in [("one_dim", one_dim), ("two_dim", two_dim)]:  # (이름, 배열) 짝마다
    print(f"{label}: ndim {array.ndim} | shape {array.shape} | size {array.size}")  # 세 속성 한 줄


# 원본 셀 15 · 16: zeros의 괄호 안 숫자 개수 = ndim
# 왜 · shape의 숫자 개수가 곧 축 개수라는 규칙을 여러 모양으로 확인하기 위해
for shape in [(5, 4, 3, 2, 1), (3, 4, 5, 6), (4, 5, 6), (5, 6), 6]:  # 원본이 차례로 만든 모양
    arr = np.zeros(shape)                                            # 0으로 채운 배열
    print(f"np.zeros({shape}) → ndim {arr.ndim}, size {arr.size}")   # 축 개수 · 값 개수
````

**결과** · 예측 정답은 `12` — 3 × 4 = 12

- `one_dim: ndim 1 | shape (4,) | size 4` — `(4,)`의 쉼표는 원소 1개짜리 튜플 표시
- `two_dim: ndim 2 | shape (3, 4) | size 12` — 3 × 4 = 12
- `np.zeros((5, 4, 3, 2, 1)) → ndim 5, size 120` — 5 × 4 × 3 × 2 × 1 = 120
- `np.zeros(6) → ndim 1, size 6` — 숫자 하나면 1차원

### 1-3. `A` 자료형 통일과 바꾸기 · `dtype` · `astype()` <sub>원본 Part 1 · 셀 8 · 10 · 11</sub>

> 🎯 **값을 섞으면 dtype은 어떻게 정해지고, `astype(int)`는 반올림인가 버림인가?**

💡 **쉽게 말하면**

- 배열은 **한 가지 규격 상자만** 씀.
- 섞이면 모두 담을 수 있는 쪽(정수 → 실수 → 문자열)으로 통일.
- `astype`은 다른 상자로 옮겨 담은 **새 배열**이고, 정수 상자에는 소수점 칸이 없어 버림.

**실행 흐름**

1. 정수 · 실수 섞기 · 문자열 섞기
2. float() 변환
3. astype(int)

**먼저 예측** · `np.array([1, 2, "3"])`의 dtype은? — 답은 코드 실행 결과 아래에

````python
# 섞인 값은 하나의 dtype으로 통일
# 왜 · 배열은 한 가지 자료형만 담는다는 규칙 때문에 값이 바뀌는 경우를 보기 위해
ints = np.array([1, 2, 3])                    # 정수만
mixed_number = np.array([1, 2, 3.5])          # 실수 하나
mixed_text = np.array([1, 2, "3"])            # 문자열 하나
for arr in [ints, mixed_number, mixed_text]:  # 세 배열마다
    print(arr, "->", arr.dtype)               # 값 · 자료형


# 원본 셀 10: Python의 float() 변환
# 왜 · 배열의 astype과 비교할 기준으로 파이썬 형변환을 먼저 보기 위해
a = 10                                                           # 정수
print(type(a).__name__, "→", type(float(a)).__name__, float(a))  # int → float


# 원본 셀 11: astype은 새 배열 · 소수점은 버림
# 왜 · 정수 변환이 반올림이 아니라 버림이라는 함정을 확인하기 위해
heights_float = np.array([171.9, 165.2, 180.8])        # 키 (실수)
heights_int = heights_float.astype(int)                # 정수로 바꾼 새 배열
print("변환 전:", heights_float, heights_float.dtype)  # 원래 값
print("변환 후:", heights_int, heights_int.dtype)      # 171.9 → 171
print("원본 유지:", heights_float)                     # 원본은 그대로
````

**결과** · 예측 정답은 문자열 `<U21` — 하나라도 문자열이면 전부 문자열

- `[1.  2.  3.5] -> float64` — 실수 하나 때문에 1도 `1.`
- `['1' '2' '3'] -> <U21` — 전부 문자열 (최대 21글자 유니코드)
- `int → float 10.0` — Python의 `float()`는 값 하나를 변환
- `변환 후: [171 165 180] int64` — 171.9 → 171 · 소수점 버림
- `원본 유지: [171.9 165.2 180.8]` — astype은 새 배열 · 원본 그대로

💡 **쉽게 말하면**

- **시작은 포함, 끝은 제외**라서 16까지 원하면 17을 적음.

**먼저 예측** · `np.arange(2, 11, 3)`의 마지막 값은? — 답은 코드 실행 결과 아래에

````python
# arange(start, stop, step): 시작 포함 · 끝 제외
# 왜 · 끝 값이 빠지는 규칙을 여러 인자 조합으로 확인하기 위해
print("np.arange(5)        ->", np.arange(5))         # 0 ~ 4
print("np.arange(2, 8)     ->", np.arange(2, 8))      # 2 ~ 7
print("np.arange(2, 11, 3) ->", np.arange(2, 11, 3))  # 2, 5, 8 (11 제외)


# 원본 미니 실습 1: 4부터 16까지 3씩 → stop은 17
# 왜 · 끝 값을 넣으려면 stop을 하나 더 크게 줘야 한다는 것을 확인하기 위해
step_array = np.arange(4, 17, 3)                          # 16을 넣으려면 stop 17
print("간격 배열:", step_array, "| 끝:", step_array[-1])  # 16 포함 확인
````

### 2-3. `A` 모양 바꾸기 · `reshape()` · `-1` <sub>원본 Part 3 · 셀 21 ~ 26</sub>

> 🎯 **`reshape`는 언제 실패하고, `-1`은 무엇을 대신 계산하는가?**

💡 **쉽게 말하면**

- `reshape`는 **같은 구슬 10개를 다른 판에 다시 꽂는 것**.
- 칸 수가 10이 아닌 판(5×3 · 4×3)에는 꽂을 수 없음.
- `-1`은 '이 칸 수는 네가 계산해'라는 빈칸.

**실행 흐름**

1. 10개 → 2×5 · 5×2
2. 15칸 · 12칸 실패
3. -1 자동 계산

**먼저 예측** · 값 10개를 `reshape(4, 3)` 하면? — 답은 코드 실행 결과 아래에

````python
# reshape: 값 순서는 그대로, 모양만 바꿈
# 왜 · 데이터를 바꾸지 않고 모델이 원하는 모양으로 맞추기 위해
base = np.arange(10)                                           # 값 10개
two_by_five = base.reshape(2, 5)                               # 10 = 2 × 5
five_by_two = base.reshape(5, 2)                               # 10 = 5 × 2
print("2행 5열:\n", two_by_five)                               # 앞에서부터 차례로 채움
print("size:", base.size, two_by_five.size, five_by_two.size)  # 모두 10


# 원본 셀 22 · 23: 칸 수가 다르면 실패
# 왜 · 전체 칸 수가 같아야만 reshape가 된다는 조건을 에러로 확인하기 위해
for shape in [(5, 3), (4, 3)]:                                      # 15칸 · 12칸
    try:
        base.reshape(*shape)                                        # 10개를 다른 칸 수에 넣기
    except ValueError as error:                                     # 모양 계약 오류
        print(f"reshape{shape} → {type(error).__name__}: {error}")  # 오류 문장


# 원본 셀 26: -1은 나머지 한 축을 자동 계산
# 왜 · 한 축 크기를 몰라도 나머지에서 자동으로 계산하게 하기 위해
print("-1, 5 ->", base.reshape(-1, 5).shape, "| 5, -1 ->", base.reshape(5, -1).shape)  # 10 ÷ 5 = 2
````

**결과** · 예측 정답은 `ValueError` — 10 ≠ 12

- `size: 10 10 10` — 모양이 달라도 개수는 같음
- `reshape(5, 3) → ValueError: cannot reshape array of size 10 into shape (5,3)` — 15칸 · 원본 셀 22의 오류
- `reshape(4, 3) → ValueError: cannot reshape array of size 10 into shape (4,3)` — 12칸 · 원본 셀 23의 의도된 오류
- `-1, 5 -> (2, 5) | 5, -1 -> (5, 2)` — 10 ÷ 5 = 2

**실행 흐름**

**먼저 예측** · `gdp > 10000`의 결과는 숫자일까? — 답은 코드 실행 결과 아래에

````python
# 1단계: 조건을 실행해 True/False 배열 보기
# 왜 · 조건식이 값이 아니라 같은 길이의 True/False 배열을 만든다는 것을 보기 위해
over_10000 = gdp > 10000    # 칸마다 비교
print("조건:", over_10000)  # 원본과 같은 길이
print("조건 dtype:", over_10000.dtype, "| 같은 shape:", gdp.shape == over_10000.shape)  # bool


# 2 · 3단계: mask로 저장 → 대괄호에 넣기
# 왜 · 조건에 맞는 값만 반복문 없이 골라내기 위해
mask = gdp > 10000                                   # 선택표
print("선택된 연도:", years[mask])                   # 조건은 gdp, 꺼내는 건 연도
print("선택된 GDP :", gdp[mask])                     # 조건에 맞는 값
print("모두 조건 만족?", np.all(gdp[mask] > 10000))  # 검증
````

> 🎯 **8,000 이상 10,000 이하 같은 범위 조건은 어떻게 묶는가?**

💡 **쉽게 말하면**

- 두 조건 모두는 **칸마다 그리고(`&`)**, 하나라도는 **칸마다 또는(`|`)**.
- 조건마다 괄호를 침.

**실행 흐름**

1. & 범위

````python
# 두 조건 모두: & (조건마다 괄호)
# 왜 · and 대신 &를 쓰고 괄호가 필요한 이유를 확인하기 위해
btw = (gdp >= 8000) & (gdp <= 10000)        # 8,000 이상 10,000 이하
print("8,000 이상 10,000 이하:", gdp[btw])  # 원본 셀 47


# 둘 중 하나: |
outside = gdp[(gdp < 8000) | (gdp > 12000)]     # 양 끝
print("8,000 미만 또는 12,000 초과:", outside)  # 원본 셀 47


# 원본 셀 48의 gap[btw]는 gdp의 오타 → 같은 mask를 연도에 적용해 확인
# 왜 · 같은 mask를 짝이 맞는 다른 배열에도 쓸 수 있음을 보이기 위해
print("해당 연도:", years[btw])  # 같은 길이 배열에 같은 mask
````

### 4-3. `B` 짝수 고르기와 빈 결과 · `numbers % 2 == 0` · `size` <sub>원본 미니 실습 (짝수) · 셀 54</sub>

💡 **쉽게 말하면**

````python
# 원본 Part 7: 기본 통계 (원본의 gdp.__dir__() 목록 출력은 생략)
# 왜 · 반복문 없이 배열 메서드로 합계 · 평균 등을 바로 구하기 위해
print("합계:", gdp.sum())                            # 10년 합
print("평균:", gdp.mean())                           # 합 ÷ 10
print("최솟값:", gdp.min(), "| 최댓값:", gdp.max())  # 가장 작은 · 큰 값
print("최댓값 연도:", years[gdp.argmax()])           # 위치로 연도 꺼내기
````

**먼저 예측** · `scores_inplace.sort()`의 반환값은? — 답은 코드 실행 결과 아래에

````python
# np.sort: 정렬된 새 배열 → 원본 유지
# 왜 · 원본 순서를 지킨 채 정렬된 결과만 받기 위해
scores = np.array([78, 95, 84, 98, 88])                       # 점수 5개
print("정렬 결과:", np.sort(scores), "| 원본 유지:", scores)  # 원본 그대로


# .sort(): 원본 자체를 정렬 → 반환값 None
# 왜 · 제자리 정렬은 원본을 바꾸고 None을 돌려준다는 함정을 보이기 위해
scores_inplace = scores.copy()                                     # 원본 대신 사본으로
returned = scores_inplace.sort()                                   # 제자리 정렬
print("sort 반환값:", returned, "| 변경된 배열:", scores_inplace)  # None · 정렬됨


# 내림차순: 정렬 후 뒤집기
print("내림차순:", np.sort(scores)[::-1])  # 전용 옵션 없음
````

### 6-4. `A` axis 방향 · `axis=0` · `axis=1` <sub>원본 셀 66 · 67</sub>

- `axis=0`은 **세로로 내려가며** 열마다 하나.
- `axis=1`은 **옆으로 가며** 행마다 하나.
- 암기 대신 **결과 개수**로 확인.

**실행 흐름**

````python
# axis=0: 행 방향으로 내려가며 → 열마다 결과 하나
# 왜 · axis가 '어느 축을 따라 계산해 없애는가'라는 것을 결과 모양으로 확인하기 위해
axis_demo = np.array([[8, 12],                               # 3행 2열
                      [7, 1],
                      [9, 5]])
print("axis=0 정렬(열마다):\n", np.sort(axis_demo, axis=0))  # 열 안에서 정렬
print("axis=0 합계(열마다):", axis_demo.sum(axis=0))         # 결과 2개


# axis=1: 열 방향으로 옆으로 → 행마다 결과 하나
# 왜 · axis=0과 결과 모양이 어떻게 다른지 비교하기 위해
print("axis=1 정렬(행마다):\n", np.sort(axis_demo, axis=1))  # 행 안에서 정렬
print("axis=1 합계(행마다):", axis_demo.sum(axis=1))         # 결과 3개
````

### 7-1. `A` broadcasting · `matrix + bonus` <sub>원본 Preview · 셀 72 ~ 74</sub>

> 🎯 **`(2, 3)`에 `(3,)`은 더해지고 `(2,)`는 왜 안 되는가?**

💡 **쉽게 말하면**

- 작은 배열을 **복사기로 늘려** 큰 배열에 맞춤.
- 오른쪽 끝부터 비교해 **같거나 1**인 축만 늘릴 수 있음.

**실행 흐름**

1. (2, 3) + (3,) 성공
2. (2, 3) + (2,) 실패

**먼저 예측** · `(2, 3) + (2,)`는 성공할까? — 답은 코드 실행 결과 아래에

````python
# (2, 3) + (3,): 오른쪽 끝 3 = 3 → 모든 행에 같은 bonus
# 왜 · 반복문 없이 모든 행에 같은 값을 더하는 브로드캐스팅 규칙을 확인하기 위해
matrix = np.array([[10, 20, 30],                                     # 2행 3열
                   [40, 50, 60]])
bonus = np.array([1, 2, 3])                                          # 열마다 하나
print("matrix shape:", matrix.shape, "| bonus shape:", bonus.shape)  # (2, 3) · (3,)
print("결과:\n", matrix + bonus)                                     # 행마다 같은 bonus


# (2, 3) + (2,): 오른쪽 끝 3 ≠ 2 → 실패
# 왜 · 오른쪽 끝 크기가 다르면 실패한다는 조건을 에러로 확인하기 위해
wrong_bonus = np.array([1, 2])  # 길이 2
try:
    matrix + wrong_bonus        # 맞출 수 없는 모양
except ValueError as error:     # broadcast 실패
    # 원본 셀 73
    print("예상한 오류 종류:", type(error).__name__, "| wrong_bonus shape:", wrong_bonus.shape)
````

**결과** · 예측 정답은 실패 — 끝자리 3 ≠ 2

- `matrix shape: (2, 3) | bonus shape: (3,)` — 끝자리 3 = 3
- `[41 52 63]]` — 둘째 행도 같은 bonus [1, 2, 3]을 더함
- `예상한 오류 종류: ValueError | wrong_bonus shape: (2,)` — 끝자리 3 ≠ 2

### 7-2. `A` 내적과 행렬 곱 · `np.dot()` <sub>원본 셀 75 ~ 77</sub>

> 🎯 **`*`와 `np.dot()`은 결과가 어떻게 다르고, 행렬 곱의 shape는?**

💡 **쉽게 말하면**

- `*`는 **같은 자리끼리 곱한 결과 여러 개**.
- `dot`은 그 곱을 **모두 더한 하나**.
- 행렬 곱은 이 내적을 칸마다 반복.

<img src="images/nl_matmul.png" alt="(2, 3) · (3, 2) → (2, 2), 첫 칸 = 58" width="747">

**실행 흐름**

1. 원소별 곱 vs 내적
2. (2, 3) · (3, 2)
3. 첫 칸 손계산

**먼저 예측** · `np.dot([1, 2, 3], [4, 5, 6])`은? — 답은 코드 실행 결과 아래에

````python
# 벡터 내적: 같은 위치끼리 곱한 뒤 모두 더함
# 왜 · 신경망 계산의 기본인 내적을 손계산과 함수 결과로 맞춰 보기 위해
vector_a = np.array([1, 2, 3])                                                         # 벡터 a
vector_b = np.array([4, 5, 6])                                                         # 벡터 b
print("원소별 곱:", vector_a * vector_b)                                               # 3개
print("곱의 합:", (vector_a * vector_b).sum(), "| 내적:", np.dot(vector_a, vector_b))  # 1개


# 행렬 곱: (2, 3) · (3, 2) → (2, 2)
# 왜 · 안쪽 크기가 같아야 곱해지고 결과는 바깥 크기가 된다는 규칙을 확인하기 위해
A = np.array([[1, 2, 3],                                                          # 2행 3열
              [4, 5, 6]])
B = np.array([[7, 8],                                                             # 3행 2열
              [9, 10],
              [11, 12]])
result = np.dot(A, B)                                                             # A @ B 와 같음
print("A shape:", A.shape, "| B shape:", B.shape, "| 결과 shape:", result.shape)  # 안쪽 3 = 3
print("행렬 내적:\n", result)                                                     # 칸마다 내적
print("첫 칸 확인:", 1 * 7 + 2 * 9 + 3 * 11)                                      # 손계산 = 58
````

**결과** · 예측 정답은 `32` — 4 + 10 + 18

- `원소별 곱: [ 4 10 18]` — 1 × 4 · 2 × 5 · 3 × 6
- `곱의 합: 32 | 내적: 32` — 4 + 10 + 18 = 32
- `A shape: (2, 3) | B shape: (3, 2) | 결과 shape: (2, 2)` — 안쪽 3 = 3 · 바깥 2, 2
- `첫 칸 확인: 58` — 1 × 7 + 2 × 9 + 3 × 11 = 58

**먼저 예측** · 결과는 몇 개이고 첫 상품(노트)의 매출은? — 답은 코드 실행 결과 아래에

````python
# Prediction 전 확인: 네 배열의 shape가 같은가
# 왜 · 계산하기 전에 짝이 맞는지 확인해 잘못된 결과를 막기 위해
final_products = np.array(["노트", "펜", "가방", "텀블러"])  # 상품 4개
final_prices = np.array([3000, 1500, 50000, 20000])          # 가격
final_quantities = np.array([120, 200, 15, 30])              # 판매량
print("상품 수:", final_products.size)                       # 4
# 계산 조건
print("shape 일치:", final_products.shape == final_prices.shape == final_quantities.shape)
````

**실행 흐름**

1. 상품별 매출 = 가격 × 수량
2. 전체 매출 합계
3. 40만 이상 상품만 고르기
4. 매출 높은 순서로 정렬해 출력

````python
# Build: 상품별 매출 · 전체 매출 · 40만 이상 · 높은 순서
# 왜 · 앞에서 배운 연산 · 조건 · 정렬을 한 흐름으로 이어 쓰기 위해
final_revenue = final_prices * final_quantities              # 같은 위치끼리 곱
final_total = final_revenue.sum()                            # 전체 합
high_mask = final_revenue >= 400000                          # 40만 이상 조건
revenue_order = np.argsort(final_revenue)[::-1]              # 매출 큰 위치부터
print("상품별 매출:", final_revenue)                         # 4개
print("전체 매출:", final_total)                             # 하나
print("40만 이상 상품:", final_products[high_mask])          # 조건 → 이름
print("매출 내림차순 상품:", final_products[revenue_order])  # 순서 → 이름
````

**실행 흐름**

1. 상품별 매출 · 합계를 원본 정답과 `assert`로 비교
2. 가방 매출 비중 계산 · 출력

````python
# Evidence: 원본 셀 83의 검증식 그대로
# 왜 · 결과가 원본 정답과 같은지 assert로 자동 확인하기 위해
assert final_revenue.tolist() == [360000, 300000, 750000, 600000]  # 상품별 매출
assert final_total == 2010000                                      # 전체 매출
assert final_products[high_mask].tolist() == ["가방", "텀블러"]    # 40만 이상
assert final_products[revenue_order].tolist() == ["가방", "텀블러", "노트", "펜"]  # 내림차순
print("모든 최종 검증을 통과했습니다.")                            # assert가 모두 통과하면 출력
print("가방 매출 비중:", round(final_revenue[2] / final_total * 100, 1), "%")  # Interpretation 근거
````

**결과** · 예측 정답은 4개 · 360000 (3000 × 120 = 360000)

- `shape 일치: True` — 네 배열 모두 `(4,)`

- `모든 최종 검증을 통과했습니다.` — 원본 검증식 4개 통과
- `가방 매출 비중: 37.3 %` — 750000 ÷ 2010000 × 100 = 37.3

**Q1. Python 리스트 대신 NumPy 배열을 쓰는 이유는?**
> - 배열은 같은 자료형의 숫자를 모아 반복문 없이 원소별 계산과 통계를 한 줄로 처리합니다.
> - 리스트의 곱하기는 이어 붙이기지만 배열은 모든 원소에 곱하기가 적용되고, Pandas와 PyTorch 텐서도 같은 구조라 데이터 처리의 기본입니다.

**Q2. reshape가 실패하는 이유와 -1의 의미는?**
> - reshape는 값의 개수를 바꾸지 않으므로 새 모양의 칸 수가 원래 size와 같아야 합니다.
> - -1은 나머지 축의 길이를 NumPy가 계산하게 하는 자리이며 한 번만 쓸 수 있습니다.

**Q3. Boolean mask로 데이터를 고르는 과정을 설명해 주세요.**
> - 조건식을 실행하면 원본과 같은 길이의 True/False 배열이 나오고, 이를 대괄호에 넣으면 True 자리의 값만 남습니다.
> - 조건 두 개는 괄호와 &, |로 묶고, 같은 mask를 길이가 같은 다른 배열에도 적용할 수 있습니다.

**Q4. broadcasting과 행렬 곱의 shape 규칙은?**
> - broadcasting은 두 shape를 오른쪽 끝부터 비교해 각 축이 같거나 1이면 늘려서 계산합니다.
> - 행렬 곱은 앞 행렬의 열 수와 뒤 행렬의 행 수가 같아야 하고, 결과는 바깥 두 숫자의 모양입니다.

**실행 흐름**

1. 1-1 리스트 vs 배열 연산
2. 1-2 `ndim` · `shape` · `size`
3. 1-3 `dtype` 통일 · `astype()`

````python
# ---------- 1-1 ----------
# 원본 Part 0: 리스트에 2를 곱하면?
# 왜 · 리스트 곱하기가 반복이라는 점을 먼저 보여 배열이 필요한 이유를 만들기 위해
prices_list = [30000, 25000, 40000]         # 상품 가격 3개 (리스트)
print("리스트 * 2 결과:", prices_list * 2)  # 곱하기가 아니라 반복
print("결과 길이:", len(prices_list * 2))   # 3개 → 6개


# 같은 값을 배열로 만들어 비교
# 왜 · 같은 연산이 배열에서는 원소마다 계산된다는 차이를 나란히 보기 위해
numbers_list = [1, 2, 3]                                # 리스트
numbers_array = np.array(numbers_list)                  # 리스트 → 배열
print("리스트 + 리스트:", numbers_list + numbers_list)  # 이어 붙이기
print("배열 + 배열:", numbers_array + numbers_array)    # 같은 위치끼리 더하기
print("타입:", type(numbers_list).__name__, "→", type(numbers_array).__name__)  # list → ndarray


# ---------- 1-2 ----------
# 1차원 · 2차원 배열의 축 개수 · 모양 · 개수
# 왜 · ndim · shape · size가 각각 무엇을 세는지 구분하기 위해
one_dim = np.array([10, 20, 30, 40])                               # 값 4개 한 줄
two_dim = np.array([[10, 20, 30, 40],                              # 3행 4열 표
                    [50, 60, 70, 80],
                    [90, 100, 110, 120]])
for label, array in [("one_dim", one_dim), ("two_dim", two_dim)]:  # (이름, 배열) 짝마다
    print(f"{label}: ndim {array.ndim} | shape {array.shape} | size {array.size}")  # 세 속성 한 줄


# 원본 셀 15 · 16: zeros의 괄호 안 숫자 개수 = ndim
# 왜 · shape의 숫자 개수가 곧 축 개수라는 규칙을 여러 모양으로 확인하기 위해
for shape in [(5, 4, 3, 2, 1), (3, 4, 5, 6), (4, 5, 6), (5, 6), 6]:  # 원본이 차례로 만든 모양
    arr = np.zeros(shape)                                            # 0으로 채운 배열
    print(f"np.zeros({shape}) → ndim {arr.ndim}, size {arr.size}")   # 축 개수 · 값 개수


# ---------- 1-3 ----------
# 섞인 값은 하나의 dtype으로 통일
# 왜 · 배열은 한 가지 자료형만 담는다는 규칙 때문에 값이 바뀌는 경우를 보기 위해
ints = np.array([1, 2, 3])                    # 정수만
mixed_number = np.array([1, 2, 3.5])          # 실수 하나
mixed_text = np.array([1, 2, "3"])            # 문자열 하나
for arr in [ints, mixed_number, mixed_text]:  # 세 배열마다
    print(arr, "->", arr.dtype)               # 값 · 자료형


# 원본 셀 10: Python의 float() 변환
# 왜 · 배열의 astype과 비교할 기준으로 파이썬 형변환을 먼저 보기 위해
a = 10                                                           # 정수
print(type(a).__name__, "→", type(float(a)).__name__, float(a))  # int → float


# 원본 셀 11: astype은 새 배열 · 소수점은 버림
# 왜 · 정수 변환이 반올림이 아니라 버림이라는 함정을 확인하기 위해
heights_float = np.array([171.9, 165.2, 180.8])        # 키 (실수)
heights_int = heights_float.astype(int)                # 정수로 바꾼 새 배열
print("변환 전:", heights_float, heights_float.dtype)  # 원래 값
print("변환 후:", heights_int, heights_int.dtype)      # 171.9 → 171
print("원본 유지:", heights_float)                     # 원본은 그대로
````

**실행 흐름**

1. 2-1 `arange()`
2. 2-2 `zeros()` · `ones()`
3. 2-3 `reshape()`와 실패 조건

````python
# ---------- 2-1 ----------
# arange(start, stop, step): 시작 포함 · 끝 제외
# 왜 · 끝 값이 빠지는 규칙을 여러 인자 조합으로 확인하기 위해
print("np.arange(5)        ->", np.arange(5))         # 0 ~ 4
print("np.arange(2, 8)     ->", np.arange(2, 8))      # 2 ~ 7
print("np.arange(2, 11, 3) ->", np.arange(2, 11, 3))  # 2, 5, 8 (11 제외)


# 원본 미니 실습 1: 4부터 16까지 3씩 → stop은 17
# 왜 · 끝 값을 넣으려면 stop을 하나 더 크게 줘야 한다는 것을 확인하기 위해
step_array = np.arange(4, 17, 3)                          # 16을 넣으려면 stop 17
print("간격 배열:", step_array, "| 끝:", step_array[-1])  # 16 포함 확인


# ---------- 2-2 ----------
# zeros · ones: shape는 괄호로 묶어 하나로 전달
# 왜 · 모양을 튜플 하나로 넘겨야 한다는 실수 포인트를 확인하기 위해
zero_table = np.zeros((2, 3), dtype=int)            # 정수 0 표
one_table = np.ones((2, 3))                         # dtype 생략 → 실수
print("0으로 채운 2행 3열:\n", zero_table)          # 정수 0
print("dtype:", zero_table.dtype, one_table.dtype)  # int64 · float64


# 원본 미니 실습 2: 3행 2열 '정수' 0 배열
# 왜 · 기본 dtype이 실수라서 정수가 필요하면 dtype을 지정해야 함을 보이기 위해
zero_answer = np.zeros((3, 2), dtype=int)  # 정수를 원하면 dtype=int
print("0 배열:\n", zero_answer)            # 3행 2열
print("dtype:", zero_answer.dtype)         # int64


# ---------- 2-3 ----------
# reshape: 값 순서는 그대로, 모양만 바꿈
# 왜 · 데이터를 바꾸지 않고 모델이 원하는 모양으로 맞추기 위해
base = np.arange(10)                                           # 값 10개
two_by_five = base.reshape(2, 5)                               # 10 = 2 × 5
five_by_two = base.reshape(5, 2)                               # 10 = 5 × 2
print("2행 5열:\n", two_by_five)                               # 앞에서부터 차례로 채움
print("size:", base.size, two_by_five.size, five_by_two.size)  # 모두 10


# 원본 셀 22 · 23: 칸 수가 다르면 실패
# 왜 · 전체 칸 수가 같아야만 reshape가 된다는 조건을 에러로 확인하기 위해
for shape in [(5, 3), (4, 3)]:                                      # 15칸 · 12칸
    try:
        base.reshape(*shape)                                        # 10개를 다른 칸 수에 넣기
    except ValueError as error:                                     # 모양 계약 오류
        print(f"reshape{shape} → {type(error).__name__}: {error}")  # 오류 문장


# 원본 셀 26: -1은 나머지 한 축을 자동 계산
# 왜 · 한 축 크기를 몰라도 나머지에서 자동으로 계산하게 하기 위해
print("-1, 5 ->", base.reshape(-1, 5).shape, "| 5, -1 ->", base.reshape(5, -1).shape)  # 10 ÷ 5 = 2
````

**실행 흐름**

1. 3-1 1차원 인덱싱 · `copy()`
2. 3-2 슬라이싱
3. 3-3 2차원 `[행, 열]`

````python
# ---------- 3-1 ----------
# 원본 Part 4: 연도와 GDP (같은 위치끼리 짝)
# 왜 · 같은 위치끼리 짝인 두 배열로 인덱싱을 연습할 데이터를 만들기 위해
years = np.arange(2010, 2020)                                          # 2010 ~ 2019
gdp = np.array([6610, 7637, 8127, 8885, 10385,                         # 연도마다 GDP
                12565, 14403, 12398, 8282, 10672])
print("같은 길이인가?", len(years) == len(gdp))                        # 짝지어 쓰기 전 확인
print("첫 번째:", gdp[0], "| 세 번째:", gdp[2], "| 마지막:", gdp[-1])  # 위치 0 · 2 · -1
print(f"2012년 GDP: {gdp[2]} (연도 {years[2]})")                       # 2012년 = 위치 2


# 원본 셀 31: 원본을 지키며 고치기
# 왜 · copy()로 원본 GDP는 그대로 두고 사본만 고치기 위해
gdp_edit = gdp.copy()                                         # 독립 복사본
gdp_edit[0] = 7000                                            # 복사본만 수정
print("원본 첫 값:", gdp[0], "| 수정본 첫 값:", gdp_edit[0])  # 원본 그대로


# ---------- 3-2 ----------
# slicing: start 포함 · stop 제외 · step
# 왜 · 구간을 자를 때 끝 위치가 빠지는 규칙을 확인하기 위해
print("gdp[2:6]   ->", gdp[2:6])    # 위치 2 ~ 5
print("gdp[:3]    ->", gdp[:3])     # 처음 3개
print("gdp[7:]    ->", gdp[7:])     # 위치 7부터 끝까지
print("gdp[1:8:3] ->", gdp[1:8:3])  # 위치 1 · 4 · 7


# 원본 미니 실습: 사람이 세는 7 ~ 9번째 → 위치 6 ~ 8 → [6:9]
# 왜 · 사람이 세는 번호와 0부터 세는 위치의 차이를 직접 맞춰 보기 위해
bitcoin = np.array([970, 1180, 1072, 1348, 2286, 2481,  # 비트코인 가격 12개
                    2875, 4703, 4339, 6468, 10234, 14156])
print("7~9번째 값:", bitcoin[6:9])                      # stop은 9


# ---------- 3-3 ----------
# 원본 Part 5: 지역 × 분기 매출표 (행 = 지역 · 열 = 분기)
# 왜 · 2차원 배열의 행 · 열 의미를 실제 표처럼 정해 두기 위해
regions = np.array(["서울", "부산", "대전"])   # 행 이름
quarters = np.array(["Q1", "Q2", "Q3", "Q4"])  # 열 이름
sales = np.array([[10, 12, 14, 16],            # 서울
                  [20, 18, 21, 25],            # 부산
                  [11, 13, 15, 17]])           # 대전
print("shape:", sales.shape)                   # (지역 3, 분기 4)
print("부산 Q3 매출:", sales[1, 2])            # 행 1 · 열 2


# 직사각형 · 열 전체 · 행 전체
# 왜 · [행, 열] 슬라이싱으로 원하는 부분 표를 꺼내는 방법을 비교하기 위해
print("부산·대전의 Q2~Q4:\n", sales[1:3, 1:4])          # 행 1 ~ 2 · 열 1 ~ 3
print("모든 지역의 Q2~Q3 shape:", sales[:, 1:3].shape)  # (3, 2)
print("서울의 모든 분기:", sales[0, :])                 # 행 0 전체


# 원본 미니 실습: 서울 · 부산 × Q2 · Q3
# 왜 · 행 · 열 구간을 함께 지정해 작은 표를 꺼내 보기 위해
selected_sales = sales[0:2, 1:3]                                # 행 0 ~ 1 · 열 1 ~ 2
print(selected_sales.tolist(), "shape:", selected_sales.shape)  # (2, 2)
````

**실행 흐름**

1. 4-1 조건 → True/False 배열
2. 4-2 `&` · `|` 두 조건
3. 4-3 짝수 고르기 · 빈 결과

````python
# ---------- 4-1 ----------
# 1단계: 조건을 실행해 True/False 배열 보기
# 왜 · 조건식이 값이 아니라 같은 길이의 True/False 배열을 만든다는 것을 보기 위해
over_10000 = gdp > 10000    # 칸마다 비교
print("조건:", over_10000)  # 원본과 같은 길이
print("조건 dtype:", over_10000.dtype, "| 같은 shape:", gdp.shape == over_10000.shape)  # bool


# 2 · 3단계: mask로 저장 → 대괄호에 넣기
# 왜 · 조건에 맞는 값만 반복문 없이 골라내기 위해
mask = gdp > 10000                                   # 선택표
print("선택된 연도:", years[mask])                   # 조건은 gdp, 꺼내는 건 연도
print("선택된 GDP :", gdp[mask])                     # 조건에 맞는 값
print("모두 조건 만족?", np.all(gdp[mask] > 10000))  # 검증


# ---------- 4-2 ----------
# 두 조건 모두: & (조건마다 괄호)
# 왜 · and 대신 &를 쓰고 괄호가 필요한 이유를 확인하기 위해
btw = (gdp >= 8000) & (gdp <= 10000)        # 8,000 이상 10,000 이하
print("8,000 이상 10,000 이하:", gdp[btw])  # 원본 셀 47


# 둘 중 하나: |
outside = gdp[(gdp < 8000) | (gdp > 12000)]     # 양 끝
print("8,000 미만 또는 12,000 초과:", outside)  # 원본 셀 47


# 원본 셀 48의 gap[btw]는 gdp의 오타 → 같은 mask를 연도에 적용해 확인
# 왜 · 같은 mask를 짝이 맞는 다른 배열에도 쓸 수 있음을 보이기 위해
print("해당 연도:", years[btw])  # 같은 길이 배열에 같은 mask


# ---------- 4-3 ----------
# 원본 미니 실습: 1 ~ 100에서 짝수만
# 왜 · 나머지 연산으로 만든 조건으로 값을 골라내는 연습을 하기 위해
numbers = np.arange(1, 101)                                   # stop 101 → 100까지
evens = numbers[numbers % 2 == 0]                             # 2로 나눈 나머지가 0
print("앞 10개 짝수:", evens[:10])                            # 2 ~ 20
print("마지막 짝수:", evens[-1], "| 짝수 개수:", evens.size)  # 100 · 50


# 원본 셀 54: 조건에 맞는 값이 없으면
# 왜 · 결과가 비어도 에러가 아니라 빈 배열이라는 것을 확인하기 위해
no_match = gdp[gdp > 20000]                                                     # 2만 넘는 해 없음
print("결과:", no_match, "| shape:", no_match.shape, "| size:", no_match.size)  # 빈 배열
````

**실행 흐름**

1. 5-1 기본 통계
2. 5-2 배열 × 숫자
3. 5-3 같은 shape 배열끼리 계산

````python
# ---------- 5-1 ----------
# 원본 Part 7: 기본 통계 (원본의 gdp.__dir__() 목록 출력은 생략)
# 왜 · 반복문 없이 배열 메서드로 합계 · 평균 등을 바로 구하기 위해
print("합계:", gdp.sum())                            # 10년 합
print("평균:", gdp.mean())                           # 합 ÷ 10
print("최솟값:", gdp.min(), "| 최댓값:", gdp.max())  # 가장 작은 · 큰 값
print("최댓값 연도:", years[gdp.argmax()])           # 위치로 연도 꺼내기


# ---------- 5-2 ----------
# 배열 × 숫자: 모든 원소에 같은 계산
# 왜 · 단위 변환처럼 모든 값에 같은 계산을 한 줄로 하기 위해
gdp_won = gdp * 1200                                                # 달러 → 원 (환율 1200 가정)
print("변환 결과 앞 3개:", gdp_won[:3], "| 원본 앞 3개:", gdp[:3])  # 원본은 그대로


# 원본 Part 0의 질문에 대한 답: 같은 shape 배열끼리
# 왜 · Part 0의 질문(가격 × 수량)을 배열 연산으로 해결하기 위해
products = np.array(["A", "B", "C", "D", "E", "F", "G"])              # 상품 7개
prices = np.array([30000, 25000, 40000, 60000, 95000, 35000, 50000])  # 가격
quantities = np.array([10, 13, 7, 20, 6, 19, 9])                      # 판매량
item_revenue = prices * quantities                                    # 같은 위치끼리 곱
print("상품별 매출:", item_revenue)                                   # 7개
print("전체 매출:", item_revenue.sum())                               # 합계 하나


# ---------- 5-3 ----------
# 원본 셀 60: 같은 shape의 두 배열로 차이 · 비율
# 왜 · 같은 위치끼리 빼고 나누어 두 나라를 연도별로 비교하기 위해
gdp_korea = np.array([12257, 11561, 13165, 14673, 16496, 19403])  # 한국
gdp_us = np.array([36335, 37133, 38023, 39496, 41713, 44115])     # 미국
gap = gdp_us - gdp_korea                                          # 같은 위치끼리 빼기
ratio = gdp_us / gdp_korea                                        # 같은 위치끼리 나누기
print("차이:", gap)                                               # 6개
print("비율(소수 둘째 자리):", np.round(ratio, 2))                # 반올림
print("shape 일치:", gdp_korea.shape == gdp_us.shape)             # 계산 조건
````

**실행 흐름**

1. 6-1 fancy indexing
2. 6-2 `np.sort()` vs `.sort()`
3. 6-3 `argsort()` · 상위 3명
4. 6-4 `axis`

````python
# ---------- 6-1 ----------
# fancy indexing: 떨어진 위치 여러 개를 한 번에
# 왜 · 연속하지 않은 위치를 위치 목록 하나로 꺼내기 위해
chosen_indices = np.array([0, 2, 4])        # 위치 목록
print("선택 연도:", years[chosen_indices])  # 2010 · 2012 · 2014
print("선택 GDP:", gdp[chosen_indices])     # 같은 위치의 GDP
print("순서 바꾸기:", gdp[[4, 0, 2]])       # 적은 순서대로 나옴


# ---------- 6-2 ----------
# np.sort: 정렬된 새 배열 → 원본 유지
# 왜 · 원본 순서를 지킨 채 정렬된 결과만 받기 위해
scores = np.array([78, 95, 84, 98, 88])                       # 점수 5개
print("정렬 결과:", np.sort(scores), "| 원본 유지:", scores)  # 원본 그대로


# .sort(): 원본 자체를 정렬 → 반환값 None
# 왜 · 제자리 정렬은 원본을 바꾸고 None을 돌려준다는 함정을 보이기 위해
scores_inplace = scores.copy()                                     # 원본 대신 사본으로
returned = scores_inplace.sort()                                   # 제자리 정렬
print("sort 반환값:", returned, "| 변경된 배열:", scores_inplace)  # None · 정렬됨


# 내림차순: 정렬 후 뒤집기
print("내림차순:", np.sort(scores)[::-1])  # 전용 옵션 없음


# ---------- 6-3 ----------
# argsort: 정렬된 값이 아니라 원래 위치의 순서
# 왜 · 점수 순서대로 이름까지 함께 정렬하기 위해
names = np.array(["John", "Mike", "Sarah", "Kate", "Samuel"])  # scores와 같은 순서
order = np.argsort(scores)                                     # 작은 점수의 위치부터
print("오름차순 위치:", order)                                 # 위치 목록
print("오름차순 이름:", names[order])                          # 같은 위치를 이름에도


# 원본 미니 실습: 상위 3명 (뒤집고 앞 3개)
# 왜 · argsort 결과를 뒤집어 상위 순위를 꺼내는 방법을 익히기 위해
top3_order = np.argsort(scores)[::-1][:3]                             # 큰 점수의 위치 3개
print("상위 3명:", names[top3_order], "| 점수:", scores[top3_order])  # 이름과 점수 짝 유지


# ---------- 6-4 ----------
# axis=0: 행 방향으로 내려가며 → 열마다 결과 하나
# 왜 · axis가 '어느 축을 따라 계산해 없애는가'라는 것을 결과 모양으로 확인하기 위해
axis_demo = np.array([[8, 12],                               # 3행 2열
                      [7, 1],
                      [9, 5]])
print("axis=0 정렬(열마다):\n", np.sort(axis_demo, axis=0))  # 열 안에서 정렬
print("axis=0 합계(열마다):", axis_demo.sum(axis=0))         # 결과 2개


# axis=1: 열 방향으로 옆으로 → 행마다 결과 하나
# 왜 · axis=0과 결과 모양이 어떻게 다른지 비교하기 위해
print("axis=1 정렬(행마다):\n", np.sort(axis_demo, axis=1))  # 행 안에서 정렬
print("axis=1 합계(행마다):", axis_demo.sum(axis=1))         # 결과 3개
````

**실행 흐름**

1. 7-1 브로드캐스팅 성공 · 실패
2. 7-2 내적 · 행렬 곱
3. 7-3 전치 · 3차원 배열

````python
# ---------- 7-1 ----------
# (2, 3) + (3,): 오른쪽 끝 3 = 3 → 모든 행에 같은 bonus
# 왜 · 반복문 없이 모든 행에 같은 값을 더하는 브로드캐스팅 규칙을 확인하기 위해
matrix = np.array([[10, 20, 30],                                     # 2행 3열
                   [40, 50, 60]])
bonus = np.array([1, 2, 3])                                          # 열마다 하나
print("matrix shape:", matrix.shape, "| bonus shape:", bonus.shape)  # (2, 3) · (3,)
print("결과:\n", matrix + bonus)                                     # 행마다 같은 bonus


# (2, 3) + (2,): 오른쪽 끝 3 ≠ 2 → 실패
# 왜 · 오른쪽 끝 크기가 다르면 실패한다는 조건을 에러로 확인하기 위해
wrong_bonus = np.array([1, 2])  # 길이 2
try:
    matrix + wrong_bonus        # 맞출 수 없는 모양
except ValueError as error:     # broadcast 실패
    # 원본 셀 73
    print("예상한 오류 종류:", type(error).__name__, "| wrong_bonus shape:", wrong_bonus.shape)


# ---------- 7-2 ----------
# 벡터 내적: 같은 위치끼리 곱한 뒤 모두 더함
# 왜 · 신경망 계산의 기본인 내적을 손계산과 함수 결과로 맞춰 보기 위해
vector_a = np.array([1, 2, 3])                                                         # 벡터 a
vector_b = np.array([4, 5, 6])                                                         # 벡터 b
print("원소별 곱:", vector_a * vector_b)                                               # 3개
print("곱의 합:", (vector_a * vector_b).sum(), "| 내적:", np.dot(vector_a, vector_b))  # 1개


# 행렬 곱: (2, 3) · (3, 2) → (2, 2)
# 왜 · 안쪽 크기가 같아야 곱해지고 결과는 바깥 크기가 된다는 규칙을 확인하기 위해
A = np.array([[1, 2, 3],                                                          # 2행 3열
              [4, 5, 6]])
B = np.array([[7, 8],                                                             # 3행 2열
              [9, 10],
              [11, 12]])
result = np.dot(A, B)                                                             # A @ B 와 같음
print("A shape:", A.shape, "| B shape:", B.shape, "| 결과 shape:", result.shape)  # 안쪽 3 = 3
print("행렬 내적:\n", result)                                                     # 칸마다 내적
print("첫 칸 확인:", 1 * 7 + 2 * 9 + 3 * 11)                                      # 손계산 = 58


# ---------- 7-3 ----------
# 전치: 행과 열을 맞바꿈
# 왜 · 행렬 곱의 모양을 맞출 때 쓰는 전치를 확인하기 위해
print("A.T:\n", A.T)                            # 3행 2열
print("shape 변화:", A.shape, "->", A.T.shape)  # (2, 3) → (3, 2)


# 3차원 배열: 2행 2열 표가 2장
# 왜 · 이미지 · 배치처럼 축이 셋인 데이터를 표 여러 장으로 이해하기 위해
cube = np.arange(8).reshape(2, 2, 2)                                     # 값 8개
print(cube)                                                              # 표 두 장
print("ndim:", cube.ndim, "| shape:", cube.shape, "| size:", cube.size)  # 3 · (2, 2, 2) · 8
````
